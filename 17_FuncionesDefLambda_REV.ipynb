## ¿Cuándo usar `def` y cuándo `lambda`?

- `def` es útil cuando la función necesita varias líneas, lógica más compleja o va a reutilizarse.
- `lambda` es ideal para expresiones sencillas, sobre todo en `filter`, `map`, `sorted` y funciones de una sola línea.

La idea central es esta: si la lógica cabe en una sola expresión, `lambda` suele ser más breve; si la lógica requiere varias instrucciones, una función con `def` es más legible.

````python
# Ejemplo típico de funciones con def en el notebook 09
def es_par(x):
    return not (x % 2)

def longitud_hasta5(x):
    return 0 if len(x) > 5 else 1

def proximo_a_cinco(x):
    return abs(5 - x)

numeros = [2, 4, 3, 5, 6, 7, 9, 1, 8]
palabras = ['bicicleta', 'casa', 'boligrafo', 'cena', 'bocadillo', 'placa']

print('Pares:', list(filter(es_par, numeros)))
print('Longitud <= 5:', list(filter(longitud_hasta5, palabras)))
print('Números más cercanos a 5:', sorted(numeros, key=proximo_a_cinco))
````

````python
# La misma lógica con lambda
numeros = [2, 4, 3, 5, 6, 7, 9, 1, 8]
palabras = ['bicicleta', 'casa', 'boligrafo', 'cena', 'bocadillo', 'placa']

print('Pares:', list(filter(lambda x: not (x % 2), numeros)))
print('Longitud <= 5:', list(filter(lambda x: 0 if len(x) > 5 else 1, palabras)))
print('Números más cercanos a 5:', sorted(numeros, key=lambda x: abs(5 - x)))
````

## Ejemplo de refactorización en `filter`, `sorted` y `map`

En `09_FilterSortedMap_REV.ipynb` aparecen varias definiciones que se usan como criterio para filtrar o ordenar.

- `longitud_hasta5` decide si una palabra debe quedarse o no.
- `proximo_a_cinco` devuelve la distancia a 5 para ordenar por cercanía.

En ambos casos, la función es muy pequeña y se puede sustituir por una `lambda` directa.

````python
# Funciones con def de otros notebooks
def doble(x):
    return x * 2

def crear_multiplicador(n):
    return lambda x: x * n

def sumar(a, b):
    return a + b

print('Doble de 6:', doble(6))
print('Por 3:', crear_multiplicador(3)(5))
print('Suma:', sumar(4, 7))
````

````python
# Versión refactorizada con lambda
doble = lambda x: x * 2
crear_multiplicador = lambda n: lambda x: x * n
sumar = lambda a, b: a + b

print('Doble de 6:', doble(6))
print('Por 3:', crear_multiplicador(3)(5))
print('Suma:', sumar(4, 7))

lista = [1, 2, 3, 4, 5]
print('Map:', list(map(lambda x: x * x, lista)))
print('Filter:', list(filter(lambda x: x % 2 == 0, lista)))
print('Sorted:', sorted(lista, key=lambda x: -x))
````

## Regla práctica para decidir

1. Si la función se usa solo una vez y la lógica es muy corta, usa `lambda`.
2. Si la función necesita varias líneas o la lógica es más compleja, usa `def`.
3. `lambda` funciona muy bien con `filter`, `map` y `sorted`, porque estos reciben una función como criterio.
4. `def` mejora la legibilidad cuando la función representa una operación con nombre claro y reutilizable.

### Resumen

Ejemplo:

```python
# Con def
def longitud_hasta5(x):
    return 0 if len(x) > 5 else 1

# Con lambda
lista = list(filter(lambda x: 0 if len(x) > 5 else 1, palabras))
```

La versión con `lambda` es más compacta, pero la versión con `def` suele ser más clara si la lógica va a volver a usarse o a expandirse en el futuro.

# Patrones didácticos: qué hace, por qué, ejemplo y lambda

Muchas funciones con `def` en este curso siguen una estructura repetida:

- reciben un valor
- aplican una condición o cálculo
- devuelven un resultado
- se usan después dentro de `filter`, `sorted` o `map`

El patrón más frecuente es este:

```python
def nombre(valor):
    return expresion
```

Y cuando la expresión es corta, normalmente se puede reescribir como una `lambda`.


````python
# Función 1: filtrar números pares

def es_par(x):
    return not (x % 2)

# Qué hace:
# Devuelve True si el número es par y False si no lo es.
# Por qué se usa:
# Porque filter necesita una condición booleana.
# Ejemplo:
numeros = [1, 2, 3, 4, 5, 6]
print(list(filter(es_par, numeros)))
# Refactorización con lambda:
print(list(filter(lambda x: not (x % 2), numeros)))
````

````python
# Función 2: filtrar palabras por longitud

def longitud_hasta5(x):
    return 0 if len(x) > 5 else 1

# Qué hace:
# Devuelve 0 cuando la palabra tiene más de 5 letras y 1 en caso contrario.
# Por qué se usa:
# En filter, cualquier valor distinto de 0 se interpreta como True, por lo que este ejemplo
# sirve para descartar palabras largas.
# Ejemplo:
palabras = ['bicicleta', 'casa', 'boligrafo', 'cena', 'bocadillo', 'placa']
print(list(filter(longitud_hasta5, palabras)))
# Refactorización con lambda:
print(list(filter(lambda x: 0 if len(x) > 5 else 1, palabras)))
````

````python
# Función 3: ordenar por proximidad a 5

def proximo_a_cinco(x):
    return abs(5 - x)

# Qué hace:
# Calcula la distancia del número a 5.
# Por qué se usa:
# sorted recibe una clave para decidir el orden; aquí esa clave es la distancia.
# Ejemplo:
numeros = [7, 2, 9, 1, 5, 10, 4, 3, 8, 6]
print(sorted(numeros, key=proximo_a_cinco))
# Refactorización con lambda:
print(sorted(numeros, key=lambda x: abs(5 - x)))
````

````python
# Función 4: doblar un valor

def doble(x):
    return x * 2

# Qué hace:
# Multiplica el número por dos.
# Por qué se usa:
# Es una transformación muy clara y muy típica con map.
# Ejemplo:
lista = [1, 2, 3, 4, 5]
print(list(map(doble, lista)))
# Refactorización con lambda:
print(list(map(lambda x: x * 2, lista)))
````

````python
# Función 5: crear un multiplicador

def crear_multiplicador(n):
    return lambda x: x * n

# Qué hace:
# Genera otra función que multiplica un valor por n.
# Por qué se usa:
# Es útil cuando necesitas una función adaptada a un número concreto.
# Ejemplo:
por_tres = crear_multiplicador(3)
print(por_tres(5))
# Refactorización con lambda:
por_tres_lambda = lambda x: x * 3
print(por_tres_lambda(5))
````

````python
# Función 6: sumar dos números

def sumar(a, b):
    return a + b

# Qué hace:
# Suma dos valores.
# Por qué se usa:
# Es la base de una operación muy habitual y se puede reutilizar en muchos contextos.
# Ejemplo:
print(sumar(4, 7))
# Refactorización con lambda:
print((lambda a, b: a + b)(4, 7))
````

### Ejemplo 9: restar dos números

```python
# Con def
def resta(a, b):
    return a - b

# Con lambda
resta_lambda = lambda a, b: a - b
```

Estos ejemplos muestran que, aunque `lambda` es muy compacto, la versión con `def` suele ser más legible cuando la función tiene un nombre claro y se va a reutilizar.


````python
# Función 7: describir una persona

def describir(nombre, edad):
    return f"{nombre} tiene {edad} años"

# Qué hace:
# Construye una frase con el nombre y la edad.
# Por qué se usa:
# Es una función pequeña, reutilizable y muy clara cuando queremos montar texto combinado.
# Ejemplo:
print(describir('Ana', 20))
# Refactorización con lambda:
print((lambda nombre, edad: f"{nombre} tiene {edad} años")('Ana', 20))

# Función 8: calcular un producto con parámetros por defecto

def producto(a=0, b=0):
    return a * b

# Qué hace:
# Multiplica dos valores y permite valores por defecto si no se pasan argumentos.
# Por qué se usa:
# Es útil para hacer una operación general y evitar errores cuando faltan argumentos.
# Ejemplo:
print(producto(7, 4))
print(producto())
# Refactorización con lambda:
print((lambda a=0, b=0: a * b)(7, 4))
print((lambda a=0, b=0: a * b)())

# Función 9: restar dos números

def resta(a, b):
    return a - b

# Qué hace:
# Calcula la diferencia entre dos valores.
# Por qué se usa:
# Es útil como ejemplo clásico de una función de cálculo breve pero que gana claridad con nombre explícito.
# Ejemplo:
print(resta(10, 3))
# Refactorización con lambda:
print((lambda a, b: a - b)(10, 3))
````

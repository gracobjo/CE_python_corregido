# Manual Completo del Módulo `math` en Python

El módulo `math` es parte de la biblioteca estándar de Python y proporciona acceso a las funciones matemáticas definidas por el estándar C. Es ideal para trabajar con operaciones matemáticas complejas que van más allá de los operadores básicos (`+`, `-`, `*`, `/`).

## 1. Importación y Constantes Matemáticas

Para usar este módulo, primero debemos importarlo. El módulo incluye varias constantes fundamentales de uso común.

```python
import math

# Principales constantes numéricas
print(f"Pi (π): {math.pi}")
print(f"Número de Euler (e): {math.e}")
print(f"Tau (τ = 2π): {math.tau}")
print(f"Infinito positivo: {math.inf}")
print(f"Not a Number (NaN): {math.nan}")
```

## 2. Funciones de Redondeo y Representación

Estas funciones permiten aproximar números decimales a enteros o analizar sus componentes básicos.

```python
numero = 4.7
negativo = -4.7

# Redondeo hacia arriba (techo)
print(f"Ceil de {numero}: {math.ceil(numero)}")

# Redondeo hacia abajo (suelo)
print(f"Floor de {numero}: {math.floor(numero)}")

# Truncar (elimina la parte decimal)
print(f"Truncar {negativo}: {math.trunc(negativo)}")

# Valor absoluto
print(f"Valor absoluto de {negativo}: {math.fabs(negativo)}")
```

## 3. Teoría de Números y Operaciones Numéricas

Funciones útiles para divisibilidad, factoriales y cálculos aritméticos precisos.

```python
# Máximo Común Divisor (MCD)
print(f"MCD de 24 y 36: {math.gcd(24, 36)}")

# Mínimo Común Múltiplo (MCM)
print(f"MCM de 12 y 18: {math.lcm(12, 18)}")

# Factorial
print(f"Factorial de 5 (5!): {math.factorial(5)}")
```

## 4. Potencias y Logaritmos

Operaciones avanzadas con exponentes, raíces cuadradas y diferentes bases logarítmicas.

```python
# Raíz cuadrada y raíz cúbica
print(f"Raíz cuadrada de 25: {math.sqrt(25)}")
print(f"Raíz cúbica de 27: {math.cbrt(27)}")

# Potencias
print(f"2 elevado a la 8: {math.pow(2, 8)}")

# Logaritmos
print(f"Logaritmo natural de e: {math.log(math.e)}")
print(f"Logaritmo en base 10 de 100: {math.log10(100)}")
print(f"Logaritmo en base 2 de 64: {math.log2(64)}")
```

## 5. Trigonometría y Conversión de Ángulos

**Nota importante**: El módulo `math` procesa todas sus funciones trigonométricas utilizando **radianes**, no grados.

```python
grados = 90

# Conversiones
radianes = math.radians(grados)
print(f"{grados}° en radianes es: {radianes}")
print(f"{radianes} rad en grados es: {math.degrees(radianes)}")

# Funciones trigonométricas básicas
print(f"Seno de 90°: {math.sin(radianes)}")
print(f"Coseno de 90°: {math.cos(radianes)} (Cercano a 0 debido a precisión float)")
print(f"Tangente de 45°: {math.tan(math.radians(45))}")
```

## 6. Geometría

Cálculo de distancias.

```python
# Hipotenusa (Teorema de Pitágoras)
cateto_a = 3
cateto_b = 4
print(f"Hipotenusa de un triángulo de lados 3 y 4: {math.hypot(cateto_a, cateto_b)}")

# Distancia euclidiana entre dos puntos (p1 y p2)
punto1 = (1, 2)
punto2 = (4, 6)
print(f"Distancia entre {punto1} y {punto2}: {math.dist(punto1, punto2)}")

```

# Funciones integradas

### abs
Devuelve el valor absoluto

````python
x = -15
print(abs(x))  # Salida: 15

y = 42
print(abs(y))  # Salida: 42

# Usado en comprensión de listas
numeros = [-10, 5, -23, 8]
absolutos = [abs(n) for n in numeros]
print(absolutos)  # Salida: [10, 5, 23, 8]
````

___
### divmod
La función divmod(a, b) en Python toma dos números, realiza una división entera y devuelve una tupla con dos elementos: el cociente y el resto (módulo)

````python
# Dividir 10 entre 3
cociente, resto = divmod(10, 3)

print("Cociente:", cociente)  # Salida: 3
print("Resto:", resto)        # Salida: 1
````

### max
Devuelve el elemento con el valor más grande dentro de un iterable (como una lista o tupla) o el mayor entre varios argumentos individuales.

````python
numeros = [10, 40, 5, 20]
resultado = max(numeros)

print(resultado)  # Salida: 40

resultado = max(10, 20, 5)

print(resultado)  # Salida: 20

letras = ('a', 'b', 'z')
print(max(letras))  # Salida: z
````

### min
Sirve para encontrar el valor más pequeño dentro de un iterable (como una lista o tupla) o entre varios argumentos independientes.

````python
# Lista de números
numeros = [8, 4, 2, 7, 3]

# Obtener el valor mínimo
menor = min(numeros)
print(menor)

### Ejemplo con valores individuales
# Pasar varios valores directamente
resultado = min(5, 10, 3)

print(resultado)  # Salida: 3
````

### pow
Calcula la potencia de un número usando dos o tres argumentos (base, exp y opcionalmente mod)

No confundir con `Math.pow`

````python
# Calcular 2 elevado a la 3 (2^3)
resultado = pow(2, 3)
print(resultado)  # Salida: 8

# Calcular (2^3) % 5 -> 8 % 5
resultado_modular = pow(2, 3, 5)
print(resultado_modular)  # Salida: 3
````

### round
Redondea un número al entero más cercano o a una cantidad específica de decimales.

Sintaxix:
```Python
round(numero, n_digitos)
```
* numero: El valor que deseas redondear.
* n_digitos (opcional): El número de decimales que quieres conservar. Si se omite, redondea al entero más cercano.

````python
# Redondear 4.6
print(round(4.6))  # Salida: 5

# Redondear 4.2
print(round(4.2))  # Salida: 4
pi = 3.14159265

# Redondear a 2 decimales
print(round(pi, 2))  # Salida: 3.14

# Redondear a 4 decimales
print(round(pi, 3))  # Salida: 3.1416
````

### sum
Calcula la suma de todos los elementos de un objeto iterable.

````python
# Definimos una lista de números
numeros = [1, 2, 3, 4, 5]

# Usamos la función sum() para sumarlos todos
total = sum(numeros)

print(total)  # Salida: 15

numeros = [1, 2, 3, 4, 5]

# Sumamos los elementos de la lista empezando a contar desde 10
total = sum(numeros, 10)

print(total)  # Salida: 25 (15 + 10)
````

### all
Devuelve True solo si todos los elementos de un iterable son verdaderos (o si el iterable está vacío); de lo contrario, devuelve False.

````python
# Todos los elementos son verdaderos
lista_1 = [True, True, True]
print(all(lista_1))  # Salida: True

# Hay un elemento falso (False)
lista_2 = [True, False, True]
print(all(lista_2))  # Salida: False

numeros = [4, 10, 25, 8]

# Verifica si todos los números son mayores que 0
todos_positivos = all(n > 0 for n in numeros)

print(todos_positivos)  # Salida: True
````

___
### any
Devuelve True si al menos un elemento de un iterable (como una lista, tupla o conjunto) es verdadero (True), y devuelve False si todos los elementos son falsos o si el iterable está vacío.

````python
# Lista de números
numeros = [3, 5, 8, 12, 1]

# Usamos una expresión generadora con any()
resultado = any(n > 10 for n in numeros)

print(resultado)  # Salida: True (porque 12 es mayor que 10)
````

### enumerate
Agrega un contador automático a un iterable (como una lista) mientras se recorre en un bucle.

````python
nombres = ['Luis', 'María', 'Carlos', 'Ana']

for indice, nombre in enumerate(nombres):
    print(f'Índice {indice}: {nombre}')

# Salida:
# Índice 0: Luis
# Índice 1: María
# Índice 2: Carlos
# Índice 3: Ana
````

___
### reversed
Devuelve un objeto iterador que recorre una secuencia (como una lista, tupla o texto) en orden inverso, sin modificar la original.

````python
numeros = [1, 2, 3, 4, 5]

# reversed() devuelve un iterador
iterador_invertido = reversed(numeros)

# Convertimos el iterador en una lista nueva
lista_invertida = list(iterador_invertido)

print(lista_invertida)  # Salida: [5, 4, 3, 2, 1]
print(numeros)          # La lista original no cambia: [1, 2, 3, 4, 5]
````

### slice
Un slice en Python es una forma de extraer una porción o subconjunto de una secuencia.

Sintaxis 
```Python
[inicio:fin:paso]
```
* **índice**: Es donde comienza el corte (incluido).
* **fin**: Es el límite del corte (no se incluye).
* **paso**: Define cuántos elementos saltar.

````python
# Lista original
numeros = [10, 20, 30, 40, 50, 60, 70]

# Aplicamos slicing: desde el índice 1 hasta el 4 (el índice 4 no se incluye)
porcion = numeros[1:4]
print(porcion)  # Salida: [20, 30, 40]

print(numeros[0:6:2])  # Salida: [10, 30, 50]

print(numeros[::-1])  # Salida: [70, 60, 50, 40, 30, 20, 10]
````

### zip
La Función zip() en Python combina elementos de dos o más iterables (como listas o tuplas) en pares o tuplas La función zip().

````python
# Definimos dos listas
nombres = ["Ana", "Carlos", "Elena"]
edades = [25, 30, 22]

# Usamos zip() para combinar ambas listas
resultado = zip(nombres, edades)

# Convertimos el objeto zip a una lista para ver su contenido
print(list(resultado))
# Salida: [('Ana', 25), ('Carlos', 30), ('Elena', 22)]
````

# callable
Un objeto callable (invocable) es cualquier objeto que se puede llamar utilizando paréntesis (), de la misma forma que llamas a una función.

El ejemplo más claro de un objeto callable personalizado es una clase que implementa el método especial `__call__`.

Además de las clases con `__call__`, en Python existen otros objetos que son nativamente invocables:
* Funciones de usuario: Creadas con def.
* Funciones lambda: Funciones anónimas de una sola línea.
* Métodos: Funciones asociadas a un objeto (ej. "texto".upper).
* Clases: Al "llamarlas", crean una nueva instancia (ej. int("10")).


````python
# Ejemplo básico con una función
def sumar(a,b):
    return a+b

print(callable(sumar))

print("=====")

#Ejemplo de clase que implementa __call__
class Multiplicador:
    def __init__(self, factor):
        self.factor = factor

    # Este método convierte a los objetos de esta clase en "callables"
    def __call__(self, numero):
        return numero * self.factor

# 1. Instanciamos el objeto de manera normal
duplicar = Multiplicador(2)

# 2. Comprobamos si es un objeto "callable"
print(callable(duplicar))  # Devuelve: True

# 3. Lo llamamos como si fuera una función utilizando paréntesis ()
resultado = duplicar(5)
print(resultado)  # Devuelve: 10

print(callable(Multiplicador))
````

### locals
Devuelve un diccionario con la tabla de símbolos locales actual, es decir, todas las variables, funciones y clases definidas en el ámbito (scope) local actual.
Su comportamiento cambia dependiendo de si se ejecuta dentro de una función o en el espacio global.

````python
def mi_funcion(nombre, edad):
    # Definimos una variable local
    ciudad = "Madrid"
    
    # Llamamos a locals() para ver qué hay en el ámbito local
    diccionario_local = locals()
    
    print("Contenido de locals():")
    print(diccionario_local)
    
    # Podemos acceder a los valores usando las claves del diccionario
    print(f"\nHola, {diccionario_local['nombre']} de {diccionario_local['ciudad']}.")

# Ejecutamos la función
mi_funcion("Carlos", 28)
````

### exec
Se utiliza para ejecutar dinámicamente código de Python almacenado en una cadena de texto o en un objeto de código.

La principal diferencia es que eval() evalúa una única expresión y devuelve un resultado, mientras que exec() ejecuta bloques de código (sentencias, bucles, definiciones, etc.) y siempre devuelve None

````python
# Definimos código en una cadena de texto
codigo = """
x = 10
y = 5
print(f'La suma es: {x + y}')
"""

# Ejecutamos el código usando exec()
exec(codigo)

x = 10
print("Antes:", x)

# Modificamos la variable 'x' mediante una cadena
exec("x += 5")

print("Después:", x)
````

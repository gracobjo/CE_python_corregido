* **Parámetros:** son variables que se definen en la función y que reciben un valor cuando se llama a la función.

````python
def describir(nombre, edad): # PARÁMETROS
    print("El nombre es: ", nombre)
    print("La edad es: ", edad)

describir("juan",30) # ARGUMENTOS
````

## Argumentos por posición y por nombre
* **Argumentos por posición:** se pasan los argumentos en el mismo orden que los parámetros de la función.
* **Argumentos por nombre:** se pasan los argumentos especificando el nombre del parámetro.

````python
# Argumentos por posición y por nombre
def resta(a, b):
    print("La resta es: ", a - b)

resta(7, 4)  # Argumentos por posición: a=7, b=4
resta(b=4, a=7)  # Argumentos por nombre: a=7, b=4
````

## Parámetros por defecto
Se pueden definir valores por defecto para los parámetros de una función. Si se llama a la función sin pasar un argumento para ese parámetro, se usará el valor por defecto.

````python
def producto(a=0, b=0):
    print(f"El producto de {a} por {b} es: ", a * b)

# Argumentos por posición
producto()
producto(9, 4)
producto(3)

# Argumentos por nombre
producto(b=9, a=4)
producto(b=7)
````

Una vez especificado un valor por defecto para un parámetro, todos los parámetros que le siguen también deben tener valores por defecto.

````python
# def multiplicacion(a=1, b=1, c): # ERROR --> c debe llevar parámetros por defecto o colocarse antes de los parámetros por defecto
#    return a * b * c

def multiplicacion_1(c, a=1, b=1): # Colocamos el parámetro obligatorio (no por defecto) antes de los parámetros por defecto
   return a * b * c

def multiplicacion_2(a=1, b=1, c=1): # El parámetro c ahora es por defecto
   return a * b * c
    
print(f"La multiplicación es: {multiplicacion_1(3,4,2)}")
print(f"La multiplicación es: {multiplicacion_2(3,4,2)}")
````

# Paso por valor y paso por referencia
En Python, los tipos de datos inmutables (como int, float, str, tuple) se pasan por valor, mientras que los tipos de datos mutables (como list, dict, set) se pasan por referencia.
* **Paso por referencia:** Se pasa una referencia al valor, por lo que los cambios realizados dentro de la función afectan al valor original.
* **Paso por valor:** Se pasa una copia del valor, por lo que los cambios realizados dentro de la función no afectan al valor original. 

___
## Paso por valor

````python
def mi_funcion(entrada):
    entrada = 0
    print(f"Dentro: {entrada} con id {id(entrada)}")
    print(f"Dentro: {x} con id {id(x)}")

x = 10
print(f"Antes: {x}")
mi_funcion(x)
print(f"Después: {x}") #10
````

___
## Paso por referencia

````python
def mi_funcion(entrada):
    entrada[1] = 25
    print(f"Dentro: {entrada}") # [10, 25, 30]

x = [10, 20, 30]
print(f"Antes: {x}")
mi_funcion(x)
print(f"Después: {x}")
````

## Uso de id para comprobar si se pasa por valor o por referencia

````python
# Paso por valor

def mi_funcion(entrada):
    entrada = 0
    print(id(entrada)) # Será un número de identificación diferente al de x, ya que se ha creado un nuevo objeto

x = 10
print(id(x)) # Dará un número de identificación único para el objeto x
mi_funcion(x)
````

````python
# Paso por referencia

def mi_funcion(entrada):
    entrada[1] = 25
    print(id(entrada)) # Será el mismo número de identificación que el de x, ya que se pasa una referencia al objeto original

x = [10, 20, 30]
print(id(x)) # Dará un número de identificación único para el objeto x
mi_funcion(x)
````

# **Ámbito de una variable**
Una variable sólo está disponible en la región en la que se ha creado. Esta región se denomina **ámbito**.

````python
def mi_funcion():
    velocidad = 1
    color = "blanco"
    print(color)


# print(velocidad) # ERROR puesto que velocidad está definida y su ámbito es de la función donde lo está
mi_funcion()
````

En el caso de funciones anidadas va al contrario.

* En este caso la variable var1 que se ha definido en la función externa, puede se accedida desde la función interna.

* En el caso del avariable var2 que se ha definico en la función interna, no puede ser accedida desde la función externa.

* Por otro lado, la función externa no puede ser llamada desde fuera de la externa.

````python
def funcion1():
    var1 = 1
    def funcion2():
        var2 = 2
        print(var1)
        print(var2)
    funcion2()
    # print(var2) # ERROR: Esta variable no está accesible desde la función externa

funcion1()
# funcion2() # ERROR: Tampoco es posible llamar a una función interna desde fuera de la externa
````

# Argumentos arbitrarios (numero de argumentos _posicionales_ variable)
## *args
* Se pueden definir funciones que reciban un número arbitrario de argumentos.
* Para ello se utiliza un asterisco (*) antes del nombre del parámetro.
* Los argumentos se pasan como una tupla.

````python
def suma_arbitraria(*args):
    resultado = 0
    for numero in args:
        resultado += numero
    print("La suma es: ", resultado)

suma_arbitraria(1, 2, 3)
suma_arbitraria(1, 2, 3, 4, 5)  # Llama a la función con 5 argumentos
# suma_arbitraria([1, 2, 3, 4, 5]) # ERROR: Se pasa un argumento que es una lista, no se puede iterar sobre ella misma como si fuera una tupla
# suma_arbitraria({1, 2, 3, 4, 5}) # ERROR: Se pasa un argumento que es un conjunto, no se puede iterar sobre el mismo como si fuera una tupla
````

````python
No confundir con lo siguiente donde pasamos una lista pero la función no está definida con argumentos arbitrarios
````

````python
def suma_arbitraria(args):
    resultado = 0
    for numero in args:
        resultado += numero
    print("La suma es: ", resultado)

# suma_arbitraria(1, 2, 3, 4, 5) # ERROR: Se pasan 5 argumentos, pero la función solo recibe 1 parámetro
suma_arbitraria([1, 2, 3, 4, 5]) 
````

Como tal tupla que es _args_ podemos imprimirla tal cual o sacar los parámetros de forma individual.

````python
def ver_datos(*args):
    print(type(args))
    print(args)
    print(f"Primero: {args[0]}")
    print(f"Segundo: {args[1]}")
    print(f"Tercero: {args[2]}")

ver_datos("A","B","C")
# ver_datos("A","B") # ERROR: Dentro de la función se llama un tercer parámetro
ver_datos("A","B","C","D") # No da error puesto que no importe que el cuarto parámetro no se use
````

Se pueden combinar parámetros parámetros arbitrarios (*args) con parámetros **normales**.

La condición está en que los normales se definen antes.

````python
def combinar_parametros(nombre,*datos):
    print(nombre)
    print(datos)
    for i in datos:
        print(i)

combinar_parametros("juan",30,"Valladolid")
print("=======")
combinar_parametros("pepe",30)
````

# Argumentos arbitrarios con nombre (numero de argumentos _con nombre_ variable)
## **kwargs
* Se pueden definir funciones que reciban un número arbitrario de argumentos por nombre.
* Para ello se utilizan dos asteriscos (**) antes del nombre del parámetro.
* Los argumentos se pasan como un diccionario.

````python
def imprimir_datos(**datos):
    print(f"{datos['nombre']}")
    print(f"{datos['edad']}")
    print(f"{datos['ciudad']}")
    print(f"{datos['telefono']}")
    
imprimir_datos(nombre="juan",edad=30,ciudad="Valladolid",telefono=454354354)
#imprimir_datos(nombre="juan",edad=30,ciudad="Valladolid") # ERROR: No conoce el nombre teléfono
imprimir_datos(nombre="juan",edad=30,ciudad="Valladolid",telefono=454354354,direccion="Calle Mirabel")
````

Podemos imprimirlo de forma genérica todo a la vez y sacar el tipo de dato

````python
def imprimir_datos(**datos):
    print(type(datos))
    print(f"{datos}")

imprimir_datos(nombre="juan",edad=30,ciudad="Valladolid",telefono=454354354)
````

````python
Se pueden combinar parámetros arbitrarios (*kwargs) con nombre con parámetros **normales**.

La condición está en que los normales se definen antes.
````

````python
def imprimir_datos(nombre,**datos):
    print(nombre)
    for i in datos: # Al ser datos un diccionario accedemos a las claves con i y a los valores con datos[i]
        print(datos[i])

imprimir_datos("juan",edad=30,ciudad="Valladolid",telefono=454354354)
````

# Combinación de argumentos arbitrarios posicionales (*args) y por nombre (**kwargs)

Se pueden combiar argumentos regulares, con arbitrarios posicionales y arbitrarior por nombre.

Orden:
1. Regulares
2. Arbitrarios posicionales
3. Arbitrarios por nombre

````python
def imprimir_datos(nombre,*args,**kwargs):
    print("Regulares")
    print(nombre)

    print("\nArbitrarios posicionales")
    for i in args:
        print(i)

    print("\nArbitrarios por nombre")
    for i in kwargs:
        print(kwargs[i])

imprimir_datos("juan",30,"Valladolid",telefono=454354354,direccion="Calle Mirabel")
````

La definición más genérica de una función sería

````python
def imprimir_datos(*args,**kwargs):
    for i in args:
        print(i)
    for i in kwargs:
        print(kwargs[i])

imprimir_datos("juan",30,"Valladolid",telefono=454354354,direccion="Calle Mirabel")
````

# Desempaquetar listas y diccionarios con * y **
Usamos el * para convertir una lista en parámetros

````python
def imprimir_parametros(p1,p2,p3):
    print(p1,p2,p3)

lista = [1,2,3]
imprimir_parametros(*lista) # Probar a quitar el * para que de error
````

Usamos el ** para convertir un diccionario en parámetros

````python
def imprimir_parametros(n1,n2,n3):
    print(n1,n2,n3)

diccionario = {"n1":"A","n2":"B","n3":"C"}
imprimir_parametros(**diccionario) 
````

# Funciones que devuelven funciones
Las funciones pueden devolver funciones, definidas dentro de las mismas o no

````python
def decrementar(n):
    return n -1
     
def operar(orden):
    def incrementar(n):
        return n + 1
    
    match orden:
        case "i":
            return incrementar
        case "d":
            return decrementar

var_funcion = operar("i")
print(var_funcion(6))

var_funcion = operar("d")
print(var_funcion(15))
````

# Pasar funciones como argumentos
La funciones pueden ser pasadas como argumentos

````python
def sumar(a,b):
    return a+b

def restar(a,b):
    return a-b

def imprimir_resultado(func,a,b):
    print(func(a,b))

imprimir_resultado(sumar,8,5)
imprimir_resultado(restar,8,5)
````

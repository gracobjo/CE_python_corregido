# Excepciones en Python

## 1. Introducción: ¿Qué es una Excepción?

En Python, los errores se dividen en dos tipos principales:
1. **Errores de sintaxis (`SyntaxError`):** El código está mal escrito y el intérprete no puede ejecutarlo.
2. **Excepciones:** Errores que ocurren durante la ejecución del programa, incluso si la sintaxis es totalmente correcta.

___
### Ejemplo de Excepción Común
Si intentamos dividir un número por cero, Python detendrá la ejecución y lanzará una excepción de tipo `ZeroDivisionError`.

````python
# Ejecuta esta celda para ver una excepción en acción
resultado = 10 / 0
````

## 2. La Estructura Fundamental: `try ... except`
Para evitar que un error rompa un programa, envolvemos el código propenso a fallar dentro de un bloque **try**. Si ocurre un error, el control pasa al bloque **except**.

````python
try:
    # Código que puede fallar
    numero = int(input("Introduce un número entero: "))
    print(f"El doble de tu número es: {numero * 2}")
except ValueError:
    # Código que se ejecuta SOLO si ocurre un ValueError
    print("¡Eso no es un número entero válido!")
````

## 3. Capturando Múltiples Excepciones
Un mismo bloque de código puede fallar por diferentes razones. Podemos encadenar múltiples bloques **except** o capturar varios errores en una sola línea usando una tupla.

````python
try:
    print("--- Intentando realizar operaciones ---")
    numerador = int(input("Dividendo: "))
    denominador = int(input("Divisor: "))
    resultado = numerador / denominador
    print(f"Resultado: {resultado}")

except ValueError:
    print("Error: Debes introducir únicamente números enteros.")

except ZeroDivisionError:
    print("Error: No se puede dividir entre cero.")

except (TypeError, NameError) as e:
    print(f"Ocurrió un error de tipo o nombre: {e}")
````

## 4. Inspeccionando el Objeto de la Excepción (`as`)
Podemos asignar el error capturado a una variable (comúnmente llamada **e**) utilizando la palabra clave **as**. Esto nos permite acceder al mensaje de error original del sistema.

````python
try:
    lista = [1, 2, 3]
    print(lista[10]) # Índice fuera de rango
except IndexError as e:
    print(f"Nombre del error: {type(e).__name__}")
    print(f"Mensaje oficial: {e}")
````

## 5. El Flujo Completo: `try`, `except`, `else` y `finally`
El sistema de control de excepciones de Python tiene dos bloques adicionales muy útiles:
* **else**: Se ejecuta solo si no ocurrió ninguna excepción en el bloque try.
* **finally**: Se ejecuta siempre, haya ocurrido un error o no. Es ideal para tareas de limpieza (como cerrar archivos o bases de datos).

````python
try:
    print("Abriendo archivo simulado...")
    archivo_abierto = True
    dato = int("10") # Cambiar esto por "hola" para ver cómo cambia el flujo
except ValueError:
    print("-> Bloque EXCEPT: Hubo un error de conversión.")
else:
    print("-> Bloque ELSE: ¡Todo salió bien! El dato es:", dato)
finally:
    print("-> Bloque FINALLY: Limpiando recursos y cerrando el archivo simulado.")
    archivo_abierto = False
````

## 6. Captura genérica de excepciones
Tenemos dos formas de capturar cualquier excepcion:
* **Usando la clase base `Exception` (Recomendado)**: Captura todos los errores estándar del programa.
    ```Python
    try:
        # Código que puede fallar
        resultado = 10 / 0
    except Exception as e:
        print(f"Ocurrió un error: {e}")
    ```
___

* **Usando un `except`: desnudo (Sin especificar tipo)**: Captura absolutamente todo, incluyendo interrupciones del sistema (KeyboardInterrupt). No se recomienda porque oculta errores imprevistos o de sintaxis que hay que corregir.
    ```Python
    try:
        # Código que puede fallar
        x = y
    except:
        print(f"Ocurrió un error desconocido")
    ```



## 7. Lanzar Excepciones Manualmente (raise)
Se puede forzar un programa a lanzar una excepción si no se cumplen ciertas condiciones utilizando la palabra clave **raise**.

````python
def registrar_edad(edad):
    if edad < 0:
        raise ValueError("La edad no puede ser un número negativo.")
    if edad < 18:
        print("Registro completado como menor de edad.")
    else:
        print("Registro completado como adulto.")

try:
    registrar_edad(-5)
except ValueError as error:
    print(f"Validación fallida: {error}")
````

## 8. Jerarquía y Tipos de Excepciones Comunes
Todas las excepciones en Python heredan de la clase base **BaseException**. Excepciones más comunes:
* `Exception`: Clase base para casi todas las excepciones que no provocan la salida del sistema. Evita capturar esta clase directamente (except Exception:) a menos que sea estrictamente necesario, ya que oculta errores imprevistos.
* `ValueError`: Un argumento tiene el tipo correcto pero un valor inapropiado.
* `TypeError`: Una operación se aplica a un objeto de tipo inapropiado.
* `IndexError`: Un índice de secuencia está fuera de rango.
* `KeyError`: Una clave de diccionario no se encuentra.
* `FileNotFoundError`: Intento de abrir un archivo que no existe.

## 8. Creación de Excepciones Personalizadas
Si estamos construyendo una librería o una aplicación compleja, podemos crear nuestros propios tipos de error heredando directamente de la clase estándar **Exception**.

### Pasos necesarios:
1. **Definir la clase**: Por convicción, el nombre de la clase debe terminar con la palabra `Error` o `Exception`.
2. **Lanzar la excepción (`raise`)**: Utilizar la palabra reservada `raise` seguida de una instancia de la clase.
3. **Capturar la excepción (`try/except`)**: Manejar el error de forma limpia en el programa usando los bloques habituales de control de errores.

````python
class SaldoInsuficienteError(Exception):
    """Excepción lanzada cuando un usuario intenta retirar más dinero del que tiene."""
    def __init__(self, saldo, cantidad_solicitada):
        self.saldo = saldo
        self.cantidad_solicitada = cantidad_solicitada
        super().__init__(f"Intento de retirar ${cantidad_solicitada} teniendo solo ${saldo}")

# Ejemplo de uso
def retirar_dinero(cuenta_saldo, monto):
    if monto > cuenta_saldo:
        raise SaldoInsuficienteError(cuenta_saldo, monto)
    return cuenta_saldo - monto

try:
    mi_saldo = 50
    retirar_dinero(mi_saldo, 120)
except SaldoInsuficienteError as e:
    print(f"Error Bancario: {e}")
````

## 9. Buenas Prácticas (Resumen)
1. **Concrección**: Capturar errores concretos (`except ValueError`) en lugar de usar bloques genéricos (`except:`).
2. **No ocultar errores**: No usar un except vacío que solo contenga un `pass` porque hará muy complicado encontrar bugs en el código.
3. **Mantener el try pequeño**: Colocar dentro del bloque `try` únicamente las líneas de código que realmente puedan fallar por una razón externa. El resto del código sin riesto debe ir en el bloque `else`.

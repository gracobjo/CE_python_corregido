# **Bucle While**
Permite iterar o repetir código mientras se cumpla una condición

___
## Sintaxis
```python
While condición:
    sentencia1
    sentencia2
    ...
    sentenciaN
```

````python
i = 1
while i < 5:
    print(i)
    i += 1
````

## Sentencia *break*
La sentencia **break** permite salirse del bucle

````python
i = 1
while i < 5:
    print(i)
    if i % 2 == 0:
        break
    i += 1
````

## Sentencia *continue*
La sentencia **continue** permite parar la ejecución de la iteración actual y seguir con la siguiente.

````python
i = 0
while i < 5:
    i += 1
    if i == 2:
        continue
    print(i)
````

## Bucles infinitos
Hay que tratar que no hagamos bucles infinitos asegurándonos que la condición en algún momento dejará de cumplirse

````python
i = 1
while i < 5:
    print(i)
    if i % 2 == 0:
        i += 1
    # Bucle infinito
````

## Sentencia *else*
La sentencia **else** después al final de un bucle **while** permite ejecutar código cuando la condición ya no es verdadera.

````python
i = 0
while i < 5:
    print(i)
    i += 1
else:
    print(f"El valor {i} ya no cumple la condición")
````

Aparentemente no sirve de nada puesto que si no se pone, lo que se ejecute fuera del while se ejecutará cuando la condición ya no se cumple, que es el propósito de este **else**. Sería una ejecución normal, es decir, cuando se termina el bucle se ejecuta el bloque **else**.

Al usarlo en combinación con **break** y al detener el bucle a la fuerza con este **break**, el bloque **else** no se ejecuta.

````python
numeros = [1, 3, 5, 7]

buscado = 5 #Probar también con un valor que no esté en la lista
i = 0

while i < len(numeros):
    if numeros[i] == buscado:
        print(f"¡Encontrado el número {buscado} en el índice {i}!")
        break #Probar a quitar el break
    i += 1
else:
    print("El número no está en la lista.")
````

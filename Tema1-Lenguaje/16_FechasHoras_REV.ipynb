### Comparación de fechas

````python
import datetime

fecha1 = datetime.date(2022,4,11)
fecha2 = datetime.date(2025,4,11)

if fecha1 < fecha2:
    print(f"{fecha1} es anterior a {fecha2}")
else:
    print(f"{fecha1} es posterior a {fecha2}")
````

___
### Diferencia entre fechas

````python
import datetime

fecha1 = datetime.date(2024,4,6)
fecha2 = datetime.date(2025,4,11)

fecha3 = fecha2 -fecha1
print(fecha3)
````

___
### Métodos principales
* **weekday()**: Devuelve un número entre 0 (lunes) y 6 (domingo)
* **replace()**: Reemplaza cualquiera de los valores (día, mes o años). Funciona como parámetros por nombre y de forma arbitraria.

````python
# WEEKDAY
import datetime

fecha = datetime.date(2022,4,11)
print(fecha.weekday())
print(datetime.date.today().weekday())
````

````python
# REPLACE
import datetime

fecha = datetime.date(2022,4,11)
print(fecha.replace(year=2018,day=9))
print(fecha.replace(month=6))
print(fecha)
````

### Métodos de formateo
* **strftime()**: Crea una cadena de sustitución personalizada

````python
# STRFTIME
import datetime

fecha = datetime.date.fromtimestamp(1790794772) # 86400 segundos en un día
fecha = datetime.date(2022,4,21)
print(fecha.strftime("El año es %Y, el mes es %m y el día del mes es %d"))
````

## timedelta
La clase timedelta se usa para representar intervalos de tiempo. Permite añadir o quitar semanas, días, horas, minutos y segundos respecto a un objeto fecha.

___
La propia diferencia entre objetos **date** o **datetime** nos genera un objeto **timedelta**

````python
import datetime
#t = datetime.datetime.now()
##res = t + datetime.timedelta(weeks=1)
#print(res)

fecha1 = datetime.date(2026,10,7)
fecha2 = datetime.date(2026,10,8)
print(fecha1,type(fecha1))
print(fecha2,type(fecha2))
print(fecha2-fecha1,type(fecha2-fecha1))

print("=====")

fecha1 = datetime.datetime(2026,10,7,15,30,0,0)
fecha2 = datetime.datetime(2026,10,8,15,35,0,0)
print(fecha1,type(fecha1))
print(fecha2,type(fecha2))
print(fecha2-fecha1,type(fecha2-fecha1))
````

___
Creación manual de un objeto **timedelta**

Sintaxis:
```Python

```

````python
import datetime

print(datetime.timedelta(days=2,hours=5))
print(datetime.timedelta(hours=5,minutes=2))
````

### `Atributos de timedelta`
#### Atributos de clase
* **min**: Valor mínimo (-999999999)
* **max**: Valor máximo (days=999999999, hours=23, minutes=59, seconds=59, microseconds=999999)
* **resolution**: La mínima diferencia que puede haber entre dos fecha, es decir, el mínimo valor que puede tomar un intervalo (1 microsegundo).
#### Atributos de instancia
* **days**: Días del intervalo actual.
* **seconds**: Segundos. Entre 0 y 86399. Se refiere al día actual, es decir, quitando *days*.
* **microseconds**: Microsegundos. Entre 0 y 999.999.

#### `Métodos de timedelta`
* **total_seconds()**: Devuelve todos los segundos del intervalo.

````python
import datetime

print(datetime.timedelta.min)
print(datetime.timedelta.max)
print(datetime.timedelta.resolution)

print("====")

intervalo = datetime.timedelta(days=3,hours=1,minutes=20,seconds=10,microseconds=2000)
print("Días",intervalo.days)
print("Segundos",intervalo.seconds)
print("Microsegundos",intervalo.microseconds)

print("Segundos totales",intervalo.total_seconds())
````

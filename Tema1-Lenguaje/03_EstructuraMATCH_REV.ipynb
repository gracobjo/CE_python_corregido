# **Estructura MATCH**
Se emplea en lugar de tener múltiples estructura **if..else**

___
## **Sintaxis**
```python
match expresión:
    case caso1:
        Bloque de código 1
    case caso2:
        Bloque de código 2
    ...
    case casoN:
        Bloque de código N
```

````python
puesto = 2
match puesto:
    case 1:
        print("Oro")
    case 2:
        print("Plata")
    case 3:
        print("Bronce")
````

## **Valor por defecto**
Si no se cumple ninguno de los casos se ejecuta el caso por defecto que se indica con guion bajo **_**

````python
mes = 27
match mes:
    case 1:
        print("Enero")
    case 2:
        print("Febrero")
    case 3:
        print("Marzo")
    case 4:
        print("Abril")
    case 5:
        print("Mayo")
    case 6:
        print("Junio")
    case 7:
        print("Julio")
    case 8:
        print("Agosto")
    case 9:
        print("Septiembre")
    case 10:
        print("Octubre")
    case 11:
        print("Noviembre")
    case 12:
        print("Diciembre")
    case _:
        print("Mes no válido")
````

## **Combinación de valores**
Es posible combinar varios casos sin repetir el bloque de código usando la tubería o barra vertical **|**.

````python
mes = 2
match mes:
    case 1|2|3:
        print("Invierno")
    case 4|5|6:
        print("Primavera")
    case 7|8|9:
        print("Verano")
    case 10|11|12:
        print("Otoño")
    case _:
        print("Mes no válido")
````

## **Estructura *if* como condición extra**
Es posible añadir a los casos una estructura **if** como condición extra

````python
mes = 9
dia = 2
match mes:
    case 1 | 2 | 3 if dia==6 or dia==7:
        print("Fin de semana de invierno")
    case 4 | 5 | 6 if dia==6 or dia==7:
        print("Fin de semana de primavera")
    case 7 | 8 | 9 if dia==6 or dia==7:
        print("Fin de semana de verano")
    case 10 | 11 | 12 if dia==6 or dia==7:
        print("Fin de semana de otoño")
    case _:
        print("Mes no válido o no es fin de semana")
````

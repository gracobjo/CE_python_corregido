# Opciones y eventos de una ventana Tkinter

Este cuaderno permite probar `iconify`, `deiconify`, `state`, `after`, `protocol` y `destroy` desde los botones de una ventana independiente.

**Uso:** ejecuta las celdas de código en orden. Tkinter abrirá una ventana de escritorio aparte; se necesita un entorno gráfico. La última celda mantiene activa la ventana. Puedes probar los botones mientras está abierta. Al cerrar la ventana o pulsar **Destruir ventana**, termina el bucle de eventos. Para empezar de nuevo, reinicia el kernel y vuelve a ejecutar las celdas.

## 1. Importar Tkinter

Tkinter forma parte de la biblioteca estándar de Python; no hace falta instalar paquetes adicionales.

```python
import tkinter as tk
```

## 2. Crear la ventana y los controles

Cada botón prueba una acción. `state()` sin argumento consulta el estado actual; con un argumento lo cambia. `zoomed` suele estar disponible en Windows. La etiqueta muestra los resultados sin cerrar la ventana.

```python
ventana = tk.Tk()
ventana.title("Demostración de varios métodos")
ventana.geometry("400x420")

estado = tk.Label(ventana, text="Elige una opción", font=("Arial", 12), wraplength=360)
estado.pack(pady=12)

tk.Button(
    ventana,
    text="Minimizar (iconify)",
    command=lambda: ventana.iconify(),
).pack(pady=4)

tk.Button(
    ventana,
    text="Restaurar (deiconify)",
    command=lambda: ventana.deiconify(),
).pack(pady=4)

tk.Button(
    ventana,
    text="Consultar estado",
    command=lambda: estado.config(text=f"Estado actual: {ventana.state()}"),
).pack(pady=4)

tk.Button(
    ventana,
    text="Estado normal",
    command=lambda: ventana.state("normal"),
).pack(pady=4)

tk.Button(
    ventana,
    text="Estado maximizado (zoomed)",
    command=lambda: ventana.state("zoomed"),
).pack(pady=4)

tk.Button(
    ventana,
    text="Cambiar etiqueta en 2 segundos (after)",
    command=lambda: ventana.after(
        2000, lambda: estado.config(text="La acción programada ya se ejecutó.")
    ),
).pack(pady=4)

tk.Button(
    ventana,
    text="Destruir ventana en 2 segundos",
    command=lambda: ventana.after(2000, ventana.destroy),
).pack(pady=4)

tk.Button(
    ventana,
    text="Cerrar ventana",
    command=lambda: ventana.destroy(),
    bg="lightcoral",
).pack(pady=4)
```

## 3. Formulario de inicio de sesión

El formulario se abre en una ventana secundaria (`Toplevel`). La contraseña se enmascara visualmente y el botón comprueba que los campos no estén vacíos. **Es solo una demostración de interfaz, no autentica usuarios**: no guarda ni imprime la contraseña. En una aplicación real, la autenticación debe validarse de forma segura en el servidor.

```python
formulario = tk.Toplevel(ventana)
formulario.title("Iniciar sesión")
formulario.geometry("340x230")

tk.Label(formulario, text="Usuario:").pack(anchor="w", padx=20, pady=(16, 4))
entrada_usuario = tk.Entry(formulario)
entrada_usuario.pack(fill="x", padx=20)

tk.Label(formulario, text="Contraseña:").pack(anchor="w", padx=20, pady=(12, 4))
entrada_contrasena = tk.Entry(formulario, show="*")
entrada_contrasena.pack(fill="x", padx=20)

resultado_login = tk.Label(formulario, text="", wraplength=300)
resultado_login.pack(pady=8)

def iniciar_sesion():
    usuario = entrada_usuario.get().strip()
    contrasena = entrada_contrasena.get()
    if not usuario or not contrasena:
        resultado_login.config(text="Completa el usuario y la contraseña.", fg="red")
        return
    resultado_login.config(
        text="Formulario completado. Aquí conectarías la autenticación real.",
        fg="green",
    )

tk.Button(
    formulario,
    text="Iniciar sesión",
    command=iniciar_sesion,
).pack(pady=4)

tk.Button(
    formulario,
    text="Cerrar ventana",
    command=lambda: formulario.destroy(),
).pack(pady=4)
```

## 4. Interceptar el botón de cerrar

`protocol("WM_DELETE_WINDOW", ...)` permite definir qué ocurre al pulsar la **X** de la ventana. En este ejemplo, muestra un aviso y la cierra después de un segundo. `after` programa la llamada sin bloquear la interfaz.

```python
def al_cerrar():
    estado.config(text="Cerrando en un segundo...")
    ventana.after(1000, ventana.destroy)


ventana.protocol("WM_DELETE_WINDOW", al_cerrar)
```

## 5. Iniciar el bucle de eventos

Ejecuta esta celda al final. La celda permanecerá activa mientras la ventana esté abierta. `destroy()` cierra la ventana y termina el bucle; por eso conviene probar esa acción al final.

```python
ventana.mainloop()
```

## Variables de control de Tkinter

Las variables de control conectan un valor de Python con uno o varios widgets. En este ejemplo se usan `StringVar` para el nombre, `IntVar` para la edad, `DoubleVar` para una escala y `BooleanVar` para una casilla de verificación. Ejecuta esta sección cuando hayas cerrado las ventanas de los ejemplos anteriores.

```python
ventana_variables = tk.Tk()
ventana_variables.title("Variables de control")
ventana_variables.geometry("420x390")

nombre_var = tk.StringVar(value="Ana")
edad_var = tk.IntVar(value=18)
nivel_var = tk.DoubleVar(value=5.0)
acepta_var = tk.BooleanVar(value=False)
datos_control = []

tk.Label(ventana_variables, text="Nombre (StringVar):").pack(anchor="w", padx=20, pady=(12, 2))
tk.Entry(ventana_variables, textvariable=nombre_var).pack(fill="x", padx=20)

tk.Label(ventana_variables, text="Edad (IntVar):").pack(anchor="w", padx=20, pady=(10, 2))
tk.Spinbox(
    ventana_variables,
    from_=0,
    to=120,
    textvariable=edad_var,
).pack(anchor="w", padx=20)

tk.Label(ventana_variables, text="Nivel (DoubleVar):").pack(anchor="w", padx=20, pady=(10, 2))
tk.Scale(
    ventana_variables,
    from_=0,
    to=10,
    orient=tk.HORIZONTAL,
    resolution=0.5,
    variable=nivel_var,
).pack(fill="x", padx=20)

tk.Checkbutton(
    ventana_variables,
    text="Acepto las condiciones (BooleanVar)",
    variable=acepta_var,
).pack(anchor="w", padx=20, pady=8)

resultado_variables = tk.Label(ventana_variables, text="Modifica los controles y consulta sus valores.")
resultado_variables.pack(pady=8)

def mostrar_valores():
    resultado_variables.config(
        text=(
            f"Nombre: {nombre_var.get()} | Edad: {edad_var.get()} | "
            f"Nivel: {nivel_var.get():.1f} | Acepta: {acepta_var.get()}"
        )
    )

def guardar_datos():
    datos_control.append({
        "nombre": nombre_var.get(),
        "edad": edad_var.get(),
        "nivel": nivel_var.get(),
        "acepta": acepta_var.get(),
    })
    resultado_variables.config(
        text=f"Registro {len(datos_control)} guardado. Puedes guardar más antes de cerrar."
    )

tk.Button(
    ventana_variables,
    text="Mostrar valores (.get())",
    command=mostrar_valores,
).pack(pady=4)

tk.Button(
    ventana_variables,
    text="Guardar datos para usarlos después",
    command=guardar_datos,
).pack(pady=4)

tk.Button(
    ventana_variables,
    text="Cambiar nombre con .set()",
    command=lambda: nombre_var.set("Luis"),
).pack(pady=4)

tk.Button(
    ventana_variables,
    text="Cerrar ventana",
    command=lambda: ventana_variables.destroy(),
).pack(pady=8)

ventana_variables.mainloop()
```

### Qué observar

- `textvariable` enlaza un `Entry` o `Spinbox` con una variable de control.
- `variable` enlaza la escala y la casilla con sus variables correspondientes.
- `.get()` lee el valor actual y `.set()` lo cambia; al pulsar **Cambiar nombre**, el campo se actualiza.
- `DoubleVar` devuelve un número decimal y `BooleanVar` devuelve `True` o `False`.

### Usar los valores después de cerrar la ventana

## Marcador gráfico de un partido de tenis

Este ejercicio adapta la clase `PartidoTenis` de `Tema2-POO/Ejercicios/Ejercicios07_PartidoTenis.ipynb` a una interfaz Tkinter. Cada botón corresponde a un jugador; al pulsarlo, se actualiza el marcador del set. **Siguiendo el ejercicio original**, cada pulsación suma 1 a los juegos del jugador (no es el tanteo reglamentario de puntos 15–30–40). También puedes suspender el partido por lluvia, guardar su resultado y consultar estadísticas. El historial se conserva en memoria mientras la aplicación siga abierta; al cerrar la ventana principal se pierde. Ejecuta las celdas de esta sección en orden. Usa **Cerrar ventana** para liberar el kernel.

### ¿Qué vamos a construir?

La imagen muestra el resultado final que obtendremos después de ejecutar todas las celdas. Las imágenes de cada paso son **maquetas ilustrativas**, no capturas automáticas de la ventana Tkinter.

### Paso 1: preparar la lógica del partido

Primero separamos las reglas del marcador de la interfaz. `PartidoTenis` guarda los nombres, el estado de suspensión y el historial. Al crear el partido, también crea el primer set. `anotar_en_set_actual()` delega el punto en ese set y devuelve `True` si lo pudo registrar o `False` si el partido está suspendido o el jugador no es válido. La clase interna `_Set` conserva los juegos de cada jugador.

Al ejecutar la celda siguiente no aparecerá una ventana: se define el modelo y sus reglas.

![Esquema del modelo del partido](imagenes_tenis/01_modelo.png)

```python
class PartidoTenis:
    def __init__(self, jugador1, jugador2):
        self.jugador1 = jugador1
        self.jugador2 = jugador2
        self.cancelado = False
        self.historial_sets = [self._Set(self, 1)]

    def anotar_en_set_actual(self, jugador_num):
        if not self.historial_sets:
            return False
        return self.historial_sets[-1].registrar_punto(jugador_num)

    def suspender_por_lluvia(self):
        self.cancelado = True

    def finalizar(self):
        self.cancelado = True

    class _Set:
        def __init__(self, partido, numero_set):
            self.partido = partido
            self.numero_set = numero_set
            self.juegos_j1 = 0
            self.juegos_j2 = 0

        def registrar_punto(self, jugador_num):
            if self.partido.cancelado:
                return False
            if jugador_num == 1:
                self.juegos_j1 += 1
            elif jugador_num == 2:
                self.juegos_j2 += 1
            else:
                return False
            return True
```

### Paso 2: crear el partido y la ventana

Instanciamos la clase con los dos nombres y guardamos una referencia al set actual. Luego creamos la ventana principal con `Tk()`, le damos un título y definimos su tamaño. Los nombres se pueden cambiar en la llamada a `PartidoTenis()`.

Después de ejecutar la celda, la ventana está creada, pero todavía vacía. Permanece abierta mientras ejecutas las siguientes celdas.

![Ventana vacía tras crearla](imagenes_tenis/02_ventana.png)

```python
partido_grafico = PartidoTenis("Carlos Alcaraz", "Jannik Sinner")
set_grafico = partido_grafico.historial_sets[-1]
historial_partidos = []

import tkinter as tk
ventana_tenis = tk.Tk()
ventana_tenis.title("Marcador de tenis")
ventana_tenis.geometry("520x500")
```

### Paso 3: construir el marcador visible

`Label` muestra títulos y nombres. El `Frame` agrupa las dos columnas del marcador y `grid()` las alinea en filas y columnas. `StringVar` conecta cada número con su etiqueta: cuando más adelante usemos `.set()`, Tkinter actualizará el texto en pantalla automáticamente. La variable `estado_tenis` sirve para mostrar mensajes.

Al ejecutar la celda, la ventana que ya estaba abierta mostrará los nombres y el tanteo inicial 0–0. Aún no se puede anotar porque todavía no hemos añadido botones.

![Marcador inicial sin botones](imagenes_tenis/03_marcador.png)

```python
tk.Label(
    ventana_tenis,
    text="Marcador - Set 1 (juegos)",
    font=("Arial", 16, "bold"),
).pack(pady=16)

marcador = tk.Frame(ventana_tenis)
marcador.pack(pady=8)

marcador_j1 = tk.StringVar(master=ventana_tenis, value="0")
marcador_j2 = tk.StringVar(master=ventana_tenis, value="0")

tk.Label(marcador, text=partido_grafico.jugador1, font=("Arial", 12)).grid(row=0, column=0, padx=20)
tk.Label(marcador, text=partido_grafico.jugador2, font=("Arial", 12)).grid(row=0, column=1, padx=20)
tk.Label(marcador, textvariable=marcador_j1, font=("Arial", 28, "bold")).grid(row=1, column=0, padx=20)
tk.Label(marcador, textvariable=marcador_j2, font=("Arial", 28, "bold")).grid(row=1, column=1, padx=20)

estado_tenis = tk.StringVar(master=ventana_tenis, value="Pulsa un botón para anotar un juego.")
tk.Label(ventana_tenis, textvariable=estado_tenis, wraplength=480).pack(pady=8)
```

### Paso 4: conectar la lógica con la pantalla

`anotar_punto_grafico()` se ejecutará al pulsar uno de los botones. Primero pide a `PartidoTenis` que anote el juego. Si la clase devuelve `False`, informa que el partido está suspendido. Si el punto se acepta, lee los contadores del set con `.juegos_j1` y `.juegos_j2` y actualiza las `StringVar` mediante `.set()`.

Esta celda define la función, pero no cambia lo que se ve todavía: la función se ejecutará cuando creemos botones que la llamen.

![El marcador aún no cambia al definir la función](imagenes_tenis/04_callback.png)

```python
def anotar_punto_grafico(jugador_num):
    if not partido_grafico.anotar_en_set_actual(jugador_num):
        estado_tenis.set("No se puede anotar: el partido está suspendido.")
        return

    marcador_j1.set(str(set_grafico.juegos_j1))
    marcador_j2.set(str(set_grafico.juegos_j2))
    nombre = partido_grafico.jugador1 if jugador_num == 1 else partido_grafico.jugador2
    estado_tenis.set(f"Juego anotado para {nombre}.")
```

### Paso 5: crear botones, cerrar el marcador y volver a jugar

Cada botón llama a la misma función con un número distinto: `lambda` permite pasar el jugador (`1` o `2`) cuando se pulse, en vez de ejecutar la función al crear el botón. **Finalizar partido** o **Suspender por lluvia** guarda una copia del resultado en `historial_partidos`. **Ver historial y estadísticas** muestra los marcadores guardados, cuántos partidos fueron finalizados o suspendidos y la media de juegos por partido (incluye los suspendidos). **Volver a jugar** inicia un marcador nuevo sin borrar el historial. Los datos solo viven en memoria y se pierden al cerrar la aplicación. El resumen del partido actual queda antes de estos botones y del botón para cerrar.

Al ejecutar esta celda aparecen los botones en la ventana. Ya puedes anotar juegos, suspender el partido o cerrarlo.

![Ventana con botones, resumen y cierre](imagenes_tenis/05_botones.png)

```python
boton_j1 = tk.Button(
    ventana_tenis,
    text=f"Anotar juego: {partido_grafico.jugador1}",
    command=lambda: anotar_punto_grafico(1),
    bg="lightblue",
)
boton_j1.pack(fill=tk.X, padx=50, pady=4)

boton_j2 = tk.Button(
    ventana_tenis,
    text=f"Anotar juego: {partido_grafico.jugador2}",
    command=lambda: anotar_punto_grafico(2),
    bg="lightgreen",
)
boton_j2.pack(fill=tk.X, padx=50, pady=4)

resultado_final = tk.StringVar(
    master=ventana_tenis,
    value="Resumen final: pendiente",
)

def cerrar_marcador(estado_final):
    if partido_grafico.cancelado:
        return

    if estado_final == "suspendido por lluvia":
        partido_grafico.suspender_por_lluvia()
    else:
        partido_grafico.finalizar()
    boton_j1.config(state=tk.DISABLED)
    boton_j2.config(state=tk.DISABLED)
    boton_suspender.config(state=tk.DISABLED)
    boton_finalizar.config(state=tk.DISABLED)
    boton_reiniciar.config(state=tk.NORMAL)

    juegos_j1 = set_grafico.juegos_j1
    juegos_j2 = set_grafico.juegos_j2
    marcador_final = (
        f"{partido_grafico.jugador1}: {juegos_j1}  -  "
        f"{juegos_j2}: {partido_grafico.jugador2}"
    )

    if juegos_j1 > juegos_j2:
        resultado = f"M\u00e1s juegos: {partido_grafico.jugador1}"
    elif juegos_j2 > juegos_j1:
        resultado = f"M\u00e1s juegos: {partido_grafico.jugador2}"
    else:
        resultado = "Marcador empatado; no hay un jugador con m\u00e1s juegos."

    resultado_final.set(
        f"Resumen ({estado_final})\n{marcador_final}\n{resultado}"
    )
    historial_partidos.append({
        "numero": len(historial_partidos) + 1,
        "jugador1": partido_grafico.jugador1,
        "jugador2": partido_grafico.jugador2,
        "juegos_j1": juegos_j1,
        "juegos_j2": juegos_j2,
        "estado": estado_final,
        "resultado": resultado,
    })
    estado_tenis.set("El marcador est\u00e1 cerrado.")

def suspender_partido_grafico():
    cerrar_marcador("suspendido por lluvia")

def finalizar_partido_grafico():
    cerrar_marcador("partido finalizado manualmente")

def volver_a_jugar():
    global partido_grafico, set_grafico
    partido_grafico = PartidoTenis(
        partido_grafico.jugador1,
        partido_grafico.jugador2,
    )
    set_grafico = partido_grafico.historial_sets[-1]
    marcador_j1.set("0")
    marcador_j2.set("0")
    resultado_final.set("Resumen final: pendiente")
    estado_tenis.set("Nuevo partido. Pulsa un botón para anotar un juego.")
    boton_j1.config(state=tk.NORMAL)
    boton_j2.config(state=tk.NORMAL)
    boton_suspender.config(state=tk.NORMAL)
    boton_finalizar.config(state=tk.NORMAL)
    boton_reiniciar.config(state=tk.DISABLED)

def abrir_estadisticas():
    ventana_estadisticas = tk.Toplevel(ventana_tenis)
    ventana_estadisticas.title("Historial y estadísticas")
    ventana_estadisticas.geometry("560x380")

    texto_estadisticas = tk.Text(ventana_estadisticas, wrap=tk.WORD, height=18, width=64)
    barra = tk.Scrollbar(ventana_estadisticas, command=texto_estadisticas.yview)
    texto_estadisticas.config(yscrollcommand=barra.set, state=tk.NORMAL)
    texto_estadisticas.pack(side=tk.LEFT, fill=tk.BOTH, expand=True, padx=(10, 0), pady=10)
    barra.pack(side=tk.RIGHT, fill=tk.Y, padx=(0, 10), pady=10)

    if not historial_partidos:
        informe = "Todav\u00eda no hay partidos finalizados o suspendidos."
    else:
        cantidad = len(historial_partidos)
        jugador1 = historial_partidos[0]["jugador1"]
        jugador2 = historial_partidos[0]["jugador2"]
        total_j1 = sum(partido["juegos_j1"] for partido in historial_partidos)
        total_j2 = sum(partido["juegos_j2"] for partido in historial_partidos)
        finalizados = sum(
            partido["estado"] == "partido finalizado manualmente"
            for partido in historial_partidos
        )
        suspendidos = cantidad - finalizados
        informe = (
            f"Partidos registrados: {cantidad}\n"
            f"Finalizados: {finalizados} | Suspendidos: {suspendidos}\n\n"
            f"Media de juegos por partido (incluye suspendidos):\n"
            f"{jugador1}: {total_j1 / cantidad:.2f}\n"
            f"{jugador2}: {total_j2 / cantidad:.2f}\n\n"
            "Historial de resultados:\n"
        )
        for partido in historial_partidos:
            informe += (
                f"Partido {partido['numero']} - {partido['estado']}\n"
                f"  {partido['jugador1']} {partido['juegos_j1']} - "
                f"{partido['juegos_j2']} {partido['jugador2']}\n"
                f"  {partido['resultado']}\n\n"
            )

    texto_estadisticas.insert("1.0", informe)
    texto_estadisticas.config(state=tk.DISABLED)

boton_suspender = tk.Button(
    ventana_tenis,
    text="Suspender por lluvia",
    command=suspender_partido_grafico,
)
boton_suspender.pack(pady=4)

boton_finalizar = tk.Button(
    ventana_tenis,
    text="Finalizar partido",
    command=finalizar_partido_grafico,
)
boton_finalizar.pack(pady=4)

tk.Label(
    ventana_tenis,
    textvariable=resultado_final,
    justify=tk.CENTER,
    wraplength=480,
).pack(pady=6)

tk.Button(
    ventana_tenis,
    text="Ver historial y estadísticas",
    command=abrir_estadisticas,
).pack(pady=4)

boton_reiniciar = tk.Button(
    ventana_tenis,
    text="Volver a jugar",
    command=volver_a_jugar,
    state=tk.DISABLED,
)
boton_reiniciar.pack(pady=4)

tk.Button(
    ventana_tenis,
    text="Cerrar ventana",
    command=lambda: ventana_tenis.destroy(),
).pack(pady=4)
```

## Simulador interactivo de La Primitiva

En el ejercicio de números aleatorios se generan apuestas de seis números entre el 1 y el 49 y se cuentan los aciertos frente a un sorteo. Aquí adaptamos esa idea a una interfaz gráfica: la combinación ganadora se elige manualmente en un tablero de 49 botones y, al pulsar **Simular**, se generan y comparan un millón de apuestas aleatorias.

### Planteamiento y resultado esperado

- Cada botón del tablero representa un número del 1 al 49. Se pueden seleccionar exactamente seis; esa selección es la combinación ganadora.
- Al pulsar **Simular 1.000.000 apuestas**, cada apuesta se crea con seis números distintos. Se cuentan las apuestas con exactamente 3, 4, 5 y 6 aciertos.
- Los cuatro recuentos aparecen debajo del tablero. Se procesan las apuestas en lotes de 10.000, sin guardarlas todas en memoria, y la ventana se mantiene actualizable durante el cálculo.
- **Limpiar selección** permite elegir otro sorteo. Una nueva simulación genera apuestas aleatorias nuevas, por lo que los resultados pueden variar.

Ejecuta esta celda para abrir la aplicación. Los recuentos que muestra son el número de apuestas simuladas con cada cantidad exacta de aciertos.

```python
import random
import tkinter as tk


def contar_aciertos_primitiva(apuesta, combinacion_ganadora):
    """Devuelve cuántos números de una apuesta están en el sorteo."""
    return sum(numero in combinacion_ganadora for numero in apuesta)


numeros_primitiva = tuple(range(1, 50))
total_apuestas_primitiva = 1_000_000

ventana_primitiva = tk.Tk()
ventana_primitiva.title("La Primitiva - simulador de apuestas")
ventana_primitiva.geometry("560x730")
ventana_primitiva.resizable(False, False)
ventana_primitiva.configure(bg="#f8f3f0")

color_azul = "#164b8b"
color_coral = "#e97868"
color_texto = "#6d514d"
color_celda = "#fffaf7"
combinacion_ganadora = set()
botones_primitiva = {}

marco_titulo = tk.Frame(ventana_primitiva, bg=color_azul, padx=12, pady=12)
marco_titulo.pack(fill=tk.X)
tk.Label(
    marco_titulo,
    text="LA PRIMITIVA - SIMULADOR",
    bg=color_azul,
    fg="white",
    font=("Arial", 15, "bold"),
).pack()

tk.Label(
    ventana_primitiva,
    text="SELECCIONA LOS 6 N\u00daMEROS GANADORES",
    bg="#f8f3f0",
    fg=color_coral,
    font=("Arial", 12, "bold"),
).pack(pady=(16, 4))

estado_primitiva = tk.StringVar(
    master=ventana_primitiva,
    value="N\u00fameros seleccionados: 0 de 6",
)
tk.Label(
    ventana_primitiva,
    textvariable=estado_primitiva,
    bg="#f8f3f0",
    fg=color_texto,
    font=("Arial", 10),
).pack(pady=(0, 10))

marco_numeros = tk.Frame(ventana_primitiva, bg="#f8f3f0")
marco_numeros.pack(padx=12)


def actualizar_tablero_primitiva():
    for numero, boton in botones_primitiva.items():
        seleccionado = numero in combinacion_ganadora
        boton.configure(
            bg=color_coral if seleccionado else color_celda,
            fg="white" if seleccionado else color_texto,
            relief=tk.SUNKEN if seleccionado else tk.RAISED,
        )


def alternar_numero_primitiva(numero):
    if numero in combinacion_ganadora:
        combinacion_ganadora.remove(numero)
    elif len(combinacion_ganadora) < 6:
        combinacion_ganadora.add(numero)
    else:
        estado_primitiva.set("Ya hay seis n\u00fameros. Desmarca uno para cambiar la selecci\u00f3n.")
        return

    actualizar_tablero_primitiva()
    for variable in resultados_primitiva.values():
        variable.set("Pendiente")
    estado_primitiva.set(
        f"N\u00fameros seleccionados: {len(combinacion_ganadora)} de 6"
    )
    boton_simular_primitiva.configure(
        state=tk.NORMAL if len(combinacion_ganadora) == 6 else tk.DISABLED
    )


for numero in numeros_primitiva:
    boton = tk.Button(
        marco_numeros,
        text=str(numero),
        width=6,
        height=2,
        font=("Arial", 10, "bold"),
        bg=color_celda,
        fg=color_texto,
        activebackground=color_coral,
        command=lambda n=numero: alternar_numero_primitiva(n),
    )
    fila, columna = divmod(numero - 1, 7)
    boton.grid(row=fila, column=columna, padx=3, pady=3)
    botones_primitiva[numero] = boton


tk.Label(
    ventana_primitiva,
    text="ACERTANTES EN LA SIMULACI\u00d3N",
    bg="#f8f3f0",
    fg=color_coral,
    font=("Arial", 11, "bold"),
).pack(pady=(14, 6))

marco_resultados = tk.Frame(ventana_primitiva, bg="#f8f3f0")
marco_resultados.pack(padx=12)
resultados_primitiva = {}
for columna, cantidad_aciertos in enumerate((3, 4, 5, 6)):
    tarjeta = tk.Frame(
        marco_resultados,
        bg=color_celda,
        highlightbackground="#efb0a6",
        highlightthickness=1,
        padx=8,
        pady=8,
    )
    tarjeta.grid(row=0, column=columna, padx=4, sticky="nsew")
    tk.Label(
        tarjeta,
        text=f"{cantidad_aciertos} aciertos",
        bg=color_celda,
        fg=color_texto,
        font=("Arial", 9, "bold"),
    ).pack()
    resultados_primitiva[cantidad_aciertos] = tk.StringVar(
        master=ventana_primitiva,
        value="Pendiente",
    )
    tk.Label(
        tarjeta,
        textvariable=resultados_primitiva[cantidad_aciertos],
        bg=color_celda,
        fg=color_azul,
        font=("Arial", 11, "bold"),
    ).pack(pady=(4, 0))


def iniciar_simulacion_primitiva():
    if len(combinacion_ganadora) != 6:
        return

    boton_simular_primitiva.configure(state=tk.DISABLED)
    boton_limpiar_primitiva.configure(state=tk.DISABLED)
    for boton in botones_primitiva.values():
        boton.configure(state=tk.DISABLED)

    recuentos = {3: 0, 4: 0, 5: 0, 6: 0}
    tamano_lote = 10_000

    def procesar_lote(primer_indice=0):
        ultimo_indice = min(
            primer_indice + tamano_lote,
            total_apuestas_primitiva,
        )
        for _ in range(primer_indice, ultimo_indice):
            apuesta = random.sample(numeros_primitiva, 6)
            aciertos = contar_aciertos_primitiva(apuesta, combinacion_ganadora)
            if aciertos in recuentos:
                recuentos[aciertos] += 1

        estado_primitiva.set(
            "Simulando: "
            f"{ultimo_indice:,} / {total_apuestas_primitiva:,} apuestas".replace(",", ".")
        )
        if ultimo_indice < total_apuestas_primitiva:
            ventana_primitiva.after(1, procesar_lote, ultimo_indice)
            return

        for cantidad_aciertos, recuento in recuentos.items():
            resultados_primitiva[cantidad_aciertos].set(f"{recuento:,}".replace(",", "."))
        estado_primitiva.set(
            "Simulaci\u00f3n finalizada: "
            f"{total_apuestas_primitiva:,} apuestas".replace(",", ".")
        )
        for boton in botones_primitiva.values():
            boton.configure(state=tk.NORMAL)
        boton_limpiar_primitiva.configure(state=tk.NORMAL)
        boton_simular_primitiva.configure(state=tk.NORMAL)

    estado_primitiva.set("Preparando la simulaci\u00f3n...")
    ventana_primitiva.after(1, procesar_lote)


def limpiar_seleccion_primitiva():
    combinacion_ganadora.clear()
    actualizar_tablero_primitiva()
    for variable in resultados_primitiva.values():
        variable.set("Pendiente")
    estado_primitiva.set("N\u00fameros seleccionados: 0 de 6")
    boton_simular_primitiva.configure(state=tk.DISABLED)


marco_acciones = tk.Frame(ventana_primitiva, bg="#f8f3f0")
marco_acciones.pack(pady=14)
boton_simular_primitiva = tk.Button(
    marco_acciones,
    text="Simular 1.000.000 apuestas",
    command=iniciar_simulacion_primitiva,
    state=tk.DISABLED,
    bg=color_azul,
    fg="white",
    activebackground="#2866a8",
    font=("Arial", 10, "bold"),
    padx=10,
    pady=7,
)
boton_simular_primitiva.grid(row=0, column=0, padx=5)
boton_limpiar_primitiva = tk.Button(
    marco_acciones,
    text="Limpiar selecci\u00f3n",
    command=limpiar_seleccion_primitiva,
    bg="#f2d6d1",
    fg=color_texto,
    font=("Arial", 10),
    padx=10,
    pady=7,
)
boton_limpiar_primitiva.grid(row=0, column=1, padx=5)
boton_cerrar_primitiva = tk.Button(
    ventana_primitiva,
    text="Cerrar aplicación",
    command=ventana_primitiva.destroy,
    bg="#d9d9d9",
    fg=color_texto,
    font=("Arial", 10),
    padx=10,
    pady=7,
)
boton_cerrar_primitiva.pack(pady=(0, 12))

ventana_primitiva.mainloop()
```

Vamos a construir un sistema de filtros que:
Detecta automáticamente el tipo de cada columna
Crea el filtro adecuado según el tipo (rango para números, selección para categorías)
Aplica todos los filtros a la vez (combinación AND)
Actualiza en tiempo real la tabla y el gráfico

🧠 El concepto clave: DataFrame original vs DataFrame filtrado
Imagina que tienes dos copias de tus datos:
df_original: La copia sagrada, nunca se toca.
df_filtrado: Una copia que se regenera cada vez que cambias un filtro.
Así nunca pierdes los datos originales. Cada filtro es una "condición" que pandas evalúa fila por fila y devuelve True o False.

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
    "edad": np.random.randint(18, 80, 1000),
    "salario": np.random.normal(30000, 10000, 1000),
    "ciudad": np.random.choice(["Madrid", "Barcelona", "Sevilla", "Valencia"], 1000),
    "departamento": np.random.choice(["Ventas", "IT", "RRHH", "Marketing", "Finanzas"], 1000),
})
df.to_csv("empleados.csv", index=False)
```

```python
import tkinter as tk
from tkinter import ttk, filedialog, messagebox
%pip install pandas
import matplotlib.pyplot as plt
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg
from matplotlib.figure import Figure


# ============================================
# 1. VENTANA Y VARIABLES GLOBALES
# ============================================
ventana = tk.Tk()
ventana.title("EDA - Explorador de Datos con Filtros")
ventana.geometry("1100x750")

df_original = None   # Los datos tal cual se cargaron
df_filtrado = None   # Los datos después de aplicar filtros
widgets_filtros = {} # Guarda los widgets de cada filtro


# ============================================
# 2. CARGAR ARCHIVO (igual que antes)
# ============================================
def cargar_archivo():
    global df_original, df_filtrado
    
    ruta = filedialog.askopenfilename(
        title="Selecciona un archivo de datos",
        filetypes=[("CSV", "*.csv"), ("Parquet", "*.parquet"), ("Todos", "*.*")]
    )
    if not ruta:
        return
    
    try:
        if ruta.endswith(".csv"):
            df_original = pd.read_csv(ruta)
        elif ruta.endswith(".parquet"):
            df_original = pd.read_parquet(ruta)
        else:
            messagebox.showerror("Error", "Formato no soportado")
            return
        
        df_filtrado = df_original.copy()
        etiqueta_estado.config(
            text=f"✅ {ruta.split('/')[-1]}  |  {len(df_original):,} filas × {len(df_original.columns)} columnas"
        )
        construir_filtros()   # ← NUEVO: crea los filtros dinámicamente
        mostrar_info()
        
    except Exception as e:
        messagebox.showerror("Error al cargar", str(e))


# ============================================
# 3. CONSTRUIR FILTROS DINÁMICOS ⭐ LA MAGIA
# ============================================
def construir_filtros():
    """Crea un filtro adecuado para cada columna según su tipo."""
    global widgets_filtros
    widgets_filtros = {}
    
    # Limpia los filtros anteriores
    for widget in marco_filtros.winfo_children():
        widget.destroy()
    
    fila, columna = 0, 0
    max_columnas = 4  # Cuántos filtros por fila
    
    for nombre_col in df_original.columns:
        # Pequeño marco para cada filtro (etiqueta + widget)
        marco_col = tk.Frame(marco_filtros, bg="#f8f3f0", padx=5, pady=3)
        marco_col.grid(row=fila, column=columna, padx=4, pady=2, sticky="ew")
        
        tk.Label(
            marco_col, text=nombre_col, bg="#f8f3f0",
            fg="#164b8b", font=("Arial", 8, "bold")
        ).pack(anchor="w")
        
        tipo = df_original[nombre_col].dtype
        
        # --- CASO A: Columna numérica → filtro de rango (min/max) ---
        if pd.api.types.is_numeric_dtype(tipo):
            minimo = float(df_original[nombre_col].min())
            maximo = float(df_original[nombre_col].max())
            
            marco_rango = tk.Frame(marco_col, bg="#f8f3f0")
            marco_rango.pack()
            
            entry_min = tk.Entry(marco_rango, width=7, font=("Arial", 9))
            entry_min.insert(0, f"{minimo:.2f}")
            entry_min.pack(side=tk.LEFT, padx=2)
            
            tk.Label(marco_rango, text="→", bg="#f8f3f0").pack(side=tk.LEFT)
            
            entry_max = tk.Entry(marco_rango, width=7, font=("Arial", 9))
            entry_max.insert(0, f"{maximo:.2f}")
            entry_max.pack(side=tk.LEFT, padx=2)
            
            # Guardamos referencias para leerlas luego
            widgets_filtros[nombre_col] = {
                "tipo": "numerico",
                "min": entry_min,
                "max": entry_max,
                "original_min": minimo,
                "original_max": maximo,
            }
        
        # --- CASO B: Columna categórica/pocas categorías → dropdown ---
        elif df_original[nombre_col].nunique() <= 15:
            valores_unicos = ["(todos)"] + sorted(
                df_original[nombre_col].dropna().unique().tolist()
            )
            combo = ttk.Combobox(
                marco_col, values=valores_unicos, state="readonly", width=12
            )
            combo.current(0)  # Selecciona "(todos)" por defecto
            combo.pack()
            
            widgets_filtros[nombre_col] = {
                "tipo": "categoria",
                "widget": combo,
            }
        
        # --- CASO C: Columna de texto con muchas categorías → búsqueda ---
        else:
            entry_texto = tk.Entry(marco_col, width=15, font=("Arial", 9))
            entry_texto.insert(0, "")
            entry_texto.pack()
            tk.Label(
                marco_col, text="(contiene...)", bg="#f8f3f0",
                fg="#999", font=("Arial", 7, "italic")
            ).pack(anchor="w")
            
            widgets_filtros[nombre_col] = {
                "tipo": "texto",
                "widget": entry_texto,
            }
        
        # Control de rejilla
        columna += 1
        if columna >= max_columnas:
            columna = 0
            fila += 1


# ============================================
# 4. APLICAR TODOS LOS FILTROS
# ============================================
def aplicar_filtros():
    """Regenera df_filtrado aplicando todas las condiciones."""
    global df_filtrado
    
    if df_original is None:
        return
    
    # Empezamos con todas las filas como "válidas" (máscara de True)
    mascara = pd.Series(True, index=df_original.index)
    
    for nombre_col, config in widgets_filtros.items():
        try:
            if config["tipo"] == "numerico":
                min_val = float(config["min"].get())
                max_val = float(config["max"].get())
                col = df_original[nombre_col]
                mascara &= (col >= min_val) & (col <= max_val)
            
            elif config["tipo"] == "categoria":
                valor = config["widget"].get()
                if valor != "(todos)":
                    mascara &= (df_original[nombre_col] == valor)
            
            elif config["tipo"] == "texto":
                texto = config["widget"].get().strip()
                if texto:
                    # Búsqueda case-insensitive
                    mascara &= df_original[nombre_col].astype(str).str.contains(
                        texto, case=False, na=False
                    )
        except ValueError:
            # Si el usuario escribió algo que no es número, ignoramos ese filtro
            continue
    
    df_filtrado = df_original[mascara].copy()
    
    etiqueta_estado.config(
        text=f"🔍 Mostrando {len(df_filtrado):,} de {len(df_original):,} filas "
             f"({len(df_filtrado)/len(df_original)*100:.1f}%)"
    )
    mostrar_info()


# ============================================
# 5. RESETEAR FILTROS
# ============================================
def resetear_filtros():
    """Vuelve todos los filtros a su valor original."""
    for nombre_col, config in widgets_filtros.items():
        if config["tipo"] == "numerico":
            config["min"].delete(0, tk.END)
            config["min"].insert(0, f"{config['original_min']:.2f}")
            config["max"].delete(0, tk.END)
            config["max"].insert(0, f"{config['original_max']:.2f}")
        elif config["tipo"] == "categoria":
            config["widget"].current(0)
        elif config["tipo"] == "texto":
            config["widget"].delete(0, tk.END)
    
    aplicar_filtros()


# ============================================
# 6. MOSTRAR INFO Y GRÁFICO (sobre df_filtrado)
# ============================================
def mostrar_info():
    texto_info.config(state=tk.NORMAL)
    texto_info.delete(1.0, tk.END)
    
    texto_info.insert(tk.END, "=== PRIMERAS FILAS FILTRADAS ===\n")
    texto_info.insert(tk.END, df_filtrado.head(10).to_string())
    texto_info.insert(tk.END, "\n\n=== ESTADÍSTICAS (datos filtrados) ===\n")
    texto_info.insert(tk.END, df_filtrado.describe().to_string())
    texto_info.config(state=tk.DISABLED)
    
    dibujar_histograma()


def dibujar_histograma():
    cols_numericas = df_filtrado.select_dtypes(include="number").columns
    if len(cols_numericas) == 0 or len(df_filtrado) == 0:
        return
    
    col = cols_numericas[0]
    fig = Figure(figsize=(5, 3), dpi=100)
    ax = fig.add_subplot(111)
    ax.hist(df_filtrado[col].dropna(), bins=30, color="#164b8b", alpha=0.8)
    ax.set_title(f"Distribución de: {col}  (n={len(df_filtrado):,})")
    ax.set_xlabel(col)
    ax.set_ylabel("Frecuencia")
    fig.tight_layout()
    
    for widget in marco_grafico.winfo_children():
        widget.destroy()
    
    canvas = FigureCanvasTkAgg(fig, master=marco_grafico)
    canvas.draw()
    canvas.get_tk_widget().pack(fill=tk.BOTH, expand=True)


# ============================================
# 7. CONSTRUIR LA INTERFAZ
# ============================================
# Barra superior
barra = tk.Frame(ventana, bg="#164b8b", padx=10, pady=10)
barra.pack(fill=tk.X)

tk.Button(
    barra, text="📂 Cargar archivo", command=cargar_archivo,
    bg="white", fg="#164b8b", font=("Arial", 10, "bold"), padx=10, pady=5
).pack(side=tk.LEFT)

etiqueta_estado = tk.Label(
    barra, text="Ningún archivo cargado",
    bg="#164b8b", fg="white", font=("Arial", 10)
)
etiqueta_estado.pack(side=tk.LEFT, padx=15)

# Zona de filtros
tk.Label(
    ventana, text="🔧 FILTROS (modifica y pulsa 'Aplicar')",
    bg="#f8f3f0", fg="#e97868", font=("Arial", 11, "bold")
).pack(anchor="w", padx=10, pady=(10, 2))

marco_filtros = tk.Frame(ventana, bg="#f8f3f0", padx=5, pady=5,
                         highlightbackground="#efb0a6", highlightthickness=1)
marco_filtros.pack(fill=tk.X, padx=10)

# Botones de acción sobre filtros
marco_botones_filtros = tk.Frame(ventana, bg="#f8f3f0")
marco_botones_filtros.pack(fill=tk.X, padx=10, pady=5)

tk.Button(
    marco_botones_filtros, text="✅ Aplicar filtros",
    command=aplicar_filtros, bg="#164b8b", fg="white",
    font=("Arial", 10, "bold"), padx=10, pady=5
).pack(side=tk.LEFT, padx=5)

tk.Button(
    marco_botones_filtros, text="🔄 Resetear",
    command=resetear_filtros, bg="#f2d6d1", fg="#6d514d",
    font=("Arial", 10), padx=10, pady=5
).pack(side=tk.LEFT, padx=5)

# Área de texto con scroll
marco_info = tk.Frame(ventana)
marco_info.pack(fill=tk.BOTH, expand=True, padx=10, pady=5)

texto_info = tk.Text(marco_info, wrap=tk.NONE, font=("Courier", 9))
scroll_y = tk.Scrollbar(marco_info, orient=tk.VERTICAL, command=texto_info.yview)
scroll_x = tk.Scrollbar(marco_info, orient=tk.HORIZONTAL, command=texto_info.xview)
texto_info.configure(yscrollcommand=scroll_y.set, xscrollcommand=scroll_x.set)

texto_info.grid(row=0, column=0, sticky="nsew")
scroll_y.grid(row=0, column=1, sticky="ns")
scroll_x.grid(row=1, column=0, sticky="ew")
marco_info.grid_rowconfigure(0, weight=1)
marco_info.grid_columnconfigure(0, weight=1)
texto_info.config(state=tk.DISABLED)

# Gráfico
marco_grafico = tk.Frame(ventana, bg="white", height=250)
marco_grafico.pack(fill=tk.X, padx=10, pady=(0, 10))

ventana.mainloop()
```

```python
%pip install seaborn scipy
import tkinter as tk
from tkinter import ttk, filedialog, messagebox
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns # ⭐ NUEVO: Gráficos más bonitos
from scipy import stats # ⭐ NUEVO: Estadística avanzada
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg
from matplotlib.figure import Figure

# Configuración visual de Seaborn
sns.set_theme(style="whitegrid")

# ============================================
# 1. VENTANA Y VARIABLES GLOBALES
# ============================================
ventana = tk.Tk()
ventana.title("EDA Pro - Estadísticas Avanzadas")
ventana.geometry("1200x800")

df_original = None
df_filtrado = None
widgets_filtros = {}


# ============================================
# 2. CARGAR ARCHIVO (Sin cambios mayores)
# ============================================
def cargar_archivo():
    global df_original, df_filtrado
    
    ruta = filedialog.askopenfilename(
        title="Selecciona archivo",
        filetypes=[("CSV", "*.csv"), ("Parquet", "*.parquet")]
    )
    if not ruta: return
    
    try:
        if ruta.endswith(".csv"): df_original = pd.read_csv(ruta)
        elif ruta.endswith(".parquet"): df_original = pd.read_parquet(ruta)
        
        df_filtrado = df_original.copy()
        etiqueta_estado.config(text=f"✅ {ruta.split('/')[-1]} | {len(df_original):,} filas")
        
        construir_filtros()
        actualizar_todo() # ⭐ NUEVO: Función maestra
        
    except Exception as e:
        messagebox.showerror("Error", str(e))


# ============================================
# 3. FILTROS (Mismo código que antes, resumido aquí)
# ============================================
def construir_filtros():
    global widgets_filtros
    widgets_filtros = {}
    for widget in marco_filtros.winfo_children(): widget.destroy()
    
    fila, col = 0, 0
    for nombre_col in df_original.columns:
        marco_col = tk.Frame(marco_filtros, bg="#f8f3f0", padx=5, pady=3)
        marco_col.grid(row=fila, column=col, padx=4, pady=2, sticky="ew")
        tk.Label(marco_col, text=nombre_col, bg="#f8f3f0", fg="#164b8b", font=("Arial", 8, "bold")).pack(anchor="w")
        
        tipo = df_original[nombre_col].dtype
        if pd.api.types.is_numeric_dtype(tipo):
            min_val, max_val = float(df_original[nombre_col].min()), float(df_original[nombre_col].max())
            f_min = tk.Entry(marco_col, width=7); f_min.insert(0, f"{min_val:.1f}"); f_min.pack(side=tk.LEFT)
            f_max = tk.Entry(marco_col, width=7); f_max.insert(0, f"{max_val:.1f}"); f_max.pack(side=tk.LEFT)
            widgets_filtros[nombre_col] = {"tipo": "num", "min": f_min, "max": f_max, "orig_min": min_val, "orig_max": max_val}
        elif df_original[nombre_col].nunique() <= 15:
            vals = ["(todos)"] + sorted(df_original[nombre_col].dropna().unique().tolist())
            combo = ttk.Combobox(marco_col, values=vals, state="readonly", width=12); combo.current(0); combo.pack()
            widgets_filtros[nombre_col] = {"tipo": "cat", "widget": combo}
        else:
            entry = tk.Entry(marco_col, width=12); entry.pack()
            widgets_filtros[nombre_col] = {"tipo": "txt", "widget": entry}
            
        col += 1
        if col >= 4: col, fila = 0, fila + 1

def aplicar_filtros():
    global df_filtrado
    if df_original is None: return
    mascara = pd.Series(True, index=df_original.index)
    for col, cfg in widgets_filtros.items():
        try:
            if cfg["tipo"] == "num":
                mn, mx = float(cfg["min"].get()), float(cfg["max"].get())
                mascara &= (df_original[col] >= mn) & (df_original[col] <= mx)
            elif cfg["tipo"] == "cat":
                v = cfg["widget"].get()
                if v != "(todos)": mascara &= (df_original[col] == v)
            elif cfg["tipo"] == "txt":
                t = cfg["widget"].get().strip()
                if t: mascara &= df_original[col].astype(str).str.contains(t, case=False, na=False)
        except: continue
    df_filtrado = df_original[mascara].copy()
    etiqueta_estado.config(text=f"🔍 {len(df_filtrado):,} / {len(df_original):,} filas")
    actualizar_todo()

def resetear_filtros():
    for col, cfg in widgets_filtros.items():
        if cfg["tipo"] == "num":
            cfg["min"].delete(0, tk.END); cfg["min"].insert(0, f"{cfg['orig_min']:.1f}")
            cfg["max"].delete(0, tk.END); cfg["max"].insert(0, f"{cfg['orig_max']:.1f}")
        elif cfg["tipo"] == "cat": cfg["widget"].current(0)
        elif cfg["tipo"] == "txt": cfg["widget"].delete(0, tk.END)
    aplicar_filtros()


# ============================================
# 4. ⭐ NUEVO: MOTOR DE ESTADÍSTICAS
# ============================================

def calcular_estadisticas_generales(df):
    """Calcula métricas globales del dataset."""
    info = {
        "Filas": len(df),
        "Columnas": len(df.columns),
        "Memoria (MB)": f"{df.memory_usage(deep=True).sum() / 1024**2:.2f}",
        "Valores Nulos Totales": int(df.isnull().sum().sum()),
        "% Nulos": f"{(df.isnull().sum().sum() / (len(df)*len(df.columns)) * 100):.2f}%",
        "Filas Duplicadas": int(df.duplicated().sum())
    }
    return info

def calcular_estadisticas_columna(df, columna):
    """Calcula estadísticas profundas para una sola columna."""
    serie = df[columna]
    es_numerica = pd.api.types.is_numeric_dtype(serie.dtype)
    
    stats_dict = {
        "Tipo de Dato": str(serie.dtype),
        "Valores Únicos": serie.nunique(),
        "Valores Nulos": int(serie.isnull().sum()),
        "% Nulos": f"{(serie.isnull().sum() / len(serie) * 100):.2f}%"
    }
    
    if es_numerica and serie.notna().any():
        stats_dict.update({
            "Media": f"{serie.mean():.4f}",
            "Mediana": f"{serie.median():.4f}",
            "Moda": f"{serie.mode()[0] if not serie.mode().empty else 'N/A'}",
            "Desv. Estándar": f"{serie.std():.4f}",
            "Mínimo": f"{serie.min():.4f}",
            "Máximo": f"{serie.max():.4f}",
            "Asimetría (Skew)": f"{stats.skew(serie.dropna()):.4f}",
            "Curtosis": f"{stats.kurtosis(serie.dropna()):.4f}"
        })
    else:
        # Para categóricas
        top_val = serie.value_counts().index[0] if not serie.value_counts().empty else "N/A"
        top_freq = serie.value_counts().iloc[0] if not serie.value_counts().empty else 0
        stats_dict.update({
            "Valor Más Frecuente": str(top_val),
            "Frecuencia Top": int(top_freq)
        })
        
    return stats_dict


# ============================================
# 5. ⭐ NUEVO: ACTUALIZAR INTERFAZ COMPLETA
# ============================================

def actualizar_todo():
    """Orquesta la actualización de todas las pestañas."""
    if df_filtrado is None or df_filtrado.empty:
        return
        
    actualizar_resumen_general()
    actualizar_analisis_columna() # Actualiza la pestaña de columna seleccionada
    dibujar_grafico_columna()     # Actualiza el gráfico


def actualizar_resumen_general():
    """Rellena la tabla de la Pestaña 1."""
    # Limpiar treeview anterior
    for item in tree_general.get_children():
        tree_general.delete(item)
        
    stats_gen = calcular_estadisticas_generales(df_filtrado)
    for key, val in stats_gen.items():
        tree_general.insert("", tk.END, values=(key, val))
        
    # También actualizamos la lista de columnas disponibles para la Pestaña 2
    combo_columnas['values'] = list(df_filtrado.columns)
    if combo_columnas.current() == -1 and len(df_filtrado.columns) > 0:
        combo_columnas.current(0)


def actualizar_analisis_columna():
    """Rellena la tabla de la Pestaña 2 según la columna elegida."""
    col_seleccionada = combo_columnas.get()
    if not col_seleccionada or col_seleccionada not in df_filtrado.columns:
        return
        
    for item in tree_columna.get_children():
        tree_columna.delete(item)
        
    stats_col = calcular_estadisticas_columna(df_filtrado, col_seleccionada)
    for key, val in stats_col.items():
        tree_columna.insert("", tk.END, values=(key, val))


def on_cambio_columna(event):
    """Se dispara cuando cambias la columna en el dropdown."""
    actualizar_analisis_columna()
    dibujar_grafico_columna()


def dibujar_grafico_columna():
    """Dibuja el gráfico adecuado según el tipo de dato."""
    col = combo_columnas.get()
    if not col or col not in df_filtrado.columns: return
    
    serie = df_filtrado[col]
    es_numerica = pd.api.types.is_numeric_dtype(serie.dtype)
    
    fig = Figure(figsize=(6, 4), dpi=100)
    ax = fig.add_subplot(111)
    
    if es_numerica:
        # Histograma con curva de densidad (KDE)
        sns.histplot(serie.dropna(), kde=True, ax=ax, color="#164b8b")
        ax.set_title(f"Distribución de {col}")
    else:
        # Gráfico de barras para categóricas (top 10)
        counts = serie.value_counts().head(10)
        sns.barplot(x=counts.index.astype(str), y=counts.values, ax=ax, palette="viridis")
        ax.set_title(f"Frecuencia de {col} (Top 10)")
        ax.tick_params(axis='x', rotation=45)
        
    fig.tight_layout()
    
    for widget in marco_grafico.winfo_children(): widget.destroy()
    canvas = FigureCanvasTkAgg(fig, master=marco_grafico)
    canvas.draw()
    canvas.get_tk_widget().pack(fill=tk.BOTH, expand=True)


# ============================================
# 6. CONSTRUCCIÓN DE LA INTERFAZ (Layout con Pestañas)
# ============================================

# Barra superior
barra = tk.Frame(ventana, bg="#164b8b", padx=10, pady=10)
barra.pack(fill=tk.X)
tk.Button(barra, text="📂 Cargar", command=cargar_archivo, bg="white", fg="#164b8b", font=("Arial", 10, "bold")).pack(side=tk.LEFT)
etiqueta_estado = tk.Label(barra, text="Esperando datos...", bg="#164b8b", fg="white", font=("Arial", 10))
etiqueta_estado.pack(side=tk.LEFT, padx=15)

# Zona de Filtros
tk.Label(ventana, text="🔧 FILTROS", bg="#f8f3f0", fg="#e97868", font=("Arial", 11, "bold")).pack(anchor="w", padx=10, pady=(10,2))
marco_filtros = tk.Frame(ventana, bg="#f8f3f0", highlightbackground="#efb0a6", highlightthickness=1)
marco_filtros.pack(fill=tk.X, padx=10)
marco_botones_filtros = tk.Frame(ventana, bg="#f8f3f0")
marco_botones_filtros.pack(fill=tk.X, padx=10, pady=5)
tk.Button(marco_botones_filtros, text="✅ Aplicar", command=aplicar_filtros, bg="#164b8b", fg="white").pack(side=tk.LEFT, padx=5)
tk.Button(marco_botones_filtros, text="🔄 Reset", command=resetear_filtros, bg="#f2d6d1", fg="#6d514d").pack(side=tk.LEFT, padx=5)

# ⭐ ZONA PRINCIPAL CON PESTAÑAS
notebook = ttk.Notebook(ventana)
notebook.pack(fill=tk.BOTH, expand=True, padx=10, pady=10)

# --- Pestaña 1: Resumen General ---
tab_general = tk.Frame(notebook, bg="white")
notebook.add(tab_general, text="📊 Resumen Dataset")

# Tabla de resumen general
cols_gen = ("Metrica", "Valor")
tree_general = ttk.Treeview(tab_general, columns=cols_gen, show="headings", height=10)
tree_general.heading("Metrica", text="Métrica Global")
tree_general.heading("Valor", text="Valor")
tree_general.column("Metrica", width=200)
tree_general.column("Valor", width=200)
tree_general.pack(padx=20, pady=20, anchor="nw")

# --- Pestaña 2: Análisis por Columna ---
tab_columna = tk.Frame(notebook, bg="white")
notebook.add(tab_columna, text="🔍 Análisis Columna")

# Selector de columna
frame_selector = tk.Frame(tab_columna, bg="white")
frame_selector.pack(fill=tk.X, padx=20, pady=10)
tk.Label(frame_selector, text="Selecciona Columna:", bg="white", font=("Arial", 10, "bold")).pack(side=tk.LEFT)
combo_columnas = ttk.Combobox(frame_selector, state="readonly", width=20)
combo_columnas.pack(side=tk.LEFT, padx=10)
combo_columnas.bind("<<ComboboxSelected>>", on_cambio_columna)

# Contenedor horizontal para Tabla de Stats y Gráfico
frame_contenido_col = tk.Frame(tab_columna, bg="white")
frame_contenido_col.pack(fill=tk.BOTH, expand=True, padx=20, pady=10)

# Izquierda: Tabla de estadísticas
cols_col = ("Metrica", "Valor")
tree_columna = ttk.Treeview(frame_contenido_col, columns=cols_col, show="headings", height=15)
tree_columna.heading("Metrica", text="Estadística")
tree_columna.heading("Valor", text="Valor")
tree_columna.column("Metrica", width=150)
tree_columna.column("Valor", width=150)
tree_columna.pack(side=tk.LEFT, fill=tk.Y, padx=(0, 10))

# Derecha: Gráfico
marco_grafico = tk.Frame(frame_contenido_col, bg="white", width=400)
marco_grafico.pack(side=tk.RIGHT, fill=tk.BOTH, expand=True)

ventana.mainloop()
```

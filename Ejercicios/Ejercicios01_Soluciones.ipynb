# Ejercicio 1

Dos conjuntos de nombres (ninguno se repite). Se pide:
1. Dos conjuntos: nombres por **consonante** y por **vocal**.
2. Lista con todos los nombres, **ordenada** alfabéticamente.
3. Conjunto de tuplas `(nombre, número)` del 1 al N.
4. Listas por **cantidad de letras**.
5. Tercer conjunto con **10 nombres de cada** conjunto original.

## Datos iniciales (Ejercicio 1)

```python
nombres_1 = {
    "Laura", "David", "Sofia", "Javier", "Elena",
    "Manuel", "Carmen", "Jose", "Isabel", "Francisco",
    "Paula", "Antonio", "Lucia", "Diego", "Martina"
}

nombres_2 = {
    "Daniel", "Marta", "Luis", "Alba", "Pablo",
    "Sara", "Adrian", "Julia", "Alvaro", "Irene",
    "Hugo", "Valeria", "Jorge", "Emma", "Ivan"
}

todos = nombres_1 | nombres_2
```

### 1. Vocal / consonante — Solución propuesta

```python
nombres_vocal = set()
nombres_consonante = set()

for x in nombres_1 | nombres_2:
    nombres_vocal.add(x) if x[0] in {"A", "E", "I", "O", "U"} else nombres_consonante.add(x)

print(nombres_vocal)
print("=====")
print(nombres_consonante)
```

#### Mejora (punto 1)

- Evitar el **ternario con efectos secundarios** (`.add` dentro del ternario): es difícil de leer.
- Usar `nombre[0].lower()` para no depender de mayúsculas.
- Preferir *set comprehensions* claras.
- Reutilizar `todos` en lugar de repetir la unión.

```python
vocales = "aeiouáéíóú"
nombres_vocal = {n for n in todos if n[0].lower() in vocales}
nombres_consonante = {n for n in todos if n[0].lower() not in vocales}

print("Vocal:", nombres_vocal)
print("Consonante:", nombres_consonante)
```

### 2. Lista ordenada — Solución propuesta

```python
print(sorted(list(nombres_1 | nombres_2)))
```

#### Mejora (punto 2)

- `sorted` ya acepta un conjunto: el `list(...)` intermedio sobra.
- Guardar el resultado en `lista_ordenada` para reutilizarlo en los puntos 3 y 4.

```python
lista_ordenada = sorted(todos)
print(lista_ordenada)
```

### 3. Conjunto de tuplas — Solución propuesta

```python
conjunto_nombres = set()
x = 1

for i in sorted(list(nombres_1 | nombres_2)):
    conjunto_nombres.add((i, x))
    x += 1

print(conjunto_nombres)
```

#### Mejora (punto 3)

- Partir de `lista_ordenada` (punto 2), sin recalcular.
- `enumerate(..., start=1)` evita el contador manual `x`.
- *Set comprehension* más compacta.
- Al imprimir, ordenar por el número (el `set` no tiene orden visual).

```python
conjunto_tuplas = {(nombre, i) for i, nombre in enumerate(lista_ordenada, start=1)}

print(f"Total: {len(conjunto_tuplas)}")
print(sorted(conjunto_tuplas, key=lambda t: t[1]))
```

### 4. Listas por longitud — Solución propuesta

```python
lista_nombres = [[], [], [], [], [], [], []]

for i in nombres_1 | nombres_2:
    match len(i):
        case 0 | 1 | 2 | 3:
            lista_nombres[0].append(i)
        case 4:
            lista_nombres[1].append(i)
        case 5:
            lista_nombres[2].append(i)
        case 6:
            lista_nombres[3].append(i)
        case 7:
            lista_nombres[4].append(i)
        case 8:
            lista_nombres[5].append(i)
        case _:
            lista_nombres[6].append(i)

print(lista_nombres)
```

#### Mejora (punto 4)

- La lista fija de 7 cajones mezcla longitudes 0–3 y obliga a adivinar índices.
- Mejor un **diccionario** `longitud → lista`, partiendo de `lista_ordenada`.
- Mostrar las claves ordenadas; cada lista queda alfabética.

```python
por_longitud = {}
for nombre in lista_ordenada:
    por_longitud.setdefault(len(nombre), []).append(nombre)

for n in sorted(por_longitud):
    print(f"{n} letras:", por_longitud[n])
```

### 5. 10 de cada conjunto — Solución propuesta

```python
print(set(list(nombres_1)[:10]) | set(list(nombres_2)[:10]))
```

#### Mejora (punto 5)

- `list(set)[:10]` es válido, pero el orden es **arbitrario** (no reproducible).
- Para los 10 primeros **alfabéticos**: `sorted(...)[:10]`.
- Guardar el resultado y comprobar `len == 20`.

```python
parte1 = set(sorted(nombres_1)[:10])
parte2 = set(sorted(nombres_2)[:10])
nombres_3 = parte1 | parte2

print("De nombres_1:", sorted(parte1))
print("De nombres_2:", sorted(parte2))
print("Unión:", sorted(nombres_3))
print("Total:", len(nombres_3))
```

### 1. Convertir fechas — Solución propuesta

```python
for k, v in contactos.items():
    v["fecha_de_nacimiento"] = (
        f'{v["fecha_de_nacimiento"][6:10]}-'
        f'{v["fecha_de_nacimiento"][3:5]}-'
        f'{v["fecha_de_nacimiento"][0:2]}'
    )

print(contactos)
```

#### Mejora (punto 1)

- El slicing por posición es frágil (rompe si el formato cambia o se re-ejecuta).
- Mejor: `split("/")` → `date(año, mes, día)` → `isoformat()`.
- Función auxiliar que acepta ambos formatos (no falla al re-ejecutar).

```python
from datetime import date

def a_fecha(cadena: str) -> date:
    if "/" in cadena:
        d, m, a = cadena.split("/")
        return date(int(a), int(m), int(d))
    return date.fromisoformat(cadena)

# Recargar datos originales por si la celda anterior ya convirtió
contactos = {
    "Alejandro Gómez": {
        "fecha_de_nacimiento": "15/04/1992",
        "ciudad": "Madrid",
        "telefono": "+34 611 223 344"
    },
    "María Rodríguez": {
        "fecha_de_nacimiento": "22/08/1988",
        "ciudad": "Barcelona",
        "telefono": "+34 622 334 455"
    },
    "Carlos Martínez": {
        "fecha_de_nacimiento": "05/12/1995",
        "ciudad": "Valencia",
        "telefono": "+34 633 445 566"
    },
    "Ana López": {
        "fecha_de_nacimiento": "30/01/1990",
        "ciudad": "Sevilla",
        "telefono": "+34 644 556 677"
    },
    "David García": {
        "fecha_de_nacimiento": "14/07/2000",
        "ciudad": "Zaragoza",
        "telefono": "+34 655 667 788"
    },
    "Laura Sánchez": {
        "fecha_de_nacimiento": "19/03/1993",
        "ciudad": "Málaga",
        "telefono": "+34 666 778 899"
    },
    "Javier Fernández": {
        "fecha_de_nacimiento": "11/11/1985",
        "ciudad": "Bilbao",
        "telefono": "+34 677 889 900"
    },
    "Elena Pérez": {
        "fecha_de_nacimiento": "25/05/1997",
        "ciudad": "Alicante",
        "telefono": "+34 688 990 011"
    },
    "Diego Ruiz": {
        "fecha_de_nacimiento": "02/09/1991",
        "ciudad": "Valladolid",
        "telefono": "+34 699 001 122"
    },
    "Sofia Torres": {
        "fecha_de_nacimiento": "08/06/1994",
        "ciudad": "Granada",
        "telefono": "+34 600 112 233"
    }
}

for datos in contactos.values():
    datos["fecha_de_nacimiento"] = a_fecha(datos["fecha_de_nacimiento"]).isoformat()

for nombre, datos in contactos.items():
    print(f"{nombre}: {datos['fecha_de_nacimiento']}")
```

### 2. Campo edad — Solución propuesta

```python
import datetime

def calcular_edad(fecha_texto):
    fecha = datetime.date.fromisoformat(fecha_texto)
    hoy = datetime.date.today()
    edad = hoy.year - fecha.year
    if hoy.month < fecha.month:
        edad -= 1
    elif hoy.month == fecha.month:
        if hoy.day <= fecha.day:
            edad -= 1
    return edad

for k, v in contactos.items():
    v["edad"] = calcular_edad(v["fecha_de_nacimiento"])

print(contactos)
```

#### Mejora (punto 2)

- Misma lógica, más compacta: comparar tuplas `(mes, día)`.
- `hoy.day <= fecha.day` resta edad también **el día del cumpleaños**; lo habitual es restar solo si aún **no** ha llegado (`<`).
- Reutilizar `a_fecha`.

```python
hoy = date.today()

for datos in contactos.values():
    nac = a_fecha(datos["fecha_de_nacimiento"])
    edad = hoy.year - nac.year
    if (hoy.month, hoy.day) < (nac.month, nac.day):
        edad -= 1
    datos["edad"] = edad

for nombre, datos in contactos.items():
    print(f"{nombre}: {datos['edad']} años")
```

### 3. Persona más joven — Solución propuesta

```python
import datetime

mas_joven_fecha = datetime.date(1, 1, 1)
mas_joven_nombre = ""

for k, v in contactos.items():
    if mas_joven_fecha < datetime.date.fromisoformat(v["fecha_de_nacimiento"]):
        mas_joven_fecha, mas_joven_nombre = datetime.date.fromisoformat(v["fecha_de_nacimiento"]), k

print(mas_joven_nombre)
```

#### Mejora (punto 3)

- Evitar sentinelas (`date(1,1,1)`) y el bucle manual.
- Con el campo `edad`: `min(..., key=...)`.

```python
mas_joven = min(contactos, key=lambda n: contactos[n]["edad"])
print(f"{mas_joven} ({contactos[mas_joven]['edad']} años)")
```

### 4. Diferencia de edades — Solución propuesta

```python
import datetime

mas_joven_fecha = datetime.date(1, 1, 1)
mas_joven_nombre = ""
mas_viejo_fecha = datetime.date(2100, 1, 1)
mas_viejo_nombre = ""

for k, v in contactos.items():
    f = datetime.date.fromisoformat(v["fecha_de_nacimiento"])
    if mas_joven_fecha < f:
        mas_joven_fecha, mas_joven_nombre = f, k
    if mas_viejo_fecha > f:
        mas_viejo_fecha, mas_viejo_nombre = f, k

print(f"El mas joven es {mas_joven_nombre} que nace en la fecha {mas_joven_fecha}")
print(f"El mas viejo es {mas_viejo_nombre} que nace en la fecha {mas_viejo_fecha}")
print(f"Diferencia {(mas_joven_fecha - mas_viejo_fecha) // 365}")
```

#### Mejora (punto 4)

- `(fecha1 - fecha2) // 365` imprime un `timedelta` confuso; no expresa bien la diferencia de **edades en años**.
- Más claro: `max(edades) - min(edades)` con el campo `edad`.

```python
edades = [d["edad"] for d in contactos.values()]
mas_vieja = max(contactos, key=lambda n: contactos[n]["edad"])
mas_joven = min(contactos, key=lambda n: contactos[n]["edad"])
diferencia = max(edades) - min(edades)

print(f"Más joven: {mas_joven} ({min(edades)} años)")
print(f"Más vieja: {mas_vieja} ({max(edades)} años)")
print(f"Diferencia: {diferencia} años")
```

### 5. Ordenar por edad — Solución propuesta

```python
print({k: v for k, v in sorted(contactos.items(), key=lambda x: x[1]["edad"])})
```

#### Mejora (punto 5)

- La idea (`sorted` + `key` + dict) ya es correcta.
- Asignar a una variable y mostrar una línea por persona.

```python
contactos_ordenados = dict(
    sorted(contactos.items(), key=lambda item: item[1]["edad"])
)

for nombre, datos in contactos_ordenados.items():
    print(f"{datos['edad']:2d} años — {nombre}")
```

### 6. Próximo cumpleaños — Solución propuesta (incorrecta)

```python
# NOTA: la solución original repetía el punto 3 (persona más joven)
# y NO calculaba el próximo cumpleaños.

import datetime

mas_joven_fecha = datetime.date(1, 1, 1)
mas_joven_nombre = ""

for k, v in contactos.items():
    if mas_joven_fecha < datetime.date.fromisoformat(v["fecha_de_nacimiento"]):
        mas_joven_fecha, mas_joven_nombre = datetime.date.fromisoformat(v["fecha_de_nacimiento"]), k

print(mas_joven_nombre)  # responde al punto 3, no al 6
```

#### Mejora (punto 6) — corrección

El enunciado pide a quién le **faltan menos días** para su cumpleaños:

1. Cumpleaños de este año: `replace(year=hoy.year)`.
2. Si ya pasó (`< hoy`), usar el año siguiente.
3. Días: `(cumple - hoy).days`.
4. Quedarse con el mínimo (`min` + `key`).

```python
hoy = date.today()
dias_hasta_cumple = {}

for nombre, datos in contactos.items():
    nac = a_fecha(datos["fecha_de_nacimiento"])
    cumple = nac.replace(year=hoy.year)
    if cumple < hoy:
        cumple = nac.replace(year=hoy.year + 1)
    dias_hasta_cumple[nombre] = (cumple - hoy).days

quien = min(dias_hasta_cumple, key=dias_hasta_cumple.get)
print(f"Hoy: {hoy}")
print(f"Más cercano: {quien} ({dias_hasta_cumple[quien]} días)")
```

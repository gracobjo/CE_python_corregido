## Ejercicio 1 — `filter`

De `nums`, obtén una lista solo con los múltiplos de 3.

```python
nums = list(range(1, 31))

# Escribe tu solución aquí

# Filtrar los múltiplos de 3 usando filter y lambda
multiplos_de_3 = list(filter(lambda x: x % 3 == 0, nums))

print(multiplos_de_3)
# Resultado: [3, 6, 9, 12, 15, 18, 21, 24, 27, 30]
```

## Ejercicio 2 — `sorted`

1. Ordena `palabras` por longitud.
2. Ordena `palabras` alfabéticamente sin distinguir mayúsculas.

```python
palabras = ["kiwi", "Manzana", "uva", "Naranja", "pera"]

# Escribe tu solución aquí
palabras_por_longitud = sorted(palabras, key=len)
palabras_alfabetico = sorted(palabras, key=str.lower)

print(palabras_por_longitud)
print(palabras_alfabetico)
```

## Ejercicio 3 — `map`

Convierte `celsius` a Fahrenheit con `map` (`F = C * 9/5 + 32`) y obtén una lista.

```python
celsius = [0, 20, 37, 100]

# Escribe tu solución aquí
fahrenheit = list(map(lambda c: c * 9 / 5 + 32, celsius))

print(fahrenheit)
```

## Ejercicio 4 — Combinar

Partiendo de `nums`:
1. Filtra los pares.
2. Mapea cada uno a su cuadrado.
3. Ordena el resultado descendente.

```python
nums = [5, 2, 9, 4, 7, 8, 1, 6]

# Escribe tu solución aquí
pares = filter(lambda x: x % 2 == 0, nums)
cuadrados = map(lambda x: x * x, pares)
resultado = sorted(cuadrados, reverse=True)

print(resultado)
```

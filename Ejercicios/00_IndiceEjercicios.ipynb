# Ejercicios del curso

Un cuaderno de práctica por cada notebook de teoría. Celdas de solución vacías para que practiques.

## Cómo usarlos

1. Abre el notebook de **teoría** correspondiente y repasa el apartado.
2. Resuelve el ejercicio en la celda vacía del cuaderno `*_Ejercicios.ipynb`.
3. Si te atascas, vuelve a la teoría (mismo número de archivo).

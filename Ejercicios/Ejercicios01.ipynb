## Ejercicio 1 — Conjuntos

Tenemos dos conjuntos de nombres (ninguno se repite). Se pide:

1. Crear dos conjuntos: nombres que empiezan por **consonante** y por **vocal**.
2. Crear una lista con todos los nombres, ordenada alfabéticamente.
3. A partir de esa lista, un conjunto de tuplas `(nombre, número)` del 1 al N.
4. Obtener listas agrupadas por cantidad de letras del nombre.
5. Crear un tercer conjunto con 20 nombres: 10 de cada conjunto original.

> Nota: el enunciado original menciona 20+20 nombres; los datos de abajo tienen 15+15. La lógica es la misma.

```python
nombres_1 = {
    "Laura", "David", "Sofia", "Javier", "Elena",
    "Manuel", "Carmen", "Jose", "Isabel", "Francisco",
    "Paula", "Antonio", "Lucia", "Diego", "Martina",
}

nombres_2 = {
    "Daniel", "Marta", "Luis", "Alba", "Pablo",
    "Sara", "Adrian", "Julia", "Alvaro", "Irene",
    "Hugo", "Valeria", "Jorge", "Emma", "Ivan",
}

todos = nombres_1 | nombres_2  # unión (sin repetidos)
```

### 1. Nombres por vocal / consonante

```python
vocales = "aeiouáéíóú"
consonantes = set()
vocales_nombres = set()

for nombre in todos:
    if nombre[0].lower() in vocales:
        vocales_nombres.add(nombre)
    else:
        consonantes.add(nombre)

print("Consonante:", consonantes)
print("Vocal:", vocales_nombres)
```

### 2. Lista de todos los nombres ordenada alfabéticamente

```python
lista_ordenada = sorted(todos)
print(lista_ordenada)
```

### 3. Conjunto de tuplas `(nombre, número)`

Los conjuntos admiten tuplas (no listas).

```python
conjunto_tuplas = {
    (nombre, i) for i, nombre in enumerate(lista_ordenada, start=1)
}

print(f"Total: {len(conjunto_tuplas)} tuplas")
print(sorted(conjunto_tuplas, key=lambda t: t[1]))  # ordenado por número para verlo claro
```

### 4. Listas por cantidad de letras

Diccionario `longitud → lista de nombres`. Las claves se muestran ordenadas.

```python
por_longitud = {}
for nombre in lista_ordenada:
    n = len(nombre)
    if n not in por_longitud:
        por_longitud[n] = []
    por_longitud[n].append(nombre)

for longitud in sorted(por_longitud):
    print(f"{longitud} letras:", por_longitud[longitud])
```

### 5. Tercer conjunto: 10 nombres de cada original

Un `set` no tiene índices: se pasa a lista, se hace slicing y se vuelve a conjunto.
Aquí se toman los **10 primeros alfabéticos** de cada uno (resultado reproducible).

```python
parte1 = set(sorted(nombres_1)[:10])
parte2 = set(sorted(nombres_2)[:10])
nombres_3 = parte1 | parte2

print("De nombres_1:", sorted(parte1))
print("De nombres_2:", sorted(parte2))
print("Unión (20):", sorted(nombres_3))
print("Total:", len(nombres_3))
```

## Ejercicio 2 — Fechas y diccionarios

Partiendo del diccionario `contactos`, se pide:

1. Convertir la fecha de `DD/MM/YYYY` a `YYYY-MM-DD`.
2. Crear un nuevo campo con la **edad**.
3. Obtener la persona más joven.
4. Diferencia de edad entre la más joven y la más vieja.
5. Ordenar el diccionario por edad ascendente.
6. Persona a la que le falta menos para su cumpleaños.

```python
from datetime import date

hoy = date.today()


def a_fecha(cadena):
    """Acepta DD/MM/YYYY o YYYY-MM-DD y devuelve un date."""
    if "/" in cadena:
        d, m, a = cadena.split("/")
        return date(int(a), int(m), int(d))
    return date.fromisoformat(cadena)


contactos = {
    "Alejandro Gómez": {
        "fecha_de_nacimiento": "15/04/1992",
        "ciudad": "Madrid",
        "telefono": "+34 611 223 344",
    },
    "María Rodríguez": {
        "fecha_de_nacimiento": "22/08/1988",
        "ciudad": "Barcelona",
        "telefono": "+34 622 334 455",
    },
    "Carlos Martínez": {
        "fecha_de_nacimiento": "05/12/1995",
        "ciudad": "Valencia",
        "telefono": "+34 633 445 566",
    },
    "Ana López": {
        "fecha_de_nacimiento": "30/01/1990",
        "ciudad": "Sevilla",
        "telefono": "+34 644 556 677",
    },
    "David García": {
        "fecha_de_nacimiento": "14/07/2000",
        "ciudad": "Zaragoza",
        "telefono": "+34 655 667 788",
    },
    "Laura Sánchez": {
        "fecha_de_nacimiento": "19/03/1993",
        "ciudad": "Málaga",
        "telefono": "+34 666 778 899",
    },
    "Javier Fernández": {
        "fecha_de_nacimiento": "11/11/1985",
        "ciudad": "Bilbao",
        "telefono": "+34 677 889 900",
    },
    "Elena Pérez": {
        "fecha_de_nacimiento": "25/05/1997",
        "ciudad": "Alicante",
        "telefono": "+34 688 990 011",
    },
    "Diego Ruiz": {
        "fecha_de_nacimiento": "02/09/1991",
        "ciudad": "Valladolid",
        "telefono": "+34 699 001 122",
    },
    "Sofia Torres": {
        "fecha_de_nacimiento": "08/06/1994",
        "ciudad": "Granada",
        "telefono": "+34 600 112 233",
    },
}

print("Hoy:", hoy)
```

### 1. Convertir fechas a `YYYY-MM-DD`

Se puede re-ejecutar sin romper: si ya está en ISO, se deja igual.

```python
for datos in contactos.values():
    f = a_fecha(datos["fecha_de_nacimiento"])
    datos["fecha_de_nacimiento"] = f.isoformat()

for nombre, datos in contactos.items():
    print(f"{nombre} → {datos['fecha_de_nacimiento']}")
```

### 2. Nuevo campo `edad`

Años cumplidos hasta hoy (si aún no ha llegado el cumpleaños, se resta 1).

```python
for datos in contactos.values():
    nac = a_fecha(datos["fecha_de_nacimiento"])
    edad = hoy.year - nac.year
    if (hoy.month, hoy.day) < (nac.month, nac.day):
        edad -= 1
    datos["edad"] = edad

for nombre, datos in contactos.items():
    print(f"{nombre}: {datos['fecha_de_nacimiento']} → {datos['edad']} años")
```

### 3. Persona más joven

```python
mas_joven = min(contactos, key=lambda n: contactos[n]["edad"])
print(
    f"La persona más joven es {mas_joven} "
    f"y tiene {contactos[mas_joven]['edad']} años"
)
```

### 4. Diferencia de edad (más joven ↔ más vieja)

```python
edades = [d["edad"] for d in contactos.values()]
mas_vieja = max(contactos, key=lambda n: contactos[n]["edad"])
diferencia = max(edades) - min(edades)

print(f"Más joven: {mas_joven} ({min(edades)} años)")
print(f"Más vieja: {mas_vieja} ({max(edades)} años)")
print(f"Diferencia: {diferencia} años")
```

### 5. Ordenar el diccionario por edad ascendente

```python
contactos_ordenados = dict(
    sorted(contactos.items(), key=lambda item: item[1]["edad"])
)

for nombre, datos in contactos_ordenados.items():
    print(f"{datos['edad']:2d} años — {nombre}")
```

### 6. Persona a la que le falta menos para el cumpleaños

`replace(year=...)` coloca el cumpleaños en el año actual (o el siguiente si ya pasó).

```python
proximo = {}
for nombre, datos in contactos.items():
    nac = a_fecha(datos["fecha_de_nacimiento"])
    cumple = nac.replace(year=hoy.year)
    if cumple < hoy:
        cumple = nac.replace(year=hoy.year + 1)
    proximo[nombre] = (cumple - hoy).days

quien = min(proximo, key=proximo.get)
print(f"Hoy: {hoy}")
print(f"Más cercano: {quien} ({proximo[quien]} días)")
```

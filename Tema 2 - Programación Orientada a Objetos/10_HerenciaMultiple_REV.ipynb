# Herencia múltiple
La herencia múltiple en Python permite que una clase hija (subclase) herede los atributos y métodos de dos o más clases padre (superclases).

___
**Orden de izquierda a derecha:** En `class Pato(Volador, Nadador)`, Python revisa primero `Volador` y luego `Nadador`.

```python
# 1. Definición de clases base (Padres)
class Volador:
    def __init__(self):
        self.velocidad_aire = 50

    def mover(self):
        return "Estoy volando por el cielo"

class Nadador:
    def __init__(self):
        self.velocidad_agua = 10

    def mover(self):
        return "Estoy nadando en el agua"

# 2. Clase que hereda de múltiples padres (Herencia Múltiple)
class Pato(Volador, Nadador):
    def __init__(self, nombre):
        self.nombre = nombre
        # super() recorre la MRO de izquierda a derecha (Volador -> Nadador -> object)
        super().__init__() # En este caso se ejecuta el __init__ de Volador pero no el de Nadador

    def presentarse(self):
        return f"Hola, soy {self.nombre}."
        # return f"Hola, soy {self.nombre} con velocidad de aire {self.velocidad_aire} y con velocidad en el agua {self.velocidad_agua}."
        # ERROR en la línea anterior puesto que velocidad_agua no se ha creado al no ejecutarse su __init__

# 3. Probando la instancia y el conflicto de métodos
pato_lucas = Pato("Lucas")

print(pato_lucas.presentarse())
# Al invocar 'mover()', Python busca según el orden MRO (primero en Volador)
print(pato_lucas.mover()) 
```

## Llamada explícita al __init__ o a otros métodos de la clase padre desde la clase hija
El objetivo es llamar a todos los init de las clases padre. De lo contrario sólo llama a las que se resuelven por MRO.
Consideraciones:
* Se llama desde la clase hija (no desde fuera)
* Se indica con `Superclase.__init__(self)` sin olvidarnos del parámetro self
* Funciona tanto para `__init__` como para cualquier otro método

```python
class Padre:
    def __init__(self):
        self.apellido_paterno = "García"

    def familia(self):
        return self.apellido_paterno

class Madre:
    def __init__(self):
        self.apellido_materno = "Pérez"

    def familia(self):
        return self.apellido_materno


class Hijo(Padre, Madre):
    def __init__(self, nombre):
        self.nombre = nombre
        
        # super().__init__() # Descomentar e intercambiar orden en la herencia y comprobar que el error se da en la otra superclase
        # Este error realmente se produce más adelante al intentar acceder a un atributo que no ha sido creado
        
        Padre.__init__(self) # Comentar  y descomentar super().__init__()
        Madre.__init__(self) # Comentar y descomentar super().__init__()

    def familia(self):
        # Llamamos al método familia de las dos superclases
        return f"Hola, soy {self.nombre} {Padre.familia(self)} {Madre.familia(self)}."


persona = Hijo("Luis")
print(persona.familia())
```

## Llamada explícita a otros métodos de la clase padre desde fuera
El objetivo es llamar a métodos de las superclases desde fuera.
Consideraciones:
* Se llama desde fuera
* Se indica con `Superclase.metodo(instancia_hija)`
* Se llama en el contexto de la clase que se pasa como parámetro

```python
class Padre:
    def __init__(self):
        self.apellido_paterno = "García"

    def familia(self):
        return self.apellido_paterno

class Madre:
    def __init__(self):
        self.apellido_materno = "Pérez"

    def familia(self):
        return self.apellido_materno


class Hijo(Madre, Padre):
    def __init__(self, nombre):
        self.nombre = nombre
        #super().__init__()
        Padre.__init__(self) # Comentar  y descomentar super().__init__()
        Madre.__init__(self) # Comentar y descomentar super().__init__()

    def familia(self):
        # Llamamos al método familia de las dos superclases
        return f"Hola, soy {self.nombre} {Padre.familia(self)} {Madre.familia(self)}."


persona = Hijo("Luis")
print(persona.familia())

Padre.familia(persona) # Modificación respecto a la celda anterior
```

# Herencia y polimorfismo: métodos de pago

Implementaremos una clase padre `MetodoPago` con los datos comunes de una operación: ordenante, destinatario, cantidad y concepto. Las clases `Tarjeta`, `Bizum` y `Transferencia` heredan esos atributos y sobrescriben `realizar_pago()` para describir cómo se procesaría cada tipo de pago.

> **Demostración educativa:** los pagos se simulan; este código no se conecta a bancos, redes de tarjetas ni Bizum. No introduzcas números de tarjeta, teléfonos ni cuentas reales.

```python
import math


class MetodoPago:
    def __init__(self, ordenante, destinatario, cantidad, concepto):
        if not ordenante or not isinstance(ordenante, str):
            raise ValueError("El ordenante debe ser un nombre no vacío.")
        if not destinatario or not isinstance(destinatario, str):
            raise ValueError("El destinatario debe ser un nombre no vacío.")
        if not concepto or not isinstance(concepto, str):
            raise ValueError("El concepto debe ser un texto no vacío.")
        if isinstance(cantidad, bool) or not isinstance(cantidad, (int, float)):
            raise TypeError("La cantidad debe ser un número.")
        if not math.isfinite(cantidad) or cantidad <= 0:
            raise ValueError("La cantidad debe ser mayor que cero y finita.")

        self.ordenante = ordenante
        self.destinatario = destinatario
        self.cantidad = float(cantidad)
        self.concepto = concepto

    def realizar_pago(self):
        raise NotImplementedError("Cada método de pago debe implementar realizar_pago().")

    def __str__(self):
        return f"{self.ordenante} → {self.destinatario}: {self.cantidad:.2f} € ({self.concepto})"


class Tarjeta(MetodoPago):
    def __init__(self, ordenante, destinatario, cantidad, concepto, ultimos_digitos):
        super().__init__(ordenante, destinatario, cantidad, concepto)
        if not (isinstance(ultimos_digitos, str) and len(ultimos_digitos) == 4 and ultimos_digitos.isdigit()):
            raise ValueError("Indica únicamente los cuatro últimos dígitos de la tarjeta.")
        self.ultimos_digitos = ultimos_digitos

    def realizar_pago(self):
        return (
            f"Pago simulado con tarjeta •••• {self.ultimos_digitos}: "
            f"{self.ordenante} paga {self.cantidad:.2f} € a {self.destinatario} "
            f"por '{self.concepto}'."
        )


class Bizum(MetodoPago):
    def __init__(self, ordenante, destinatario, cantidad, concepto, telefono):
        super().__init__(ordenante, destinatario, cantidad, concepto)
        if not (isinstance(telefono, str) and telefono.isdigit() and len(telefono) >= 3):
            raise ValueError("El teléfono de Bizum debe contener al menos tres dígitos.")
        self.telefono = telefono

    def realizar_pago(self):
        telefono_oculto = "•" * (len(self.telefono) - 2) + self.telefono[-2:]
        return (
            f"Bizum simulado al teléfono {telefono_oculto}: "
            f"{self.ordenante} envía {self.cantidad:.2f} € a {self.destinatario} "
            f"por '{self.concepto}'."
        )


class Transferencia(MetodoPago):
    def __init__(self, ordenante, destinatario, cantidad, concepto, cuenta_ultimos4):
        super().__init__(ordenante, destinatario, cantidad, concepto)
        if not (isinstance(cuenta_ultimos4, str) and len(cuenta_ultimos4) == 4 and cuenta_ultimos4.isalnum()):
            raise ValueError("Indica únicamente los cuatro últimos caracteres de la cuenta.")
        self.cuenta_ultimos4 = cuenta_ultimos4.upper()

    def realizar_pago(self):
        return (
            f"Transferencia simulada a la cuenta •••• {self.cuenta_ultimos4}: "
            f"{self.ordenante} transfiere {self.cantidad:.2f} € a {self.destinatario} "
            f"por '{self.concepto}'."
        )
```

## Ejemplo: una interfaz, tres comportamientos

Creamos un objeto de cada clase hija y los guardamos en la misma lista. El bucle llama a `realizar_pago()` sin tener que comprobar el tipo concreto: Python selecciona la versión sobrescrita correspondiente. Eso es **polimorfismo**.

```python
pagos = [
    Tarjeta("Ana", "Librería Central", 29.90, "Compra de libros", "1234"),
    Bizum("Luis", "Marta", 12.50, "Cena compartida", "600123456"),
    Transferencia("Sofía", "Academia Python", 85.00, "Curso de Python", "A1B2"),
]

for pago in pagos:
    print(pago.realizar_pago())

print("\nLos tres pagos comparten los atributos heredados ordenante, destinatario y cantidad:")
for pago in pagos:
    print(f"- {pago}")
```

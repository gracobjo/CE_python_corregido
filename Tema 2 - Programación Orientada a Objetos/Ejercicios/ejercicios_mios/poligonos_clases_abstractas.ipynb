# Ejercicio: polígonos y clases abstractas

Vamos a modelar varios polígonos usando una clase abstracta. Todos los polígonos deben poder calcular su área, aunque la fórmula dependa de cada figura.

## Enunciado

1. Crea una clase abstracta `Poligono` que herede de `ABC` e importe `abstractmethod` desde `abc`.
2. Declara en `Poligono` el método abstracto `calcular_area()`. No se debe poder crear directamente un objeto `Poligono`.
3. Implementa estas clases concretas:
   - `Triangulo`: recibe la base y la altura. Su área es `base * altura / 2`.
   - `Rectangulo`: recibe la base y la altura. Su área es `base * altura`.
   - `PentagonoRegular`: recibe la longitud de un lado y la apotema. Su área es `perimetro * apotema / 2`, donde el perímetro es `5 * lado`.
4. Las medidas deben ser números positivos y finitos.
5. Comprueba las áreas y crea una lista de polígonos para mostrar el uso polimórfico del método `calcular_area()`.

**Nota:** se considera un pentágono regular. Para calcular el área de un pentágono irregular harían falta otros datos, por ejemplo, las coordenadas de sus vértices.

## Solución: clase abstracta

`Poligono` define el comportamiento común: todas sus subclases deben proporcionar `calcular_area()`. También centralizamos la validación de las medidas para reutilizarla.

```python
from abc import ABC, abstractmethod
import math


class Poligono(ABC):
    @abstractmethod
    def calcular_area(self):
        """Devuelve el área del polígono."""
        raise NotImplementedError

    @staticmethod
    def _validar_medida(nombre, valor):
        if isinstance(valor, bool) or not isinstance(valor, (int, float)):
            raise TypeError(f"{nombre} debe ser un número.")
        if not math.isfinite(valor) or valor <= 0:
            raise ValueError(f"{nombre} debe ser positivo y finito.")
```

## Solución: figuras concretas

Cada clase guarda sus medidas y redefine `calcular_area()` con la fórmula correspondiente.

```python
class Triangulo(Poligono):
    def __init__(self, base, altura):
        self._validar_medida("La base", base)
        self._validar_medida("La altura", altura)
        self.base = base
        self.altura = altura

    def calcular_area(self):
        return self.base * self.altura / 2


class Rectangulo(Poligono):
    def __init__(self, base, altura):
        self._validar_medida("La base", base)
        self._validar_medida("La altura", altura)
        self.base = base
        self.altura = altura

    def calcular_area(self):
        return self.base * self.altura


class PentagonoRegular(Poligono):
    def __init__(self, lado, apotema):
        self._validar_medida("El lado", lado)
        self._validar_medida("La apotema", apotema)
        self.lado = lado
        self.apotema = apotema

    def calcular_area(self):
        perimetro = 5 * self.lado
        return perimetro * self.apotema / 2
```

## Comprobaciones y polimorfismo

Los `assert` comprueban las fórmulas y la validación. Al recorrer la lista, llamamos al mismo método en figuras distintas y cada objeto aplica su propia fórmula.

```python
triangulo = Triangulo(10, 6)
rectangulo = Rectangulo(4, 3)
pentagono = PentagonoRegular(2, 1.376)

assert triangulo.calcular_area() == 30
assert rectangulo.calcular_area() == 12
assert math.isclose(pentagono.calcular_area(), 6.88)

poligonos = [triangulo, rectangulo, pentagono]
for poligono in poligonos:
    print(f"{type(poligono).__name__}: área = {poligono.calcular_area()}")
```

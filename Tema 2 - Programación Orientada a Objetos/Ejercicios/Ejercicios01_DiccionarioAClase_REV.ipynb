# Ejercicios 01
Crear un programa en Python que a partir de un diccionario de atributos, cree objetos de una clase conteniendo dichos atributos.

Cada atributo almacena:
* **tipo**: Si es atributo de clase o de objeto
* **valor**: Un valor por defecto

Ejemplo de diccionario:
```Python
clase_diccionario = {
    "nombre":{
        "tipo":"atributo_clase",
        "valor":"generico"
    },
    "edad":{
        "tipo":"atributo_objeto",
        "valor":50
    }
}
```

````python
clase_diccionario = {
    "nombre":{
        "tipo":"atributo_clase",
        "valor":"generico"
    },
    "edad":{
        "tipo":"atributo_objeto",
        "valor":50
    }
}

class MiClase:
    pass

mc = MiClase()

for k,v in clase_diccionario.items():
    if v["tipo"] == "atributo_clase":
        setattr(MiClase,k,v["valor"])
    if v["tipo"] == "atributo_objeto":
        setattr(mc,k,v["valor"])

print(mc.nombre,mc.edad)

dir(mc)
````

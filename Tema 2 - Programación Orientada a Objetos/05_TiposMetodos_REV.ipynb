# Tipos de métodos en Python

* **Métodos de instancia**: Son los métodos tradicionales que se ejecutan en el contexto de un objeto ya instanciado. Reciben el parámetro `self` para hacer referencia dentro del método a la propia instancia.
* **Métodos de clase**: Un método de clase en Python es un método que opera a nivel de clase en lugar de operar a nivel de objeto o instancia. Consideraciones a tener en cuenta:
    * Utiliza el decorador `@classmethod`
    * Recibe la propia clase como primer parámetro (llamado por convención `cls`) en lugar de una instancia (`self`).
    * Se usa para:
        1. Modificar el estado de la clase
        2. Ofrecer constructores alternativos
* **Métodos estáticos**: Es un método que se define dentro de una clase pero no necesita acceder ni a la instancia ni a la clase en sí. Cuestiones a tener en cuenta:
    *  Se emplea anteponiendo el decorador `@staticmethod`
    *  No recibe ni `self` ni `cls`
    *  Funciona como una función normal aunque se agrupe dentro de una clase
    *  Se puede llamar tanto en el contexto de la instancia como en el contexto de la clase

___
## Métodos de instancia

```python
class Persona:
    def __init__(self, nombre, edad, logro):
        # Atributos de instancia
        self.nombre = nombre
        self.edad = edad
        self.logros = []

    # Ejemplo de método de instancia tradicional
    def presentarse(self):
        return f"Hola, me llamo {self.nombre} y tengo {self.edad} años y he logrado {self.logros}."

    # Otro método de instancia que modifica un atributo
    def cumplir_anos(self):
        self.edad += 1
        return f"¡Felicidades! Ahora tienes {self.edad} años."

    def lograr_algo(self,logro):
        self.logros.append(logro)

# 1. Crear una instancia (objeto) de la clase Persona
usuario = Persona("Carlos", 28,[])

# 2. Llamar a los métodos de instancia
print(usuario.cumplir_anos())  # Output: ¡Felicidades! Ahora tienes 29 años.
usuario.lograr_algo("Bachillerato")
usuario.lograr_algo("Carnet conducir")
print(usuario.presentarse())   # Output: Hola, me llamo Carlos y tengo 28 años.
```

## Métodos de clase

```python
class Coche:
  uso = "mobilidad"  # Atributo de clase

  def __init__(self, marca, modelo):
    self.marca = marca
    self.modelo = modelo

  @classmethod
  def cambiar_uso(cls, nuevo_uso):
    cls.uso = nuevo_uso  # Modifica el atributo de clase

  @classmethod
  def desde_texto(cls, mar, mod):
    # Constructor alternativo
    marca, modelo = mar, mod
    return cls(marca, modelo)

# Uso del método de clase para cambiar un atributo de la clase entera
Coche.cambiar_uso("competición")
print(Coche.uso)  # Salida: Homo Sapiens

# Uso del método de clase como constructor alternativo
c1 = Coche.desde_texto("seat","ibiza")
print(c1.marca, c1.modelo, c1.uso, sep="-----")

# Uso de __init__ como constructor normal
c2 = Coche("renault","clio")
print(c2.marca, c2.modelo, c2.uso, sep="-----")
```

## Métodos estáticos

```python
class Utilidades:
    @staticmethod
    def sumar(a, b):
        return a + b

# Llamada desde la clase (sin crear un objeto)
resultado = Utilidades.sumar(3, 5)
print(resultado)  # Salida: 8

# Llamada creando el objeto
mi_utilidad = Utilidades()
resultado = mi_utilidad.sumar(2,4)
print(resultado)  # Salida: 6
```

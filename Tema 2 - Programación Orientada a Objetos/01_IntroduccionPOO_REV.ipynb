# Programación Orientada a Objetos (POO) en Python

Este notebook es una guía práctica para entender los pilares fundamentales de la **Programación Orientada a Objetos (POO)** en Python. 

La POO es un paradigma de programación que utiliza **clases** y **objetos** para estructurar el software de forma modular, reutilizable y fácil de mantener.

---

## 1. Clases y Objetos
* **Clase:** Es el "molde" o plantilla para crear objetos. Define los atributos y métodos comunes.
* **Objeto:** Es la instancia real de una clase.
* **Constructor (`__init__`):** Método especial que se ejecuta automáticamente al crear un objeto para inicializar sus atributos.


```python
# Definición de una clase básica
class Perro:
    # El método constructor
    def __init__(self, nombre, raza):
        self.nombre = nombre  # Atributo de instancia
        self.raza = raza      # Atributo de instancia

    # Un método (función dentro de una clase)
    def ladrar(self):
        return f"¡Guau! Me llamo {self.nombre}."

# Instanciar objetos de la clase Perro
mi_perro = Perro("Tobbie", "Labrador")
otro_perro = Perro("Luna", "Pastor alemán")

# Acceder a atributos y métodos
print(f"Mi perro es un {mi_perro.raza} llamado {mi_perro.nombre}.")
print(mi_perro.ladrar())
print(otro_perro.ladrar())
```

___
## 2. Los 4 conceptos fundamentales de la POO
1. **Herencia** para reutilizar código.
2. **Polimorfismo** para dar múltiples formas a un mismo método.
3. **Encapsulamiento** para proteger los datos con atributos privados.
4. **Abstracción** para definir interfaces y plantillas obligatorias.


___
### Concepto 1: Herencia
Permite que una clase nueva (clase hija o subclase) adopte los atributos y métodos de una clase existente (clase padre o superclase). Evita la duplicación de código.

```python
# Clase Padre
class Animal:
    def __init__(self, nombre):
        self.nombre = nombre

    def comer(self):
        return f"{self.nombre} está comiendo."

# Clase Hija que hereda de Animal
class Gato(Animal):
    def maullar(self):
        return f"{self.nombre} dice: ¡Miau!"

# Probando la herencia
mi_gato = Gato("Michi")
print(mi_gato.comer())   # Método heredado de la clase Animal
print(mi_gato.maullar()) # Método propio de la clase Gato
```

___
### Concepto 2: Polimorfismo
Es la capacidad de que diferentes clases tengan métodos con el **mismo nombre** pero con comportamientos distintos.


```python
class Leon:
    def hacer_sonido(self):
        return "¡Rugido!"

class Vaca:
    def hacer_sonido(self):
        return "¡Muuu!"

# Función que demuestra el polimorfismo
def emitir_sonido_animal(animal):
    print(animal.hacer_sonido())

# Instancias
simba = Leon()
lola = Vaca()

# El mismo método produce resultados diferentes según el objeto
emitir_sonido_animal(simba)
emitir_sonido_animal(lola)
```

___
### Pilar 3: Encapsulamiento
Consiste en **restringir el acceso directo** a los atributos y métodos de un objeto para evitar modificaciones accidentales. En Python se logra usando guiones bajos:
* `_atributo`: Un guion bajo. Privado por convención (sugiere no modificarlo desde fuera).
* `__atributo`: Dos guiones bajos. Privado estricto (activa el *Name Mangling* para protegerlo).

```python
class CuentaBancaria:
    def __init__(self, iban, titular, saldo_inicial):
        self.iban = iban
        self._titular = titular
        self.__saldo = saldo_inicial # Atributo privado

    # Getter: Método para obtener el saldo de forma segura
    def obtener_saldo(self):
        return self.__saldo

    # Setter: Método para modificar el saldo con validación
    def depositar(self, cantidad):
        if cantidad > 0:
            self.__saldo += cantidad
            print(f"Depósito exitoso. Nuevo saldo: ${self.__saldo}")
        else:
            print("La cantidad debe ser mayor a 0.")

# Probando el encapsulamiento
cuenta = CuentaBancaria("ES11 1111 1111 1111 1111","Carlos", 1000)

print(cuenta.iban)
print(cuenta._titular)
# Intentar acceder directamente al atributo privado causará un error:
try:
    print(cuenta.__saldo) # Descomentar esta línea dará AttributeError
except AttributeError as ae:
    print(f"Error: {ae}")

# Acceso correcto mediante métodos
print(f"Saldo actual de {cuenta._titular}: ${cuenta.obtener_saldo()}")
cuenta.depositar(500)
```

### Pilar 4: Abstracción
Oculta los detalles de implementación complejos y solo muestra la funcionalidad esencial al usuario. En Python se implementa utilizando el módulo nativo `abc` (Abstract Base Classes) para crear **clases abstractas** que no se pueden instanciar directamente.


```python
from abc import ABC, abstractmethod

# Clase Abstracta (Molde estricto)
class Vehiculo(ABC):
    @abstractmethod
    def arrancar(self):
        pass

# Clase Hija obligada a implementar el método abstracto
class Moto(Vehiculo):
    def arrancar(self):
        return "La moto ha encendido. ¡Rum-rum!"

# mi_vehiculo = Vehiculo() # Error: No se puede instanciar una clase abstracta

mi_moto = Moto()
print(mi_moto.arrancar())
```

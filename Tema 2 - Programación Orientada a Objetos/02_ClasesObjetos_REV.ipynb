# Clases y objetos
* **Clase:** Es un "molde" o plano arquitectónico para crear cosas. Define las características y comportamientos que tendrán los elementos creados con ese molde.
* **Objeto:** Es la instancia real creada a partir de la clase. Es el elemento físico con datos propios.

**Ejemplo:** 
La clase es el plano de una casa (diseño). El objeto es la casa física construida en una calle específica (con su propio color, dueño, etc.).

___
## 1. Creación de clases

Creamos la clase o "molde" para un `Perro`. Definimos todas las características y acciones que puede hacer un perro:
* **Características**: Son los atributos.
* **Acciones**: Son los métodos.

````python
# Definimos la clase 'Perro'
class Perro:
    # El método __init__ es el constructor. Se ejecuta automáticamente al crear un objeto.
    # 'self' hace referencia al objeto específico que estamos creando.
    def __init__(self, nombre, raza, edad):
        # Atributos (Características del objeto)
        self.nombre = nombre
        self.raza = raza
        self.edad = edad
        
    # Métodos (Funciones/Comportamientos del objeto)
    def ladrar(self):
        return f"¡Guau! Me llamo {self.nombre}."
    
    def cumplir_anos(self):
        self.edad += 1
        return f"¡Feliz cumpleaños {self.nombre}! Ahora tienes {self.edad} años."
````

___
## 2. Creación de objetos a partir de la clase (Instanciación)

Ahora que tenemos el "molde" (`Perro`), podemos crear tantos perros individuales (objetos) como queramos. Cada uno tendrá su propio nombre, raza y edad de forma independiente. Cada perro individual es una *`instanciación`* de la clase a partir de la que se crea.


````python
# Creamos dos objetos (instancias) de la clase Perro
perro_uno = Perro(nombre="Fido", raza="Labrador", edad=3)
perro_dos = Perro(nombre="Luna", raza="Poodle", edad=5)
perro_tres = Perro("Trueno","Salchicha",6)

# Acceder a los atributos de los objetos
print(f"El primer perro se llama {perro_uno.nombre} y es un {perro_uno.raza}.")
print(f"El segundo perro se llama {perro_dos.nombre} y tiene {perro_dos.edad} años.\n")
print(f"El tercer perro se llama {perro_tres.nombre} y tiene {perro_tres.edad} años.\n")

# Llamar a los métodos de los objetos
print(perro_uno.ladrar())
print(perro_dos.ladrar())
print(perro_tres.ladrar())
````

___
## 3. Acceder a los atributos y métodos de un objeto

Se emplea el operados punto ` . ` entre el objeto y el atributo o método.
```Python
objeto.atributo
objeto.metodo(parámetros)
```

````python
# Accedemos a la propiedad edad
print(f"Edad inicial de Luna: {perro_dos.edad}")

# Ejecutamos el método cumplir_ano
print(perro_dos.cumplir_anos())
````

___
## 4. Modificar el estado de un objeto

Los objetos no son estáticos. Podemos alterar sus atributos directamente o a través de sus propios métodos (lo cual es una mejor práctica).

````python
# Hacemos que cumpla años usando su método
print(perro_dos.cumplir_anos())

# Hacemos que cumpla años accediendo directamente desde fuera de la clase a un atributo (no recomendado)
perro_dos.edad += 1

# Comprobamos que el perro_uno (Fido) no ha cambiado su edad
print(f"Edad de Fido sigue siendo: {perro_uno.edad}")
````

## 5. El método `__init()__`

Es un método que se ejecuta siempre que se crea un objeto.
Funciones principales:
* Asignar valores a los atributos de la clase
* Realizar operaciones necesarias cuando se crea un objeto.

````python
class Coche:
    def __init__(self,marca,modelo):
        self.marca = marca
        self.modelo = modelo
    
    def mostrar_datos(self):
        print(f"El coche es de la marca {self.marca} y el modelo es el {self.modelo}")

coche1 = Coche("seat","ibiza")
coche1.mostrar_datos()
````

Sin el método `__init()__` habría que ponerlos **a mano** pero **no se recomienda**.

````python
class Moto:
    def mostrar_datos(self):
        print(f"El coche es de la marca {self.marca} y el precio es {self.precio}")

moto1 = Moto()
moto1.marca = "Honda"
moto1.precio = 15000
moto1.mostrar_datos()

moto2 = Moto()
# moto2.mostrar_datos() # ERROR: No reconoce los atributos

# CONCLUSIÓN: Poner los atributos en la construcción del objeto
````

El método __init()__ puede recibir valores por defecto.

````python
class Bicicleta:
    def __init__(self,marca,precio=1000):
        self.marca = marca
        self.precio = precio

    def mostrar_datos(self):
        print(f"La bicicleta es de la marca {self.marca} y el precio es {self.precio}")

bicicleta1 = Bicicleta("BH")
bicicleta1.mostrar_datos()
````

El método __init()__ **no** puede devolver valores, es decir, **no puede tener un return** salvo que sea **None**.

````python
class Bicicleta:
    def __init__(self,marca,precio=1000):
        self.marca = marca
        self.precio = precio
        # return self.marca # ERROR: No puede tener un return
        return None

bicicleta1 = Bicicleta("BH")
````

## 6. El parámetro `self`

* Es un parámetro que hace referencia al objeto o instancia actual dentro de la clase.
* Dentro de la clase accedemos a los atributos con `self.` y a continuación el atributo.
* Se usa para acceder a los atributos y métodos que pertenecen al objeto.
* No es un error sintáctico no ponerlo pero al usar los métodos, estos fallarán si no se pone tanto en el __init()__ como en el resto de métodos.

````python
# Por sí sólo no falla pero cuando se use sí fallará
class Bicicleta:
    def __init__(marca,precio=1000):
        self.marca = marca
        self.precio = precio

    def mostrar_datos():
        print(f"Es una bicicleta") 
````

````python
class Bicicleta:
    def __init__(marca,precio=1000): # Fallará
        self.marca = marca
        self.precio = precio

    def mostrar_datos():
        print(f"Es una bicicleta") 

# bicicleta1 = Bicicleta("Orbea",500) # ERROR
# bicicleta1.mostrar_datos() # ERROR: Fallará al ejecutarse aunque el propio método no llame a self
````

No tiene por qué llamarse self en particular. Es una simple convención.

````python
class Bicicleta:
    def __init__(mismo,marca,precio=1000):
        mismo.marca = marca
        mismo.precio = precio

    def mostrar_datos(mismo):
        print(f"La bicicleta es de la marca {mismo.marca} y el precio es {mismo.precio}") 

bicicleta1 = Bicicleta("Orbea",500)
bicicleta1.mostrar_datos()
````

## 7. Eliminación de objetos con `del`

Podemos elminar un objeto usando **del**.

````python
class Bicicleta:
    def __init__(self,marca,precio=1000): # Fallará
        self.marca = marca
        self.precio = precio

    def mostrar_datos(self):
        print(f"La bicicleta es de la marca {self.marca} y el precio es {self.precio}") 

bicicleta1 = Bicicleta("Orbea",500)
bicicleta1.mostrar_datos()

del bicicleta1

# bicicleta1.mostrar_datos() # ERROR: No está definida puesto que se ha elimindao
````

## 8. Atributos de clase y atributos de objeto (instancia)
* Un atributo de un objeto es particular para cada objeto o instancia de esa clase.
  * Se definen dentro de __init()__
  * Cada instancia tendrá un valor concreto de dicho atributo
* Un atributo de clase es común a todos los objetos derivados de dicha clase.
  * Se definen fuera de __init()__
  * Todas las instancias tendrán el mismo valor concreto de dicho atributo

````python
class Autobus:
    uso = "Mobilidad general" # Atributo de clase

    def __init__(self,matricula,capacidad): # Atributos de objeto
        self.matricula = matricula
        self.capacidad = capacidad

autobus1 = Autobus("1111AAA",50)
autobus2 = Autobus("2222BBB",60)

print(autobus1.matricula,autobus1.capacidad,autobus1.uso) # Aunque se puede acceder a 'uso' con la instancia, poner Autobus.uso
print(autobus2.matricula,autobus2.capacidad,autobus2.uso) # Aunque se puede acceder a 'uso' con la instancia, poner Autobus.uso
print(Autobus.uso)
````

* Los atributos de clase los modifico en el *contexto* de la clase.
  * `clase.atributo`
* Los atributos de objeto los modifico en el *conteto* del objeto.
  * `objeto.atributo`

````python
class Autobus:
    uso = "Mobilidad general" # Atributo de clase

    def __init__(self,matricula,capacidad): # Atributos de objeto
        self.matricula = matricula
        self.capacidad = capacidad

autobus1 = Autobus("1111AAA",50)
autobus2 = Autobus("2222BBB",60)

#Modificaciones
Autobus.uso = "Transporte escolar"
autobus1.capacidad = 70

print(autobus1.matricula,autobus1.capacidad,autobus1.uso)
print(autobus2.matricula,autobus2.capacidad,autobus2.uso)
print(Autobus.uso)
````

# Encapsulamiento en Python

El **encapsulamiento** es uno de los cuatro pilares fundamentales de la Programación Orientada a Objetos (POO), junto con la herencia, el polimorfismo y la abstracción. 

Consiste en **agrupar los datos (atributos) y los métodos (funciones) que operan sobre esos datos en una sola unidad (la clase)**, y restringir el acceso directo a algunos de los componentes del objeto. 

La propia definición de una clase está proporcionando un cierto nivel de encapsulamiento por el hecho de definir los atributos y los métodos dentro de la misma clase.

Esto evita la modificación accidental de los datos y protege la integridad de la aplicación.

## Conceptos Clave
1. **Niveles de acceso (Público, Protegido, Privado):** A diferencia de otros lenguajes como Java o C++, Python no tiene palabras clave como `private` o `public`. En su lugar, utiliza una **convención de nombres** basada en guiones bajos (`_`).
2. **Métodos de acceso (Getters y Setters):** Métodos públicos utilizados para leer y modificar de forma segura los atributos restringidos.
3. **El decorador `@property`:** La forma idiomática (o "pythonica") de implementar Getters y Setters.

## Resumen Gráfico de Niveles de Acceso
| Sintaxis | Tipo de Acceso | ¿Se puede acceder desde fuera? | Propósito |
|---|---|---|---|
| `objeto.atributo` | **Público** | Sí | Uso general sin restricciones. |
| `objeto._atributo` | **Protegido** | Sí (Por convención no se debe) | Uso interno de la clase y subclases. |
| `objeto.__atributo` | **Privado** | No (Genera `AttributeError`) | Uso exclusivo y estricto de la clase actual. |


## Atributos y Métodos Públicos
Por defecto, todos los miembros de una clase en Python son **públicos**. Esto significa que se puede acceder a ellos y modificarlos desde cualquier lugar fuera de la clase.


```python
class CuentaBancariaPublica:
    def __init__(self, titular, saldo):
        self.titular = titular  # Atributo público
        self.saldo = saldo      # Atributo público

    def mostrar_info(self):    # Método público
        return f"Titular: {self.titular}, Saldo: ${self.saldo}"

# Uso de la clase pública
cuenta = CuentaBancariaPublica("Ana López", 1000)
print(cuenta.mostrar_info())

# Al ser público, cualquiera puede modificar el saldo directamente de forma errónea
cuenta.saldo = -5000  # Modificación no deseada o sin validar
print("Saldo modificado directamente:", cuenta.saldo)
```

## Atributos y Métodos Protegidos (`_`)
Se definen anteponiendo **un guion bajo (`_`)** al nombre del miembro. 

* **Significado:** Es un aviso para otros desarrolladores que indica: *"Este atributo es interno, no deberías acceder a él desde fuera de la clase o de sus subclases"*.
* **Realidad en Python:** Python **no prohíbe** el acceso. Es puramente una convención.

```python
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self._edad = edad  # Atributo protegido

# Instanciación
persona = Persona("Carlos", 30)

# Python permite el acceso, pero el editor (IDE) te advertirá que no deberías hacerlo
print(f"Accediendo al atributo protegido: {persona._edad}")
```

## Atributos y Métodos Privados (`__`)
Se definen anteponiendo **dos guiones bajos (`__`)** al nombre del miembro.

Aquí Python sí aplica un mecanismo llamado **Name Mangling** (Sustitución de nombres). Transforma internamente el nombre del atributo añadiéndole el prefijo `_NombreDeLaClase`, dificultando el acceso directo desde el exterior.

```python
class CuentaBancariaPrivada:
    def __init__(self, titular, saldo_inicial):
        self.titular = titular
        self.__saldo = saldo_inicial  # Atributo estrictamente privado

    def __calcular_interes(self):    # Método privado
        return self.__saldo * 0.02

    def mostrar_saldo_real(self):
        # Los métodos de la misma clase sí pueden acceder a lo privado
        return self.__saldo + self.__calcular_interes()

cuenta_privada = CuentaBancariaPrivada("Pedro", 5000)

# Intentar acceder directamente provocará un error de atributo (AttributeError)
try:
    print(cuenta_privada.__saldo)
except AttributeError as e:
    print(f"Error esperado: {e}")
```

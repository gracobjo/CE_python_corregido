# Nane mangling en Python

En Python el `Name mangling`se aplica a atributos y métodos que empiezan con dos subrayados (underscores) pero que no acaban con dos subrayados como ocurre con otros métodos.

Python automáticamente renombra estos identificadores internamente de tal forma que la transformación se hace por el intérprete, no por el programador.

El nombre se transforma al siguiente formato:

`_NombreClase__identificador`

```python
class Persona:
    def __init__(self, nombre):
        self.__nombre = nombre

    def mostrar(self):
        print(self.__nombre)

p = Persona("Juan")
p.mostrar()
#print(p.__nombre) # ERROR: Te indica que Persona no tiene el atributo __nombre
```

___
### Explicación
* El atributo `__nombre` se ha cambiado internamente a `_Persona__nombre`
* Acceder a `__nombre` desde fuera produce un error

```python
class Persona:
    def __init__(self, nombre):
        self.__nombre = nombre

    def mostrar(self):
        print(self.__nombre)

p = Persona("Juan")
print(dir(p))

print("=====")

print(p._Persona__nombre) # NO DA ERROR PORQUE SE HA TRANSFORMADO INTERNAMENTE
```

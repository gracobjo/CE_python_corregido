# Atributos especiales
* `__name__`: Nombre de la clase
* `__class__`: Clase a la que pertenece el objeto
* `__dict__`: Espacio de nombres del objeto
* `__module__`: Módulo al que pertenece la clase


```python
import modulocoche

class Perro:
    # El método __init__ es el constructor. Se ejecuta automáticamente al crear un objeto.
    # 'self' hace referencia al objeto específico que estamos creando.
    def __init__(self, nombre, raza, edad):
        # Atributos (Características del objeto)
        self.nombre = nombre
        self.raza = raza
        self.edad = edad
        
    # Métodos (Funciones/Comportamientos del objeto)
    def ladrar(self):
        return f"¡Guau! Me llamo {self.nombre}."
    
    def cumplir_anos(self):
        self.edad += 1
        return f"¡Feliz cumpleaños {self.nombre}! Ahora tienes {self.edad} años."

perro1 = Perro("Trueno","Pastor alemán",5)

print(f"Atributo Clase.__name__: {Perro.__name__}") # Nombre de la clase a la que pertenece el objeto
# print(f"Atributo Objeto.__name__: {perro1.__name__}") # ERROR: No se puede usar en el contexto del objeto

print("=====")

print(f"Atributo Clase.__class__: {Perro.__class__}") # Devuelve type. No tiene sentido usarlo en el contexto de la clase
print(f"Atributo Objeto.__class__: {type(perro1.__class__)}") # Igual que el anterior
print(f"Atributo Objeto.__class__: {perro1.__class__}") # Clase a la que pertenece el objeto

print("=====")

print(f"Atributo Clase.__dict__: {Perro.__dict__}") # Espacio de nombres de la clase incluyendo atributos, métodos, etc.
print(f"Atributo Objeto.__dict__: {perro1.__dict__}") # Espacio de nombres del objeto. Contiene únicamente atributos de ese objeto
# siendo un diccionario mutable por lo que podriamos añadir atributos modificando este diccionario
perro1.__dict__["peso"] = 20
print(f"Atributo Objeto.__dict__: {perro1.__dict__}")
print(perro1.peso)

print("=====")

print(f"Atributo Clase.__module__: {Perro.__module__}")
print(f"Atributo Objeto.__module__: {perro1.__module__}")

print("=====")

# Empleamos un módulo de otro fichero que importamos
coche1 = modulocoche.Coche("seat","ibiza")
print(f"Atributo Clase.__module__: {modulocoche.Coche.__module__}")
print(f"Atributo Objeto.__module__: {coche1.__module__}")
```

## Atributo __bases__
Nos devuelve las clases directamente superiores en la jerarquía de clases.

```python
class Volador:
    def __init__(self):
        self.velocidad_aire = 50

    def mover(self):
        return "Estoy volando por el cielo"

class Nadador:
    def __init__(self):
        self.velocidad_agua = 10

    def mover(self):
        return "Estoy nadando en el agua"

class Pato(Volador, Nadador):
    def __init__(self, nombre):
        self.nombre = nombre
        super().__init__()

    def presentarse(self):
        return f"Hola, soy {self.nombre}."

pato1 = Pato("Lucas")

print(f"Atributo Clase.__bases__: {Pato.__bases__}")
# print(f"Atributo Objeto.__bases__: {pato1.__bases__}") # ERROR: No se puede usar en el contexto del objeto

print("=====")
for i in Pato.__bases__:
    print(f"{i.__name__}")
```

```python
# El atributo __bases__ sólo da la jerarquía superior directa
class A:
    pass

class B(A):
    pass

class C(B):
    pass
    
print(C.__bases__)
```

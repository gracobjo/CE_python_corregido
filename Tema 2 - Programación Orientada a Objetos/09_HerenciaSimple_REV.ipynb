# Herencia simple
La herencia permite que una clase hija (**subclase**) reutilice los atributos y métodos de una clase padre (**superclase**).

Conceptos importantes:
* `super()`
* `__init()__`
* Sobreescritura de métodos

```python
# Definimos la clase padre (superclase)
class Animal:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def comer(self):
        return f"{self.nombre} está comiendo."

    def dormir(self):
        return f"{self.nombre} está durmiendo."

# Definimos la clase hija que hereda de Animal
class Perro(Animal):
    pass

# Creamos una instancia de la clase hija
mi_perro = Perro("Tobi", 3)

print(mi_perro.comer())   # Heredado de Animal
print(mi_perro.dormir())  # Heredado de Animal
```

___
## La función `super()`
La función `super()` en Python permite llamar a métodos y constructores de una clase padre (o superclase) desde una clase hija (o subclase) sin tener que escribir su nombre de forma explícita.

¿Para qué sirve?

* **Evita repetir código**: Puedes reutilizar lo que ya hace el padre y solo agregar o modificar lo nuevo en el hijo.
* **Mantenimiento fácil**: Si cambias el nombre de la clase padre, no tienes que actualizar cada llamada manual dentro de la clase hija.
* **Orden de herencia (MRO)**: En herencia múltiple, super() se encarga de seguir el orden correcto de resolución de métodos (Method Resolution Order).

Utilizamos `super()` para llamar al constructor de la clase padre e inicializar sus atributos, además de agregar características propias de la clase hija.

```python
class Perro(Animal):
    def __init__(self, nombre, edad, raza):
        # Llamamos al constructor de la clase padre (Animal)
        super().__init__(nombre, edad)
        self.raza = raza

    # Método propio de la clase Perro
    def ladrar(self):
        return f"{self.nombre} dice: ¡Guau!"

# Instanciamos el nuevo perro
perro_1 = Perro("Bobby", 5, "Labrador")

print(f"Nombre: {perro_1.nombre}, Edad: {perro_1.edad}, Raza: {perro_1.raza}")
print(perro_1.comer())
print(perro_1.ladrar())
```

Una clase hija puede modificar o reemplazar un método que viene de la clase padre (**sobreescritura de métodos**).

```python
class Gato(Animal):
    def __init__(self, nombre, edad, vidas=7):
        super().__init__(nombre, edad)
        self.vidas = vidas

    # Sobrescribimos el método comer
    def comer(self):
        return f"{self.cat_sound()} {self.nombre} come pescado."

    def cat_sound(self):
        return "¡Miau!"

mi_gato = Gato("Michi", 2)
print(mi_gato.comer())
```

___
Python incluye funciones integradas como `isinstance()` e `issubclass()` para verificar las relaciones de herencia

```python
print("¿mi_gato es instancia de Gato?:", isinstance(mi_gato, Gato))
print("¿mi_gato es instancia de Animal?:", isinstance(mi_gato, Animal))
print("¿Gato es subclase de Animal?:", issubclass(Gato, Animal))
print("¿Perro es subclase de Gato?:", issubclass(Perro, Gato))
```

## El uso de `__init()__`
__init__ en la clase hija (subclase), este sobrescribe por completo al método __init__ de la clase padre. El intérprete no ejecutará el inicializador del padre a menos que se llame de forma explícita.

El método __init()__ de la clase padre no se llama automáticamente sino que debe llamarse desde la clase hija usando `super.__init()__`

```python
class Animal:
    def __init__(self, nombre):
        self.nombre = nombre
        print(f"Animal creado: {self.nombre}")


class Perro(Animal):
    def __init__(self, nombre, raza):
        # Llamamos al __init__ de la clase padre (Animal)
        super().__init__(nombre)
        self.raza = raza
        print(f"Perro de raza: {self.raza}")


# Creamos un objeto
mi_perro = Perro("Relámpago", "Labrador")
```

Una excepción a lo anterior es donde la clase hija no sobreescribe **__init()__** y en consecuencia se llama el **__init()__** de la superclase de forma automática.


Resumen de herencia y uso de **`__init__`**:

* **Si la clase hija no define su propio `__init__`**: Python hereda el __init__ de la superclase y sí lo ejecuta automáticamente al crear una instancia de la clase hija.
* **Si la clase hija define su propio `__init__`**: Este sobrescribe (reemplaza) al de la superclase. Por lo tanto, el constructor de la clase padre no se ejecuta a menos que se llame de forma explícita. La cuestión es que vamos a necesitar que se ejecute si estamos creando atributos en dicha clase padre.

```python
class Animal:
    def __init__(self, nombre):
        self.nombre = nombre
        print(f"Animal creado: {self.nombre}")


class Perro(Animal):
    pass

# Creamos un objeto
mi_perro = Perro("Relámpago")
```

# Sobreescritura de métodos
La sobrescritura de métodos (en inglés, method overriding) en Python es una técnica de la Programación Orientada a Objetos (POO) donde una clase hija redefine un método que ya existe en su clase padre, con el objetivo de ofrecer una implementación específica o adaptada.

Funcionamiento:
* Cuando se llama a un método en un objeto de la clase hija, Python busca primero si ese método está definido en la propia clase hija.
* Si lo encuentra, ejecuta esa versión y reemplaza o ignora la del padre.
* Si no lo encuentra en la hija, busca automáticamente en la clase padre.

```python
class Animal:
    def hacer_sonido(self):
        return "Sonido genérico"

class Perro(Animal):
    # Sobrescribimos el método del padre
    def hacer_sonido(self):
        return "¡Guau!"

# Uso
animal = Animal()
print(animal.hacer_sonido())  # Salida: Sonido genérico

perro = Perro()
print(perro.hacer_sonido())    # Salida: ¡Guau!
```

___
Podemos usar **super()** para extender el método.

A veces no se desea reemplazar por completo el código del padre, sino reutilizarlo y ampliarlo. Para ello se usa la función **super()**.

```python
class Persona:
    def __init__(self, nombre):
        self.nombre = nombre

    def presentar(self):
        return f"Hola, soy {self.nombre}"

class Estudiante(Persona):
    def __init__(self, nombre, grado):
        super().__init__(nombre)  # Llama al __init__ de Persona
        self.grado = grado

    def presentar(self):
        # Reutilizamos el método del padre y añadimos más información
        texto_base = super().presentar()
        return f"{texto_base} y estoy en el grado {self.grado}"

est = Estudiante("Lucía", 5)
print(est.presentar()) 
# Salida: Hola, soy Lucía y estoy en el grado 5
```

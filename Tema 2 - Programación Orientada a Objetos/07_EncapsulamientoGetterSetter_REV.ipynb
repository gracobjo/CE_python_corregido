# Getter y _Setter en Python

Los getters y setters son métodos que permiten obtener y modificar de forma controlada los atributos privados de una clase.
* **Getter**: Método para leer o recuperar el valor de un atributo.
* **Setter**: Método para asignar o actualizar el valor de un atributo, permitiendo añadir reglas o validaciones (por ejemplo, evitar números negativos).

Existen dos formas de implementarlos:
* **Estilo tracicional**: Se define un atributo con guion bajo (`_atributo`) para indicar que es de uso interno y se crean funciones `get_*` y `set_*`.
* **Usar el decorador `@property`**: Esto permite acceder y modificar los atributos como si fueran públicos, pero pasando por una capa de control por debajo.

___
## Estilo tradicional

```python
class CuentaBancaria:
    def __init__(self, balance):
        self._balance = balance  # Atributo protegido/privado

    # Getter
    def get_balance(self):
        return self._balance

    # Setter con validación
    def set_balance(self, valor):
        if valor < 0:
            print("El balance no puede ser negativo.")
        else:
            self._balance = valor

# Uso:
cuenta = CuentaBancaria(100)
print(cuenta.get_balance())  # Lee el valor -> 100
cuenta.set_balance(250)      # Modifica el valor
```

___
## Usar el decorador `@property`

Para interactuar de forma segura con los atributos privados, usamos **Getters** (para leer) y **Setters** (para validar y modificar). La forma estándar en Python es utilizar el decorador `@property`.

Pasos:
1. Definir el atributo como "privado o protegido" en el constructor (__init__) anteponiendo un guion bajo _ para no producir un error de recursión infinita.
2. Crear el getter usando el decorador `@property` sobre un método con el nombre del atributo.
3. Crear el setter usando el decorador `@nombre_del_metodo.setter`

**Nota**: Podemos variar alguna cuestión de lo anterior (p.e. el nombre del setter) pero vamos a seguir estas normas.

```python
class Persona:
    def __init__(self, nombre, edad):
        self._nombre = nombre
        self._edad = edad

    # Getter con @property
    @property
    def nombre(self):
        return self._nombre

    # Setter con @nombre.setter
    @nombre.setter
    def nombre(self, nuevo_nombre):
        if not nuevo_nombre:
            raise ValueError("El nombre no puede estar vacío")
        self._nombre = nuevo_nombre

    # Getter con @property
    @property
    def edad(self):
        return self._edad

    # Setter con @nombre.setter
    @edad.setter
    def edad(self, nueva_edad):
        if not nueva_edad:
            raise ValueError("La edad no puede estar vacía")
        self._edad = nueva_edad

# Uso:
p = Persona("Ana",20)
print(p.nombre,p.edad)      # Llama al getter -> "Ana",20

p.nombre = "Lucía"   # Llama al setter
p.edad = 30 # Llama al setter
print(p.nombre,p.edad)      # Llama al getter -> "Lucía",30
```

___
Podríamos escribir el `setter`con un nombre distinto pero por convención lo dejaremos tal cual.

```python
class Persona:
    def __init__(self, nombre, edad):
        self._nombre = nombre
        self._edad = edad

    # Getter con @property
    @property
    def nombre(self):
        return self._nombre

    # Setter con @nombre.setter
    @nombre.setter
    def nombre(self, nuevo_nombre):
        if not nuevo_nombre:
            raise ValueError("El nombre no puede estar vacío")
        self._nombre = nuevo_nombre

    # Getter con @property
    @property
    def edad(self):
        return self._edad

    # Setter con @nombre.setter
    @edad.setter
    def edad_escribir(self, nueva_edad): # NOMBRE DISTINTO
        if not nueva_edad:
            raise ValueError("La edad no puede estar vacía")
        self._edad = nueva_edad

# Uso:
p = Persona("Ana",20)
print(p.nombre,p.edad)      # Llama al getter -> "Ana",20

p.nombre = "Lucía"   # Llama al setter
p.edad_escribir = 30 # Llama al setter con el nombre DISTINTO
print(p.nombre,p.edad)      # Llama al getter -> "Lucía",30
```

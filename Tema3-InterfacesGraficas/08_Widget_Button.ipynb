# Widget `Button`

Permite al usuario interactuar con la aplicación ejecutando una acción o función cuando se hace clic en él.

Atributos:
* `text`: El texto que se muestra dentro del botón.
* `command`: La función o método que se ejecutará cuando el usuario presione el botón.
* `state`: Controla el estado del botón. Puede ser "normal" (activo) o "disabled" (desactivado/gris).

```python
import tkinter as tk

ventana = tk.Tk()
ventana.title("Widget Button")
ventana.geometry("300x150")

# Función que se ejecuta al presionar el botón
def saludar():
  print("¡Hola! Has presionado el botón.")

# Se crea el botón
boton = tk.Button(
    ventana,
    text="Haz clic aquí",
    command=saludar,
    bg="#4CAF50",
    fg="white",
    font=("Arial", 12),
    state="normal"
)

boton.pack(expand=True)

ventana.mainloop()
```

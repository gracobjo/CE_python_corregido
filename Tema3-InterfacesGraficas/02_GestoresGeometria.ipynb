# Gestores de geometría (Layouts)
`Tkinte`r tiene tres formas de organizar los widgets: `grid()`, `pack()` y `place()`.

No se deden mezclar `pack()` y `grid()` en la misma ventana contenedora.

___
## El gestor de geometría `grid()`
Organiza los elementos en filas (`row`) y columnas (`column`). Es ideal para formularios.

* `row` y `column`: Indican la fila y la columna exactas donde se ubica cada elemento. Comienza en 0 en ambos casos.
* `columnspan` y `rowspan`: Permite que un widget (como el botón "Enviar") abarque más de una columna o fila.
* `padx` y `pady`: Añaden espacio de separación (en píxeles) alrededor de los widgets.
* `ipadx` y `ipady`: Espacio interno adicional (dentro del widget) en el eje horizontal (haciéndolo más ancho) y en el eje vertical (haciéndolo más alto).
* `sticky`: Alinea el elemento dentro de su celda (por ejemplo, "e" para east o derecha). Los valores permitidos: 'n', 'ne', 'e', 'se', 's', 'sw', 'w', 'nw', o 'center' pudiendo usarse como cadena de texto o constante de `tkinter`.

| Constante Tkinter | Cadena de Texto | Dirección Cardinal | Posición en el Contenedor |
| :--- | :--- | :--- | :--- |
| `tk.NW` | `"nw"` | Noroeste (North-West) | Esquina superior izquierda |
| `tk.N` | `"n"` | Norte (North) | Centro superior |
| `tk.NE` | `"ne"` | Noreste (North-East) | Esquina superior derecha |
| `tk.W` | `"w"` | Oeste (West) | Centro izquierdo |
| `tk.CENTER` | `"center"` | Centro (Center) | Centro exacto (Por defecto) |
| `tk.E` | `"e"` | Este (East) | Centro derecho |
| `tk.SW` | `"sw"` | Suroeste (South-West) | Esquina inferior izquierda |
| `tk.S` | `"s"` | Sur (South) | Centro inferior |
| `tk.SE` | `"se"` | Sureste (South-East) | Esquina inferior derecha |

___
### Ejemplo de `grid` para una típica ventana de usuario y contraseña

```python
import tkinter as tk

ventana = tk.Tk()
ventana.title("Gestor de geometría Grid")
ventana.geometry("400x300")

# Fila 0
tk.Label(ventana, text="Usuario:").grid(row=0, column=0, padx=10, pady=10, sticky="e")
entrada_user = tk.Entry(ventana)
entrada_user.grid(row=0, column=1, padx=10, pady=10)

# Fila 1
tk.Label(ventana, text="Contraseña:").grid(row=1, column=0, padx=10, pady=10, sticky="e")
entrada_pass = tk.Entry(ventana, show="*") # Oculta el texto
entrada_pass.grid(row=1, column=1, padx=10, pady=10)

# Fila 2 (El botón ocupa 2 columnas gracias a columnspan)
boton_login = tk.Button(ventana, text="Iniciar Sesión")
boton_login.grid(row=2, column=0, columnspan=2, pady=10, ipadx=40)

ventana.mainloop()
```

## El gestor de gemoetría `pack()`
Organiza los elementos (widgets) colocándolos en bloques dentro de una ventana o un contenedor (Frame).

¿Cómo funciona?
* **Apilamiento automático**: Por defecto, coloca los widgets uno debajo del otro, de arriba hacia abajo (centrados).
* **Orden de aparición**: Se empaquetan en el orden exacto en el que se llama al método `pack()`.
* **No se mezcla con grid**: No se deben usar `pack()` y `grid()` en el mismo contenedor padre, ya que genera errores. Hay que usar contenedores (Frame) separados para usar ambos.

Propiedades:
* `side`: Define en qué borde del contenedor se ancla el widget.
    * `tk.TOP` (por defecto: arriba hacia abajo)
	* `tk.BOTTOM` (abajo)
	* `tk.LEFT` (izquierda)
	* `tk.RIGHT` (derecha)
* `fill`: Indica si el widget debe expandirse para ocupar espacio extra:
	* `tk.NONE` (por defecto, no se expande)
	* `tk.X` (llena horizontalmente el ancho disponible)
	* `tk.Y` (llena verticalmente el alto disponible)
	* `tk.BOTH` (llena tanto en horizontal como en vertical)
* `expand`: Es un booleano. Si es True, el widget toma y reparte el espacio libre disponible en la ventana cuando ésta se redimensiona.
* `padx` y `pady`: Añaden espacio de relleno (padding) externo en píxeles alrededor del widget (en horizontal **x** y en vertical **y**).
* `anchor`: Si el widget no ocupa todo el espacio, define hacia qué punto cardinal se alinea (**tk.N, tk.S, tk.E, tk.W, tk.NW**, etc.). Usa el mismo esquema de valores que la propiedad _sticky_ vista anteriormente (_ver tabla_) .

___
### Ejemplo de `pack` con distintos valores de `side`, `fill`, `padx`y `pady`

```python
import tkinter as tk

ventana = tk.Tk()
ventana.title("Gestor de geometría Pack")
ventana.geometry("400x300")

# Botón arriba (por defecto)
btn1 = tk.Button(ventana, text="Botón 1")
btn1.pack(side=tk.TOP, padx=0, pady=0)

# Botón alineado a la izquierda y sin expandirse
btn2 = tk.Button(ventana, text="Botón 2")
btn2.pack(side=tk.LEFT, fill=tk.NONE, padx=0, pady=0)

# Botón alineado a la izquierda y con expansión en ambos sentidos
btn3 = tk.Button(ventana, text="Botón 3")
btn3.pack(side=tk.LEFT, fill=tk.BOTH, padx=0, pady=0)

# Botón alineado a la izquierda y con expansión horizontal aunque no se perciba
btn4 = tk.Button(ventana, text="Botón 4")
btn4.pack(side=tk.LEFT, fill=tk.X, padx=0, pady=0)

# Botón alineado a la derecha y con expansión vertical pero más ancho para adaptarse a su texto de contenido
btn5 = tk.Button(ventana, text="Botón 5 con más texto")
btn5.pack(side=tk.RIGHT, fill=tk.Y, padx=5, pady=0)

ventana.mainloop()
```

___
### Uso de expand
`expand=True` (o `expand=1`) del gestor de geometría `pack` indica al contenedor que asigne espacio extra o sobrante a un widget cuando la ventana principal cambia de tamaño. Por defecto, su valor es False.

Consideraciones:
* **Asignación de espacio**: Si la ventana crece, el espacio libre se reparte entre los widgets que tengan `expand=True`. Si varios lo tienen activado, se dividen ese espacio de forma equitativa.
* **No rellena el espacio por sí solo**: `expand` sólo indica si requiere espacio adicional, pero **por sí sólo no hace que el widget crezca visualmente** para ocupar ese espacio. Para que crezca visualmente, se debe combinar con la propiedad `fill`.
* **Interacción con fill**:
    * `expand=True + fill=tk.BOTH`: El widget ocupa todo el espacio extra tanto a lo ancho como a lo alto.
	* `expand=True + fill=tk.X`: El widget se expande horizontalmente cubriendo el ancho extra, pero mantiene su altura original.
    * `expand=True + fill=tk.Y`: El widget se expande verticalmente cubriendo el ancho extra, pero mantiene su anchura original.
	* `expand=True + fill=tk.NONE (por defecto)`: El widget no cambia de tamaño visualmente, pero el espacio que hay a su alrededor se expande (centrándolo o dejándolo en su sitio según el ancla).

```python
import tkinter as tk

ventana = tk.Tk()
ventana.title("Uso de expand")
ventana.geometry("400x300")

# Se crean 3 botones
boton1 = tk.Button(ventana, text="Botón 1")
boton2 = tk.Button(ventana, text="Botón 2")
boton3 = tk.Button(ventana, text="Botón 3")
boton4 = tk.Button(ventana, text="Botón 4")

boton1.pack() # No se expande
boton2.pack(expand=True,fill=tk.BOTH) # Se expande en ambos sentidos
boton3.pack(expand=True,fill=tk.X) # Se expande horizontalmente
boton4.pack(expand=True,fill=tk.Y) # Se expande verticalmente

# Redimensionar la ventana para comprobar el efecto
ventana.mainloop()
```

## El gestor de gemoetría `place()`
Permite posicionar widgets de forma absoluta o relativa respecto a su contenedor (ventana o frame). Es el gestor más preciso, ya que funciona como un sistema de coordenadas", aunque es el menos responsivo si cambia el tamaño de la ventana.

Propiedades:
* `x`: Posición horizontal absoluta en píxeles desde el borde izquierdo del contenedor. (Por defecto: **0**).
* `y`: Posición vertical absoluta en píxeles desde el borde superior del contenedor. (Por defecto: **0**).
* `relx`: Posición horizontal relativa. Se expresa como un número flotante entre **0.0** y **1.0** (donde **0.0** es el borde izquierdo y **1.0** el derecho). Si la ventana cambia de tamaño, el widget se moverá proporcionalmente.
* `rely`: Posición vertical relativa. Funciona igual que `relx`, donde **0.0** es el borde superior y **1.0** el inferior.
* `anchor`: Determina qué parte del widget se alinea con las coordenadas x/y especificadas. Por defecto es **'nw'** (**North-West / Noroeste**, es decir, la esquina superior izquierda del widget). Los valores permitidos: **'n', 'ne', 'e', 'se', 's', 'sw', 'w', 'nw', o 'center'** pudiendo usarse como cadena de texto o constante de `tkinter`. (_Ver tabla anterior_)

Dimensiones:
Estas propiedades controlan el tamaño del widget. Si no se especifican, el widget tomará su tamaño natural.
* `width`: Ancho absoluto del widget en píxeles.
* `height`: Alto absoluto del widget en píxeles.
* `relwidth`: Ancho relativo del widget respecto al ancho de su contenedor. Se expresa entre **0.0** y **1.0** (por ejemplo, **0.5** significa que el widget siempre ocupará la mitad del ancho del contenedor).
* `relheight`: Alto relativo del widget respecto al alto de su contenedor (entre **0.0** y **1.0**).

```python
import tkinter as tk

ventana = tk.Tk()
ventana.title("Gestor de geometría Place")
ventana.geometry("400x300")

# Ejemplo 1: Posicionamiento Absoluto (Fijo)
btn_fijo = tk.Button(ventana, text="Botón Fijo (x=50, y=50)")
btn_fijo.place(x=50, y=50, width=150, height=40,anchor="nw")

# Ejemplo 2: Posicionamiento Relativo (Centrado y Responsivo)
# Se ancla en el centro (anchor='center') y se pone a la mitad (0.5) de la ventana
btn_relativo = tk.Button(ventana, text="Botón Centrado")
btn_relativo.place(relx=0.5, rely=0.5, relwidth=0.4, relheight=0.15, anchor="center")

ventana.mainloop()
```

### Contenedor de Referencia:
* **in_** (o **in**): Permite posicionar el widget con el que estamos trabajando respecto un widget maestro o contenedor diferente al que se usó al crear el widget (normalmente la ventana principal).

Se trata de posicionar el widget teniendo en cuenta un elemento de referencia diferente.

```python
import tkinter as tk

ventana = tk.Tk()
ventana.title("Uso de in_ en Place")
ventana.geometry("400x300")

# Se crea un contenedor (Frame) de color azul. Los Frames se verán a continuación.
marco_contenedor = tk.Frame(ventana, bg="lightblue", bd=2, relief="solid")
# Posicionamos el marco en la ventana principal
marco_contenedor.place(x=50, y=50, width=200, height=100)

# Se crea un botón
# Nota: Su padre original es 'ventana', pero lo posicionaremos con referencia al frame 'marco_contenedor' con 'in_'
boton = tk.Button(ventana, text="Botón Interno", bg="white")

# Se posiciona el botón usando 'in_' respecto al marco_contenedor
# relx=0.5 y rely=0.5 con anchor="center" lo colocan exactamente en el centro de 'marco_contenedor'
boton.place(in_=marco_contenedor, relx=0.5, rely=0.5, anchor="center")
#boton.place(relx=0.5, rely=0.5, anchor="center") # De este modo estaría centrado respecto a la ventana principal

ventana.mainloop()
```

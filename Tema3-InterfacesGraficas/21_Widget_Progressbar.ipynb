# Widget `Progressbar`
Este control se utiliza para ofrecer una información al usuario sobre el progreso de una tarea.

Propiedades:
* `orient`: Define la orientación de la barra. Puede ser `tk.HORIZONTAL` u `horizontal` (por defecto) o `tk.VERTICAL` o `vertical`.
* `length`: Especifica la longitud del eje principal de la barra (el ancho si es horizontal, o el alto si es vertical) en píxeles.
* `mode`: Determina el comportamiento del widget. Tiene dos opciones:
	* `determinate`: Muestra una progresión acumulativa hacia un indicador final (se usa cuando se conoce el tiempo o progreso total).
	* `indeterminate`: Muestra un indicador que se mueve de un lado a otro (se usa cuando la duración del proceso es desconocida).
* `maximum`: El valor máximo de la barra (por defecto es 100.0). Representa el estado de "completado" en el modo determinate.
* `value`: El valor actual de la barra de progreso.
* `variable`: Una variable de control (normalmente DoubleVar o IntVar) vinculada al valor de la barra para actualizarla automáticamente.
* `style`: El nombre del estilo utilizado para personalizar la apariencia visual del widget.
* `cursor`: Cambia el aspecto del cursor del ratón cuando pasa por encima del widget (por ejemplo, `arrow`, `watch`).

Métodos:
* `step(incremento)`: Incrementa el valor actual de la barra en la cantidad indicada (por defecto 1.0). Si llega al máximo, vuelve a empezar desde 0.
* `start(intervalo)`: Inicia el modo de animación automática. El intervalo se da en milisegundos (por defecto 50ms). Cada intervalo, el valor se incrementa automáticamente. Es ideal para el modo `indeterminate`.
* `stop()`: Detiene la animación automática que se inició con el método `start()`.

### Barra de progreso de tipo `determinate`

```python
import time
import tkinter as tk
from tkinter import ttk

def iniciar_progreso():
  # Simula un proceso que avanza hasta 100
  for i in range(1, 101):
    time.sleep(0.03)  # Pausa corta para simular carga
    barra.step()
    ventana.update() # Necesario actualizar la ventana para ver el progreso de la barra

ventana = tk.Tk()
ventana.title("Widget Progressbar en modo determinate")
ventana.geometry("400x300")

etiqueta = ttk.Label(ventana, text="Presiona el botón para iniciar la carga:")
etiqueta.pack(pady=10)

valor_barra = tk.DoubleVar()
valor_barra.set(0)

# Se crea la barra de progreso (modo determinado de 0 a 100)
barra = ttk.Progressbar(
    ventana, orient=tk.HORIZONTAL, length=280, mode="determinate", variable=valor_barra, maximum=100
)
barra.pack(pady=10)

boton = ttk.Button(
    ventana, text="Iniciar Progreso", command=iniciar_progreso
)
boton.pack(pady=10)

# Empleamos para la etiqueta la misma variable de control que para la barra de progreso
etiqueta_valor = ttk.Label(ventana, font=('Courier',14,'bold'), textvariable=valor_barra)
etiqueta_valor.pack(pady=10)

ventana.mainloop()
```

### Barra de progreso de tipo `indeterminate`

```python
import tkinter as tk
from tkinter import ttk

def simular_tarea():
    global pasos_restantes
    if pasos_restantes > 0:
        pasos_restantes -= 1
        valor_barra.set(f"Trabajando... pasos restantes: {pasos_restantes}")
        # Se vuelve a ejecutar la función en 100 milisegundos
        ventana.after(100, simular_tarea)
    else:
        # Cuando la tarea termina, se detiene
        barra.stop()
        boton_inicio.config(state="normal")
        etiqueta_estado.config(text="¡Proceso terminado!")

def iniciar_proceso():
    global pasos_restantes
    pasos_restantes = 50  # Simula una tarea que requiere 50 'pasos'
    boton_inicio.config(state="disabled")
    etiqueta_estado.config(text="Procesando...")

    # Se inicia el movimiento de la barra indeterminada
    barra.start(10)  # El argumento es la velocidad del movimiento

    # Se lanza la simulación del proceso
    simular_tarea()

ventana = tk.Tk()
ventana.title("Widget Progressbar en modo indeterminate")
ventana.geometry("400x300")

# Variable global para controlar el bucle simulado
pasos_restantes = 50
valor_barra = tk.StringVar()
valor_barra.set(f"Trabajando... pasos restantes: {pasos_restantes}")

etiqueta_estado = ttk.Label(ventana, text="Presiona iniciar para comenzar")
etiqueta_estado.pack(pady=10)

# ProgressBar en modo 'indeterminate'
barra = ttk.Progressbar(ventana, mode="indeterminate", length=250)
barra.pack(pady=10)

boton_inicio = ttk.Button(ventana, text="Iniciar Tarea", command=iniciar_proceso)
boton_inicio.pack(pady=10)

etiqueta_valor = ttk.Label(ventana, font=('Courier',14,'bold'), textvariable=valor_barra)
etiqueta_valor.pack(pady=10)

ventana.mainloop()
```

```python
import tkinter as tk

sorteo = []

def finalizar_sorteo():
    pass

def realizar_apuestas():
    pass

def comprobar_resultados():
    pass

def manejar_clic(a,event):
    if a not in sorteo:
        if len(sorteo)<6:
            sorteo.append(a)
            print("¡Hiciste clic en la etiqueta!",a)
            event.widget.config(bg="red")
        else:
            print("Ya has seleccionado 6")
    else:
        sorteo.remove(a)
        event.widget.config(bg="yellow")
        print("Has quitado el 6")

ventana = tk.Tk()
ventana.title("Loteria primitiva")
ventana.geometry("400x900")

tk.Label(ventana, text="LOTERÍA PRIMITIVA", font=('Arial',18,'bold'), bg="lightblue", bd=2, relief="sunken").pack(pady=20)

frame = tk.Frame(ventana, bg="lightblue", bd=2, relief="sunken")
frame.pack(pady=20)
numeros = []
num = 0
for columna in range(0,5):
    for fila in range(0,10):
        etiqueta = tk.Label(frame, font=("Arial", 14), fg="blue",bd=2,relief="solid",width=2,height=2)
        etiqueta.grid(row=fila,column=columna)
        if num != 0:
            etiqueta.config(text=str(num),bg="yellow",)
            etiqueta.bind("<Button-1>", lambda event, v=num: manejar_clic(v, event))
        num +=1


btn_finalizar = tk.Button(ventana, text="Finalizar sorteo",command=finalizar_sorteo)
btn_finalizar.pack(pady=20)

btn_apostar = tk.Button(ventana, text="Realizar apuestas",command=realizar_apuestas)
btn_apostar.pack(pady=20)

btn_resultados = tk.Button(ventana, text="Comprobar resultados",command=comprobar_resultados)
btn_resultados.pack(pady=20)
    



ventana.mainloop()
```

# Ejercicios 01
Partiendo del ejercico 7 del marcador de tenis del tema 2 (programación orientada a objetos), crear una aplicación gráfica con los siguientes componentes:
* Etiqueta que sea un marcador de resultados
* Dos botones, uno por jugador, que incremente el marcador para cada uno de los jugadores

```python
import tkinter as tk

class PartidoTenis:
    def __init__(self, jugador1, jugador2):
        self.jugador1 = jugador1
        self.jugador2 = jugador2
        self.marcador = ""
        self.cancelado = False
        self.historial_sets = []
        
        primer_set = self._Set(self, 1) 
        self.historial_sets.append(primer_set)

    def anotar_en_set_actual(self, jugador_num):
        if self.historial_sets:
            set_actual = self.historial_sets[-1]
            set_actual.registrar_punto(jugador_num)
        else:
            print("[Error] No hay ningún set activo en juego.")

    def suspender_por_lluvia(self):
        print(f"\n[Torneo] Tormenta eléctrica detectada. Partido SUSPENDIDO.")
        self.cancelado = True

    # --- CLASE INTERNA ---
    class _Set:
        def __init__(self, partido_padre, numero_set):
            self.partido = partido_padre
            self.numero_set = numero_set
            self.juegos_j1 = 0
            self.juegos_j2 = 0

        def registrar_punto(self, jugador_num):
            if self.partido and not self.partido.cancelado:
                if jugador_num == 1:
                    self.juegos_j1 += 1
                elif jugador_num == 2:
                    self.juegos_j2 += 1
                
                self.partido.marcador = f"[Set {self.numero_set}]: {self.partido.jugador1} [{self.juegos_j1}] vs [{self.juegos_j2}] {self.partido.jugador2}"
                resultado_etiqueta.config(text=self.partido.marcador)
            else:
                print(f"[Árbitro] Punto denegado en Set {self.numero_set}: El partido no está activo.")


ventana = tk.Tk()

final_open = PartidoTenis("Carlos Alcaraz", "Jannik Sinner")

ventana.title("Partido de tenis")
ventana.geometry("400x300")

resultado_etiqueta = tk.Label(ventana, text="NO HA EMPEZADO",font=("Arial", 14))
resultado_etiqueta.grid(row=0, column=0, columnspan=2, padx=10, pady=10)

tk.Button(ventana, text=final_open.jugador1,command=lambda: final_open.anotar_en_set_actual(1)).grid(row=1, column=0, padx=10, pady=10, ipadx=40)

tk.Button(ventana, text=final_open.jugador2,command=lambda: final_open.anotar_en_set_actual(2)).grid(row=1, column=1, padx=10, pady=10, ipadx=40)

ventana.mainloop()
```

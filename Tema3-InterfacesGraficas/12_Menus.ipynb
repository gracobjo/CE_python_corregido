# Menús (Widget `Menu`)
Se utiliza para crear barras de menú (en la parte superior de la ventana) o menús desplegables (pop-up/contextuales).

Propiedades:
* `tearoff`: Controla si el menú puede "desprenderse" de la ventana en una ventana flotante independiente. Por defecto es `True` (muestra una línea punteada). Se suele configurar en `False` (`tearoff=0`) para un diseño moderno.
* `bg` / `background`: Cambia el color de fondo del menú.
* `fg` / `foreground`: Cambia el color del texto de las opciones.
* `activebackground`: Define el color de fondo del elemento sobre el que está el cursor.
* `activeforeground`: Define el color del texto del elemento sobre el que está el cursor.
* `font`: Define la tipografía y tamaño del texto del menú.
* `postcommand`: Ejecuta una función automáticamente justo antes de que el menú se despliegue. Útil para activar/desactivar opciones dinámicamente.

Métodos:
* `add_cascade`: Conecta un menú desplegable o submenú a la barra principal o a otro menú padre.
* `add_command`: Añade las opciones finales que ejecutan una función cuando el usuario hace clic en ellas
* `add_separator`: Añade una línea horizontal divisoria para organizar las opciones.

```python
import tkinter as tk

def abrir_archivo():
    print("Abriendo archivo...")

ventana = tk.Tk()
ventana.title("Widget Menu")
ventana.geometry("400x300")

# Barra de menú principal
menubar = tk.Menu(ventana, postcommand=lambda: print("Ejecución de postcommand"))
ventana.config(menu=menubar) # La propiedad menu de la ventana se asocia con la barra de menú

# ARICHIVO
archivo_menu = tk.Menu(menubar, tearoff=0) # Paso 1: Se crea el menú
menubar.add_cascade(label="Archivo", menu=archivo_menu) # Paso 2: Se inserta en otro menú
archivo_menu.add_command(label="Nuevo", command=lambda: print("Nuevo")) # Paso 3: Se añaden opciones del menú
archivo_menu.add_command(label="Abrir", command=abrir_archivo)
archivo_menu.add_separator()  # Línea divisoria
archivo_menu.add_command(label="Salir", command=ventana.quit)

# EDICION
edicion_menu = tk.Menu(menubar, tearoff=0)
menubar.add_cascade(label="Edición", menu=edicion_menu)
edicion_menu.add_command(label="Copiar", activeforeground="red", command=lambda: print("Copiar"))
edicion_menu.add_command(label="Cortar", activebackground="yellow",command=lambda: print("Cortar"))
edicion_menu.add_command(label="Pegar", command=lambda: print("Pegar"))

# VER
ver_menu = tk.Menu(menubar, tearoff=1)
menubar.add_cascade(label="Ver", menu=ver_menu)
zoom_menu = tk.Menu(menubar, tearoff=0)
ver_menu.add_cascade(label="Zoom", menu=zoom_menu)
zoom_menu.add_command(label="Alejar", command=lambda: print("Alejar"))
zoom_menu.add_command(label="Acercar", command=lambda: print("Acercar"))

# AYUDA
ayuda_menu = tk.Menu(menubar, tearoff=0)
menubar.add_cascade(label="Ayuda", menu=ayuda_menu)
ayuda_menu.add_command(label="Acerca de...", command=lambda: print("Acerca de..."))

ventana.mainloop()
```

## Menú contextual
Aparece en cualquier punto en lugar de estar anclado a una barra principal.

Se emplea el método `tk_popup` indicando el punto donde aparece.

```python
import tkinter as tk

def mostrar_menu(event):
    menu_contextual.tk_popup(event.x_root, event.y_root)

def opcion_1():
  print("Seleccionaste la opción 1")

def opcion_2():
  print("Seleccionaste la opción 2")

ventana = tk.Tk()
ventana.title("Widget Menu (contextutal)")
ventana.geometry("400x300")

# Se crea el menú contextual (tearoff=0 quita la línea punteada superior)
menu_contextual = tk.Menu(ventana, tearoff=0)
menu_contextual.add_command(label="Opción 1", command=opcion_1)
menu_contextual.add_command(label="Opción 2", command=opcion_2)
menu_contextual.add_separator()
menu_contextual.add_command(label="Salir", command=lambda: ventana.destroy())

# Se vincular el clic derecho
ventana.bind("<Button-3>", mostrar_menu)

ventana.mainloop()
```

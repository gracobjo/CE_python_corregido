# Widget `Text`
Se utiliza para insertar y editar texto multilínea (a diferencia de `Entry`, que es solo para una línea).

Permite aplicar formatos avanzados como colores, fuentes y estilos a partes específicas del texto mediante el uso de "etiquetas" (tags).

#### Propiedades:
* `width` y `height`: Definen el tamaño del widget donde `width` se mide en número de caracteres por línea y `height` en número de líneas visibles (no en píxeles).
* `font`: Define la familia tipográfica, el tamaño y el estilo del texto por defecto.
* `wrap`: Controla cómo se comporta el texto cuando llega al borde derecho del widget.
    * `none`: El texto sigue de largo (requiere scrollbar horizontal).
    * `char`": Se corta exactamente en el último carácter que quepa.
    * `word`: Rompe la línea respetando las palabras completas.
* `padx` y `pady`: Añaden espacio de colchón (margen interno) entre el texto y los bordes del widget.
* `state`: Controla si el usuario puede editar el texto. Puede ser `normal` (por defecto, editable) o `disabled` (solo lectura).

#### Métodos generales:
Para interactuar con el texto de forma genérica tenemos los siguientes métodos.

**IMPORTANTE**: Tkinter maneja la posición del texto en formato "línea.carácter". Por ejemplo, **"1.0"** significa **línea 1, carácter 0**. La posición `end` representa el final absoluto del texto.
* `insert(índice, texto)`: Inserta una cadena de texto en la posición indicada.
* `get(inicio, fin)`: Devuelve el texto que se encuentra entre los dos índices.
* `delete(inicio, fin)`: Borra el texto dentro del rango especificado.
* `focus_set()`: Centra el foco en el Widget.

#### Métodos de Tags (etiquetas):
Para trabajar con texto con formato empleamos etiquetas o `Tags`.
* `tag_config(nombre_tag, opciones...)`: Define el estilo visual de una etiqueta (color de fondo, de letra, subrayado, etc.).
* `tag_add(nombre_tag, inicio, fin)`: Aplica el estilo definido a un fragmento de texto específico.
* `tag_remove(nombre_tag, inicio, fin)`: Quita la marca de selección de todo el documento.

**IMPORTANTE**: Existe una etiquta denominada `sel` que representa el texto seleccionado. A partir de esta etiqueta tenemos los índices del texto seleccinado:
* `sel.first`: Índice de inicio del texto seleccionado.
* `sel.last`: Índice de final del texto seleccionado.

___
### Ejemplo básico para obtener texto e insertar texto

```python
import tkinter as tk

ventana = tk.Tk()
ventana.title("Widget Text")
ventana.geometry("400x300")

# Crear el widget Text (multilínea). Es de tipo clásico. No exite versión temática.
caja_texto = tk.Text(
    ventana,
    width=40,
    height=10,
    font=("Consolas", 11),
    padx=10,
    pady=10,
    wrap="word" # Cambiar a "char" y "word"
)
caja_texto.pack(pady=10)

caja_texto.insert(tk.END, "Hola, este es un widget de texto en Tkinter que tiene que ocupar mucho espacio en la línea.\nPuedes escribir varias líneas aquí.\n")

def obtener_texto():
  contenido = caja_texto.get("2.4", tk.END)
  print("Texto ingresado:\n", contenido)

boton = tk.Button(ventana, text="Obtener texto", command=obtener_texto)
boton.pack(pady=5)

ventana.mainloop()
```

### Ejemplo con tags para el formato del texto

```python
import tkinter as tk

# Crear la ventana principal
ventana = tk.Tk()
ventana.title("Widget Text con Tags")
ventana.geometry("500x400")

# 1. Crear el widget Text con algunas propiedades iniciales
area_texto = tk.Text(
    ventana, 
    width=50, 
    height=15, 
    font=("Arial", 12), 
    padx=10, 
    pady=10, 
    wrap="word"
)
area_texto.pack(pady=20)

# Se inserta texto
area_texto.insert("1.0", "Este es un widget Text de Tkinter.\n")
area_texto.insert("end", "Puedes escribir múltiples líneas aquí.\n")
area_texto.insert("end", "Texto de prueba para formatear.")

# Se crean tags o formatos de texto
area_texto.tag_config("resaltado", background="yellow", foreground="red")
area_texto.tag_config("negrita", font=("Arial", 12, "bold"))

# Se aplican los tags a determinadas partes del texto
area_texto.tag_add("resaltado", "1.18", "1.23")  
area_texto.tag_add("negrita", "2.7", "2.24")

# Obtiene todo el texto
def obtener_contenido():
    contenido = area_texto.get("1.0", "end")
    print("Contenido actual:\n", contenido)

btn_obtener = tk.Button(ventana, text="Obtener Texto", command=obtener_contenido)
btn_obtener.pack(side="left", padx=20)

def quitar_formato():
    area_texto.tag_remove("resaltado", "1.18", "1.23")

btn_quitarformato = tk.Button(ventana, text="Quitar el formato a la línea 1", command=quitar_formato)
btn_quitarformato.pack(side="left", padx=40)

# Borrar toto el texto
def borrar_todo():
    area_texto.delete("1.0", "end")
    
btn_borrar = tk.Button(ventana, text="Borrar Todo", command=borrar_todo)
btn_borrar.pack(side="right", padx=20)

ventana.mainloop()
```

### Ejemplo con selección de texto

```python
import tkinter as tk
from tkinter import messagebox

def seleccionar_todo():
    txt_area.tag_add("sel", "1.0", "end")
    txt_area.focus_set()  # Da el foco a Text para que se vea la selección

def obtener_seleccion():
    try:
        # Recupera el texto que está bajo la etiqueta de selección ("sel")
        texto_seleccionado = txt_area.get("sel.first", "sel.last")
        messagebox.showinfo("Texto Seleccionado", f"Has seleccionado:\n\n{texto_seleccionado}")
    except tk.TclError:
        # Si no hay texto seleccionado, tkinter lanza un error TclError
        messagebox.showwarning("Sin selección", "Por favor, selecciona primero algún texto con el ratón.")

def borrar_seleccion():
    # Elimina la etiqueta de selección, desmarcando el texto
    txt_area.tag_remove("sel", "1.0", "end")

# Configuración de la ventana principal
ventana = tk.Tk()
ventana.title("Widget Text con selección de texto")
ventana.geometry("500x350")

# Widget de Texto
txt_area = tk.Text(ventana, wrap="word", height=10)
txt_area.pack(padx=10, pady=10, fill="both", expand=True)

# Insertar un texto por defecto
texto_inicial = """Haz clic y arrastra el ratón para seleccionar cualquier parte de este texto.
También puedes usar los botones de abajo para interactuar con las funciones de selección de Tkinter."""

txt_area.insert("1.0", texto_inicial)

# Contenedor para los botones
frame_botones = tk.Frame(ventana)
frame_botones.pack(pady=10)

# Botones
btn_seleccionar_todo = tk.Button(frame_botones, text="Seleccionar Todo", command=seleccionar_todo)
btn_seleccionar_todo.pack(side="left", padx=5)

btn_obtener_sel = tk.Button(frame_botones, text="Ver Texto Seleccionado", command=obtener_seleccion)
btn_obtener_sel.pack(side="left", padx=5)

btn_borrar_sel = tk.Button(frame_botones, text="Desmarcar Selección", command=borrar_seleccion)
btn_borrar_sel.pack(side="left", padx=5)

ventana.mainloop()
```

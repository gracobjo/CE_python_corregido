# Eventos

En Tkinter, los eventos se vinculan usando el método bind('<secuencia>', funcion) para capturar acciones del teclado o ratón (como <Button-1> o <Key>), o mediante el parámetro command=funcion en botones. La función recibe un objeto event con los detalles de la acción.

Métodos principales de vinculación:

* **widget.bind(evento, funcion)**: Asocia cualquier evento del sistema (clics, teclas, foco) a un widget específico.
* **Button(..., command=funcion)**: Forma directa y sencilla para el clic izquierdo en botones.

Ejemplos comunes de eventos

|Evento|Descripción|
|---|---|
|`<Button-1>`|Clic con el botón izquierdo del ratón|
|`<Return>`|Pulsar la tecla Intro (Enter)|
|`<Key>`|Pulsar cualquier tecla del teclado|
|`<FocusIn>`|Un widget recibe el foco|

```python
import tkinter as tk

def manejar_clic(event):
    print("¡Hiciste clic en la etiqueta!")

def manejar_return(event):
    print("¡Pulsaste intro!")

def manejar_key(event):
    print("¡Pulsaste una tecla!")

def manejar_focus(event):
    print("¡Recibe el foco!")

ventana = tk.Tk()
ventana.title("Eventos básicos")
ventana.geometry("400x300")

etiqueta = tk.Label(ventana, text="Hacer click")
etiqueta.pack(pady=10)

texto = tk.Entry(ventana, width=40)
texto.pack(pady=10)

etiqueta.bind("<Button-1>", manejar_clic)

texto.bind("<Return>", manejar_return)
texto.bind("<Key>", manejar_key)
texto.bind("<FocusIn>", manejar_focus)

ventana.mainloop()
```

___
### Información del evento producido
El objeto event que Tkinter pasa automáticamente a las funciones vinculadas con .bind() es una instancia de la clase Event. Este objeto actúa como un contenedor que almacena todos los detalles del evento que acaba de ocurrir (quién lo generó, dónde y cuándo).

Los atributos disponibles dependen del tipo de evento (ratón o teclado), pero los más utilizados se dividen en las siguientes categorías:

Información del origen y estado
* **event.widget**: El componente exacto (botón, etiqueta, ventana) que originó el evento. Permite modificar el widget directamente (ej. event.widget.config(bg="red")).
* **event.type**: El tipo de evento numérico (representado internamente por Tkinter como cadenas como ButtonPress o KeyPress).
* **event.state**: Un número entero que describe el estado de los modificadores del teclado (si Shift, Ctrl o Alt estaban presionados durante el evento).

Posición del ratón
* **event.x** y **event.y**: Las coordenadas del puntero del ratón en píxeles, relativas al widget que disparó el evento.
* **event.x_root** y **event.y_root**: Las coordenadas del puntero relativas a la pantalla completa.

Teclado y caracteres
* **event.char**: El carácter de la tecla presionada como una cadena de texto (ej. 'a', '7', 'A').
* **event.keysym**: El nombre de la tecla en formato de texto (ej. 'Return', 'BackSpace', 'space', 'Shift_L'). Es ideal para detectar teclas especiales que no tienen representación de texto.
* **event.keycode**: El código numérico interno del sistema operativo para esa tecla.

```python
import tkinter as tk

def detallar_evento_clic(event):
    cuadro_texto.delete("1.0", "end")
    c = f"Evento Detectado: {event.type}"
    c += f"\nWidget origen: {event.widget}"
    c += f"\nWidget state: {event.state}"
    c += f"\nWidget botón: {event.num}"
    c += f"\nPosición relativa al widget: x={event.x}, y={event.y}"
    c += f"\nPosición en pantalla: x={event.x_root}, y={event.y_root}"
    event.widget.config(bg="red")
    cuadro_texto.insert("end",c)

ventana = tk.Tk()
ventana.title("Detectar posición clic de ratón")
ventana.geometry("500x400")

cuadro_texto = tk.Text(ventana, wrap="word", width=50, height=15)
cuadro_texto.pack(pady=10)

ventana.bind("<Button-1>", detallar_evento_clic)
ventana.bind("<Button-2>", detallar_evento_clic)
ventana.bind("<Button-3>", detallar_evento_clic)

ventana.mainloop()
```

```python
import tkinter as tk

def detectar_tecla(event):
    texto = f"Tecla pulsada: {event.keysym} (Carácter: '{event.char}') (Código: '{event.keycode}')"
    etiqueta.config(text=texto)

ventana = tk.Tk()
ventana.title("Detectar tecla pulsada")
ventana.geometry("400x300")

etiqueta = tk.Label(ventana, text="Pulsa cualquier tecla...", font=("Arial", 12))
etiqueta.pack(expand=True)

ventana.bind("<Key>", detectar_tecla)

ventana.mainloop()
```

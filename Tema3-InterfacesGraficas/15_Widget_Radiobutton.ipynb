# Widget `Radiobutton`
Se utiliza para permitir al usuario seleccionar una sola opción de un grupo de opciones mutuamente excluyentes. Tenemos tanto versión clásica como temática.

Consideraciones importantes:
* Para que funcionen en grupo, todos los widgets `Radiobutton` deben compartir la misma variable de control (_StringVar_ o _IntVar_).
* Esta variable es algo diferente a la de otros controles puesto que los `Radiobutton` no funcionan por sí sólos sino que lo hacen en conjunción con otros `Radiobutton`.
* Para que la variable de control tome un valor diferente para cada control `Radiobutton` usamos la propiedad `value`. 

Propiedades:
* `variable`: La variable de control de Tkinter (`IntVar`, `StringVar`) que comparte el grupo de botones para rastrear cuál está seleccionado.
* `value`: El valor único que se asignará a la variable cuando ese `radiobutton` específico sea seleccionado.
* `text`: El texto que se muestra junto al botón de opción.
* `command`: La función que se ejecutará automáticamente cuando el usuario seleccione el botón.
* `state`: El estado del control. Puede ser "normal", "disabled" (desactivado/gris) o "active" (bajo el cursor).
* `indicatoron`: Si se establece en `True` (por defecto), muestra el círculo físico de selección. Si se establece en `False`, el `radiobutton` se convierte en un botón regular que se mantiene presionado al ser seleccionado. Sólo funciona con en modo clásico.

___
### Ejemplo básico
Se puede cambiar cualquier `Radiobutton` a tipo clásico para comparar.

```python
import tkinter as tk
from tkinter import ttk

def mostrar_seleccion():
    etiqueta.configure(text=f"Seleccionaste la opción: {opcion.get()}")

ventana = tk.Tk()
ventana.title("Widget Radiobutton")
ventana.geometry("400x300")

# Variable para almacenar el valor seleccionado (entero)
opcion = tk.IntVar()
opcion.set(1) # Opción seleccionada por defecto

# Se crean los Radiobuttons
ttk.Radiobutton(ventana, text="Opción 1", variable=opcion, value=1, command=mostrar_seleccion).pack(anchor="w", padx=20, pady=5)
ttk.Radiobutton(ventana, text="Opción 2", variable=opcion, value=2, command=mostrar_seleccion).pack(anchor="w", padx=20, pady=5)
ttk.Radiobutton(ventana, text="Opción 3", variable=opcion, value=3, command=mostrar_seleccion).pack(anchor="w", padx=20, pady=5) #Cambiar a tk en lugar de ttk y ver diferencia
ttk.Radiobutton(ventana, text="Opción 4", variable=opcion, value=4, command=mostrar_seleccion, state="active").pack(anchor="w", padx=20, pady=5)

# Etiqueta para mostrar el resultado
etiqueta = ttk.Label(ventana, text="Seleccionaste la opción: 1")
etiqueta.pack(padx=20, pady=5)

ventana.mainloop()
```

### indicatoron
Sólo es posible en modo clásico. Valores:
* `True - 1`: Se ve como un radiobutton normal
* `False - 0`: Se ve como un botón pulsado

```python
import tkinter as tk
from tkinter import ttk

def mostrar_seleccion():
    etiqueta.configure(text=f"Seleccionaste la opción: {opcion.get()}")

ventana = tk.Tk()
ventana.title("Propiedad indicatoron")
ventana.geometry("400x300")

# Variable para almacenar el valor seleccionado (entero)
opcion = tk.IntVar()
opcion.set(1) # Opción seleccionada por defecto

# Se crean los Radiobuttons
tk.Radiobutton(ventana, text="Opción 1", variable=opcion, value=1, command=mostrar_seleccion, indicatoron=0).pack(anchor="w", padx=20, pady=5)
tk.Radiobutton(ventana, text="Opción 2", variable=opcion, value=2, command=mostrar_seleccion, indicatoron=0).pack(anchor="w", padx=20, pady=5)
tk.Radiobutton(ventana, text="Opción 3", variable=opcion, value=3, command=mostrar_seleccion, indicatoron=0).pack(anchor="w", padx=20, pady=5) #Cambiar a tk en lugar de ttk y ver diferencia
tk.Radiobutton(ventana, text="Opción 4", variable=opcion, value=4, command=mostrar_seleccion, state="active", indicatoron=0).pack(anchor="w", padx=20, pady=5)

# Etiqueta para mostrar el resultado
etiqueta = ttk.Label(ventana, text="Seleccionaste la opción: 1")
etiqueta.pack(padx=20, pady=5)

ventana.mainloop()
```

### Métodos
* `invoke()`: Simula un clic del usuario en el botón. Ejecuta la función asociada en command y cambia su estado.

```python
import tkinter as tk
from tkinter import ttk

def mostrar_seleccion():
    etiqueta.configure(text=f"Seleccionaste la opción: {opcion.get()}")

ventana = tk.Tk()
ventana.title("Ejemplo de Radiobutton")
ventana.geometry("300x200")

# Variable para almacenar el valor seleccionado (entero)
opcion = tk.IntVar()
opcion.set(1) # Opción seleccionada por defecto

# Se crean los Radiobuttons
rb1 = ttk.Radiobutton(ventana, text="Opción 1", variable=opcion, value=1, command=mostrar_seleccion)
rb1.pack(anchor="w", padx=20, pady=5)

rb2 = ttk.Radiobutton(ventana, text="Opción 2", variable=opcion, value=2, command=mostrar_seleccion)
rb2.pack(anchor="w", padx=20, pady=5)

rb3 = ttk.Radiobutton(ventana, text="Opción 3", variable=opcion, value=3, command=mostrar_seleccion)
rb3.pack(anchor="w", padx=20, pady=5)

# Etiqueta para mostrar el resultado
etiqueta = ttk.Label(ventana, text="Seleccionaste la opción: 1")
etiqueta.pack(padx=20, pady=5)

# Se selecciona el 1 y se ejecuta su command
ttk.Button(ventana, text="Invoke para seleccionar la 1", command=lambda: rb1.invoke()).pack(padx=20, pady=5)

ventana.mainloop()
```

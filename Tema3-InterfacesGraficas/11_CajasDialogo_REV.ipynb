### showinfo

```python
import tkinter as tk
from tkinter import messagebox

def mostrar_informacion():
    messagebox.showinfo("Información", "¡Operación completada con éxito!")    

ventana = tk.Tk()
ventana.title("Ventana principal")
ventana.geometry("400x300")

btn = tk.Button(ventana, text="Ejecutar showinfo", command=mostrar_informacion)
btn.pack(pady=50)

ventana.mainloop()
```

### askquestion
Devuelve un str que puede ser `yes` o `no`

```python
import tkinter as tk
from tkinter import messagebox

def comprobar_salir():
    # Muestra la ventana de pregunta y guarda la respuesta ('yes' o 'no')
    respuesta = messagebox.askquestion("Confirmación", "¿Estás seguro de que quieres salir?")
    
    if respuesta == 'yes':
        print("El usuario seleccionó: Sí")
        ventana.destroy()  # Cierra la ventana principal
    else:
        print("El usuario seleccionó: No")


ventana = tk.Tk()
ventana.title("Ventana principal")
ventana.geometry("400x300")

btn_exit = tk.Button(ventana, text="Ejecutar askquestion", command=comprobar_salir)
btn_exit.pack(pady=50)

ventana.mainloop()
```

```python
import tkinter as tk
from tkinter import messagebox

def comprobar_salir():
    # Muestra la ventana de pregunta y guarda la respuesta ('yes' o 'no')
    respuesta = messagebox.askokcancel("Confirmación", "¿Estás seguro de que quieres salir?")
    
    if respuesta:
        print("El usuario seleccionó: Aceptar")
        ventana.destroy()  # Cierra la ventana principal
    else:
        print("El usuario seleccionó: Cancelar")


ventana = tk.Tk()
ventana.title("Ventana principal")
ventana.geometry("400x300")

btn_exit = tk.Button(ventana, text="Ejecutar askokcancel", command=comprobar_salir)
btn_exit.pack(pady=50)

ventana.mainloop()
```

### askyesnocancel
Devuelve un booleano que puede ser `True` si pulsa **Sí**, `False` si pulsa **No** y `None` si pulsa **Cancelar**.

```python
import tkinter as tk
from tkinter import messagebox

def comprobar_salir():
    respuesta = messagebox.askyesnocancel("Confirmación", "¿Estás seguro de que quieres salir?")
    
    if respuesta:
        print("El usuario seleccionó: Sí")
        ventana.destroy()  # Cierra la ventana principal
    
    if not respuesta and respuesta != None:
        print("El usuario seleccionó: No")
        
    if respuesta == None:
        print("El usuario seleccionó: Cancelar")


ventana = tk.Tk()
ventana.title("Ventana principal")
ventana.geometry("400x300")

btn_exit = tk.Button(ventana, text="Ejecutar askyesnocancel", command=comprobar_salir)
btn_exit.pack(pady=50)

ventana.mainloop()
```

## Archivos y directorios
* **askopenfilename**: Abre el diálogo para seleccionar un solo archivo existente. Devuelve la ruta completa como un string (vacío si cancela).
* **askopenfilenames**: Permite seleccionar múltiples archivos a la vez. Devuelve una tupla con las rutas de los archivos.
* **asksaveasfilename**: Abre el diálogo para guardar un archivo. Devuelve la ruta y el nombre que el usuario haya asignado.
* **askdirectory**: Abre un diálogo para seleccionar una carpeta o directorio. Devuelve la ruta de la carpeta.
* **askopenfile**: Igual que askopenfilename, pero en lugar de la ruta, abre el archivo directamente y devuelve el objeto de archivo listo para leer.
* **askopenfiles**: Abre múltiples archivos y devuelve una lista de objetos de archivo.
* **asksaveasfile**: Abre el archivo seleccionado en modo escritura para guardar los datos directamente.

___
### askopenfilename

```python
import tkinter as tk
from tkinter import filedialog

def seleccionar_archivo():
    # Opciones de configuración (opcional)
    tipos_archivos = [
        ('Archivos de texto', '*.txt'),
        ('Documentos PDF', '*.pdf'),
        ('Todos los archivos', '*.*')
    ]
    
    # Abre el cuadro de diálogo y guarda la ruta del archivo seleccionado
    ruta_archivo = filedialog.askopenfilename(
        title="Selecciona un archivo",
        initialdir="/",
        filetypes=tipos_archivos
    )
    
    # Verifica si el usuario seleccionó un archivo o canceló
    if ruta_archivo:
        label_resultado.config(text=f"Archivo seleccionado:\n{ruta_archivo}")
    else:
        label_resultado.config(text="No se seleccionó ningún archivo.")


ventana = tk.Tk()
ventana.title("Ventana principal")
ventana.geometry("400x300")

boton_abrir = tk.Tk.Button if False else tk.Button(ventana, text="Buscar Archivo", command=seleccionar_archivo)
boton_abrir.pack(pady=20)

label_resultado = tk.Label(ventana, text="Ningún archivo seleccionado", wraplength=350)
label_resultado.pack(pady=20)

ventana.mainloop()
```

### askopenfilenames

```python
import tkinter as tk
from tkinter import filedialog

def seleccionar_archivos():
    # Abrir el cuadro de diálogo para seleccionar múltiples archivos
    archivos_seleccionados = filedialog.askopenfilenames(
        title="Selecciona uno o más archivos",
        filetypes=[("Archivos de texto", "*.txt"), ("Todos los archivos", "*.*")]
    )
    
    # askopenfilenames devuelve una tupla con las rutas de los archivos
    if archivos_seleccionados:
        print("Archivos seleccionados:")
        for archivo in archivos_seleccionados:
            print(archivo)
    else:
        print("No se seleccionó ningún archivo.")

ventana = tk.Tk()
ventana.title("Ventana principal")
ventana.geometry("400x300")

boton = tk.Button(ventana, text="Buscar Archivos", command=seleccionar_archivos)
boton.pack(pady=50)

ventana.mainloop()
```

### asksaveasfilename

```python
import tkinter as tk
from tkinter import filedialog

def guardar_archivo():
    ruta_archivo = filedialog.asksaveasfilename(
        title="Guardar archivo como",
        defaultextension=".docx",
        filetypes=[("Archivos de texto", "*.txt"), ("Archivos de word", "*.docx"), ("Todos los archivos", "*.*")]
    )
    
    if ruta_archivo:
        print(f"Archivo guardado con éxito en: {ruta_archivo}")


ventana = tk.Tk()
ventana.title("Ventana principal")
ventana.geometry("400x300")

boton_guardar = tk.Button(ventana, text="Guardar como...", command=guardar_archivo)
boton_guardar.pack(pady=10)

ventana.mainloop()
```

## Entrada de datos
* askstring
* askinteger
* askfloat
* askcolor

```python
import tkinter as tk
from tkinter import colorchooser, simpledialog

def pedir_cadena():
    cadena = simpledialog.askstring("Entrada de cadenas", "Escribe una cadena")
    print(cadena)

def pedir_entero():
    entero = simpledialog.askinteger("Entrada de enteros", "Escribe un entero")
    print(entero)

def pedir_decimal():
    decimal = simpledialog.askfloat("Entrada de decimales", "Escribe un decimal")
    print(decimal)

def cambiar_color():
    color = colorchooser.askcolor(title="Selecciona un color")
    # Devuelve una tupla con los colores RGB en formato decimal y hexadecimal ((255, 0, 0), '#ff0000'))
    # color[1] contiene el código Hexadecimal (ej: '#ffffff')
    if color[1]:
        ventana.config(bg=color[1])  # Cambia el fondo de la ventana
        print(f"Color seleccionado: {color[1]} (RGB: {color[0]})")


ventana = tk.Tk()
ventana.title("Ventana principal")
ventana.geometry("400x300")

boton1 = tk.Button(ventana, text="Pedir cadena", command=pedir_cadena)
boton1.pack(pady=20)

boton2 = tk.Button(ventana, text="Pedir entero", command=pedir_entero)
boton2.pack(pady=20)

boton3 = tk.Button(ventana, text="Pedir decimal", command=pedir_decimal)
boton3.pack(pady=20)

boton4 = tk.Button(ventana, text="Elegir color de Fondo", command=cambiar_color)
boton4.pack(pady=20)


ventana.mainloop()
```

# Widget `Button`

permite al usuario interactuar con la aplicación ejecutando una acción o función cuando se hace clic en él.

Atributos:
* **text**: El texto que se muestra dentro del botón.
* **command**: La función o método que se ejecutará cuando el usuario presione el botón (sin usar paréntesis ()).
* **state**: Controla el estado del botón. Puede ser NORMAL (activo) o DISABLED (desactivado/gris).

```python
import tkinter as tk

# Función que se ejecuta al presionar el botón
def saludar():
  print("¡Hola! Has presionado el botón.")

ventana = tk.Tk()
ventana.title("Ejemplo de Button en Tkinter")
ventana.geometry("300x150")

# Crear el botón
boton = tk.Button(
    ventana,
    text="Haz clic aquí",
    command=saludar,
    bg="#4CAF50",
    fg="white",
    font=("Arial", 12),
    state="disabled"
)

# Mostrar el botón en la ventana
boton.pack(expand=True)

# Iniciar el bucle principal de la aplicación
ventana.mainloop()
```

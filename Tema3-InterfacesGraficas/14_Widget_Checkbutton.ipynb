# Widget `Checkbutton`

Permite crear casillas de verificación para seleccionar una o varias opciones de forma simultánea. Tenemos tanto versión clásica como temática.

Propiedades:
* `text`: Texto que se muestra junto a la casilla.
* `variable`: Enlaza una variable de control (BooleanVar, IntVar o StringVar) para rastrear el estado del botón.
* `onvalue`: Valor que toma la variable asociada cuando el Checkbutton está marcado (por defecto suele ser `1` o `True`).
* `offvalue`: Valor que toma la variable asociada cuando el Checkbutton está desmarcado (por defecto suele ser `0` or `False`).
* `command`: Función o método que se ejecuta automáticamente cada vez que el usuario hace clic en el botón.
* `state`: Controla el estado del widget. Puede ser "normal" (activo), "disabled" (deshabilitado/gris) o "active" (enfoque/hover).
* `selectcolor`: Color del interior de la casilla cuando está marcada.

___
### Control Checkbutton clásico y temático

```python
import tkinter as tk
from tkinter import ttk

def mostrar_estado():
    if var_control1.get(): # cambiar
        print("Estado", f"La casilla1 sí está seleccionada: {var_control1.get()}")
        print("Estado", f"La casilla2 sí está seleccionada: {var_control2.get()}")
    else:
        print("Estado", f"La casilla1 no está seleccionada: {var_control1.get()}")
        print("Estado", f"La casilla2 sí está seleccionada: {var_control2.get()}")

ventana = tk.Tk()
ventana.title("Widget Checkbutton")
ventana.geometry("400x300")

# Contról de la librería estándar tk
var_control1 = tk.IntVar()

casilla1 = tk.Checkbutton(
    ventana,
    text="Aceptar términos y condiciones",
    variable=var_control1,
    onvalue=1,
    offvalue=0,
    #selectcolor="red",
    state="normal", # Cambiar a "disabled" para deshabilitar
    command=mostrar_estado # Permite poner la acción de pulsado    
)

casilla1.pack(pady=20)


# Contról de la librería temática ttk
var_control2 = tk.IntVar()

casilla2 = ttk.Checkbutton(
    ventana,
    text="Aceptar términos y condiciones",
    variable=var_control2,
    onvalue=1,
    offvalue=0,
    state="active", # Cambiar a "disabled" para deshabilitar
    command=mostrar_estado # Permite poner la acción de pulsado    
)

casilla2.pack(pady=20)

boton = tk.Button(ventana, text="Verificar", command=mostrar_estado)
boton.pack(pady=10)

ventana.mainloop()
```

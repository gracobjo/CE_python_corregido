# El contenedor PanedWindow
Es un contenedor que permite organizar otros widgets (como marcos, textos o botones) en paneles ajustables.

Su característica principal es que introduce una barra de separación (llamada sash) que el usuario puede arrastrar con el ratón para cambiar el tamaño de los paneles de forma dinámica.

Propiedades:

* **orient**: Define la disposición de los componentes internos. Puede ser `tk.HORIZONTAL` (los paneles se colocan uno al lado del otro) o `tk.VERTICAL` (uno encima de otro). El valor por defecto suele ser horizontal.
* **showhandle**: Un valor booleano (True o False). Si se activa, muestra un pequeño cuadro o "tirador" (handle) sobre la línea divisoria para que sea más fácil de clicar y arrastrar.
* **handlepad** y **handlesize**: Controlan la estética del tirador. handlesize define el tamaño en píxeles del cuadro y handlepad la distancia entre el tirador y el borde del widget.
* **sashwidth** y **sashpad**: Controlan la barra de separación. sashwidth ajusta el grosor de la línea divisoria y sashpad añade espacio de relleno a ambos lados de la barra.
* **sashrelief**: Define el tipo de relieve tridimensional de la barra divisoria (por ejemplo, tk.RAISED, tk.SUNKEN, tk.FLAT).
* **opaqueresize**: Determina cómo se redimensionan los paneles al arrastrar la barra. Si es True (por defecto), los widgets se redimensionan en tiempo real mientras arrastras. Si es False, solo se mueve una línea guía y los widgets cambian de tamaño al soltar el ratón.
* **borderwidth** (o bd) y **relief**: Definen el grosor del borde exterior del PanedWindow y su estilo tridimensional.

Propiedades de los paneles internos (Frames):
* **minsize**: Define el tamaño mínimo (en píxeles) que puede tener ese panel. Evita que el usuario oculte por completo un panel al arrastrar la barra.
* **weight**: Determina cómo se distribuye el espacio adicional cuando la ventana principal cambia de tamaño. Un panel con weight=1 se expandirá, mientras que uno con weight=0 mantendrá su tamaño fijo.
* **padx** y **pady**: Añaden espacio de separación (relleno) externo alrededor del widget dentro de su panel.
* **sticky**: Controla cómo se alinea o estira el widget si el panel es más grande que él (utiliza puntos cardinales como N, S, E, W).



```python
import tkinter as tk

# Crear la ventana principal
ventana = tk.Tk()
ventana.title("Ejemplo de PanedWindow")
ventana.geometry("600x400")

# 1. Crear el PanedWindow principal (de orientación horizontal)
panel_principal = tk.PanedWindow(ventana, orient=tk.HORIZONTAL, sashrelief=tk.RAISED, opaqueresize=True)

# panel_principal.config(showhandle=True, opaqueresize=False,handlesize=40,handlepad=150)
# panel_principal.config(sashwidth=3,sashpad=40)
# panel_principal.config(bd=50,relief=tk.RAISED)

panel_principal.pack(fill=tk.BOTH, expand=True)

# 2. Crear los componentes que irán dentro del panel
# Panel Izquierdo (un contenedor Frame con botones)
marco_izquierdo = tk.Frame(panel_principal, bg="#d3d3d3", width=150)

btn1 = tk.Button(marco_izquierdo, text="Opción 1")
btn1.pack(pady=5, fill=tk.X, padx=10)

# Panel Derecho (un contenedor Frame con un área de texto)
marco_derecho = tk.Frame(panel_principal, bg="white")
txt_contenido = tk.Text(marco_derecho, wrap=tk.WORD)
txt_contenido.pack(fill=tk.BOTH, expand=True, padx=5, pady=5)
txt_contenido.insert(tk.END, "Arrastra la barra gris de la izquierda para cambiar el tamaño de los paneles.")

# 3. Agregar los marcos al PanedWindow
panel_principal.add(marco_izquierdo)
panel_principal.add(marco_derecho)

ventana.mainloop()
```

# Variables de control

Las variables de control (también llamadas variables de Tkinter o Variable Objects) son objetos especiales que se utilizan para conectar (vincular) los datos de los widgets (como campos de texto, botones de opción o casillas de verificación) con el código de Python.
* `StringVar`: Maneja cadenas de texto (str). Es la más utilizada para campos de entrada de texto (Entry) o etiquetas dinámicas (Label).
* `IntVar`: Maneja números enteros (int). Se usa habitualmente para almacenar el estado de casillas de verificación (Checkbutton) o selecciones numéricas.
* `DoubleVar`: Maneja números de punto flotante o decimales (float). Es ideal para widgets como barras de deslizamiento (Scale).
* `BooleanVar`: Maneja valores booleanos (True o False). Muy útil para conocer si un Checkbutton está marcado o no.

Todas estas variables comparten los mismos métodos esenciales para gestionar su contenido:
1. `set(valor)`: Establece o cambia el valor de la variable. Al hacerlo, todos los widgets vinculados se actualizan automáticamente.
2. `get()`: Devuelve el valor actual almacenado en la variable (con su tipo de dato correspondiente).
3. trace_add(mode, callback): Permite "observar" la variable y ejecutar una función automáticamente cuando el valor se lee ("read"), se modifica ("write") o se borra ("unset").

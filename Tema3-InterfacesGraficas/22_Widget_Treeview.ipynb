# Widget `Treeview`

El widget `ttk.TreeView` se utiliza para mostrar datos en formatos jerárquicos (árboles) o tabulares (listas/tablas).

Propiedades:
* `columns`: Lista de cadenas de texto que definen los identificadores de las columnas (ej. columns=('nombre', 'edad')). Puede ser una tupla o lista principalmente.
* `displaycolumns`: Define qué columnas se muestran y en qué orden. Puede ser `#all` o una lista de índices/identificadores.
* `height`: Número de filas que el widget mostrará sin necesidad de hacer scroll.
* `padding`: Espacio interno alrededor del contenido del widget.
* `selectmode`: Controla cómo se seleccionan los elementos. Valores: `extended` (múltiple, por defecto), `browse` (solo uno) o `none` (desactivado).
* `show`: Tipo de control. Valores: `tree`' (árbol), `headings` (tabla) o `tree headings` (ambos).

Métodos:
* `insert`: Inserta un nuevo elemento indicando los siguientes campos:
	* `padre`: El ID del elemento padre (usa "" para la raíz).
	* `indice`: Posición de inserción (índice numérico o 'end').
	* `iid`: ID único opcional para el elemento. Si no se indica se le asigna uno automáticamente.
	* `**kw`: Propiedades como `text`, `values` (tupla de datos para las columnas) y tags.
* `delete`: Elimina uno o más elementos del árbol.
* `detach`: Desvincula temporalmente uno o más elementos del árbol (no los borra de la memoria y se pueden volver a insertar con `move`).
* `move`: Mueve un elemento a una nueva posición o a un nuevo padre.
* `exists`: Devuelve `True` si el ID del elemento existe en el árbol.
* `item`: Consulta o modifica las propiedades de un elemento específico (como su text, values, tags o open).
* `set`: Lee o cambia el valor de una celda específica. Si solo se pasa item y column, devuelve el valor actual.
* `get_children`: Devuelve una tupla con los IDs de los hijos del elemento especificado. Si se omite, devuelve los de la raíz ("").
* `parent`: Devuelve el ID del elemento padre del ítem especificado.
* `selection`: Devuelve una tupla con los IDs de todos los elementos actualmente seleccionados.
* `column`: Consulta o modifica las propiedades de una columna (ej. width, anchor, stretch, minwidth).
* `heading`: Consulta o modifica las opciones del encabezado de una columna (ej. text para el título, image, command para asignar una función al hacer clic).

Eventos:
* `<<TreeviewSelect>>`: Se ejecuta cuando cambia la selección de los elementos (filas). Es el evento más utilizado.
* `<<TreeviewOpen>>`: Se ejecuta justo cuando un elemento con hijos se despliega (se abre con el símbolo +).
* `<<TreeviewClose>>`: Se ejecuta cuando un elemento con hijos se cierra (con el símbolo -).

**IMPORTANTE**: Los elementos, tanto de la lista como del arbol, llevan una estructura de datos personalizable (lista, tupla...) por el usuario a través de la propiedad `values`.

### Lista básica
Para selección de un único elemento

```python
import tkinter as tk
from tkinter import ttk
from tkinter import messagebox

ventana = tk.Tk()
ventana.title("Widget Treeview como lista")
ventana.geometry("500x400")

# Se definen las columnas de la tabla
columnas = ["id", "nombre", "edad", "profesion"]

# Se crea el widget Treeview
lista = ttk.Treeview(ventana, columns=columnas, show="headings", displaycolumns="#all", selectmode="browse")

# Se configuran los encabezados (headings) de las columnas
lista.heading("id", text="ID",command=lambda: print("Clic en el encabezado de id"))
lista.heading("nombre", text="Nombre",command=lambda: print("Clic en el encabezado de nombre"))
lista.heading("edad", text="Edad",command=lambda: print("Clic en el encabezado de edad"))
lista.heading("profesion", text="Profesión",command=lambda: print("Clic en el encabezado de profesión"))

# Se configura el ancho y la alineación de las columnas
lista.column("id", width=50, anchor="center")
lista.column("nombre", width=150, anchor="w")
lista.column("edad", width=80, anchor="center")
lista.column("profesion", width=150, anchor="w")

# Se insertan datos en la tabla
datos = [
    (1, "Ana Pérez", 28, "Ingeniera"),
    (2, "Carlos Gómez", 34, "Diseñador"),
    (3, "María López", 41, "Arquitecta"),
    (4, "Luis Torres", 22, "Estudiante"),
    (6, "Isabel Martínez", 53, "Policía"),
    (7, "Juan Mediavilla", 36, "Programador"),
    (8, "Francisco Álvarez", 46, "Taxista"),
    (10, "Susana Alonso", 19, "Estudiante"),
    (11, "Marcos Fuentes", 43, "Autónomo"),
    (12, "Belén Sánchez", 55, "Comercial")
]

# Insert
for persona in datos:
    lista.insert("", tk.END, values=persona)
    
lista.insert("",4,iid="5",values=(5, "Marta Fernández", 41, "Profesor")) # Inserción en posición
lista.insert("",8,values=(9, "Fernando García", 38, "Bombero")) #Inserción en posición y con id personalizado

lista.pack(fill=tk.BOTH, expand=True, padx=10, pady=10)


def seleccionar(event):
    elementos = lista.selection()
    if len(elementos) > 0:
        texto = ""
        identificador = elementos[0]
        datos = lista.item(identificador) 
        texto += "\nIdentificaxor interno:" + str(identificador)
        texto += "\nId:" + str(datos["values"][0])
        texto += "\nNombre:" + str(datos["values"][1])
        texto += "\nEdad:" + str(datos["values"][2])
        texto += "\nProfesión:" + str(datos["values"][3])
        print(texto)  

lista.bind("<<TreeviewSelect>>", seleccionar)


# Eliminar
def eliminar_seleccionado():
    lista.delete(lista.selection())
    
boton = tk.Button(ventana, text="Eliminar seleccionado", command=eliminar_seleccionado)
boton.pack(pady=5)

# Mover
def mover_seleccionado():
    lista.move(lista.selection(),parent="",index=0)
    
boton = tk.Button(ventana, text="Mover seleccionado al principio", command=mover_seleccionado)
boton.pack(pady=5)


# Detach
def detach_seleccionado():
    quitado = lista.selection()
    if lista.exists(quitado): # Exists
        messagebox.showinfo("Información", "Sí existe y se va a desasociar")
        lista.detach(quitado)
        lista.move(quitado,parent="",index=0) # Descomentar para comprobar que se reengancha con move y se pone al principio en este caso
    
boton = tk.Button(ventana, text="Detach seleccionado", command=detach_seleccionado)
boton.pack(pady=5)


# Set
def cambiar_seleccionado():
    lista.set(lista.selection(),"profesion","Jubilado")
    
boton = tk.Button(ventana, text="Cambiar seleccionado y jubilarlo", command=cambiar_seleccionado)
boton.pack(pady=5)

ventana.mainloop()
```

### Árbol básico

```python
import tkinter as tk
from tkinter import ttk

ventana = tk.Tk()
ventana.title("Widget Treeview como árbol")
ventana.geometry("300x350")

# Se crea el Treeview especificando show="tree"
arbol = ttk.Treeview(ventana, show="tree")
arbol.pack(fill=tk.BOTH, expand=True, padx=10, pady=10)

# Se agregan nodos principales (Raíces)
# El primer argumento es el elemento padre (vacío "" significa que está en la raíz)
# 'end' indica que se añade al final de la lista
nodo_personas = arbol.insert("", "end", text="Personas")

datos = [
    (1, "Ana Pérez", 28, "Ingeniera"),
    (2, "Carlos Gómez", 34, "Diseñador"),
    (3, "María López", 41, "Arquitecta"),
    (4, "Luis Torres", 22, "Estudiante")
]

for i in datos:
    nodo_id = arbol.insert(nodo_personas, "end", text=i[0],values=i)
    arbol.insert(nodo_id, "end", text=i[1],values=i)
    arbol.insert(nodo_id, "end", text=i[2],values=i)
    arbol.insert(nodo_id, "end", text=i[3],values=i)

arbol.item(nodo_personas, open=True)

# Se especifican los eventos
def seleccionar(event):
    elementos = arbol.selection()
    print(elementos)
    if len(elementos) > 0:
        identificador = elementos[0]
        datos = arbol.item(identificador) 
        if datos["values"] != "":
            print(datos["values"])
        else:
            print("Haz clic en el ID de cada persona")
        print(datos)

arbol.bind("<<TreeviewSelect>>", seleccionar)

def abrir(event):
    print("Se abre un nodo")

arbol.bind("<<TreeviewOpen>>", abrir)

def cerrar(event):
    print("Se cierra un nodo")
    
arbol.bind("<<TreeviewClose>>", cerrar)
    
# Eliminar
def eliminar_seleccionado():
    arbol.delete(arbol.selection())
    
boton = tk.Button(ventana, text="Eliminar seleccionado", command=eliminar_seleccionado)
boton.pack(pady=5)

# Obtener padre
def obtener_padre():
    identificador = arbol.selection()[0]
    datos = arbol.item(arbol.parent(identificador))
    messagebox.showinfo("Padre", datos["text"])
    
boton = tk.Button(ventana, text="Obtener padre", command=obtener_padre)
boton.pack(pady=5)

# Obtener hijos
def obtener_hijos():
    identificador = arbol.selection()[0]
    tupla_hijos = arbol.get_children(identificador)
    cadena = ""
    for i in tupla_hijos:
        datos = arbol.item(i)
        cadena += str(datos["text"]) + "\n"
    messagebox.showinfo("Hijos", cadena)
    
boton = tk.Button(ventana, text="Obtener hijos", command=obtener_hijos)
boton.pack(pady=5)

ventana.mainloop()
```

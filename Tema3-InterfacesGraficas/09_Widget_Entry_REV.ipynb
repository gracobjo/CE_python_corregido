# Widget `Entry`
Es un campo de texto de una sola línea que permite al usuario escribir y editar datos, ideal para formularios, contraseñas o búsquedas. Se define mediante la clase `tk.Entry`.

Parámetros
* **width**: Define el ancho del campo medido en número de caracteres.
* **show**: Oculta el texto real con un símbolo (por ejemplo, `show="*"` para contraseñas).
* **state**: Controla si se puede interactuar con el campo; sus valores son `tk.NORMAL`, `tk.DISABLED` o `tk.READONLY`.
* **textvariable**: Enlaza el contenido del Entry con una variable de control de Tkinter, como `tk.StringVar()`.
* **justify**: Alinea el texto en el interior (`tk.LEFT`, `tk.CENTER` o `tk.RIGHT`).
* **font**: Define la tipografía, estilo y tamaño de la letra.
* **bg** / **fg**: Cambian el color de fondo (background) y el color del texto (foreground).
* **bd**: Grosor del borde.
* **relief**: Tipo de borde

Métodos:
* **config(opciones)** / **configure(opciones)**: Modifica parámetros del widget después de haber sido creado.
* **get()**: Retorna un `str` con el texto actual que contiene la caja de entrada.
* **insert(indice, cadena)**: Inserta texto en una posición numérica específica.
* **delete(primero, ultimo=None)**: Borra el texto comprendido entre el índice inicial y final.
* **focus_set()**: Asigna el foco del teclado directamente al widget Entry.
* **icursor(indice)**: Mueve el cursor de escritura a la posición indicada por el índice.
* **select_range(inicio, fin)**: Selecciona visualmente el texto entre las posiciones indicadas.
* **select_clear()**: Deselecciona cualquier texto que estuviera marcado.

```python
import tkinter as tk

ventana = tk.Tk()
ventana.title("Widget Entry")
ventana.geometry("400x300")

tk.Label(ventana, text="Introduce tu nombre:").pack(pady=5)
entrada_usuario = tk.Entry(
    ventana, 
    width=30, 
    font=("Helvetica", 12), 
    bg="#ffffff", 
    fg="blue", 
    bd=3, 
    relief="sunken",
    justify="center"
)
entrada_usuario.pack(pady=5)

ventana.mainloop()
```

### Métodos `icursor(indice)`, `select_range(inicio,fin)` `focus_set()` y `select_clear()`

```python
import tkinter as tk

def mostrar_texto():
    # .get() obtiene el contenido del widget
    contenido = entrada_usuario.get()
    etiqueta_resultado.config(text=f"Escribiste: {contenido}")

def borrar_texto():
    # .delete(0, "end") borra desde el primer carácter hasta el final
    entrada_usuario.delete(0, tk.END)

def insertar_hola():
    # .insert("insert", ...) añade texto donde esté el cursor parpadeando
    entrada_usuario.insert(tk.INSERT, "¡Hola! ")

# Configuración de la ventana principal
ventana = tk.Tk()
ventana.title("Ejemplo de Tkinter Entry")
ventana.geometry("400x300")

# 1. Entrada de texto con diseño personalizado
tk.Label(ventana, text="Introduce tu nombre:").pack(pady=5)
entrada_usuario = tk.Entry(
    ventana, 
    width=30, 
    font=("Helvetica", 12), 
    bg="#f0f0f0", 
    fg="blue", 
    bd=3, 
    relief="sunken",
    justify="center"
)
entrada_usuario.pack(pady=5)
entrada_usuario.insert(0, "Escribe aquí...") # Texto por defecto

# 2. Entrada de texto tipo Contraseña (Propiedad 'show')
tk.Label(ventana, text="Introduce tu contraseña:").pack(pady=5)
entrada_password = tk.Entry(ventana, width=30, show="*")
entrada_password.pack(pady=5)

# Botones para ejecutar los métodos
btn_mostrar = tk.Button(ventana, text="Obtener Texto (get)", command=mostrar_texto)
btn_mostrar.pack(pady=5)

btn_insertar = tk.Button(ventana, text="Insertar 'Hola' (insert)", command=insertar_hola)
btn_insertar.pack(pady=5)

btn_borrar = tk.Button(ventana, text="Limpiar Campo (delete)", command=borrar_texto)
btn_borrar.pack(pady=5)

# Etiqueta para mostrar los resultados de .get()
etiqueta_resultado = tk.Label(ventana, text="", fg="green", font=("Arial", 11, "bold"))
etiqueta_resultado.pack(pady=10)

ventana.mainloop()
```

### Métodos `icursor(indice)`, `select_range(inicio,fin)` y `select_clear()`

```python
import tkinter as tk

def seleccionar_texto():
    entrada_usuario.focus_set()
    entrada_usuario.select_range(5,14)

def quitar_seleccion_texto():
    entrada_usuario.focus_set()
    entrada_usuario.select_clear()

def posicionarse():
    entrada_usuario.focus_set()
    entrada_usuario.icursor(3)

# Configuración de la ventana principal
ventana = tk.Tk()
ventana.title("Control Entry")
ventana.geometry("400x300")

texto = tk.StringVar()
texto.set("Escribe algo abajo...")
entrada_usuario = tk.Entry(
    ventana,
    textvariable=texto,
    width=30, 
    font=("Helvetica", 12), 
    bg="#f0f0f0", 
    fg="blue", 
    bd=3, 
    relief="sunken",
    justify="left"
)
entrada_usuario.pack(pady=5)

btn1 = tk.Button(ventana, text="Seleccionar", command=seleccionar_texto)
btn1.pack(pady=5)

btn2 = tk.Button(ventana, text="Quitar selección", command=quitar_seleccion_texto)
btn2.pack(pady=5)

btn3 = tk.Button(ventana, text="Posicionarse", command=posicionarse)
btn3.pack(pady=5)

entrada_auxiliar = tk.Entry(ventana,width=30)
entrada_auxiliar.pack(pady=5)

ventana.mainloop()
```

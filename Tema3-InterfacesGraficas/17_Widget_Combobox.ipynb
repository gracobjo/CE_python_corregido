# Widget `Combobox`

Este widget es una combinación de un campo de texto (`Entry`) y una lista desplegable (`Listbox`). Permite al usuario seleccionar una opción de una lista predefinida o escribir su propia opción.

Propiedades:
* `values`: Una lista o tupla con las opciones que se mostrarán en el menú desplegable.
* `state`: Controla si el usuario puede escribir.
	* `normal` (por defecto): El usuario puede seleccionar o escribir texto libre.
	* `readonly`: El usuario solo puede seleccionar opciones de la lista (no puede escribir).
	* `disabled`: El control se desactiva por completo.
* `textvariable`: Se le asigna una variable de control para rastrear y modificar el texto seleccionado de forma automática.
* `width`: El ancho del cuadro de texto en caracteres.

Métodos:
* `get`: Devuelve el texto que está actualmente visible en el Combobox.
* `set`: Cambia el texto visible del Combobox al valor especificado.
* `current`: Selecciona el elemento de la lista según su posición numérica (empezando desde 0). Si no se pasa ningún argumento, devuelve el índice del elemento seleccionado actualmente.

Eventos:
* `<<ComboboxSelected>>`: Este evento virtual se dispara únicamente cuando el usuario selecciona una opción de la lista desplegable. No se dispara si el usuario escribe caracteres manualmente (en modo normal).

___
### Ejemplo de Combobox para leer valores y seleccionar elementos

```python
import tkinter as tk
from tkinter import ttk

def mostrar_seleccion(event):
  etiqueta.config(text=f"Seleccionaste: {combo.get()} - {texto_dinamico.get()}")

ventana = tk.Tk()
ventana.title("Widget Combobox")
ventana.geometry("400x300")

texto_dinamico = tk.StringVar()
texto_dinamico.set("Escribe algo abajo...")

label = ttk.Label(ventana, text="Elige una opción:")
label.pack(pady=10)

# Se crea el Combobox
opciones = ["Opción 1", "Opción 2", "Opción 3", "Opción 4"]
combo = ttk.Combobox(ventana, values=opciones, textvariable=texto_dinamico, state="normal") # normal, readonly y disabled

combo.set("Opción 1")  # Valor por defecto
combo.pack(pady=5)

# Se asocia el evento de selección a un método
combo.bind("<<ComboboxSelected>>", mostrar_seleccion)

etiqueta = ttk.Label(ventana, text="Seleccionaste: Opción 1")
etiqueta.pack(pady=20)

def seleccionar(indice):
    combo.current(indice)
    
btn1 = ttk.Button(ventana, text="Seleccionar la tercera opción", command=lambda: seleccionar(2))
btn1.pack(pady=10)

def pulsa_informacion():
    print(f"Seleccionaste: {combo.get()} - {texto_dinamico.get()}")

btn2 = ttk.Button(ventana, text="Mostrar lo seleccionado", command=pulsa_informacion)
btn2.pack(pady=10)

ventana.mainloop()
```

# Los contenedores `Frame` y `LabelFrame`
___
# `Frame`
Es un contenedor invisible o con borde que sirve para agrupar y organizar otros elementos (como botones o etiquetas) dentro de una ventana principal.

Aspectos clave:
* **Contenedor**: Recibe como primer parámetro la ventana principal a la que se asocia.
* **Asignación de controles**: Los controles se asignan al frame en lugar de a la ventana principal
* **Organización**: Permite tener su geometría al igual que la ventana principal.

Las *propiedades principales* ya se han visto en otros controles y su funcionamiento es el mismo:
* `bg`: Color de fondo.
* `bd`: Grosor del borde.
* `relief`: Tipo de borde. Posibles valores:
    * `flat`: Plano, sin efecto de relieve (es el valor por defecto).
    * `raised`: Apariencia elevada o tridimensional hacia afuera (como un botón tradicional sin pulsar).
    * `sunken`: Apariencia hundida o de bajo relieve (como un campo de texto o un botón presionado).
    * `groove`: Borde tallado o en relieve hacia adentro.
    * `ridge`: Borde estriado o saliente (lo opuesto a groove).
    * `solid`: Borde de línea simple, continua y uniforme.
* `cursor`: Tipo de icono del ratón. Algunos valores serían: `arrow`, `hand1`, `hand2`, `cross`, etc.
* `width`: Ancho.
* `heigth`: Alto.
* `padx`: Margen horizontal externo.
* `pady`: Margen vertical externo.

```python
import tkinter as tk

# Ventana principal
ventana = tk.Tk()
ventana.title("Ventana principal")
ventana.geometry("400x300")

# Se crea el Frame indicando que pertenece a la ventana
frame = tk.Frame(ventana, bg="lightblue", bd=2, relief="sunken",cursor="cross")
frame.pack(pady=20, padx=20, fill="both", expand=True)

# Se añaden elementos dentro del Frame
label = tk.Label(frame, text="Estoy dentro del Frame", bg="lightblue")
label.pack(pady=10)

boton = tk.Button(frame, text="Botón de ejemplo")
boton.pack(pady=10)

ventana.mainloop()
```

___
### Método 
Por defecto, un frame se encoge de forma automática para ajustarse exactamente al tamaño de los elementos que tiene dentro (botones, etiquetas, etc.), ignorando las propiedades `width` y `height`.

Para forzar a que el frame mantenga un tamaño fijo establecido, se debe desactivar la propagación de tamaño con:

* **Gestor de tipo pack**: `frame.pack_propagate(False)`
* **Gestor de tipo grid**: `frame.grid_propagate(False)`

```python
import tkinter as tk

ventana = tk.Tk()
ventana.title("Uso de pack_propagate")
ventana.geometry("400x300")

# FRAME 1: Con propagación activa (Comportamiento por defecto)
frame_automatico = tk.Frame(ventana, bg="lightblue", width=150, height=150)
frame_automatico.pack(side="left", padx=20, pady=20)

# Aunque el frame mide 150x150, se encogerá para envolver justo a este botón y nada más
btn1 = tk.Button(frame_automatico, text="Ajuste Automático")
btn1.pack(padx=10, pady=10)


# FRAME 2: Con propagación desactivada (Tamaño fijo)
frame_fijo = tk.Frame(ventana, bg="lightgreen", width=150, height=150)
frame_fijo.pack(side="right", padx=20, pady=20)

# Desactivamos la propagación para que mantenga sus 150x150 píxeles
frame_fijo.pack_propagate(False) # Comentar y descomentar para explicar

# El botón se coloca dentro, pero el frame no cambia su tamaño original
btn2 = tk.Button(frame_fijo, text="Tamaño Fijo")
btn2.pack(padx=10, pady=10)

ventana.mainloop()
```

# `LabelFrame`
Funciona como `Frame` pero contiene un borde visual y un texto en dicho borde.
* `text`: Texto del `LabelFrame` que aparece en el borde del mismo.

```python
import tkinter as tk

ventana = tk.Tk()
ventana.title("Ejemplo de LabelFrame")
ventana.geometry("400x300")

# Se crea el LabelFrame
# El argumento 'text' asigna el título que se mostrará en el borde
contenedor_perfil = tk.LabelFrame(ventana, text=" Información de Perfil ", padx=10, pady=10)
contenedor_perfil.pack(padx=20, pady=20, fill="both", expand="yes")

# Se crean widgets DENTRO del LabelFrame
# Nota cómo pasamos 'contenedor_perfil' como el widget padre (master)
etiqueta_nombre = tk.Label(contenedor_perfil, text="Nombre:")
etiqueta_nombre.grid(row=0, column=0, sticky="w", pady=5)

campo_nombre = tk.Entry(contenedor_perfil)
campo_nombre.grid(row=0, column=1, pady=5)

boton_guardar = tk.Button(contenedor_perfil, text="Guardar")
boton_guardar.grid(row=1, column=0, columnspan=2, pady=10)

ventana.mainloop()
```

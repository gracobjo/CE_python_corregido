# El contenedor `Frame`
Es un contenedor invisible o con borde que sirve para agrupar y organizar otros elementos (como botones o etiquetas) dentro de una ventana principal.

Aspectos clave:
* **Contenedor**: recibe como primer parámetro la ventana principal a la que se asocia.
* **Asignación de controles**: Los controles se asignan al frame en lugar de a la ventan principal
* **Organización**: Permite tener su geometría al igual que la ventana principal.

Las *propiedades principales* ya se han visto en otros controles y su funcionamiento es el mismo:
* bg
* bd
* relief
* cursor
* width y height: Por defecto, un frame se encoge de forma automática para ajustarse exactamente al tamaño de los elementos que tiene dentro (botones, etiquetas, etc.), ignorando las propiedades width y height
* heigth
* padx
* pady


```python
import tkinter as tk

# Ventana principal
root = tk.Tk()
root.title("Ejemplo de Frame")
root.geometry("400x300")

# Crear el Frame
frame = tk.Frame(root, bg="lightblue", bd=2, relief="sunken")
frame.pack(pady=20, padx=20, fill="both", expand=True)
#frame.pack_propagate(False)

# Añadir elementos dentro del Frame
label = tk.Label(frame, text="Estoy dentro del Frame", bg="lightblue")
label.pack(pady=10)

boton = tk.Button(frame, text="Botón de ejemplo")
boton.pack(pady=10)

root.mainloop()
```

```python
import tkinter as tk

# Crear la ventana principal
ventana = tk.Tk()
ventana.title("Ejemplo de pack_propagate")
ventana.geometry("400x300")

# --- FRAME 1: Con propagación activa (Comportamiento por defecto) ---
frame_automatico = tk.Frame(ventana, bg="lightblue", width=150, height=150)
frame_automatico.pack(side="left", padx=20, pady=20)

# Aunque el frame mide 150x150, se encogerá para envolver justo a este botón
btn1 = tk.Button(frame_automatico, text="Ajuste Automático")
btn1.pack(padx=10, pady=10)


# --- FRAME 2: Con propagación desactivada (Tamaño fijo) ---
frame_fijo = tk.Frame(ventana, bg="lightgreen", width=150, height=150)
frame_fijo.pack(side="right", padx=20, pady=20)

# Desactivamos la propagación para que mantenga sus 150x150 píxeles
frame_fijo.pack_propagate(False)

# El botón se coloca dentro, pero el frame no cambia su tamaño original
btn2 = tk.Button(frame_fijo, text="Tamaño Fijo")
btn2.pack(padx=10, pady=10)

# Iniciar la aplicación
ventana.mainloop()
```

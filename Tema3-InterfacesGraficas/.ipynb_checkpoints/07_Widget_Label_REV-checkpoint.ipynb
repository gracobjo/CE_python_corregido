## Widget `Label`

Parámetros
* text: Define el texto plano que mostrará la etiqueta.
* textvariable: Permite asociar una variable de control (como StringVar) para cambiar el texto de forma dinámica.
* bg o background: Cambia el color de fondo del widget.
* fg o foreground: Cambia el color de la letra.
* font: Permite especificar la familia tipográfica, el tamaño y el estilo de la fuente (ejemplo: ("Helvetica", 12, "bold")).
* wraplength: Define el ancho en píxeles a partir del cual el texto se divide automáticamente en varias líneas

```python
from tkinter import *

root = Tk()
root.title("Ejemplo de Label")

# Crear y empaquetar una etiqueta
etiqueta = Label(root, text="¡Hola, este es un Label en Tkinter!")
etiqueta.pack(pady=20)

root.mainloop()
```

#### Propiedad `textvariable`

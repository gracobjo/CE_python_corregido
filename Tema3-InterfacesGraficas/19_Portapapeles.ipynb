# El portapapeles

El portapapeles copiar y pegar texto entre distintos componentes.

Métodos:
* `clipboard_clear()`: Borra por completo el contenido actual del portapapeles del sistema. Es necesario llamarlo antes de agregar texto nuevo si se quiere reemplazar el contenido anterior.
* `clipboard_append(texto)`: Añade la cadena de texto especificada al portapapeles. Si no se usa `clipboard_clear` primero, el nuevo texto se pegará a continuación del que ya existía.
* `clipboard_get()`: Recupera y devuelve el texto que esté almacenado actualmente en el portapapeles del sistema. Lanza una excepción (`_tkinter.TclError`) si el portapapeles está vacío o no contiene texto.

```python
import tkinter as tk
from tkinter import messagebox

def copiar_texto():
  ventana.clipboard_clear()
  ventana.clipboard_append(texto_dinamico.get())
  messagebox.showinfo("Copiado", "Texto guardado en el portapapeles.")

def pegar_texto():
  try:
    texto = ventana.clipboard_get()
    entrada.delete(0, tk.END)
    entrada.insert(0, texto)
  except tk.TclError:
    messagebox.showwarning("Error", "El portapapeles está vacío o no contiene texto.")

ventana = tk.Tk()
ventana.title("Portapapeles")
ventana.geometry("400x300")

texto_dinamico = tk.StringVar()

entrada = tk.Entry(ventana, textvariable=texto_dinamico, width=30)
entrada.pack(pady=20)

btn_copiar = tk.Button(ventana, text="Copiar", command=copiar_texto)
btn_copiar.pack(pady=5)

btn_pegar = tk.Button(ventana, text="Pegar", command=pegar_texto)
btn_pegar.pack(pady=5)

ventana.mainloop()
```

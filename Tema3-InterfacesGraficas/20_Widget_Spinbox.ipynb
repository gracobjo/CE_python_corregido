# Widget `Spinbox`

Permite al usuario seleccionar un número (o un valor de una lista) incrementando o decrementando el valor con dos flechas ascendente/descendente.

Propiedades:
* `from_`: Define el límite inferior del rango numérico (lleva guión bajo porque from es palabra reservada en Python).
* `to`: Define el límite superior del rango numérico.
* `increment`: El valor que aumenta o disminuye al hacer clic en las flechas (por defecto es **1.0**).
* `format`: Formato de cadena para los números flotantes (ej. '%0.2f' para dos decimales).
* `values`: Una tupla o lista de cadenas de texto. Si se usa, el Spinbox alternará entre estos elementos en lugar de usar un rango numérico.
* `textvariable`: Enlaza el valor del widget a una variable de control (como `IntVar` o `StringVar`).
* `state`: Estado del widget. Puede ser 'normal', 'disabled' (deshabilitado) o 'readonly' (solo lectura, el usuario no puede escribir directamente, solo usar las flechas).
* `command`: Función que se ejecuta cada vez que el usuario presiona una de las flechas.
* `wrap`: Booleano (`True/False`). Si es `True`, cuando llegue al valor máximo y se pulse subir, volverá al valor mínimo (y viceversa).
* `justify`: Alineación del texto dentro del campo (`left`, `center`, `right`).
* `width`: Ancho del widget en caracteres (por defecto suele ser 20).

Métodos principales:
* `get`: Devuelve el valor actual del Spinbox como una cadena de texto.
* `set`: Establece el valor del Spinbox al dato indicado. (sólo en ttk)

```python
import tkinter as tk
from tkinter import ttk
from tkinter import messagebox

def mostrar_valor():
    edad_seleccionada = spin_edad.get()
    messagebox.showinfo("Valor Seleccionado", f"Tu edad seleccionada es: {edad_seleccionada}")

ventana = tk.Tk()
ventana.title("Ejemplo de Spinbox")
ventana.geometry("400x300")

etiqueta = tk.Label(ventana, text="Selecciona tu edad:", font=("Arial", 12))
etiqueta.pack(pady=10)

# Se crea el Spinbox
spin_edad = ttk.Spinbox(ventana, from_=18, to=30, increment=0.5, font=("Arial", 12), width=5, wrap=False, format='%0.2f')
spin_edad.pack(pady=10)
spin_edad.set("20.00")

# Se crea un botón para obtener el valor
boton1 = tk.Button(ventana, text="Confirmar", command=mostrar_valor)
boton1.pack(pady=10)

# Se crea un Spinbox no numérico
dias = ('LUNES','MARTES','MIERCOLES','JUEVES','VIERNES','SABADO','DOMINGO')
spin_dias = ttk.Spinbox(ventana, values=dias, font=("Arial", 12), width=15, wrap=True)
spin_dias.pack(pady=10)

def cambiar_valor():
    spin_dias.set("---") # Método sólo disponible en la versión temática del widget
    
boton2 = tk.Button(ventana, text="Cambiar el texto que aparece por 3 guiones", command=cambiar_valor)
boton2.pack(pady=10)

ventana.mainloop()
```

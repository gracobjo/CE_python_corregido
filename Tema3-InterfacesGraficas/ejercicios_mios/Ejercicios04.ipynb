## Cómo se ha resuelto paso a paso

El ejercicio combina datos en memoria, una interfaz gráfica y un historial de operaciones. Este recorrido explica qué aporta cada bloque y por qué se organiza así; más abajo se detallan las funciones una por una.

### 1. Los datos iniciales

El diccionario `personas` es la fuente de datos durante la ejecución. Cada clave es el DNI y su valor es otro diccionario con los datos del alumno, como nombre, edad, ciclo, curso y módulos. Así se puede buscar y modificar un alumno mediante su DNI sin configurar una base de datos. El diccionario vive en memoria: al cerrar o reiniciar el kernel, los cambios no se recuperan automáticamente desde él.

### 2. Preparación de Tkinter y del historial

La línea mágica `%gui tk` integra el bucle de eventos de Tkinter con el de Jupyter. Las importaciones proporcionan los widgets y avisos (`tkinter`, `ttk` y `messagebox`), la lectura y escritura JSON, la fecha de cada operación y el manejo de rutas con `Path`. `archivo_log` identifica `alumnos.log`, donde se guarda el historial acumulado.

### 3. Funciones de apoyo y sincronización de la lista

`ciclos_formativos_disponibles()` reúne los ciclos existentes, elimina duplicados y los ordena sin distinguir mayúsculas. `valor_legible()` adapta valores internos para que se entiendan en los mensajes. `actualizar_lista()` ordena los alumnos por nombre y reconstruye el `Listbox`; si recibe un DNI, también selecciona y muestra esa fila. De esta forma, la lista visible se mantiene sincronizada con `personas`.

### 4. Formulario y estado de edición

El formulario de la derecha usa variables de Tkinter (`StringVar` y `BooleanVar`) enlazadas a sus widgets. Las entradas permiten escribir nombre, DNI y edad; la casilla representa si el alumno repite; los botones de opción seleccionan curso y el `Combobox` ofrece los ciclos disponibles en modo de solo lectura. `abrir_formulario(dni)` carga los datos de un alumno para editarlo; sin DNI, prepara un alta. El diccionario `alumno_en_edicion` conserva el DNI que se está editando y permite distinguir ese caso de un alta nueva.

### 5. Acciones, validaciones y persistencia

Los botones llaman a funciones y los eventos de la lista permiten abrir la selección. `abrir_alumno_seleccionado()` traduce la fila elegida a su DNI y carga sus datos en el panel derecho; tanto el botón «Abrir» como el doble clic (`<Double-Button-1>`, conectado con `bind()`) usan esa misma acción.

`guardar_alumno()` valida los campos antes de guardar. Para editar, parte de una copia de los datos existentes y conserva los campos que no están en el formulario. Después compara los valores anteriores y nuevos para resumir los cambios. Cada operación se añade a `alumnos.log` como una línea JSON independiente (formato JSONL). El registro se escribe antes de actualizar `personas`; si falla la escritura, se informa del error y se conservan los datos en memoria. Al eliminar, también se pide confirmación y se registra la baja antes de borrar al alumno.

### 6. Construcción de la ventana

La ventana principal agrupa la lista y el formulario en paneles. `grid()` coloca los elementos por filas y columnas, mientras que los `Frame` y `LabelFrame` organizan las zonas y sus márgenes. Los botones conectan las acciones de alta, apertura, eliminación y guardado con sus funciones. Al final se llena la lista y se prepara el formulario para una nueva alta.

### 7. Consulta visual del historial

La última celda comprueba si existe `alumnos.log` y, si existe, lee sus líneas JSON para reconstruir los registros. Después genera una tabla HTML con las operaciones más recientes primero y presenta el resultado dentro del notebook mediante `display(HTML(...))`. `escape()` protege la tabla al representar como texto los datos introducidos por los usuarios, y los estilos CSS diferencian visualmente las altas, modificaciones y bajas.

### Ideas clave

- El diccionario mantiene el estado actual mientras se ejecuta el notebook; el log conserva la secuencia de operaciones entre ejecuciones.
- La interfaz responde a eventos: los widgets cambian variables y sus acciones llaman a funciones.
- Registrar el cambio antes de modificar los datos evita que la aplicación muestre una operación como guardada si no pudo escribirla en el historial.
- Jupyter puede mostrar contenido enriquecido, como una tabla HTML, además de texto.

### 1. `ciclos_formativos_disponibles()`

1. Recorre los alumnos guardados en `personas`.
2. Obtiene los valores de `ciclo_formativo` y descarta los vacíos.
3. Usa un conjunto para quitar duplicados y devuelve una lista ordenada sin distinguir mayúsculas y minúsculas.

La lista resultante alimenta el desplegable del formulario.

### 2. `valor_legible(valor)`

Convierte los valores internos del diccionario a texto más claro para los mensajes de confirmación:

- `None` se muestra como «(sin dato)».
- Los booleanos se muestran como «Sí» o «No».
- Los cursos internos `primero` y `segundo` se muestran como «1º curso» y «2º curso».
- Las listas se unen con comas; si están vacías, se muestra «(ninguno)».
- Los demás valores se convierten directamente a texto.

### 3. `actualizar_lista(dni_seleccionado=None)`

1. Ordena los pares DNI-alumno alfabéticamente por nombre, sin distinguir mayúsculas y minúsculas.
2. Vacía el `Listbox` y lo rellena con el nombre y el DNI de cada alumno.
3. Si recibe un DNI, selecciona y desplaza la lista hasta ese alumno.

Se llama al iniciar la ventana y después de guardar o eliminar para que la lista refleje el estado actual del diccionario.

### 4. `abrir_formulario(dni=None)`

Esta función prepara el panel de la derecha; no crea otra ventana.

1. Si recibe un DNI existente, obtiene sus datos y prepara el modo de edición. Sin DNI, prepara un alta nueva.
2. Guarda el DNI que se está editando y cambia el título del formulario.
3. Carga nombre, DNI, edad, repetición y curso en sus variables de Tkinter.
4. Actualiza el desplegable con los ciclos formativos disponibles.
5. En edición selecciona el ciclo del alumno; en un alta propone el primer ciclo disponible.

### 5. `guardar_alumno()`

Es el paso central de alta y edición, ejecutado por el botón «Guardar los cambios».

1. Lee y valida los campos: nombre y DNI obligatorios, edad entera no negativa, DNI no duplicado y ciclo seleccionado. Si hay un error, avisa y detiene el guardado.
2. Para un alta crea el registro inicial. Para una edición copia los datos existentes y actualiza los campos del formulario, conservando el resto, como los módulos.
3. Compara los datos anteriores y nuevos para construir el detalle de cambios.
4. Añade a `alumnos.log` una línea JSON con fecha, operación, DNI y valores anteriores/nuevos. El fichero conserva un historial acumulado.
5. Si no puede escribir el log, muestra un error y no modifica `personas`.
6. Una vez registrado el cambio, actualiza el diccionario en memoria, refresca la lista y muestra una confirmación con los cambios legibles.

El diccionario de la celda inicial no se reescribe: los datos del CRUD viven en memoria mientras se ejecuta el notebook; el log conserva la trazabilidad.

### 6. `eliminar_alumno_seleccionado()`

1. Comprueba que haya un alumno seleccionado; si no, muestra un aviso.
2. Pide confirmación indicando el nombre y el DNI. Si se cancela, no hace cambios.
3. Antes de borrar, registra en `alumnos.log` una baja con los datos del alumno.
4. Si el log no se puede escribir, informa del error y conserva al alumno en el diccionario.
5. Si el registro se guarda, elimina al alumno de `personas`, actualiza la lista, limpia el formulario y confirma la operación.

### 7. `abrir_alumno_seleccionado()`

1. Consulta qué fila está seleccionada en la lista.
2. Si no hay selección, muestra un aviso y termina.
3. Si hay selección, recupera el DNI asociado a esa fila y llama a `abrir_formulario(dni)` para cargar sus datos en el panel de edición. También se puede abrir con doble clic sobre el alumno.

```python
import tkinter as tk
import json
from datetime import datetime
from pathlib import Path
from tkinter import messagebox, ttk

archivo_log = Path("alumnos.log")


def ciclos_formativos_disponibles():
    return sorted(
        {alumno["ciclo_formativo"] for alumno in personas.values() if alumno.get("ciclo_formativo")},
        key=str.casefold,
    )


def valor_legible(valor):
    if valor is None:
        return "(sin dato)"
    if isinstance(valor, bool):
        return "Sí" if valor else "No"
    if valor == "primero":
        return "1º curso"
    if valor == "segundo":
        return "2º curso"
    if isinstance(valor, list):
        return ", ".join(valor) if valor else "(ninguno)"
    return str(valor)


def actualizar_lista(dni_seleccionado=None):
    global alumnos_ordenados
    alumnos_ordenados = sorted(
        personas.items(), key=lambda elemento: elemento[1]["nombre"].casefold()
    )
    lista_alumnos.delete(0, tk.END)
    for indice, (dni, alumno) in enumerate(alumnos_ordenados):
        lista_alumnos.insert(tk.END, f"{alumno['nombre']} ({dni})")
        if dni == dni_seleccionado:
            lista_alumnos.selection_set(indice)
            lista_alumnos.see(indice)


def abrir_formulario(dni=None):
    alumno = personas.get(dni) if dni is not None else None
    es_nuevo = alumno is None
    alumno_en_edicion["dni"] = None if es_nuevo else dni
    titulo_formulario.set("Nuevo alumno" if es_nuevo else "Modificar alumno")
    nombre_var.set("" if es_nuevo else alumno.get("nombre", ""))
    dni_var.set("" if es_nuevo else dni)
    edad_var.set("" if es_nuevo else str(alumno.get("edad", "")))
    repite_var.set(False if es_nuevo else alumno.get("repite", False))
    curso_var.set("primero" if es_nuevo else alumno.get("curso", "primero"))
    opciones = ciclos_formativos_disponibles()
    selector_ciclos["values"] = opciones
    if es_nuevo:
        ciclo_var.set(opciones[0] if opciones else "")
    else:
        ciclo_var.set(alumno.get("ciclo_formativo", ""))


def guardar_alumno():
    nombre = nombre_var.get().strip()
    nuevo_dni = dni_var.get().strip()
    dni_original = alumno_en_edicion["dni"]
    es_nuevo = dni_original is None
    if not nombre or not nuevo_dni:
        messagebox.showerror(
            "Datos incompletos", "El nombre y el DNI son obligatorios.", parent=ventana
        )
        return
    try:
        edad = int(edad_var.get())
    except ValueError:
        messagebox.showerror(
            "Edad no válida", "La edad debe ser un número entero.", parent=ventana
        )
        return
    if edad < 0:
        messagebox.showerror(
            "Edad no válida", "La edad no puede ser negativa.", parent=ventana
        )
        return
    if nuevo_dni in personas and nuevo_dni != dni_original:
        messagebox.showerror(
            "DNI duplicado", "Ya existe un alumno con ese DNI.", parent=ventana
        )
        return
    ciclo = ciclo_var.get()
    if not ciclo:
        messagebox.showerror(
            "Ciclo formativo no seleccionado",
            "Selecciona un ciclo formativo para el alumno.",
            parent=ventana,
        )
        return

    if es_nuevo:
        alumno = {
            "ciudad": "",
            "telefono": "",
            "ciclo_formativo": ciclo,
            "modulos": [],
        }
    else:
        alumno = personas[dni_original].copy()
        alumno["ciclo_formativo"] = ciclo
    alumno.update({
        "nombre": nombre,
        "edad": edad,
        "repite": repite_var.get(),
        "curso": curso_var.get(),
    })
    alumno_anterior = {} if es_nuevo else personas[dni_original]
    cambios = {
        campo: {
            "antes": alumno_anterior.get(campo),
            "despues": alumno.get(campo),
        }
        for campo in sorted(set(alumno_anterior) | set(alumno))
        if alumno_anterior.get(campo) != alumno.get(campo)
    }
    registro = {
        "fecha": datetime.now().astimezone().isoformat(timespec="seconds"),
        "accion": "alta" if es_nuevo else "modificacion",
        "dni_anterior": dni_original,
        "dni": nuevo_dni,
        "cambios": cambios,
    }
    try:
        with archivo_log.open("a", encoding="utf-8") as fichero_log:
            fichero_log.write(json.dumps(registro, ensure_ascii=False) + "\n")
    except OSError as error:
        messagebox.showerror(
            "No se pudo guardar el registro",
            f"No se ha aplicado el cambio porque no se pudo escribir {archivo_log}:\n{error}",
            parent=ventana,
        )
        return

    if not es_nuevo and nuevo_dni != dni_original:
        del personas[dni_original]
    personas[nuevo_dni] = alumno
    alumno_en_edicion["dni"] = nuevo_dni
    titulo_formulario.set("Modificar alumno")
    actualizar_lista(nuevo_dni)
    accion = "Alumno añadido" if es_nuevo else "Cambios guardados"
    etiquetas = {
        "nombre": "Nombre",
        "dni": "DNI",
        "edad": "Edad",
        "ciudad": "Ciudad",
        "telefono": "Teléfono",
        "ciclo_formativo": "Ciclo formativo",
        "modulos": "Módulos",
        "repite": "Repite curso",
        "curso": "Curso",
    }
    lineas_cambios = []
    for campo, valores in cambios.items():
        etiqueta = etiquetas.get(campo, campo)
        despues = valor_legible(valores["despues"])
        if es_nuevo:
            if despues in ("", "(ninguno)", "(sin dato)"):
                continue
            lineas_cambios.append(f"- {etiqueta}: {despues}")
        else:
            antes = valor_legible(valores["antes"])
            lineas_cambios.append(f"- {etiqueta}: {antes} → {despues}")
    detalle_cambios = "\n".join(lineas_cambios) or "- No había cambios en los datos."
    messagebox.showinfo(
        accion,
        f"{accion}\n\n{nombre} ({nuevo_dni})\n\n"
        f"Cambios:\n{detalle_cambios}\n\n"
        f"Registro guardado en:\n{archivo_log.resolve()}",
        parent=ventana,
    )


def eliminar_alumno_seleccionado():
    seleccion = lista_alumnos.curselection()
    if not seleccion:
        messagebox.showwarning(
            "Sin selección", "Selecciona un alumno para eliminarlo.", parent=ventana
        )
        return

    dni = alumnos_ordenados[seleccion[0]][0]
    alumno = personas[dni]
    confirmado = messagebox.askyesno(
        "Confirmar eliminación",
        f"¿Quieres eliminar a {alumno['nombre']} ({dni})?\n\nEsta acción no se puede deshacer.",
        parent=ventana,
    )
    if not confirmado:
        return

    registro = {
        "fecha": datetime.now().astimezone().isoformat(timespec="seconds"),
        "accion": "baja",
        "dni_anterior": dni,
        "dni": dni,
        "cambios": {
            campo: {"antes": valor, "despues": None}
            for campo, valor in sorted(alumno.items())
        },
    }
    try:
        with archivo_log.open("a", encoding="utf-8") as fichero_log:
            fichero_log.write(json.dumps(registro, ensure_ascii=False) + "\n")
    except OSError as error:
        messagebox.showerror(
            "No se pudo guardar el registro",
            f"No se ha eliminado el alumno porque no se pudo escribir {archivo_log}:\n{error}",
            parent=ventana,
        )
        return

    del personas[dni]
    actualizar_lista()
    abrir_formulario()
    messagebox.showinfo(
        "Alumno eliminado",
        f"Alumno eliminado\n\n{alumno['nombre']} ({dni})\n\n"
        f"Baja registrada en:\n{archivo_log.resolve()}",
        parent=ventana,
    )


def abrir_alumno_seleccionado():
    seleccion = lista_alumnos.curselection()
    if not seleccion:
        messagebox.showwarning(
            "Sin selección", "Selecciona un alumno para abrirlo.", parent=ventana
        )
        return
    abrir_formulario(alumnos_ordenados[seleccion[0]][0])


```

```python
import json
import tempfile
import unittest
from pathlib import Path
from unittest.mock import Mock


class VariablePrueba:
    def __init__(self, value=""):
        self.value = value

    def get(self):
        return self.value

    def set(self, value):
        self.value = value


class ListaPrueba:
    def __init__(self):
        self.items = []
        self.seleccion = ()

    def delete(self, inicio, fin):
        self.items.clear()
        self.seleccion = ()

    def insert(self, indice, valor):
        if indice == tk.END:
            self.items.append(valor)
        else:
            self.items.insert(indice, valor)

    def selection_set(self, indice):
        self.seleccion = (indice,)

    def see(self, indice):
        pass

    def curselection(self):
        return self.seleccion


class SelectorPrueba:
    def __setitem__(self, clave, valor):
        if clave == "values":
            self.values = valor


class PruebasBotonesAlumnos(unittest.TestCase):
    variables_globales = (
        "personas", "alumno_en_edicion", "titulo_formulario", "nombre_var",
        "dni_var", "edad_var", "repite_var", "curso_var", "ciclo_var",
        "selector_ciclos", "lista_alumnos", "alumnos_ordenados", "archivo_log",
        "ventana", "messagebox",
    )

    def setUp(self):
        self.estado_anterior = {
            nombre: (nombre in globals(), globals().get(nombre))
            for nombre in self.variables_globales
        }
        self.addCleanup(self.restaurar_estado)
        self.directorio_temporal = tempfile.TemporaryDirectory()
        self.addCleanup(self.directorio_temporal.cleanup)
        globals().update({
            "personas": {
                "12345678A": {
                    "nombre": "Ana Garc?a", "edad": 20, "ciudad": "Madrid",
                    "telefono": "600111222", "ciclo_formativo": "Ciclo A",
                    "modulos": ["Programaci?n"], "repite": False, "curso": "primero",
                }
            },
            "alumno_en_edicion": {"dni": None},
            "titulo_formulario": VariablePrueba("Nuevo alumno"),
            "nombre_var": VariablePrueba(),
            "dni_var": VariablePrueba(),
            "edad_var": VariablePrueba(),
            "repite_var": VariablePrueba(False),
            "curso_var": VariablePrueba("primero"),
            "ciclo_var": VariablePrueba(),
            "selector_ciclos": SelectorPrueba(),
            "lista_alumnos": ListaPrueba(),
            "alumnos_ordenados": [],
            "archivo_log": Path(self.directorio_temporal.name) / "alumnos.log",
            "ventana": object(),
            "messagebox": Mock(),
        })
        messagebox.askyesno.return_value = True
        actualizar_lista()

    def restaurar_estado(self):
        for nombre, (existia, valor) in self.estado_anterior.items():
            if existia:
                globals()[nombre] = valor
            else:
                globals().pop(nombre, None)

    def leer_registro(self):
        return json.loads(archivo_log.read_text(encoding="utf-8").splitlines()[0])

    def test_boton_nuevo_limpia_el_formulario(self):
        nombre_var.set("Dato anterior")
        dni_var.set("11111111B")
        edad_var.set("99")
        abrir_formulario()
        self.assertIsNone(alumno_en_edicion["dni"])
        self.assertEqual(titulo_formulario.get(), "Nuevo alumno")
        self.assertEqual(nombre_var.get(), "")
        self.assertEqual(dni_var.get(), "")
        self.assertEqual(edad_var.get(), "")
        self.assertEqual(ciclo_var.get(), "Ciclo A")

    def test_boton_abrir_carga_el_alumno_seleccionado(self):
        lista_alumnos.selection_set(0)
        abrir_alumno_seleccionado()
        self.assertEqual(alumno_en_edicion["dni"], "12345678A")
        self.assertEqual(nombre_var.get(), "Ana Garc?a")
        self.assertEqual(dni_var.get(), "12345678A")
        self.assertEqual(edad_var.get(), "20")
        self.assertEqual(ciclo_var.get(), "Ciclo A")

    def test_boton_guardar_registra_un_alta(self):
        nombre_var.set("Luis P?rez")
        dni_var.set("87654321B")
        edad_var.set("22")
        repite_var.set(True)
        curso_var.set("segundo")
        ciclo_var.set("Ciclo A")
        guardar_alumno()
        self.assertEqual(personas["87654321B"]["nombre"], "Luis P?rez")
        self.assertEqual(personas["87654321B"]["edad"], 22)
        self.assertEqual(self.leer_registro()["accion"], "alta")
        self.assertEqual(self.leer_registro()["dni"], "87654321B")
        messagebox.showinfo.assert_called_once()

    def test_boton_eliminar_registra_la_baja(self):
        lista_alumnos.selection_set(0)
        eliminar_alumno_seleccionado()
        self.assertNotIn("12345678A", personas)
        self.assertEqual(self.leer_registro()["accion"], "baja")
        self.assertEqual(self.leer_registro()["dni"], "12345678A")
        messagebox.askyesno.assert_called_once()


suite = unittest.defaultTestLoader.loadTestsFromTestCase(PruebasBotonesAlumnos)
resultado_pruebas = unittest.TextTestRunner(verbosity=2).run(suite)
assert resultado_pruebas.wasSuccessful(), "Han fallado uno o m?s casos de prueba."
```

## Historial de cambios

El fichero [`alumnos.log`](./alumnos.log) conserva una línea por cada alta, modificación o baja. La celda siguiente lee ese historial y lo presenta como un resumen ordenado, con la fecha, la operación, el alumno y los valores que cambiaron. El archivo original sigue disponible para consultar el detalle completo.

La tabla muestra primero las operaciones más recientes. En cada una, el resumen indica los valores añadidos, modificados o eliminados; la etiqueta de color diferencia altas, modificaciones y bajas.

La función `mostrar_valor(valor)` de la celda convierte los valores internos (por ejemplo, `True`, `primero` o una lista vacía) a expresiones más fáciles de leer antes de construir el resumen.

```python
from html import escape
from IPython.display import HTML, Markdown, display

if not archivo_log.exists():
    display(Markdown(
        f"No se ha encontrado el fichero de historial: `{archivo_log.resolve()}`. "
        "Guarda primero una operación en la aplicación."
    ))
else:
    registros = []
    with archivo_log.open("r", encoding="utf-8") as fichero_log:
        for numero_linea, linea in enumerate(fichero_log, start=1):
            if linea.strip():
                try:
                    registros.append(json.loads(linea))
                except json.JSONDecodeError as error:
                    raise ValueError(
                        f"Registro JSON no válido en {archivo_log}, línea {numero_linea}."
                    ) from error

    display(Markdown(
        f"### Resumen del historial\n\n"
        f"**Fichero:** `{archivo_log.resolve()}`  \n"
        f"**Operaciones registradas:** {len(registros)}"
    ))

    if not registros:
        display(Markdown("Todavía no hay operaciones registradas."))
    else:
        etiquetas_campos = {
            "nombre": "Nombre",
            "edad": "Edad",
            "ciudad": "Ciudad",
            "telefono": "Teléfono",
            "ciclo_formativo": "Ciclo formativo",
            "modulos": "Módulos",
            "repite": "Repite curso",
            "curso": "Curso",
        }
        etiquetas_acciones = {
            "alta": "Alta",
            "modificacion": "Modificación",
            "baja": "Baja",
        }

        def mostrar_valor(valor):
            if valor is None:
                return "—"
            if isinstance(valor, bool):
                return "Sí" if valor else "No"
            if valor == "primero":
                return "1º curso"
            if valor == "segundo":
                return "2º curso"
            if isinstance(valor, list):
                return ", ".join(valor) if valor else "(ninguno)"
            return str(valor)

        filas = []
        for registro in reversed(registros):
            cambios = registro.get("cambios", {})
            accion = registro.get("accion", "operación")
            valores_nombre = cambios.get("nombre", {})
            nombre = valores_nombre.get("despues") or valores_nombre.get("antes") or "Alumno"
            detalles = []
            for campo, valores in cambios.items():
                etiqueta = etiquetas_campos.get(campo, campo.replace("_", " ").capitalize())
                antes = valores.get("antes")
                despues = valores.get("despues")
                if accion == "alta":
                    if despues is not None and despues != "" and despues != []:
                        detalles.append(f"<li><strong>{escape(etiqueta)}:</strong> {escape(mostrar_valor(despues))}</li>")
                elif accion == "baja":
                    detalles.append(f"<li><strong>{escape(etiqueta)}:</strong> {escape(mostrar_valor(antes))} → eliminado</li>")
                elif antes != despues:
                    detalles.append(f"<li><strong>{escape(etiqueta)}:</strong> {escape(mostrar_valor(antes))} → {escape(mostrar_valor(despues))}</li>")

            dni_anterior = registro.get("dni_anterior")
            dni = registro.get("dni", "—")
            if dni_anterior and dni_anterior != dni:
                detalles.append(f"<li><strong>DNI:</strong> {escape(dni_anterior)} → {escape(dni)}</li>")
            if not detalles:
                detalles.append("<li>Sin cambios de campos.</li>")

            fecha = datetime.fromisoformat(registro["fecha"]).strftime("%d/%m/%Y %H:%M:%S")
            filas.append(
                "<tr>"
                f"<td>{escape(fecha)}</td>"
                f"<td><span class='accion {escape(accion)}'>{escape(etiquetas_acciones.get(accion, accion.capitalize()))}</span></td>"
                f"<td><strong>{escape(nombre)}</strong><br><small>DNI: {escape(dni)}</small></td>"
                f"<td><ul>{''.join(detalles)}</ul></td>"
                "</tr>"
            )

        display(HTML(
            "<style>"
            ".historial{border-collapse:collapse;width:100%;font-family:Arial,sans-serif;font-size:14px}"
            ".historial th,.historial td{border:1px solid #d8dee8;padding:10px;text-align:left;vertical-align:top}"
            ".historial th{background:#f1f5f9}.historial tr:nth-child(even){background:#fafbfc}"
            ".historial ul{margin:0;padding-left:18px}.historial li{margin:3px 0}"
            ".accion{display:inline-block;padding:3px 8px;border-radius:12px;background:#e8eef7}"
            ".accion.alta{background:#dcfce7}.accion.modificacion{background:#dbeafe}"
            ".accion.baja{background:#fee2e2}"
            "</style>"
            "<table class='historial'><thead><tr>"
            "<th>Fecha</th><th>Operación</th><th>Alumno</th><th>Resumen de cambios</th>"
            "</tr></thead><tbody>"
            f"{''.join(filas)}"
            "</tbody></table>"
        ))
```

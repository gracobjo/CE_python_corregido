## ---------- Treeview ----------

Este programa muestra las personas del diccionario `personas` en un `Treeview` de Tkinter. Los ciclos formativos son nodos padre y cada persona aparece como nodo hijo con sus datos en columnas; el DNI se conserva como identificador de esa fila.


### El DNI como identificador de fila

Al insertar cada persona se usa `iid=dni`. El `iid` es un identificador único dentro del `Treeview`, independiente de los valores visibles. Como el DNI identifica a cada persona, permite localizar su fila para consultarla o modificarla.


## Puntos clave del código

1. **`show="tree headings"`**: muestra la columna jerárquica del árbol y los encabezados de las columnas de datos.
2. **`iid=dni`**: asigna el DNI como identificador único. Por ejemplo, `tabla.delete(dni)` elimina esa fila y `tabla.item(dni)` permite consultar sus datos. `tabla.selection()` devuelve los identificadores de las filas seleccionadas.
3. **Se omite `modulos`**: no se incluye en las columnas ni en los valores que se insertan en la tabla.
4. **Booleanos formateados**: `repite` se muestra como `"Sí"` o `"No"`, que resulta más claro para quien consulta la tabla.
5. **Barras de desplazamiento**: la barra vertical permite recorrer las filas y la horizontal ver columnas que no caben en la ventana.

### 2. Ventana principal

`ventana` guarda la ventana raíz de Tkinter y se pasa como contenedor padre a la tabla y a las barras. `title()` define el título visible. `geometry()` establece el tamaño inicial en píxeles para que las columnas tengan espacio.


```python
ventana = tk.Tk()
ventana.title("Listado de personas")
ventana.geometry("1100x400")

```

### 3. Columnas y tabla en árbol

`columnas` es una tupla porque los identificadores y su orden son fijos: ese orden determina cómo se mostrarán los valores de cada fila. `tabla` referencia el `Treeview`; el nombre describe su función y permite reutilizarla al configurar, rellenar y conectar la tabla con las barras.

`show="tree headings"` muestra tanto la columna especial `#0` como los encabezados de datos. En este ejemplo, `#0` muestra el ciclo formativo en los nodos padre y el nombre de cada persona en los nodos hijo; el resto de datos aparece en columnas. El ciclo no se repite en otra columna. `titulos` asocia cada identificador interno con el encabezado visible.


### Opciones de visualización

El argumento `show` permite elegir qué partes del `Treeview` se muestran:

- `show="tree"`: muestra solo la columna jerárquica `#0`, sin encabezados de datos.
- `show="headings"`: muestra solo las columnas con encabezados; se comporta como una tabla plana.
- `show="tree headings"`: muestra la columna `#0` y las columnas con encabezados. Es la opción de este ejemplo porque permite ver a la vez la jerarquía y los datos.

La columna `#0` es especial: su texto se asigna con `text=...`. Las columnas declaradas en `columns` reciben sus datos mediante `values=(...)`.


```python
columnas = ("dni", "edad", "ciudad", "telefono", "repite", "curso")
tabla = ttk.Treeview(ventana, columns=columnas, show="tree headings")
tabla.heading("#0", text="Ciclo formativo / Persona")
tabla.column("#0", anchor="w", width=300)

titulos = {
    "dni": "DNI",
    "edad": "Edad",
    "ciudad": "Ciudad",
    "telefono": "Teléfono",
    "repite": "Repite",
    "curso": "Curso"
}

for columna, titulo in titulos.items():
    tabla.heading(columna, text=titulo)
    tabla.column(columna, anchor="center", width=130)


```

### 4. Barras de desplazamiento y distribución

`barra_vertical` y `barra_horizontal` guardan las barras y usan `tabla.yview` y `tabla.xview` para desplazar filas y columnas. `tabla.configure()` conecta la posición de la tabla con las barras.

`grid()` coloca los widgets en una cuadrícula: la tabla y la barra vertical en la fila 0, y la barra horizontal en la fila 1. `rowconfigure()` y `columnconfigure()` permiten que la tabla se estire cuando cambia el tamaño de la ventana.


### Conexión de las barras y distribución en la ventana

`ttk.Scrollbar(ventana, ...)` crea una barra dentro de la ventana principal. `orient="vertical"` indica que se desplaza de arriba abajo y `orient="horizontal"` que se desplaza de izquierda a derecha.

El argumento `command` indica qué función llamar cuando el usuario mueve la barra. `command=tabla.yview` hace que la barra vertical desplace las filas del `Treeview`; `command=tabla.xview` hace que la horizontal desplace las columnas. Se pasa el método como función para que Tkinter lo llame cuando haga falta.

`tabla.configure(yscrollcommand=barra_vertical.set, xscrollcommand=barra_horizontal.set)` completa la conexión en sentido contrario: cuando cambia la parte visible de la tabla, Tkinter llama a `set()` para actualizar la posición del indicador de la barra. Así, `command` mueve la tabla y `yscrollcommand`/`xscrollcommand` mantienen sincronizado el indicador.

### Colocación con `grid()`

`grid(row, column)` ubica cada widget en una cuadrícula numerada desde cero:

- La tabla va en la fila 0, columna 0. `sticky="nsew"` hace que ocupe todo el espacio de su celda: norte, sur, este y oeste.
- La barra vertical va en la fila 0, columna 1, a la derecha de la tabla. `sticky="ns"` la estira de arriba abajo.
- La barra horizontal va en la fila 1, columna 0, debajo de la tabla. `sticky="ew"` la estira de izquierda a derecha.

`ventana.rowconfigure(0, weight=1)` y `ventana.columnconfigure(0, weight=1)` permiten que la fila 0 y la columna 0 reciban el espacio sobrante al redimensionar la ventana. Como la tabla ocupa esa celda, se amplía en alto y ancho; las barras permanecen junto a sus bordes.


```python
barra_vertical = ttk.Scrollbar(ventana, orient="vertical", command=tabla.yview)
barra_horizontal = ttk.Scrollbar(ventana, orient="horizontal", command=tabla.xview)
tabla.configure(yscrollcommand=barra_vertical.set, xscrollcommand=barra_horizontal.set)

tabla.grid(row=0, column=0, sticky="nsew")
barra_vertical.grid(row=0, column=1, sticky="ns")
barra_horizontal.grid(row=1, column=0, sticky="ew")
ventana.rowconfigure(0, weight=1)
ventana.columnconfigure(0, weight=1)

```

### 5. Insertar las personas

`personas.items()` entrega cada par clave-valor. Se crea un nodo padre por cada ciclo formativo y, debajo, se inserta cada persona como nodo hijo. Así el `Treeview` representa una jerarquía real, no solo una tabla plana. En el bucle, `dni` recibe la clave y `persona` el diccionario con los datos.

`tabla.insert("", "end", ...)` agrega cada ciclo formativo en la raíz; `tabla.insert(iid_ciclo, "end", ...)` agrega cada persona como hija del ciclo correspondiente. `iid=dni` conserva el DNI como identificador único de cada persona. El nombre se muestra en la columna de árbol `#0`; `values` mantiene el mismo orden que `columnas`. El booleano `repite` se convierte en `"Sí"` o `"No"`. No se incluyen los módulos.


### ¿Por qué agrupar por ciclo formativo?

En estos datos, varias personas comparten el valor `ciclo_formativo`. Usarlo como nodo padre permite reunirlas bajo una categoría común y expandir o contraer cada grupo. El nombre de la persona aparece como nodo hijo y los demás datos como columnas; por eso el ciclo no se repite en cada fila.

`grupos` relaciona el nombre de cada ciclo con el identificador de su nodo padre. La primera persona de un ciclo crea el nodo; las siguientes reutilizan su identificador. `open=True` hace que los grupos aparezcan desplegados al abrir la ventana; con `open=False` comenzarían contraídos.

La clave de agrupación puede cambiar según lo que se quiera consultar. Por ejemplo, se puede agrupar por `persona["curso"]` para separar primero y segundo, por `persona["ciudad"]` para reunir localidades, o por `persona["repite"]` para distinguir quién repite. También se pueden crear niveles anidados, por ejemplo ciclo formativo y, dentro de cada ciclo, curso. Conviene elegir un campo que tenga valores compartidos y forme categorías útiles para la persona que consulta la tabla.


```python
grupos = {}

for dni, persona in personas.items():
    ciclo = persona["ciclo_formativo"]

    if ciclo not in grupos:
        grupos[ciclo] = tabla.insert("", "end", text=ciclo, open=True)

    tabla.insert(
        grupos[ciclo],
        "end",
        iid=dni,
        text=persona["nombre"],
        values=(
            dni,
            persona["edad"],
            persona["ciudad"],
            persona["telefono"],
            "Sí" if persona["repite"] else "No",
            persona["curso"]
        )
    )

```

### 6. Operaciones CRUD

CRUD reúne las operaciones de crear, leer, actualizar y eliminar. En este ejemplo, **leer** consiste en seleccionar y consultar una persona en la tabla; **crear** y **actualizar** usan el mismo formulario; **eliminar** pide confirmación antes de borrar.

Los cambios se guardan en el diccionario `personas` y después se reconstruye el `Treeview`. Esto es necesario porque una persona editada puede cambiar de ciclo formativo y, por tanto, debe aparecer bajo otro nodo padre. El DNI no se modifica en el formulario de edición: es a la vez la clave del diccionario y el `iid` del elemento, así se mantienen sincronizados.


```python
def reconstruir_treeview(dni_seleccionado=None):
    for iid in tabla.get_children(""):
        tabla.delete(iid)

    grupos.clear()
    for dni, persona in personas.items():
        ciclo = persona["ciclo_formativo"]
        if ciclo not in grupos:
            grupos[ciclo] = tabla.insert("", "end", text=ciclo, open=True)

        tabla.insert(
            grupos[ciclo],
            "end",
            iid=dni,
            text=persona["nombre"],
            values=(
                dni,
                persona["edad"],
                persona["ciudad"],
                persona["telefono"],
                "Sí" if persona["repite"] else "No",
                persona["curso"]
            )
        )

    if dni_seleccionado in personas:
        tabla.selection_set(dni_seleccionado)
        tabla.focus(dni_seleccionado)
        tabla.see(dni_seleccionado)


def abrir_formulario(persona=None, dni_original=None):
    es_nueva = persona is None
    datos = persona or {
        "nombre": "",
        "edad": "",
        "ciudad": "",
        "telefono": "",
        "ciclo_formativo": "",
        "repite": False,
        "curso": "primero"
    }

    dialogo = tk.Toplevel(ventana)
    dialogo.title("Añadir persona" if es_nueva else "Modificar persona")
    dialogo.transient(ventana)
    dialogo.resizable(False, False)
    dialogo.grab_set()

    campos = (
        ("dni", "DNI"),
        ("nombre", "Nombre"),
        ("edad", "Edad"),
        ("ciudad", "Ciudad"),
        ("telefono", "Teléfono"),
        ("ciclo_formativo", "Ciclo formativo"),
        ("repite", "¿Repite?"),
        ("curso", "Curso")
    )
    entradas = {}

    for fila, (clave, etiqueta) in enumerate(campos):
        ttk.Label(dialogo, text=etiqueta).grid(
            row=fila, column=0, padx=10, pady=5, sticky="w"
        )

        if clave == "repite":
            entrada = ttk.Combobox(
                dialogo, values=("No", "Sí"), state="readonly", width=37
            )
            entrada.set("Sí" if datos.get(clave, False) else "No")
        elif clave == "curso":
            entrada = ttk.Combobox(
                dialogo,
                values=("primero", "segundo"),
                state="readonly",
                width=37
            )
            entrada.set(datos.get(clave, "primero"))
        else:
            entrada = ttk.Entry(dialogo, width=40)
            valor = dni_original if clave == "dni" and dni_original else datos.get(clave, "")
            entrada.insert(0, str(valor))
            if clave == "dni" and not es_nueva:
                entrada.configure(state="readonly")

        entrada.grid(row=fila, column=1, padx=10, pady=5, sticky="ew")
        entradas[clave] = entrada

    def guardar():
        valores = {clave: entrada.get().strip() for clave, entrada in entradas.items()}
        if not all(valores.values()):
            messagebox.showerror("Datos incompletos", "Rellena todos los campos.", parent=dialogo)
            return

        try:
            edad = int(valores["edad"])
        except ValueError:
            messagebox.showerror("Edad no válida", "La edad debe ser un número entero.", parent=dialogo)
            return

        if edad < 1:
            messagebox.showerror("Edad no válida", "La edad debe ser mayor que cero.", parent=dialogo)
            return

        dni = valores["dni"]
        if es_nueva and dni in personas:
            messagebox.showerror("DNI duplicado", "Ya existe una persona con ese DNI.", parent=dialogo)
            return

        personas[dni] = {
            "nombre": valores["nombre"],
            "edad": edad,
            "ciudad": valores["ciudad"],
            "telefono": valores["telefono"],
            "ciclo_formativo": valores["ciclo_formativo"],
            "repite": valores["repite"] == "Sí",
            "curso": valores["curso"]
        }
        reconstruir_treeview(dni)
        dialogo.destroy()

    botones = ttk.Frame(dialogo)
    botones.grid(row=len(campos), column=0, columnspan=2, pady=10)
    ttk.Button(botones, text="Guardar", command=guardar).pack(side="left", padx=5)
    ttk.Button(botones, text="Cancelar", command=dialogo.destroy).pack(side="left", padx=5)
    dialogo.bind("<Return>", lambda evento: guardar())
    dialogo.bind("<Escape>", lambda evento: dialogo.destroy())
    dialogo.wait_window()


def obtener_dni_seleccionado():
    seleccion = tabla.selection()
    if not seleccion:
        messagebox.showwarning("Sin selección", "Selecciona una persona de la tabla.", parent=ventana)
        return None

    dni = seleccion[0]
    if dni not in personas:
        messagebox.showwarning("Selección no válida", "Selecciona una persona, no un grupo.", parent=ventana)
        return None
    return dni


def agregar_persona():
    abrir_formulario()


def modificar_persona():
    dni = obtener_dni_seleccionado()
    if dni is not None:
        abrir_formulario(personas[dni], dni)


def eliminar_persona():
    dni = obtener_dni_seleccionado()
    if dni is None:
        return

    nombre = personas[dni]["nombre"]
    confirmar = messagebox.askyesno(
        "Confirmar eliminación",
        f"¿Quieres eliminar a {nombre} ({dni})?",
        parent=ventana
    )
    if confirmar:
        del personas[dni]
        reconstruir_treeview()
        estado_seleccion.set("Persona eliminada.")


def actualizar_estado_seleccion(evento=None):
    seleccion = tabla.selection()
    if not seleccion:
        estado_seleccion.set("Selecciona una persona para consultar sus datos.")
        return

    iid = seleccion[0]
    if iid in personas:
        persona = personas[iid]
        estado_seleccion.set(f"Seleccionada: {persona['nombre']} — DNI: {iid}")
    else:
        estado_seleccion.set(f"Grupo seleccionado: {tabla.item(iid, 'text')}")

```

### 7. Controles y eventos

Los botones conectan las acciones de crear, actualizar y eliminar con sus funciones mediante `command`. La lectura se realiza al seleccionar una persona: el evento `<<TreeviewSelect>>` actualiza el texto de estado, mientras que los datos completos siguen visibles en sus columnas. Las acciones de modificar y eliminar solo aceptan nodos de persona, no los nodos padre de los ciclos.


```python
barra_acciones = ttk.Frame(ventana)
barra_acciones.grid(row=2, column=0, sticky="ew", pady=8)

ttk.Button(barra_acciones, text="Añadir persona", command=agregar_persona).pack(side="left", padx=5)
ttk.Button(barra_acciones, text="Modificar selección", command=modificar_persona).pack(side="left", padx=5)
ttk.Button(barra_acciones, text="Eliminar selección", command=eliminar_persona).pack(side="left", padx=5)

estado_seleccion = tk.StringVar(
    value="Selecciona una persona para consultar sus datos."
)
ttk.Label(barra_acciones, textvariable=estado_seleccion).pack(
    side="left", padx=12
)
tabla.bind("<<TreeviewSelect>>", actualizar_estado_seleccion)

```

`ventana.mainloop()` inicia el bucle de eventos de Tkinter. Mantiene abierta la ventana y atiende las acciones del usuario, como redimensionarla o desplazarse por la tabla. Se ejecuta al final, cuando ya se han creado y rellenado los widgets.


```python
ventana.mainloop()
```

## ---------- Treeview ----------

Este programa muestra las personas del diccionario `personas` en una ventana de Tkinter. Cada persona ocupa una fila y sus datos aparecen en columnas; el DNI se conserva como identificador de esa fila.


### El DNI como identificador de fila

Al insertar cada persona se usa `iid=dni`. El `iid` es un identificador único dentro del `Treeview`, independiente de los valores visibles. Como el DNI identifica a cada persona, permite localizar su fila para consultarla o modificarla.


## Puntos clave del código

1. **`show="headings"`**: muestra solo los encabezados y las filas, sin la columna jerárquica del árbol. Así el `Treeview` funciona como una tabla plana.
2. **`iid=dni`**: asigna el DNI como identificador único. Por ejemplo, `tabla.delete(dni)` elimina esa fila y `tabla.item(dni)` permite consultar sus datos. `tabla.selection()` devuelve los identificadores de las filas seleccionadas.
3. **Se omite `modulos`**: no se incluye en las columnas ni en los valores que se insertan en la tabla.
4. **Booleanos formateados**: `repite` se muestra como `"Sí"` o `"No"`, que resulta más claro para quien consulta la tabla.
5. **Barras de desplazamiento**: la barra vertical permite recorrer las filas y la horizontal ver columnas que no caben en la ventana.

### 2. Ventana principal

`ventana` guarda la ventana raíz de Tkinter y se pasa como contenedor padre a la tabla y a las barras. `title()` define el título visible. `geometry()` establece el tamaño inicial en píxeles para que las columnas tengan espacio.


```python
ventana = tk.Tk()
ventana.title("Listado de personas")
ventana.geometry("1100x400")

```

### 3. Columnas y tabla

`columnas` es una tupla porque los identificadores y su orden son fijos: ese orden determina cómo se mostrarán los valores de cada fila. `tabla` referencia el `Treeview`; el nombre describe su función y permite reutilizarla al configurar, rellenar y conectar la tabla con las barras.

`show="headings"` oculta la columna jerárquica para dejar una tabla plana. `titulos` asocia cada identificador interno con el encabezado visible. En el bucle, `columna` recibe la clave y `titulo` el texto del encabezado. Todas las columnas empiezan centradas con ancho 130; después se amplían `nombre` y `ciclo`, cuyos textos suelen ser más largos.


```python
columnas = ("dni", "nombre", "edad", "ciudad", "telefono", "ciclo", "repite", "curso")
tabla = ttk.Treeview(ventana, columns=columnas, show="headings")

titulos = {
    "dni": "DNI",
    "nombre": "Nombre",
    "edad": "Edad",
    "ciudad": "Ciudad",
    "telefono": "Teléfono",
    "ciclo": "Ciclo formativo",
    "repite": "Repite",
    "curso": "Curso"
}

for columna, titulo in titulos.items():
    tabla.heading(columna, text=titulo)
    tabla.column(columna, anchor="center", width=130)

tabla.column("nombre", width=160)
tabla.column("ciclo", width=360)

```

### 4. Barras de desplazamiento y distribución

`barra_vertical` y `barra_horizontal` guardan las barras y usan `tabla.yview` y `tabla.xview` para desplazar filas y columnas. `tabla.configure()` conecta la posición de la tabla con las barras.

`grid()` coloca los widgets en una cuadrícula: la tabla y la barra vertical en la fila 0, y la barra horizontal en la fila 1. `rowconfigure()` y `columnconfigure()` permiten que la tabla se estire cuando cambia el tamaño de la ventana.


### Conexión de las barras y distribución en la ventana

`ttk.Scrollbar(ventana, ...)` crea una barra dentro de la ventana principal. `orient="vertical"` indica que se desplaza de arriba abajo y `orient="horizontal"` que se desplaza de izquierda a derecha.

El argumento `command` indica qué función llamar cuando el usuario mueve la barra. `command=tabla.yview` hace que la barra vertical desplace las filas del `Treeview`; `command=tabla.xview` hace que la horizontal desplace las columnas. Se pasa el método como función para que Tkinter lo llame cuando haga falta.

`tabla.configure(yscrollcommand=barra_vertical.set, xscrollcommand=barra_horizontal.set)` completa la conexión en sentido contrario: cuando cambia la parte visible de la tabla, Tkinter llama a `set()` para actualizar la posición del indicador de la barra. Así, `command` mueve la tabla y `yscrollcommand`/`xscrollcommand` mantienen sincronizado el indicador.

### Colocación con `grid()`

`grid(row, column)` ubica cada widget en una cuadrícula numerada desde cero:

- La tabla va en la fila 0, columna 0. `sticky="nsew"` hace que ocupe todo el espacio de su celda: norte, sur, este y oeste.
- La barra vertical va en la fila 0, columna 1, a la derecha de la tabla. `sticky="ns"` la estira de arriba abajo.
- La barra horizontal va en la fila 1, columna 0, debajo de la tabla. `sticky="ew"` la estira de izquierda a derecha.

`ventana.rowconfigure(0, weight=1)` y `ventana.columnconfigure(0, weight=1)` permiten que la fila 0 y la columna 0 reciban el espacio sobrante al redimensionar la ventana. Como la tabla ocupa esa celda, se amplía en alto y ancho; las barras permanecen junto a sus bordes.


```python
barra_vertical = ttk.Scrollbar(ventana, orient="vertical", command=tabla.yview)
barra_horizontal = ttk.Scrollbar(ventana, orient="horizontal", command=tabla.xview)
tabla.configure(yscrollcommand=barra_vertical.set, xscrollcommand=barra_horizontal.set)

tabla.grid(row=0, column=0, sticky="nsew")
barra_vertical.grid(row=0, column=1, sticky="ns")
barra_horizontal.grid(row=1, column=0, sticky="ew")
ventana.rowconfigure(0, weight=1)
ventana.columnconfigure(0, weight=1)

```

### 5. Insertar las personas

`personas.items()` entrega cada par clave-valor. En el bucle, `dni` recibe la clave y `persona` el diccionario con los datos; estos nombres dejan claro qué representa cada variable.

`tabla.insert("", "end", ...)` agrega una fila en el nivel principal de la lista. `iid=dni` usa la clave como identificador estable. La tupla `values` mantiene el mismo orden que `columnas`. El booleano `repite` se convierte en `"Sí"` o `"No"` para que sea más legible. No se incluyen los módulos.


```python
for dni, persona in personas.items():
    tabla.insert(
        "",
        "end",
        iid=dni,
        values=(
            dni,
            persona["nombre"],
            persona["edad"],
            persona["ciudad"],
            persona["telefono"],
            persona["ciclo_formativo"],
            "Sí" if persona["repite"] else "No",
            persona["curso"]
        )
    )

```

`ventana.mainloop()` inicia el bucle de eventos de Tkinter. Mantiene abierta la ventana y atiende las acciones del usuario, como redimensionarla o desplazarse por la tabla. Se ejecuta al final, cuando ya se han creado y rellenado los widgets.


```python
ventana.mainloop()
```

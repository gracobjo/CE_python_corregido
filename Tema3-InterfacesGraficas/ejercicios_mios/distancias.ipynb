## 1. Datos de las distancias

La primera celda define `distancias_capitales`, un diccionario de diccionarios. La clave exterior es la ciudad de origen; dentro, cada clave es una ciudad de destino y su valor es la distancia en kilómetros. Por ejemplo, `distancias_capitales["Madrid"]["Toledo"]` devuelve `75`. La tabla guarda también las distancias en sentido contrario.

La siguiente celda crea la interfaz y la lógica que responde a las selecciones. Ejecútala después de la celda de datos, porque utiliza `distancias_capitales`.

````python
%gui tk

import tkinter as tk
from tkinter import ttk

ciudades = sorted(distancias_capitales)

def mostrar_distancia(event=None):
    origen_seleccionado = lista_origen.curselection()
    destino_seleccionado = lista_destino.curselection()

    if not origen_seleccionado or not destino_seleccionado:
        texto_distancia.set("Selecciona un origen y un destino")
        return

    origen = lista_origen.get(origen_seleccionado[0])
    destino = lista_destino.get(destino_seleccionado[0])

    if origen == destino:
        distancia = 0
    else:
        distancia = distancias_capitales[origen].get(destino)

    if distancia is None:
        texto_distancia.set(f"No hay distancia registrada entre\n{origen} y {destino}")
    else:
        texto_distancia.set(f"{origen} - {destino}\n{distancia} km")


ventana = tk.Tk()
ventana.title("Distancia entre ciudades")
ventana.resizable(False, False)

contenedor = ttk.Frame(ventana, padding=16)
contenedor.grid(row=0, column=0)

ttk.Label(contenedor, text="Ciudad de origen").grid(row=0, column=0, padx=8, pady=6)
ttk.Label(contenedor, text="Distancia").grid(row=0, column=1, padx=12, pady=6)
ttk.Label(contenedor, text="Ciudad de destino").grid(row=0, column=2, padx=8, pady=6)

lista_origen = tk.Listbox(contenedor, height=15, width=25, exportselection=False)
lista_origen.grid(row=1, column=0, padx=8)
lista_destino = tk.Listbox(contenedor, height=15, width=25, exportselection=False)
lista_destino.grid(row=1, column=2, padx=8)

for ciudad in ciudades:
    lista_origen.insert(tk.END, ciudad)
    lista_destino.insert(tk.END, ciudad)

texto_distancia = tk.StringVar(value="Selecciona un origen y un destino")
ttk.Label(
    contenedor,
    textvariable=texto_distancia,
    anchor="center",
    justify="center",
    wraplength=180,
).grid(row=1, column=1, padx=12)

lista_origen.bind("<<ListboxSelect>>", mostrar_distancia)
lista_destino.bind("<<ListboxSelect>>", mostrar_distancia);
````

## 2. Antes de leer el código: ¿qué es Tkinter?

Tkinter es una biblioteca de Python para crear ventanas y elementos con los que una persona puede interactuar: botones, etiquetas, listas, cuadros de texto, etc. A cada elemento visual lo llamaremos **widget**. En esta aplicación, la ventana contiene etiquetas, dos listas y un texto con el resultado.

La interfaz es **interactiva**: el programa no pregunta primero por el origen y luego por el destino en la consola. En cambio, espera a que la persona haga clic y responde a esos cambios. A esa forma de trabajar se la llama programación basada en eventos.

## 3. Preparar Tkinter y obtener las ciudades

### `%gui tk`

Es una instrucción especial de IPython/Jupyter, no una instrucción normal de Python. Le permite a Jupyter atender los eventos de Tkinter mientras mantiene abierto el notebook. Por eso esta aplicación puede funcionar en una celda interactiva sin escribir `ventana.mainloop()`.

### `import tkinter as tk` y `from tkinter import ttk`

`import` carga una biblioteca para poder utilizarla. `as tk` crea el nombre corto `tk`: así escribimos `tk.Tk()` en lugar de `tkinter.Tk()`. `ttk` es otro módulo de Tkinter con widgets como `Frame` y `Label`.

### `ciudades = sorted(distancias_capitales)`

Un diccionario se puede recorrer por sus claves; aquí las claves exteriores son nombres de ciudades. `sorted(...)` las ordena alfabéticamente y devuelve una lista. Esa misma lista se utilizará para llenar las listas de origen y destino.

## 4. Definir la función que calculará y mostrará la distancia

### `def mostrar_distancia(event=None):`

`def` define una función: agrupa instrucciones que se ejecutarán cuando alguien las llame. Aquí Tkinter llamará a `mostrar_distancia` cuando cambie una selección. Al detectar el clic, Tkinter entrega información del evento como argumento; por eso existe el parámetro `event`. El valor predeterminado `None` permite que el parámetro también pueda omitirse.

La función se define antes de crear las listas, pero no intenta leerlas en ese momento. Las instrucciones indentadas dentro de la función se ejecutan más tarde, cuando la función es llamada; para entonces las listas ya están creadas.

### Averiguar qué elementos están seleccionados

```python
origen_seleccionado = lista_origen.curselection()
destino_seleccionado = lista_destino.curselection()
```

`lista_origen` y `lista_destino` serán las dos listas visibles. El método `curselection()` pregunta a cada lista qué elementos están seleccionados. Devuelve los índices de selección; si no hay ninguno, devuelve una tupla vacía. Un índice indica la posición del elemento: el primero está en la posición 0, el siguiente en la 1, etc.

### Comprobar si falta una selección

```python
if not origen_seleccionado or not destino_seleccionado:
    texto_distancia.set("Selecciona un origen y un destino")
    return
```

`if` permite ejecutar instrucciones solo cuando se cumple una condición. `not` significa “no”; una tupla vacía se considera falsa. `or` significa “o”, así que el bloque se ejecuta si falta el origen o falta el destino. Se cambia el mensaje de la interfaz y `return` termina la función en ese punto, evitando intentar calcular con una selección incompleta.

### Convertir los índices en nombres de ciudad

```python
origen = lista_origen.get(origen_seleccionado[0])
destino = lista_destino.get(destino_seleccionado[0])
```

Cada selección está guardada como una secuencia de índices. `[0]` obtiene el primer índice seleccionado y `.get(indice)` lee el nombre que aparece en esa posición de la lista. Por ejemplo, si el índice seleccionado es 2, se lee la tercera ciudad.

### Buscar los kilómetros

```python
if origen == destino:
    distancia = 0
else:
    distancia = distancias_capitales[origen].get(destino)
```

`==` compara si dos valores son iguales. Si el origen y el destino son la misma ciudad, la distancia se fija en 0. Si son diferentes, primero `distancias_capitales[origen]` obtiene el diccionario de destinos de esa ciudad; después `.get(destino)` busca la distancia correspondiente.

Se utiliza `.get()` en lugar de corchetes para la ciudad de destino porque, si la distancia no está registrada, `.get()` devuelve `None` en vez de producir un error. `None` representa que no se encontró un valor.

### Elegir el mensaje que verá la persona

```python
if distancia is None:
    texto_distancia.set(f"No hay distancia registrada entre\n{origen} y {destino}")
else:
    texto_distancia.set(f"{origen} - {destino}\n{distancia} km")
```

`is None` comprueba específicamente si no se encontró una distancia. Si es así, se muestra un aviso. En caso contrario, se muestran las ciudades y los kilómetros.

La letra `f` delante del texto crea una **f-string**: permite insertar valores dentro de las llaves, por ejemplo `{origen}`. `\n` representa un salto de línea. `texto_distancia.set(...)` cambia el texto asociado a la etiqueta central; veremos más adelante cómo queda conectada esa variable con la etiqueta.

## 5. Crear la ventana y organizar los elementos

### Ventana principal y contenedor

```python
ventana = tk.Tk()
ventana.title("Distancia entre ciudades")
ventana.resizable(False, False)
contenedor = ttk.Frame(ventana, padding=16)
contenedor.grid(row=0, column=0)
```

`tk.Tk()` crea la ventana principal. `title(...)` le pone el título que aparece en la barra superior. `resizable(False, False)` impide cambiar el ancho y el alto. Un `Frame` es un widget contenedor: ayuda a agrupar y ordenar otros widgets. `padding=16` deja espacio interior. `grid(row=0, column=0)` coloca el contenedor en la fila 0 y columna 0 de la ventana.

### La cuadrícula `grid`

`grid` organiza los widgets en filas y columnas, como una tabla. En este programa se usan tres columnas: origen, resultado y destino. La fila 0 contiene los títulos; la fila 1 contiene las dos listas y el resultado. `row` significa fila y `column` significa columna. `padx` añade espacio horizontal alrededor de un widget y `pady` añade espacio vertical.

## 6. Crear las etiquetas y las listas

### Etiquetas (`Label`)

`ttk.Label(contenedor, text="Ciudad de origen")` crea un texto fijo dentro del contenedor. Los tres `Label` de la fila superior identifican las columnas: origen, distancia y destino. Al final de cada línea, `.grid(...)` coloca esa etiqueta en la fila y columna indicadas.

### Listas (`Listbox`)

```python
lista_origen = tk.Listbox(contenedor, height=15, width=25, exportselection=False)
lista_destino = tk.Listbox(contenedor, height=15, width=25, exportselection=False)
```

`Listbox` crea una lista visible en la que se puede seleccionar un elemento. `height=15` define su altura en filas de texto y `width=25` su anchura aproximada en caracteres. `exportselection=False` permite mantener seleccionada la ciudad de origen al seleccionar también una ciudad en la otra lista; así pueden quedar elegidos ambos valores.

Las llamadas `.grid(...)` siguientes colocan las listas en la fila 1: origen en la columna 0 y destino en la columna 2.

## 7. Llenar las listas con las ciudades

```python
for ciudad in ciudades:
    lista_origen.insert(tk.END, ciudad)
    lista_destino.insert(tk.END, ciudad)
```

`for` repite las instrucciones indentadas una vez por cada elemento de `ciudades`. En cada repetición, `ciudad` contiene un nombre distinto. `.insert(tk.END, ciudad)` añade ese nombre al final de la lista; `tk.END` quiere decir “al final”. Como se inserta en las dos listas, ambas muestran las mismas ciudades.

Este bucle evita escribir a mano una instrucción por ciudad. Si se añade una nueva ciudad al diccionario, también aparecerá en las listas.

## 8. Mostrar el resultado y reaccionar a los clics

### Variable de texto y etiqueta central

`tk.StringVar(value="Selecciona un origen y un destino")` crea una variable especial de Tkinter para guardar texto. Se usa para que el programa pueda cambiar el mensaje de la ventana mientras está funcionando. La etiqueta central recibe `textvariable=texto_distancia`; eso la conecta a la variable, de modo que cada llamada a `.set(...)` actualiza lo que aparece en pantalla.

En esa etiqueta, `anchor="center"` sitúa el contenido en el centro, `justify="center"` centra las líneas entre sí y `wraplength=180` hace que un texto largo pueda continuar en otra línea. Su llamada a `.grid(row=1, column=1, ...)` la coloca entre las dos listas.

### Eventos: `bind`

```python
lista_origen.bind("<<ListboxSelect>>", mostrar_distancia)
lista_destino.bind("<<ListboxSelect>>", mostrar_distancia)
```

`bind` conecta un evento de un widget con una función. `<<ListboxSelect>>` es el evento que indica que cambió la selección de una lista. Al conectarlo en las dos listas, Tkinter llama a `mostrar_distancia` tanto cuando se elige el origen como cuando se elige el destino. La función revisa ambas listas cada vez; si todavía falta una selección, muestra la instrucción para elegir las dos.

En el notebook, un punto y coma al final de la última línea evita que Jupyter imprima el identificador que devuelve `bind`; no cambia el comportamiento de la interfaz.

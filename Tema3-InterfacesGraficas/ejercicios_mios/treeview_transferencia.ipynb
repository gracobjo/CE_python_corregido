## Crear las dos vistas

La vista de la izquierda usa `show="headings"` y presenta cada persona en una fila con todas sus columnas. La vista de la derecha usa `show="tree headings"`: el ciclo formativo es el nodo padre y cada persona es un nodo hijo. El DNI (`iid`) identifica a la persona en ambas vistas.

Cada panel tiene barras de desplazamiento propias. La columna especial `#0` del árbol muestra el nombre de la persona en los nodos hijo y el ciclo en los nodos padre.


```python
columnas_tabla = ("dni", "nombre", "edad", "ciudad", "telefono", "ciclo", "repite", "curso")
columnas_arbol = ("dni", "edad", "ciudad", "telefono", "repite", "curso")

titulos_tabla = {
    "dni": "DNI",
    "nombre": "Nombre",
    "edad": "Edad",
    "ciudad": "Ciudad",
    "telefono": "Teléfono",
    "ciclo": "Ciclo formativo",
    "repite": "Repite",
    "curso": "Curso"
}

titulos_arbol = {
    "dni": "DNI",
    "edad": "Edad",
    "ciudad": "Ciudad",
    "telefono": "Teléfono",
    "repite": "Repite",
    "curso": "Curso"
}

ventana = tk.Tk()
ventana.title("Transferir personas entre Treeview")
ventana.geometry("1600x620")
ventana.rowconfigure(0, weight=1)
ventana.columnconfigure(0, weight=1)
ventana.columnconfigure(2, weight=1)

marco_tabla = ttk.LabelFrame(ventana, text="Treeview tipo tabla")
marco_tabla.grid(row=0, column=0, sticky="nsew", padx=8, pady=8)
marco_intercambio = ttk.Frame(ventana)
marco_intercambio.grid(row=0, column=1, padx=8, pady=8)
marco_arbol = ttk.LabelFrame(ventana, text="Treeview tipo árbol")
marco_arbol.grid(row=0, column=2, sticky="nsew", padx=8, pady=8)

marco_tabla.rowconfigure(0, weight=1)
marco_tabla.columnconfigure(0, weight=1)
tabla = ttk.Treeview(
    marco_tabla,
    columns=columnas_tabla,
    show="headings",
    selectmode="browse"
)
for columna, titulo in titulos_tabla.items():
    tabla.heading(columna, text=titulo)
    tabla.column(columna, anchor="center", width=110)
tabla.column("nombre", width=150)
tabla.column("ciclo", width=290)

barra_tabla_vertical = ttk.Scrollbar(marco_tabla, orient="vertical", command=tabla.yview)
barra_tabla_horizontal = ttk.Scrollbar(marco_tabla, orient="horizontal", command=tabla.xview)
tabla.configure(
    yscrollcommand=barra_tabla_vertical.set,
    xscrollcommand=barra_tabla_horizontal.set
)
tabla.grid(row=0, column=0, sticky="nsew")
barra_tabla_vertical.grid(row=0, column=1, sticky="ns")
barra_tabla_horizontal.grid(row=1, column=0, sticky="ew")

marco_arbol.rowconfigure(0, weight=1)
marco_arbol.columnconfigure(0, weight=1)
tabla_arbol = ttk.Treeview(
    marco_arbol,
    columns=columnas_arbol,
    show="tree headings",
    selectmode="browse"
)
tabla_arbol.heading("#0", text="Ciclo formativo / Persona")
tabla_arbol.column("#0", anchor="w", width=270)
for columna, titulo in titulos_arbol.items():
    tabla_arbol.heading(columna, text=titulo)
    tabla_arbol.column(columna, anchor="center", width=90)

barra_arbol_vertical = ttk.Scrollbar(marco_arbol, orient="vertical", command=tabla_arbol.yview)
barra_arbol_horizontal = ttk.Scrollbar(marco_arbol, orient="horizontal", command=tabla_arbol.xview)
tabla_arbol.configure(
    yscrollcommand=barra_arbol_vertical.set,
    xscrollcommand=barra_arbol_horizontal.set
)
tabla_arbol.grid(row=0, column=0, sticky="nsew")
barra_arbol_vertical.grid(row=0, column=1, sticky="ns")
barra_arbol_horizontal.grid(row=1, column=0, sticky="ew")
```

## Mostrar las personas

La tabla guarda los datos directamente en sus columnas. En el árbol, `grupos_arbol` asocia cada ciclo formativo con el identificador de su nodo padre; al insertar una persona se crea el grupo solo si todavía no existe.

Al reconstruir el árbol se eliminan los nodos anteriores y se vuelven a crear a partir de `personas_arbol`. Así desaparecen también los grupos que se quedan vacíos después de una transferencia.


```python
def actualizar_tabla(dni_seleccionado=None):
    for iid in tabla.get_children(""):
        tabla.delete(iid)

    for dni, persona in personas_tabla.items():
        tabla.insert(
            "",
            "end",
            iid=dni,
            values=(
                dni,
                persona["nombre"],
                persona["edad"],
                persona["ciudad"],
                persona["telefono"],
                persona["ciclo_formativo"],
                "Sí" if persona["repite"] else "No",
                persona["curso"]
            )
        )

    if dni_seleccionado in personas_tabla:
        tabla.selection_set(dni_seleccionado)
        tabla.focus(dni_seleccionado)
        tabla.see(dni_seleccionado)


def reconstruir_arbol(dni_seleccionado=None):
    for iid in tabla_arbol.get_children(""):
        tabla_arbol.delete(iid)

    grupos_arbol.clear()
    for dni, persona in personas_arbol.items():
        ciclo = persona["ciclo_formativo"]
        if ciclo not in grupos_arbol:
            grupos_arbol[ciclo] = tabla_arbol.insert(
                "", "end", text=ciclo, open=True
            )

        tabla_arbol.insert(
            grupos_arbol[ciclo],
            "end",
            iid=dni,
            text=persona["nombre"],
            values=(
                dni,
                persona["edad"],
                persona["ciudad"],
                persona["telefono"],
                "Sí" if persona["repite"] else "No",
                persona["curso"]
            )
        )

    if dni_seleccionado in personas_arbol:
        tabla_arbol.selection_set(dni_seleccionado)
        tabla_arbol.focus(dni_seleccionado)
        tabla_arbol.see(dni_seleccionado)


grupos_arbol = {}
actualizar_tabla()
reconstruir_arbol()
```

## Botones de transferencia

Selecciona una persona (no un nodo de ciclo) y pulsa la flecha que apunta a la otra vista. `selectmode="browse"` limita cada selección a una persona. Al moverla, se quita del diccionario de origen y se añade al de destino; después se actualizan las dos vistas y se conserva la selección en el destino.

La flecha `→` envía de la tabla al árbol y la flecha `←` devuelve del árbol a la tabla. Si no hay una persona seleccionada o se selecciona un nodo padre, se muestra un aviso y no se modifica ningún dato.


```python
estado = tk.StringVar(value="Selecciona una persona y pulsa una flecha para transferirla.")


def enviar_a_arbol():
    seleccion = tabla.selection()
    if not seleccion:
        messagebox.showwarning(
            "Sin selección", "Selecciona una persona de la tabla.", parent=ventana
        )
        return

    dni = seleccion[0]
    if dni not in personas_tabla:
        messagebox.showwarning(
            "Selección no válida", "La selección no corresponde a una persona.", parent=ventana
        )
        return

    persona = personas_tabla.pop(dni)
    personas_arbol[dni] = persona
    actualizar_tabla()
    reconstruir_arbol(dni)
    estado.set(f"{persona['nombre']} se ha enviado al árbol.")


def enviar_a_tabla():
    seleccion = tabla_arbol.selection()
    if not seleccion:
        messagebox.showwarning(
            "Sin selección", "Selecciona una persona del árbol.", parent=ventana
        )
        return

    dni = seleccion[0]
    if dni not in personas_arbol:
        messagebox.showwarning(
            "Selección no válida", "Selecciona una persona, no un ciclo formativo.", parent=ventana
        )
        return

    persona = personas_arbol.pop(dni)
    personas_tabla[dni] = persona
    reconstruir_arbol()
    actualizar_tabla(dni)
    estado.set(f"{persona['nombre']} se ha enviado a la tabla.")


def mostrar_seleccion_tabla(evento=None):
    seleccion = tabla.selection()
    if seleccion and seleccion[0] in personas_tabla:
        persona = personas_tabla[seleccion[0]]
        estado.set(f"Tabla: {persona['nombre']} ({seleccion[0]}).")


def mostrar_seleccion_arbol(evento=None):
    seleccion = tabla_arbol.selection()
    if not seleccion:
        return

    iid = seleccion[0]
    if iid in personas_arbol:
        persona = personas_arbol[iid]
        estado.set(f"árbol: {persona['nombre']} ({iid}).")
    else:
        estado.set(f"Ciclo seleccionado: {tabla_arbol.item(iid, 'text')}.")


etiqueta_estado = ttk.Label(ventana, textvariable=estado)
etiqueta_estado.grid(row=1, column=0, columnspan=3, sticky="w", padx=12, pady=(0, 8))
tabla.bind("<<TreeviewSelect>>", mostrar_seleccion_tabla)
tabla_arbol.bind("<<TreeviewSelect>>", mostrar_seleccion_arbol)

ttk.Label(marco_intercambio, text="Mover selección").grid(row=0, column=0, pady=8)
ttk.Button(
    marco_intercambio,
    text="→",
    command=enviar_a_arbol,
    width=5
).grid(row=1, column=0, pady=8)
ttk.Button(
    marco_intercambio,
    text="←",
    command=enviar_a_tabla,
    width=5
).grid(row=2, column=0, pady=8)
```

## Bucle de eventos

`mainloop()` mantiene la ventana abierta y permite seleccionar personas, consultar el estado y activar los botones de transferencia.


```python
ventana.mainloop()
```

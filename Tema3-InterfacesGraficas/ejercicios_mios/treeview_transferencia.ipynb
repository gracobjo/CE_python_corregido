### Explicación de la interfaz, bloque a bloque

Esta celda crea y configura la ventana, los marcos, los dos `Treeview` y sus barras de desplazamiento. **Todavía no inserta las personas**: eso se hace en la sección siguiente, «Mostrar las personas».

#### Bloque 1: columnas y encabezados

`columnas_tabla` y `columnas_arbol` fijan los identificadores y el orden de las columnas de datos. La tabla incluye todos los campos; el árbol no incluye `nombre` porque el nombre se mostrará en la columna especial `#0`, ni `ciclo` porque se usa para formar los nodos padre. En el árbol sí se conserva `curso` como columna.

`titulos_tabla` y `titulos_arbol` relacionan cada identificador interno con el texto visible del encabezado. Por ejemplo, la clave `telefono` se muestra como `Teléfono`. Más adelante, al insertar los datos, el valor de `ciclo` se obtiene de `persona["ciclo_formativo"]`.

#### Bloque 2: ventana principal y distribución

`ventana = tk.Tk()` crea la ventana raíz de la aplicación; los demás widgets se construyen dentro de ella. `title()` establece el texto de la barra superior y `geometry("1600x620")` define el tamaño inicial en píxeles.

`rowconfigure(0, weight=1)` permite que la fila principal crezca al redimensionar la ventana. Las columnas 0 y 2 reciben `weight=1`, de modo que los paneles de la tabla y el árbol pueden repartirse el espacio adicional. La columna 1, que contiene los botones de transferencia, no recibe peso y conserva su tamaño solicitado.

#### Bloque 3: marcos contenedores

`ttk.LabelFrame` crea los paneles con borde y título para la tabla y el árbol. `ttk.Frame` crea el panel central, sin título, para los botones de transferencia.

Los tres paneles se colocan con `grid()` en la fila 0: tabla en la columna 0, controles en la columna 1 y árbol en la columna 2. `sticky="nsew"` hace que los paneles laterales ocupen los cuatro lados de su celda; `padx` y `pady` añaden márgenes exteriores. No se asigna `sticky` al panel central porque su contenido puede conservar el tamaño natural.

#### Bloque 4: Treeview de la tabla

Las llamadas a `rowconfigure()` y `columnconfigure()` hacen que la celda del `Treeview` crezca dentro de `marco_tabla`. El widget se crea así:

```python
tabla = ttk.Treeview(
    marco_tabla,
    columns=columnas_tabla,
    show="headings",
    selectmode="browse"
)
```

`show="headings"` muestra los encabezados y las filas, pero oculta la columna jerárquica `#0`; por eso esta vista se comporta como una tabla plana. `selectmode="browse"` limita la selección a una persona cada vez, que es lo que esperan los botones de transferencia.

El bucle `for columna, titulo in titulos_tabla.items()` establece el encabezado visible y centra cada columna con un ancho inicial de 110 píxeles. Después, `nombre` y `ciclo` reciben anchos mayores para que quepan mejor sus textos.

#### Bloque 5: barras de desplazamiento de la tabla

Se crea una barra vertical conectada a `tabla.yview` y otra horizontal conectada a `tabla.xview`. El argumento `command` permite que mover la barra desplace el `Treeview`. A su vez, `yscrollcommand=...set` y `xscrollcommand=...set` permiten que el `Treeview` actualice la posición de las barras cuando cambia la vista, por ejemplo, al desplazarse por las filas.

Finalmente, `grid()` coloca el `Treeview` en la celda (0, 0), la barra vertical a su derecha y la horizontal debajo. Los valores `sticky="nsew"`, `"ns"` y `"ew"` hacen que cada widget se estire en las direcciones necesarias.

#### Bloque 6: Treeview del árbol

El segundo `Treeview` se construye dentro de `marco_arbol`. `show="tree headings"` muestra tanto la columna especial `#0` como los encabezados de datos: a diferencia de la tabla plana, esta vista puede mostrar padres e hijos.

`heading("#0", ...)` pone el encabezado de esa columna y `column("#0", anchor="w", width=270)` alinea su contenido a la izquierda y fija su ancho. En la sección siguiente, los ciclos formativos se insertan como nodos padre y los nombres como nodos hijo en `#0`. El bucle sobre `titulos_arbol` configura los encabezados y anchos de las demás columnas.

#### Bloque 7: barras de desplazamiento del árbol

El árbol tiene sus propias barras vertical y horizontal. Se conectan con `tabla_arbol.yview` y `tabla_arbol.xview`, y el `Treeview` las actualiza mediante `yscrollcommand` y `xscrollcommand`. Luego se colocan junto al árbol con el mismo patrón de `grid()` usado para la tabla.

### Otras decisiones de diseño

- **`grid()` frente a `pack()`:** `grid()` permite especificar filas y columnas, útil para alinear cada vista con sus barras. `pack()` también podría organizar los tres paneles lateralmente, pero suele requerir marcos adicionales para colocar y alinear las barras. No se deben mezclar `grid()` y `pack()` dentro de un mismo contenedor.
- **`place()`:** permite indicar posiciones y tamaños concretos, pero habría que recalcularlos al redimensionar la ventana; por eso no es la opción adecuada aquí.
- **`ttk.PanedWindow`:** permitiría que el usuario arrastrase separadores para cambiar el tamaño de los paneles. Es una alternativa si ese comportamiento es un requisito; esta interfaz mantiene una columna central fija para los botones.
- **Uno o dos `Treeview`:** se usan dos para comparar simultáneamente la tabla plana y el árbol. Con uno solo se podría cambiar entre ambas presentaciones, pero no verlas a la vez.
- **`selectmode`:** `browse` limita la selección a una fila. `extended` o `multiple` permiten seleccionar varias; para usarlos habría que adaptar la transferencia, que actualmente procesa una sola selección.
- **`Frame` o `LabelFrame`:** `LabelFrame` añade un título visible al panel; `Frame` es más sencillo visualmente, pero no muestra ese título.
- **Columnas:** `stretch` controla si una columna puede crecer al redimensionar el widget; `minwidth` establece un ancho mínimo. No se configuran aquí porque el ejemplo asigna anchos iniciales y usa desplazamiento horizontal.
- **Estilo:** `ttk.Style()` permitiría personalizar fuentes, colores y otros aspectos visuales sin cambiar la estructura de los widgets.

### Preguntas habituales para explicar el diseño

- **¿Qué hace `sticky="nsew"`?** Estira el widget hacia los cuatro lados de su celda. Para que el espacio adicional se reparta al redimensionar, también se configuran los pesos (`weight`) de las filas y columnas correspondientes.
- **¿Por qué la columna central no tiene `weight=1`?** Para que los paneles de datos reciban el espacio adicional y los botones centrales mantengan su tamaño solicitado.
- **¿Qué pasa si se quita `yscrollcommand` o `xscrollcommand`?** La barra puede seguir enviando desplazamientos al `Treeview` mediante `command`, pero el `Treeview` ya no actualizará la posición del indicador de la barra. Si se quita `command`, mover la barra deja de desplazar el `Treeview`.
- **¿Por qué `selectmode="browse"`?** Porque el código de transferencia mueve una persona cada vez. La selección múltiple exigiría recorrer y mover todos los identificadores seleccionados.


## Mostrar las personas

La tabla guarda los datos directamente en sus columnas. En el árbol, `grupos_arbol` asocia cada ciclo formativo con el identificador de su nodo padre; al insertar una persona se crea el grupo solo si todavía no existe.

Al reconstruir el árbol se eliminan los nodos anteriores y se vuelven a crear a partir de `personas_arbol`. Así desaparecen también los grupos que se quedan vacíos después de una transferencia.


````python
def actualizar_tabla(dni_seleccionado=None):
    for iid in tabla.get_children(""):
        tabla.delete(iid)

    for dni, persona in personas_tabla.items():
        tabla.insert(
            "",
            "end",
            iid=dni,
            values=(
                dni,
                persona["nombre"],
                persona["edad"],
                persona["ciudad"],
                persona["telefono"],
                persona["ciclo_formativo"],
                "Sí" if persona["repite"] else "No",
                persona["curso"]
            )
        )

    if dni_seleccionado in personas_tabla:
        tabla.selection_set(dni_seleccionado)
        tabla.focus(dni_seleccionado)
        tabla.see(dni_seleccionado)


def reconstruir_arbol(dni_seleccionado=None):
    for iid in tabla_arbol.get_children(""):
        tabla_arbol.delete(iid)

    grupos_arbol.clear()
    for dni, persona in personas_arbol.items():
        ciclo = persona["ciclo_formativo"]
        if ciclo not in grupos_arbol:
            grupos_arbol[ciclo] = tabla_arbol.insert(
                "", "end", text=ciclo, open=True
            )

        tabla_arbol.insert(
            grupos_arbol[ciclo],
            "end",
            iid=dni,
            text=persona["nombre"],
            values=(
                dni,
                persona["edad"],
                persona["ciudad"],
                persona["telefono"],
                "Sí" if persona["repite"] else "No",
                persona["curso"]
            )
        )

    if dni_seleccionado in personas_arbol:
        tabla_arbol.selection_set(dni_seleccionado)
        tabla_arbol.focus(dni_seleccionado)
        tabla_arbol.see(dni_seleccionado)


grupos_arbol = {}
actualizar_tabla()
reconstruir_arbol()
````

### Explicación de cómo se muestran las personas

Los diccionarios `personas_tabla` y `personas_arbol` son la fuente de los datos de cada vista. Las funciones de esta sección no mueven personas entre diccionarios: borran los elementos visuales existentes y los vuelven a crear a partir de los datos actuales. Las transferencias entre diccionarios se realizan en la sección siguiente.

#### Bloque 1: actualizar la tabla plana

```python
def actualizar_tabla(dni_seleccionado=None):
```

La función reconstruye las filas de `tabla`. El parámetro opcional permite indicar un DNI cuya fila se volverá a seleccionar después de actualizar la vista.

```python
for iid in tabla.get_children(""):
    tabla.delete(iid)
```

`get_children("")` devuelve los elementos que cuelgan de la raíz del `Treeview`. Como la tabla es plana, esas filas son todas las personas. Se eliminan para evitar duplicados; luego se insertarán de nuevo según el diccionario actualizado.

```python
for dni, persona in personas_tabla.items():
    tabla.insert(...)
```

El bucle recorre cada par del diccionario: `dni` es la clave y `persona` contiene sus campos. `insert("", "end", ...)` añade la fila directamente en la raíz y al final de la tabla. `iid=dni` asigna a la fila un identificador estable, que permite encontrarla y seleccionarla sin depender del número de fila.

La tupla de `values` debe seguir exactamente el orden de `columnas_tabla`, definido en «Crear las dos vistas». Por eso incluye DNI, nombre, edad, ciudad, teléfono, ciclo, si repite y curso, en ese orden. La expresión `"Sí" if persona["repite"] else "No"` convierte el valor booleano en texto más legible.

```python
if dni_seleccionado in personas_tabla:
    tabla.selection_set(dni_seleccionado)
    tabla.focus(dni_seleccionado)
    tabla.see(dni_seleccionado)
```

Este bloque solo actúa si el DNI sigue en la tabla. `selection_set()` selecciona la fila, `focus()` establece el elemento enfocado y `see()` desplaza la vista si hace falta para que la fila quede visible. Si no se pasa un DNI, la tabla se actualiza sin seleccionar ninguna fila.

#### Bloque 2: reconstruir el árbol agrupado

```python
def reconstruir_arbol(dni_seleccionado=None):
```

Esta función vuelve a crear `tabla_arbol` y acepta el mismo parámetro opcional para restaurar la selección de una persona después de una transferencia.

```python
for iid in tabla_arbol.get_children(""):
    tabla_arbol.delete(iid)
grupos_arbol.clear()
```

Se eliminan los nodos raíz anteriores. Al borrar un nodo padre, también desaparecen sus descendientes. Después se vacía `grupos_arbol`, que relaciona cada nombre de ciclo con el identificador (`iid`) de su nodo padre. Esto evita conservar identificadores de nodos que ya no existen y permite eliminar grupos que hayan quedado vacíos.

```python
for dni, persona in personas_arbol.items():
    ciclo = persona["ciclo_formativo"]
    if ciclo not in grupos_arbol:
        grupos_arbol[ciclo] = tabla_arbol.insert(
            "", "end", text=ciclo, open=True
        )
```

El bucle recorre las personas que pertenecen al árbol. Si todavía no hay un nodo para su ciclo, se crea uno en la raíz (`parent=""`). `text=ciclo` muestra el ciclo en la columna `#0`; `open=True` hace que el grupo aparezca desplegado. El `iid` que devuelve `insert()` se guarda en `grupos_arbol` para reutilizar ese mismo padre con las demás personas del ciclo.

```python
tabla_arbol.insert(
    grupos_arbol[ciclo],
    "end",
    iid=dni,
    text=persona["nombre"],
    values=(...)
)
```

La persona se inserta como hija del nodo de su ciclo. Su DNI vuelve a ser el `iid`, pero `text` muestra el nombre en `#0`. La tupla `values` sigue el orden de `columnas_arbol`: DNI, edad, ciudad, teléfono, repite y curso. El ciclo y el nombre no se repiten en esa tupla porque se muestran en la jerarquía del árbol.

Al final se restaura la selección solo si el DNI indicado sigue en `personas_arbol`. Así se selecciona la persona después de enviarla al árbol; si no se proporciona un DNI, el árbol queda sin selección explícita.

#### Bloque 3: inicialización de las dos vistas

```python
grupos_arbol = {}
actualizar_tabla()
reconstruir_arbol()
```

Se crea vacío el mapa de ciclos a nodos padre. Después se dibuja cada vista usando el contenido actual de `personas_tabla` y `personas_arbol`, preparado al inicio del ejemplo. Estas llamadas iniciales no mueven ni duplican datos: solo generan su representación visual.

#### Qué ocurre durante una transferencia

La sección siguiente modifica los diccionarios: quita a la persona del origen y la añade al destino. Después llama a `actualizar_tabla()` y `reconstruir_arbol()` para reflejar el nuevo reparto. Si se transfiere una persona, se pasa su DNI a la función de la vista destino para dejarla seleccionada allí. Como el árbol se reconstruye desde cero, los grupos que se queden sin personas desaparecen automáticamente.


## Botones de transferencia

Selecciona una persona (no un nodo de ciclo) y pulsa la flecha que apunta a la otra vista. `selectmode="browse"` limita cada selección a una persona. Al moverla, se quita del diccionario de origen y se añade al de destino; después se actualizan las dos vistas y se conserva la selección en el destino.

La flecha `→` envía de la tabla al árbol y la flecha `←` devuelve del árbol a la tabla. Si no hay una persona seleccionada o se selecciona un nodo de agrupación, se muestra un aviso y no se modifica ningún dato.


````python
estado = tk.StringVar(value="Selecciona una persona y pulsa una flecha para transferirla.")


def enviar_a_arbol():
    seleccion = tabla.selection()
    if not seleccion:
        messagebox.showwarning(
            "Sin selección", "Selecciona una persona de la tabla.", parent=ventana
        )
        return

    dni = seleccion[0]
    if dni not in personas_tabla:
        messagebox.showwarning(
            "Selección no válida", "La selección no corresponde a una persona.", parent=ventana
        )
        return

    persona = personas_tabla.pop(dni)
    personas_arbol[dni] = persona
    actualizar_tabla()
    reconstruir_arbol(dni)
    estado.set(f"{persona['nombre']} se ha enviado al árbol.")


def enviar_a_tabla():
    seleccion = tabla_arbol.selection()
    if not seleccion:
        messagebox.showwarning(
            "Sin selección", "Selecciona una persona del árbol.", parent=ventana
        )
        return

    dni = seleccion[0]
    if dni not in personas_arbol:
        messagebox.showwarning(
            "Selección no válida", "Selecciona una persona, no un nodo de agrupación.", parent=ventana
        )
        return

    persona = personas_arbol.pop(dni)
    personas_tabla[dni] = persona
    reconstruir_arbol()
    actualizar_tabla(dni)
    estado.set(f"{persona['nombre']} se ha enviado a la tabla.")


def mostrar_seleccion_tabla(evento=None):
    seleccion = tabla.selection()
    if seleccion and seleccion[0] in personas_tabla:
        persona = personas_tabla[seleccion[0]]
        estado.set(f"Tabla: {persona['nombre']} ({seleccion[0]}).")


def mostrar_seleccion_arbol(evento=None):
    seleccion = tabla_arbol.selection()
    if not seleccion:
        return

    iid = seleccion[0]
    if iid in personas_arbol:
        persona = personas_arbol[iid]
        estado.set(f"árbol: {persona['nombre']} ({iid}).")
    else:
        estado.set(f"Grupo seleccionado: {tabla_arbol.item(iid, 'text')}.")


etiqueta_estado = ttk.Label(ventana, textvariable=estado)
etiqueta_estado.grid(row=1, column=0, columnspan=3, sticky="w", padx=12, pady=(0, 8))
tabla.bind("<<TreeviewSelect>>", mostrar_seleccion_tabla)
tabla_arbol.bind("<<TreeviewSelect>>", mostrar_seleccion_arbol)

ttk.Label(marco_intercambio, text="Mover selección").grid(row=0, column=0, pady=8)
ttk.Button(
    marco_intercambio,
    text="→",
    command=enviar_a_arbol,
    width=5
).grid(row=1, column=0, pady=8)
ttk.Button(
    marco_intercambio,
    text="←",
    command=enviar_a_tabla,
    width=5
).grid(row=2, column=0, pady=8)
````

### Explicación de los botones y eventos de transferencia

Los diccionarios `personas_tabla` y `personas_arbol` indican qué personas pertenecen a cada vista. Los botones no copian una fila visual de un `Treeview` a otro: mueven la persona entre esos diccionarios y después reconstruyen las vistas para reflejar el nuevo reparto.

#### Bloque 1: mensaje de estado

```python
estado = tk.StringVar(value="Selecciona una persona y pulsa una flecha para transferirla.")
```

`StringVar` es una variable de Tkinter que puede enlazarse a un widget. Más adelante, la etiqueta `etiqueta_estado` la usa mediante `textvariable=estado`; llamar a `estado.set(...)` cambia el texto visible.

#### Bloque 2: mover una persona de la tabla al árbol

```python
seleccion = tabla.selection()
```

`selection()` devuelve una tupla con los identificadores (`iid`) seleccionados. Como el `Treeview` se creó con `selectmode="browse"`, normalmente contiene cero o un identificador. Si está vacía, se muestra un aviso y `return` termina la función sin cambiar los datos.

El identificador de una fila de persona es su DNI. Se comprueba que ese DNI todavía pertenezca a `personas_tabla`; si no, se muestra un aviso y se sale. Esta comprobación evita modificar los diccionarios ante una selección que no corresponde a una persona.

```python
persona = personas_tabla.pop(dni)
personas_arbol[dni] = persona
```

`pop()` obtiene la persona y la elimina del diccionario de origen. La siguiente asignación la incorpora al diccionario del árbol con el mismo DNI. Así la persona queda en una sola vista, sin crear una copia.

Después, `actualizar_tabla()` y `reconstruir_arbol(dni)` vuelven a dibujar ambas vistas con los diccionarios actualizados. Se pasa el DNI a `reconstruir_arbol` para seleccionar y mostrar la persona en su nueva vista. Por último, `estado.set(...)` informa de que se ha enviado al árbol.

#### Bloque 3: devolver una persona del árbol a la tabla

`enviar_a_tabla()` sigue el mismo flujo en sentido contrario: obtiene la selección con `tabla_arbol.selection()`, avisa si no hay selección y comprueba que el identificador pertenezca a `personas_arbol`.

Esta última comprobación también descarta los nodos padre del árbol. Sus identificadores los genera el `Treeview` y no aparecen como claves en `personas_arbol`; solo las personas tienen un `iid` igual a su DNI. Por eso, si se selecciona un grupo en vez de una persona, se muestra un aviso y no se modifica nada.

Si la selección es válida, se mueve el registro con `personas_arbol.pop(dni)` y `personas_tabla[dni] = persona`. Luego se reconstruye el árbol y se actualiza la tabla con `actualizar_tabla(dni)`, que deja seleccionada la fila de destino. El mensaje de estado confirma el movimiento.

#### Bloque 4: mostrar qué elemento está seleccionado

`mostrar_seleccion_tabla(evento=None)` y `mostrar_seleccion_arbol(evento=None)` son manejadores de eventos. El parámetro opcional `evento` permite que Tkinter los invoque cuando cambia la selección. Cada función consulta el elemento seleccionado y actualiza `estado` con información de la persona.

En el árbol se comprueba si el `iid` está en `personas_arbol`. Si está, es una persona y el mensaje muestra su nombre y DNI. Si no está, se trata de un nodo de agrupación; `tabla_arbol.item(iid, "text")` obtiene el texto del grupo para mostrarlo en el estado.

#### Bloque 5: etiqueta, eventos y botones

```python
etiqueta_estado = ttk.Label(ventana, textvariable=estado)
etiqueta_estado.grid(row=1, column=0, columnspan=3, ...)
```

La etiqueta presenta el mensaje de `estado` debajo de las tres columnas de la interfaz. `columnspan=3` hace que ocupe las columnas de la tabla, los controles y el árbol.

```python
tabla.bind("<<TreeviewSelect>>", mostrar_seleccion_tabla)
tabla_arbol.bind("<<TreeviewSelect>>", mostrar_seleccion_arbol)
```

`bind()` conecta el evento virtual `<<TreeviewSelect>>` con el manejador correspondiente. Cuando cambia la selección en una vista, se actualiza el mensaje de estado.

El `Label` del marco central identifica la zona de transferencia. El botón `→` usa `command=enviar_a_arbol` y mueve la selección de la tabla al árbol; el botón `←` usa `command=enviar_a_tabla` y realiza el movimiento inverso. Los botones se colocan en el marco central con `grid()`. `width=5` fija un ancho cómodo para mostrar cada flecha.

#### Resumen del flujo

1. Seleccionar una persona en una de las vistas.
2. Pulsar la flecha que apunta a la vista de destino.
3. Validar que la selección sea una persona y no un grupo.
4. Quitarla del diccionario de origen y añadirla al de destino.
5. Reconstruir las dos vistas y seleccionar la persona en el destino.

Si no hay una persona seleccionada o se selecciona un nodo de agrupación, se muestra un aviso y el reparto de los datos queda intacto.


## Probar otras formas de agrupar

El árbol se crea inicialmente agrupado por ciclo formativo. Para comparar otras opciones, ejecuta primero la celda de ayuda común y luego **una sola** de las celdas de opción siguientes. Hazlo después de ejecutar las celdas anteriores y antes de `ventana.mainloop()`. Cada opción actualiza el árbol y redefine `reconstruir_arbol`, por lo que las transferencias posteriores mantienen el agrupamiento elegido. Ejecutar otra opción cambia el agrupamiento.


### Función común para reconstruir el árbol

La función recibe cómo obtener el nombre del grupo y qué columna debe ocultarse para no repetir el mismo dato en cada persona. Después crea un nodo por grupo e inserta debajo las personas correspondientes. Las celdas de opción reutilizan esta función y solo cambian esos parámetros.


````python
columnas_base_arbol = ("dni", "edad", "ciudad", "telefono", "repite", "curso")
titulos_base_arbol = {
    "dni": "DNI",
    "edad": "Edad",
    "ciudad": "Ciudad",
    "telefono": "Teléfono",
    "repite": "Repite",
    "curso": "Curso"
}
anchos_base_arbol = {
    "dni": 90,
    "edad": 70,
    "ciudad": 120,
    "telefono": 110,
    "repite": 90,
    "curso": 90
}


def crear_reconstructor_arbol(clave_grupo, etiqueta_grupo, columna_grupo=None):
    def reconstruir_arbol(dni_seleccionado=None):
        for iid in tabla_arbol.get_children(""):
            tabla_arbol.delete(iid)

        columnas = tuple(
            columna for columna in columnas_base_arbol
            if columna != columna_grupo
        )
        tabla_arbol.configure(columns=columnas)
        tabla_arbol.heading("#0", text=f"{etiqueta_grupo} / Persona")
        tabla_arbol.column("#0", anchor="w", width=270)
        for columna in columnas:
            tabla_arbol.heading(columna, text=titulos_base_arbol[columna])
            tabla_arbol.column(
                columna,
                anchor="center",
                width=anchos_base_arbol[columna]
            )

        grupos = {}
        for dni, persona in personas_arbol.items():
            grupo = clave_grupo(persona)
            if grupo not in grupos:
                grupos[grupo] = tabla_arbol.insert(
                    "", "end", text=grupo, open=True
                )

            valores = {
                "dni": dni,
                "edad": persona["edad"],
                "ciudad": persona["ciudad"],
                "telefono": persona["telefono"],
                "repite": "Sí" if persona["repite"] else "No",
                "curso": persona["curso"]
            }
            tabla_arbol.insert(
                grupos[grupo],
                "end",
                iid=dni,
                text=persona["nombre"],
                values=tuple(valores[columna] for columna in columnas)
            )

        if dni_seleccionado in personas_arbol:
            tabla_arbol.selection_set(dni_seleccionado)
            tabla_arbol.focus(dni_seleccionado)
            tabla_arbol.see(dni_seleccionado)

    return reconstruir_arbol
````

### Explicación de la función común

Esta celda define una fábrica de funciones: `crear_reconstructor_arbol(...)` recibe la regla de agrupación y devuelve una función llamada `reconstruir_arbol`. La celda por sí sola define la lógica, pero no dibuja el árbol hasta que se ejecuta una celda de opción que llama a la función devuelta.

**Dependencias y orden de ejecución:** antes de ejecutar una opción, deben haberse ejecutado las celdas que crean `tabla_arbol` y preparan `personas_arbol`. Ejecuta el notebook desde el principio hasta las celdas anteriores a `ventana.mainloop()`, después esta función común y luego **una sola** opción. Las opciones no son programas independientes: son configuraciones alternativas de la interfaz y los datos ya creados.

#### Bloque 1: columnas, encabezados y anchos base

`columnas_base_arbol` enumera las columnas de información que pueden mostrarse en las filas de personas. `titulos_base_arbol` traduce sus identificadores a encabezados visibles y `anchos_base_arbol` establece un ancho inicial para cada una.

Estos diccionarios comparten las mismas claves para que la función pueda configurar cada columna por su identificador. La columna especial `#0` no se incluye aquí: se reserva para el texto de los nodos de grupo y el nombre de cada persona.

#### Bloque 2: parámetros de `crear_reconstructor_arbol`

```python
def crear_reconstructor_arbol(clave_grupo, etiqueta_grupo, columna_grupo=None):
```

- `clave_grupo` es una función que recibe una persona y devuelve el valor que la agrupa. Por ejemplo, `lambda persona: persona["curso"]` agrupa por curso.
- `etiqueta_grupo` es el texto que se usa en el encabezado especial `#0`, como `"Curso / Persona"`.
- `columna_grupo` indica qué columna normal debe ocultarse porque su dato ya aparece como nodo padre. Para el curso se pasa `"curso"`; para el ciclo se deja `None`, pues `ciclo_formativo` no es una columna de datos del árbol.

#### Bloque 3: la función interior `reconstruir_arbol`

```python
def reconstruir_arbol(dni_seleccionado=None):
```

Esta función se ejecuta cada vez que se quiere volver a dibujar el árbol. El DNI opcional permite restaurar la selección de una persona después de transferirla.

Primero elimina los nodos que cuelgan de la raíz del `Treeview`. Borrar los nodos padre también elimina sus descendientes, así que no quedan filas duplicadas al reconstruir. Los diccionarios de personas no se borran ni se modifican: son la fuente de los datos.

#### Bloque 4: configurar las columnas para la opción

```python
columnas = tuple(
    columna for columna in columnas_base_arbol
    if columna != columna_grupo
)
```

La comprensión conserva las columnas base excepto la que se está usando como agrupación. Así, al agrupar por curso, ciudad o si repite, no se muestra ese mismo dato también en cada fila. Al agrupar por ciclo, `columna_grupo` es `None` y se mantienen todas las columnas base.

`tabla_arbol.configure(columns=columnas)` actualiza las columnas del widget. A continuación, se cambia el encabezado de `#0` y se vuelven a establecer los títulos, alineaciones y anchos de las columnas visibles. Esto permite reutilizar el mismo `Treeview` para las distintas opciones.

#### Bloque 5: crear los grupos y las filas

`grupos = {}` es un mapa local que relaciona el valor de cada grupo con el `iid` del nodo padre creado en el `Treeview`. Al ser local, comienza vacío en cada reconstrucción y no conserva identificadores de nodos eliminados.

Para cada persona, `clave_grupo(persona)` calcula el grupo. Si ese grupo aún no está en el mapa, `tabla_arbol.insert("", "end", ...)` crea un nodo padre en la raíz y su `iid` se guarda en `grupos`. Si ya existe, se reutiliza el padre existente. De esta manera se crea un nodo por grupo, aunque tenga varias personas.

El diccionario `valores` prepara los datos de una persona por identificador de columna. La tupla que se pasa a `values` se forma siguiendo el orden de `columnas`; así los valores coinciden con las columnas incluso cuando una de ellas se oculta.

Cada persona se inserta como hija de `grupos[grupo]`. `iid=dni` hace que su identificador sea el DNI; `text=persona["nombre"]` muestra el nombre en `#0`; `values` muestra el resto de datos. El booleano `repite` se representa como `"Sí"` o `"No"`.

#### Bloque 6: restaurar la selección y devolver la función

Si `dni_seleccionado` todavía pertenece a `personas_arbol`, se selecciona, enfoca y hace visible su fila con `selection_set()`, `focus()` y `see()`. La comprobación evita intentar seleccionar una persona que ya no esté en esta vista.

`return reconstruir_arbol` devuelve la función interior sin ejecutarla en ese momento. La celda de cada opción asigna esa función a `reconstruir_arbol` y después la llama para dibujar el árbol. Como los botones también consultan el nombre `reconstruir_arbol` cuando se pulsan, las transferencias posteriores continúan usando la opción elegida.

#### Cómo se aplica una opción

Por ejemplo, la celda para agrupar por curso hace dos cosas: crea la función con una regla que lee `persona["curso"]` y oculta la columna `curso`; después llama a la función para reconstruir el árbol. Ejecutar otra celda de opción reemplaza esa configuración. Por eso se debe ejecutar una sola opción antes de `ventana.mainloop()`.


### Opción 1: agrupar por ciclo formativo

Cada ciclo aparece como nodo padre. Es el agrupamiento inicial del ejemplo y resulta útil porque varias personas comparten ciclo. No se repite el ciclo en las columnas de cada persona.


````python
reconstruir_arbol = crear_reconstructor_arbol(
    lambda persona: persona["ciclo_formativo"],
    "Ciclo formativo"
)
reconstruir_arbol()
````

### Opción 2: agrupar por curso

El árbol tendrá grupos como `primero` y `segundo`. La columna `curso` se oculta en las filas porque el curso ya está indicado en el nodo padre.


````python
reconstruir_arbol = crear_reconstructor_arbol(
    lambda persona: persona["curso"],
    "Curso",
    "curso"
)
reconstruir_arbol()
````

### Opción 3: agrupar por si repite

El `bool` de `repite` se convierte en los textos `Repite` y `No repite`, más claros como categorías. Se oculta la columna `repite` porque esa información ya aparece en el nodo padre.


````python
reconstruir_arbol = crear_reconstructor_arbol(
    lambda persona: "Repite" if persona["repite"] else "No repite",
    "Repite",
    "repite"
)
reconstruir_arbol()
````

### Opción 4: agrupar por ciudad

Cada ciudad se convierte en un nodo padre. En este conjunto hay ciudades con una sola persona, así que esta opción sirve para probar el código, aunque agrupa menos que hacerlo por ciclo o curso. La columna `ciudad` se oculta para evitar repetir el nombre.


````python
reconstruir_arbol = crear_reconstructor_arbol(
    lambda persona: persona["ciudad"],
    "Ciudad",
    "ciudad"
)
reconstruir_arbol()
````

### Opción 5: agrupar por ciclo y luego por curso

Esta opción crea dos niveles de grupos: primero el ciclo formativo y, dentro, el curso. Las personas quedan en el tercer nivel. Se elimina la columna `curso` porque el curso ya aparece como nodo intermedio.


````python
def reconstruir_arbol_por_ciclo_y_curso(dni_seleccionado=None):
    for iid in tabla_arbol.get_children(""):
        tabla_arbol.delete(iid)

    columnas = tuple(
        columna for columna in columnas_base_arbol
        if columna != "curso"
    )
    tabla_arbol.configure(columns=columnas)
    tabla_arbol.heading("#0", text="Ciclo / Curso / Persona")
    tabla_arbol.column("#0", anchor="w", width=300)
    for columna in columnas:
        tabla_arbol.heading(columna, text=titulos_base_arbol[columna])
        tabla_arbol.column(
            columna,
            anchor="center",
            width=anchos_base_arbol[columna]
        )

    grupos = {}
    for dni, persona in personas_arbol.items():
        ciclo = persona["ciclo_formativo"]
        curso = persona["curso"]

        if ciclo not in grupos:
            iid_ciclo = tabla_arbol.insert(
                "", "end", text=ciclo, open=True
            )
            grupos[ciclo] = {"iid": iid_ciclo, "cursos": {}}

        cursos = grupos[ciclo]["cursos"]
        if curso not in cursos:
            cursos[curso] = tabla_arbol.insert(
                grupos[ciclo]["iid"], "end", text=curso, open=True
            )

        valores = {
            "dni": dni,
            "edad": persona["edad"],
            "ciudad": persona["ciudad"],
            "telefono": persona["telefono"],
            "repite": "Sí" if persona["repite"] else "No"
        }
        tabla_arbol.insert(
            cursos[curso],
            "end",
            iid=dni,
            text=persona["nombre"],
            values=tuple(valores[columna] for columna in columnas)
        )

    if dni_seleccionado in personas_arbol:
        tabla_arbol.selection_set(dni_seleccionado)
        tabla_arbol.focus(dni_seleccionado)
        tabla_arbol.see(dni_seleccionado)


reconstruir_arbol = reconstruir_arbol_por_ciclo_y_curso
reconstruir_arbol()
````

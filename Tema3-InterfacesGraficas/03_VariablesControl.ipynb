# Variables de control

Las variables de control son objetos especiales que se utilizan para conectar (vincular) los datos de los widgets (como campos de texto, botones de opción o casillas de verificación) con el código de Python.

Si cambia la variable cambiará el control y viceversa.

Tenemos las siguientes variables de control:
* `StringVar`: Maneja cadenas de texto (str). Es la más utilizada para campos de entrada de texto (Entry) o etiquetas dinámicas (Label).
* `IntVar`: Maneja números enteros (int). Se usa habitualmente para almacenar el estado de casillas de verificación (Checkbutton) o selecciones numéricas.
* `DoubleVar`: Maneja números de punto flotante o decimales (float). Es ideal para widgets como barras de deslizamiento (Scale).
* `BooleanVar`: Maneja valores booleanos (True o False). Muy útil para conocer si un Checkbutton está marcado o no.

Todas estas variables comparten los mismos métodos esenciales para gestionar su contenido que básicamente son:
* `set(valor)`: Establece o cambia el valor de la variable. Al hacerlo, todos los widgets vinculados se actualizan automáticamente.
* `get()`: Devuelve el valor actual almacenado en la variable (con su tipo de dato correspondiente).

Se irán viendo según hablemos de los distintos widgets.

### Ejemplo
Tenemos una etiqueta y una caja de texto asociadas a la misma variable de control (bidireccional).
* Al cambiar el texto en la caja se modifica la variable de control (del widget a la variable).
* Al cambiar la variable de control se modifica la etiqueta (de la variable al widget).

```python
import tkinter as tk

ventana = tk.Tk()
ventana.title("Ventana principal")
ventana.geometry("400x300")

# Se crear la variable de control de tipo texto
texto_variable = tk.StringVar(value="Escribe algo aquí...")

# Se asocia la variable a un widget Entry (entrada de texto)
entrada = tk.Entry(ventana, textvariable=texto_variable, width=25)
entrada.pack(pady=20)

# Se asociar la misma variable a un widget Label (etiqueta)
etiqueta = tk.Label(ventana, textvariable=texto_variable, font=("Arial", 12))
etiqueta.pack(pady=10)

ventana.mainloop()
```

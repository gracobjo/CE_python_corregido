# Widgets temáticos

En tkinter existen dos tipos de widgets:
* **Clásicos**
* **Temáticos**

La diferencia principal entre los widgets clásicos de `Tk (tkinter)` y los widgets temáticos de `Ttk (tkinter.ttk)` radica en su **apariencia**, **diseño** y **sistema de estilos**.

___
* Los controles clásicos más sencillos nos los podemos encontrar con su versión temática y, por supuesto, los avanzados.
* De momento, los widgets clásicos que se han visto en clase son únicamente `Label`, `Entry` y `Button`.
* A partir de este punto vamos a usar en algunos casos la versión temática.
* En el ejemplo siguiente vemos un ejercicio ya hecho en clase (segunda celda de código del fichero 07_Widget_Label) pero con los controles `Label`, `Entry` y `Frame` en su versión temática. Podemos ver que el cambio es mínimo en cuanto a código.
* Es útil importar ttk por separado para abreviar: `form tkinter impor ttk`

```python
# COMPARAR CON EL EJERCICIO DE LA SEGUNDA CELDA DEL FICHERO 07
import tkinter as tk
from tkinter import ttk # Se importa ttk

ventana = tk.Tk()
ventana.title("Widgets temáticos")
ventana.geometry("400x300")

texto_dinamico = tk.StringVar()
texto_dinamico.set("Escribe algo abajo...")

# Se crea un Label temático (ttk) en este caso. No acepta font como parámetro
etiqueta = ttk.Label(ventana, textvariable=texto_dinamico) #, font=("Arial", 14), fg="blue")
etiqueta.pack(pady=20)

# Se crea un Entry temático (ttk). En este caso Entry si admitiría font aunque se recomienda separar estilo de funcionamiento.
caja_texto = ttk.Entry(ventana, textvariable=texto_dinamico, font=("Courier", 12))
caja_texto.pack(pady=10)

def reiniciar():
    texto_dinamico.set("¡Texto reiniciado!")

# Se crea un Button temático (ttk). No acepta bg como parámetro
boton = ttk.Button(ventana, text="Reiniciar Texto", command=reiniciar) #, bg="yellow")
boton.pack(pady=10)

ventana.mainloop()
```

## Estilos con la clase `ttk.Style`

Para personalizar la apariencia de los widgets temáticos en Tkinter, se utiliza la clase `ttk.Style()`, que funciona de manera similar a una hoja de estilos (CSS) para definir colores, fuentes y la aparencia general sin modificar cada widget individualmente.

El método `style.configure` de `tkinter.ttk` sirve para definir o modificar las propiedades visuales por defecto de un estilo.

Algunas propiedades cambian de nombre:
* fg es `foreground`
* bg es `background`
* `padding`

Los nombres de los estilos de los componentes (widgets) siguen una convención de nomenclatura muy clara.

La regla general es usar el nombre de la clase del widget precedido por un prefijo modificador (ej. `MiEstilo.TButton`) o modificar el estilo raíz del widget. 

**`[NombrePersonalizado].T[ClaseBaseWidget]`**

* `[NombrePersonalizado]`: Prefijo. **Opcional**. Es el nombre que se inventa para crear una variante de ese widget sin alterar los widgets por defecto.
* `T[ClaseBaseWidget]`: Sufijo o clase base. Es el nombre estándar del componente en `ttk`. Le dice a Tkinter a qué tipo de widget se aplica el estilo (por ejemplo, `TButton` para botones, `TLabel` para etiquetas).

Los "sufijos" son, en realidad, los nombres de las clases base de los widgets de `ttk`, los cuales siempre empiezan con una **T mayúscula** (que significa Themed). Salvo alguna excepción, ocurre con todos los widgets. La estructura del nombre es la siguiente:
* `TLabel`
* `TButton`
* `TEntry`
* ...

```python
# COMPARAR CON EL EJERCICIO DE LA SEGUNDA CELDA DEL FICHERO 07
import tkinter as tk
from tkinter import ttk # Se importa ttk

ventana = tk.Tk()
ventana.title("Widgets temáticos")
ventana.geometry("400x400")

texto_dinamico = tk.StringVar()
texto_dinamico.set("Escribe algo abajo...")

# CREAMOS LOS ESTILOS
# Se instancila la clase Style
estilo = ttk.Style()

# Se crean diferentes estilos
estilo.configure("TLabel",font=("Arial", 14), foreground="blue") # Se crea el estilo sin el prefijo
estilo.configure("estilo_caja.TEntry", padding=20) # Estilo con prefijo personalizado
estilo.configure("estilo_boton.TButton", background="red") # Estilo con prefijo personalizado

# Se crea un Label temático (ttk) en este caso. Emplamos el parámetro style.
etiqueta = ttk.Label(ventana, textvariable=texto_dinamico, style="TLabel")
etiqueta.pack(pady=20)

# Se crea un Entry temático (ttk). Emplamos el parámetro style.
caja_texto = ttk.Entry(ventana, textvariable=texto_dinamico, style="estilo_caja.TEntry")
caja_texto.pack(pady=10)

def reiniciar():
    texto_dinamico.set("¡Texto reiniciado!")

# Se crea un Button temático (ttk). Emplamos el parámetro style.
boton = ttk.Button(ventana, text="Reiniciar Texto", command=reiniciar, style="estilo_boton.TButton")
boton.pack(pady=10)

# CONTROLES SECUNDARIOS PARA COMPROBAR LA NOMENCLARUTA DE LOS ESTILOS
# Como el estilo se creó SIN sufijo personalizado, no es necesario aplicar style
etiqueta_secundaria = ttk.Label(ventana, textvariable=texto_dinamico)
etiqueta_secundaria.pack(pady=20)

# Como el estilo se creó CON sufijo personalizado, si no se aplica style no se verá el efecto
caja_texto_secundaria = ttk.Entry(ventana, textvariable=texto_dinamico)
caja_texto_secundaria.pack(pady=10)

ventana.mainloop()
```

## Tabla resumen de widgets clásicos (tk) y temáticos (ttk)

| Tipo de Widget | Nombre Clásico (`tkinter.Widget`) | Nombre Temático (`ttk.Widget`) | Descripción y uso principal |
| :--- | :--- | :--- | :--- |
| **Etiqueta** | `Label` | `Label` | Muestra texto estático o imágenes en la interfaz. |
| **Entrada de texto** | `Entry` | `Entry` | Campo de texto de una sola línea para que el usuario escriba. |
| **Botón estándar** | `Button` | `Button` | Botón común para ejecutar una acción al hacer clic. |
| **Marco / Contenedor** | `Frame` | `Frame` | Contenedor invisible para agrupar y organizar otros widgets. |
| **Marco con título** | `LabelFrame` | `LabelFrame` | Contenedor que incluye un borde y un título en la parte superior. |
| **Paneles divisorios** | `PanedWindow` | `PanedWindow` | Contenedor que permite al usuario redimensionar el espacio de sus hijos. |
| **Casilla de verificación** | `Checkbutton` | `Checkbutton` | Cuadro de selección múltiple (activa/desactiva una opción). |
| **Botón de opción** | `Radiobutton` | `Radiobutton` | Botón de selección única dentro de un grupo de opciones. |
| **Menú desplegable** | `OptionMenu` | `Combobox` | Cuadro de texto combinado con una lista desplegable de opciones. |
| **Barra de desplazamiento** | `Scrollbar` | `Scrollbar` | Permite desplazarse vertical u horizontalmente por otros widgets. |
| **Barra de estado / Ajuste** | `Scale` | `Scale` | Deslizador numérico para seleccionar un valor dentro de un rango. |
| **Contenedor con pestañas** | *No existe en clásico* | `Notebook` | Permite organizar múltiples pantallas mediante pestañas superiores. |
| **Vista de árbol / Tabla** | *No existe en clásico* | `Treeview` | Muestra datos en formato jerárquico (árbol) o tabular (columnas). |
| **Barra de progreso** | *No existe en clásico* | `Progressbar` | Muestra visualmente el estado de avance de una tarea. |
| **Separador** | *No existe en clásico* | `Separator` | Línea horizontal o vertical para dividir secciones visuales. |
| **Selector numérico** | `Spinbox` | `Spinbox` | Campo de texto con flechas para incrementar o decrementar un valor. |
| **Indicador de tamaño** | *No existe en clásico* | `Sizegrip` | Pequeño control en la esquina inferior para redimensionar la ventana. |
| **Texto multilínea** | `Text` | *No existe en TTK* | Editor de texto enriquecido de múltiples líneas. |
| **Lienzo / Gráficos** | `Canvas` | *No existe en TTK* | Área para dibujar formas geométricas, imágenes o gráficos complejos. |
| **Lista de elementos** | `Listbox` | *No existe en TTK* | Lista fija de elementos de texto donde el usuario puede seleccionar uno o varios. |
| **Menú de aplicación** | `Menu` | *No existe en TTK* | Barras de menú superiores y menús contextuales desplegables. |
| **Botón de menú** | `Menubutton` | `Menubutton` | Botón que despliega un menú al ser pulsado (en desuso por Combobox). |
| **Mensaje corto** | `Message` | *No existe en TTK* | Similar a Label, pero formatea automáticamente texto largo en párrafos. |

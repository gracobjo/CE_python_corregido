# tkinter
Tkinter es la biblioteca estándar de Python utilizada para la creación de interfaces gráficas de usuario (GUI). 

Viene integrada por defecto con la instalación de Python, lo que significa que no requiere ninguna instalación adicional y permite desarrollar aplicaciones de escritorio visuales de forma rápida y sencilla.

A través de un conjunto de componentes llamados widgets como botones, etiquetas, cuadros de texto, menús, etc.

Tkinter actúa como un puente entre el código de Python y las ventanas nativas del sistema operativo (Windows, macOS o Linux).

## La ventana principal y el bucle gráfico
Toda aplicación de Tkinter necesita una ventana raíz (`Tk()`) y un bucle principal (`mainloop()`) que mantiene la ventana abierta y escucha los eventos (clics, teclado, etc.).

Tres pasos fundamentales:
* Crear ventana principal
* Configurar ventana principal
* Ejecutar ventana principal

````python
import tkinter as tk

# 1. Crear la ventana principal
ventana = tk.Tk()

# 2. Configurar propiedades básicas
ventana.title("Mi Primera App con Tkinter")
ventana.geometry("400x300") # Ancho x Alto en píxeles

# 3. Iniciar el bucle de la aplicación
ventana.mainloop()
````

___
### Añadimos algún control y evento a la ventana

````python
import tkinter as tk

def saludar():
    etiqueta.config(text="Has hecho clic en el botón.")

ventana = tk.Tk()
ventana.title("Aplicación básica")
ventana.geometry("400x200")

etiqueta = tk.Label(ventana, text="Etiqueta antes de pulsar el botón", font=("Arial", 14))
etiqueta.pack(pady=20) # 'pady' añade espacio vertical interno

boton = tk.Button(ventana, text="Haz clic aquí", command=saludar, bg="lightblue")
boton.pack(pady=10)

ventana.mainloop()
````

## resizable
Se utiliza para controlar si el usuario puede cambiar el tamaño de una ventana (tanto a lo ancho como a lo alto). Por defecto, todas las ventanas que creas en Tkinter se pueden estirar o encoger arrastrando sus bordes. Con este método se puede bloquear esa capacidad por completo o limitarla a una sola dirección.

* **ancho**: Valor booleano que indica si se puede modificar el ancho.
* **alto**: Valor booleano que indica si se puede modificar el alto.

Sin parámetro se puede consultar el valor devolviendo una tupla.

````python
import tkinter as tk

ventana = tk.Tk()
ventana.geometry("400x300")

# La ventana no se puede cambiar de tamaño en ninguna dirección
ventana.resizable(True, False)

# print(ventana.resizable()) #Escribe por consola fuera de la ventana.

ventana.mainloop()
````

## minsize y maxsize
Sirven para establecer los límites de dimensión de una ventana, controlando el tamaño mínimo y máximo al que un usuario puede arrastrar o redimensionar la ventana.

1. El método `minsize()`

Define las dimensiones mínimas que puede tener la ventana. El usuario no podrá encoger la ventana más allá de estos valores.
* **Sintaxis**: objeto_ventana.minsize(width, height)
* **Parámetros**: Tanto width (ancho) como height (alto) se expresan en píxeles.
* **Uso común**: Evita que el usuario oculte accidentalmente componentes importantes (como botones o textos) al hacer la ventana demasiado pequeña.

2. El método `maxsize()`

Define las dimensiones máximas que puede tener la ventana. El usuario no podrá estirar la ventana ni agrandarla más allá de estos límites.
* **Sintaxis**: objeto_ventana.maxsize(width, height)
* **Parámetros**: Al igual que el anterior, width y height se definen en píxeles.
* **Uso común**: Útil cuando tu diseño de interfaz no es responsivo o se deforma si la ventana se vuelve gigantesca. También limita el comportamiento del botón "Maximizar" del sistema operativo a esas medidas exactas.

También es psoible consultar los valores actuales obteniendo una tupla en ambos casos.

La diferencia con `resizable`es que `minsize`y `maxsize`establecen topes numéricos pero `resizable` bloquea por completo la capacidad de cambio de tamaño.

````python
import tkinter as tk

ventana = tk.Tk()
ventana.title("Ejemplo de Límites de Tamaño")

ventana.geometry("400x300")

# Aplicar restricciones de tamaño (Ancho, Alto)
ventana.minsize(300, 200)  # No se puede hacer más chica que 300x200
ventana.maxsize(600, 500)  # No se puede hacer más grande que 600x500

ventana.mainloop()
````

## iconify, destroy, state, after

````python
import tkinter as tk

ventana = tk.Tk()

ventana.title("Demostración de varios métodos")
ventana.geometry("400x300")

boton1 = tk.Button(ventana,text="iconify",command=lambda: ventana.iconify())
boton1.pack(pady=10)

boton2 = tk.Button(ventana,text="destroy",command=lambda: ventana.destroy())
boton2.pack(pady=10)

boton3 = tk.Button(ventana,text="state",command=lambda: ventana.state("zoomed")) # normal, iconic, zoomed
boton3.pack(pady=10)

def despues():
    ventana.destroy()

boton4 = tk.Button(ventana,text="after: destroy en 2 seg",command=lambda: ventana.after(2000,despues))
boton4.pack(pady=10)

ventana.mainloop()
````

## deiconify y withdraw

````python
import tkinter as tk

ventana = tk.Tk()

ventana.title("Demostración de iconify()")
ventana.geometry("400x300")

def restaurar():
    ventana.deiconify()
    
def despues():
    ventana.withdraw()
    ventana.after(4000,restaurar)

boton1 = tk.Button(ventana,text="withdraw",command=despues)
boton1.pack(pady=10)

ventana.mainloop()
````

## protocol

````python
import tkinter as tk
from tkinter import messagebox

def confirmar_salida():
    if messagebox.askokcancel("Salir", "¿Seguro que quieres cerrar la aplicación?"):
        ventana.destroy()

ventana = tk.Tk()
ventana.title("Ejemplo de protocol")
ventana.geometry("300x200")

ventana.protocol("WM_DELETE_WINDOW", confirmar_salida)

ventana.mainloop()
````

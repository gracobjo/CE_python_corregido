# Widget `Listbox`

El control `ListBox` muestra una lista de elementos de texto de la que se pueden seleccionar uno o varios.

Propiedades:
* `selectmode`: Define cuántos elementos se pueden seleccionar y cómo. Los valores principales son:
	* `tk.BROWSE` (Por defecto): Permite seleccionar sólo un elemento. Si se arrastra el ratón, la selección se mueve.
	* `tk.SINGLE`: Permite seleccionar solo un elemento, pero no se puede arrastrar la selección.
	* `tk.MULTIPLE`: Permite seleccionar varios elementos haciendo clic en cada uno (actúan como interruptores).
	* `tk.EXTENDED`: Permite seleccionar múltiples elementos usando las teclas `Shift` y `Ctrl`.
* `height`: Define el número de filas (líneas) visibles. Si hay más elementos, se ocultan a menos que se use una barra de desplazamiento.
* `width`: Define el ancho del control en caracteres (no en píxeles).
* `exportselection`: Por defecto es `True`. Si hay varios `ListBox`, al seleccionar algo en uno se deselecciona el otro. Si se pone a `False`, cada lista mantiene su propia selección de forma independiente.
* `bg` / `fg`: Cambian el color de fondo (bg) y el color del texto (fg).

Métodos:
* `insert`: Añade elementos. Con `tk.END` como índice se añaden al final.
* `curselection`: Devuelve una tupla con los índices (posiciones numéricas, empezando en 0) de los elementos seleccionados.
* `get`: Devuelve el texto del elemento en esa posición.

Eventos:
* `<<ListboxSelect>>`: Se dispara automáticamente cada vez que la selección cambia (ya sea porque el usuario hizo clic en otro elemento o usó las flechas del teclado).

```python
import tkinter as tk
from tkinter import ttk

ventana = tk.Tk()
ventana.title("Widget Listbox")
ventana.geometry("400x300")

# Etiqueta de título
label = ttk.Label(ventana, text="Selecciona un lenguaje:")
label.pack(anchor=tk.W, pady=(0, 5))

# Se crea el Listbox (del módulo tk básico, ya que ttk no tiene Listbox)
lista_elementos = tk.Listbox(
    ventana,
    height=8,
    selectmode=tk.EXTENDED,
    font=("Arial", 10)
)
lista_elementos.pack(pady=10)

# Se insertan elementos en la lista
lenguajes = ["Python", "JavaScript", "C++", "Java", "Ruby", "TypeScript", "Go", "Rust"]
for item in lenguajes:
    lista_elementos.insert(tk.END, item)

# Función para obtener la selección
def mostrar_seleccion():
    seleccion = lista_elementos.curselection()
    if seleccion:
        elemento = lista_elementos.get(seleccion[0]) # Obtiene sólo el primero de los que estén seleccionados
        resultado_label.config(text=f"Seleccionaste: {elemento}")
    else:
        resultado_label.config(text="Por favor, selecciona un elemento.")

btn = ttk.Button(ventana, text="Mostrar selección", command=mostrar_seleccion)
btn.pack(pady=10)

# Etiqueta para mostrar el resultado
resultado_label = ttk.Label(ventana, text="Selecciona un elemento de la lista")
resultado_label.pack(anchor=tk.W, pady=5)

def al_seleccionar(event):
    seleccion = lista_elementos.curselection()
    if seleccion:
        print(f"Primero de los seleccionados:{event.widget.get(seleccion[0])}")
    
lista_elementos.bind("<<ListboxSelect>>", al_seleccionar) # Evento que se dispara al seleccionar un elemento

ventana.mainloop()
```

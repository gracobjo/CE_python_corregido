# Widget `Label`

Muestra texto o imágenes.

Propiedades:
* `text`: Define el texto plano que mostrará la etiqueta.
* `textvariable`: Permite asociar una variable de control para cambiar el texto de forma dinámica.
* `bg` o `background`: Cambia el color de fondo del widget.
* `fg` o `foreground`: Cambia el color de la letra.
* `font`: Permite especificar la familia tipográfica, el tamaño y el estilo de la fuente. Por ejemplo: ("Helvetica", 12, "bold").
* `width` y `height`: Ancho y alto (en caracteres para texto y en pixels para imágenes).
* `wraplength`: Define el ancho en píxeles a partir del cual el texto se divide automáticamente en varias líneas.
* `justify`: Es la alineación pudiendo ser `right`, `left` o `center`.
* `padx `y `pady`: Relleno en horizontal y en vertical.
* `relief`: Estilo del borde: `flat`, `raised`, `sunken`, `groove` y `ridge`.
* `bd` o `borderwidth`: Grosor del borde.
* `image`: Permite mostrar una imagen.

Métodos:
* `config` o `configure`: Establece las propiedades que se le indican. Es el mismo método que de hecho se usa en todos los widgets.

___
### Propiedades `text`, `bg`, `fg`, `font`, `wraplength`, `relief` y `bd`

```python
import tkinter as tk

ventana = tk.Tk()
ventana.title("Widget Label")
ventana.geometry("400x300")

# Podemos poner propiedades a la hora de crear el componente
etiqueta1 = tk.Label(ventana,
                 text="Este es el texto de la etiqueta",
                 bg="yellow",
                 fg="green",
                 font=("courier",12,"bold"),
                 wraplength=40)

# Podemos poner propiedades con el componente ya creado usando el método config
etiqueta1.configure(relief="groove",bd=15)

etiqueta1.pack(pady=20)

ventana.mainloop()
```

___
### Etiqueta con una imagen

```python
import tkinter as tk

ventana = tk.Tk()
ventana.title("Widget Label")
ventana.geometry("400x300")

imagen = tk.PhotoImage(file="..\\Imagenes\\python_small.png")
etiqueta_imagen = tk.Label(ventana, image=imagen)
etiqueta_imagen.pack(expand=True)

ventana.mainloop()
```

### Propiedades `textvariable`

```python
import tkinter as tk

ventana = tk.Tk()
ventana.title("Widget Label")
ventana.geometry("400x300")

# Se crea la variable de control (StringVar)
texto_dinamico = tk.StringVar()
texto_dinamico.set("Escribe algo abajo...")

# Se crea el Label y se vincula con la variable de control
etiqueta = tk.Label(ventana, textvariable=texto_dinamico, font=("Arial", 14), fg="blue")
etiqueta.pack(pady=20)

# Se crea un Entry (caja de texto) y se vincula a la misma variable
# Así, al escribir en la caja de texto se modificará la variable y se actualizará el Label automáticamente
caja_texto = tk.Entry(ventana, textvariable=texto_dinamico, font=("Arial", 12))
caja_texto.pack(pady=10)

# Botón para demostrar cómo cambiar el texto desde el código
def reiniciar():
    texto_dinamico.set("¡Texto reiniciado!")

boton = tk.Button(ventana, text="Reiniciar Texto", command=reiniciar,bg="yellow")
boton.pack(pady=10)

ventana.mainloop()
```

## Fechas
Es necesario importar desde el módulo **datetime** la clase **date**

```python
from datetime import date

hoy = date.today()
print(hoy)
print(type(hoy))
```

___
### Craeación de fechas
* **today():** Crea la fecha actual
* **date(año,mes,dia):** Crea una fecha especificada

```python
import datetime

hoy = date.today()
print(hoy)

fecha = date(2025,5,23)
print(fecha)
```

___
### Atributos principales
* year
* month
* day

Son de sólo lectura

```python
import datetime

fecha = date(2025,5,23)

print(fecha)

print("Año",fecha.year)
print("Mes",fecha.month)
print("Día",fecha.day)

# fecha.year = 2027 # ERROR: No se puede modificar directamente
```

___
### Comparación de fechas

```python
import datetime

fecha1 = date(2022,4,11)
fecha2 = date(2025,4,11)

if fecha1 < fecha2:
    print(f"{fecha1} es anterior a {fecha2}")
else:
    print(f"{fecha1} es posterior a {fecha2}")
```

___
### Diferencia entre fechas

```python
import datetime

fecha1 = date(2024,4,6)
fecha2 = date(2025,4,11)

fecha3 = fecha2 -fecha1
print(fecha3)
```

___
### Métodos principales
* **weekday()**: Devuelve un número entre 0 (lunes) y 6 (domingo)
* **replace()**: Reemplaza cualquiera de los valores (día, mes o años). Funciona como parámetros por nombre y de forma arbitraria.

```python
# WEEKDAY
import datetime

fecha = date(2022,4,11)
print(fecha.weekday())
print(date.today().weekday())
```

```python
# REPLACE
import datetime

fecha = date(2022,4,11)
print(fecha.replace(year=2018,day=9))
print(fecha.replace(month=6))
```

___
### Otros métodos de creación
* **fromtimestamp()**: Crea una fecha con los segundos especificados desde el 1 de enero de 1970
* **fromisoformat()**: Crea una fecha a partir de una cadena YYYY-MM-DD

```python
# FROMTIMESTAMP
import datetime

fecha = date.fromtimestamp(100) # 86400 segundos en un día
print(fecha)

fecha = date.fromtimestamp(90000) # 86400 segundos en un día
print(fecha)

fecha = date.fromtimestamp(1790794772) # 86400 segundos en un día
print(fecha)
```

```python
# FROMISOFORMAT
import datetime

fecha = date.fromisoformat("2025-04-10")
print("Año",fecha.year)
print("Mes",fecha.month)
print("Día",fecha.day)
```

___
### Métodos de formateo
* **strftime()**: Crea una cadena de sustitución personalizada
* **isoformat()**: Devuelve una cadena de fecha con formato YYYY-MM-DD
* **ctime()**: Devuelve una cadena de fecha con formato de texto legible en inglés
* **toordinal()**: Número gregoriano de la fecha. El ordinal 1 es el 1 de enero del año 1

```python
# STRFTIME
import datetime

fecha = date(2022,4,21)
print(fecha.strftime("El año es %Y, el mes es %m y el día del mes es %d"))
```

```python
# ISOFORMAT
import datetime

fecha = date(2022,4,21)
print(fecha.isoformat())
print(type(fecha.isoformat()))
```

```python
# CTIME
import datetime

fecha = date(2022,4,21)
print(fecha.ctime())
print(type(fecha.ctime()))
```

```python
# TOORDINAL
import datetime

fecha = date(2022,4,21)
print(fecha.toordinal())

print(date.today().toordinal())
```

## Horas

```python
from datetime import time

ahora = time()
print(ahora)

hora1 = time(14,30,12,345)
print(hora1)
print(hora1.hour)
print(hora1.minute)
print(hora1.second)
print(hora1.microsecond)

# hora1.hour = 4 # ERROR: No se puede modificar

# hora = time(42,30,12,345) # ERROR: No puedes dar un valor no válido. Hora entre 0 y 23

# Método replace para modificar una hora
hora3 = hora1.replace(hour=6,minute=34,second=56,microsecond=62455)
print(hora3)

# Formatos
print("\nFormatos")
print(hora3.strftime("La hora es %H y los minutos %M y los segundos %S"))
print(hora3.isoformat())

# Comparar horas
hora4 = time(15,30,0)
hora5 = time(15,35,0)
if hora4 < hora5:
    print("Es anterior")
else:
    print("Es posterior")

# hora6 = hora5 - hora4 # ERROR: No se pueden restar horas así
# 
fechahora1 = datetime.datetime(2026,8,20,15,30,0)
fechahora2 = datetime.datetime(2026,8,20,15,35,0)
fechahora3 = fechahora2 -fechahora1
print(fechahora3)

hora7 = datetime.time(15,30,40) # time está realmente detro de datetime
```

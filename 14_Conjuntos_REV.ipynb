# Conjuntos
Almacenan múltiples elementos en una variable:
* Se crean con llaves {}
* Sus elementos están:
    * No están ordenados
    * No son modificables (podemos añadir y quitar elementos pero no cambiarlos).
    * No son accesibles mediante índices
    * Nermite almacenar valores duplicados
    * Permite elementos de distinto tipo **salvo compuestos**

___
### Definición básica de un conjunto

```python
conjunto1 = {1,2,3,4,5}
conjunto2 = set({1,2,3,4,5}) # A partir de un conjunto
conjunto3 = set([1,2,3,4,5]) # A partir de una lista
conjunto4 = set((1,2,3,4,5)) # A partir de una tupla
print(type(conjunto1),type(conjunto2),type(conjunto3),type(conjunto4))

print("=======")

conjunto5 = {1} # Conjunto de un elemento
conjunto6 = {1,"cadena"} # Conjunto con elementos de distinto tipo
print(type(conjunto5),type(conjunto6))

print("=======")

conjunto7 = {} # NO ES UN CONJUNTO VACÍO. ES UN DICCIONARIO
conjunto8 = set() # Conjunto vacío
print(type(conjunto7),type(conjunto8))
```

___
### Podemos tener diferentes tipos de datos incluso mezclados
Los conjuntos **no pueden contener listas, conjuntos o diccionarios** pero **sí tuplas o frozensets**

```python
conjunto9 = {1,2,3}
conjunto10 = {"texto1","texto2","texto3"}
conjunto11 = {"texto1",6,7,8.5,"texto2"}

#conjunto12 = {1,2,["a","b"]} # ERROR: Contiene una lista
conjunto13 = {1,2,("a","b")} # TUPLAS SI
# conjunto14 = {1,2,{"a","b"}} # ERROR: Contiene un conjunto
# conjunto15 = {1,2,{"nombre":"juan","edad":30}} # ERROR: Contiene un diccionario
conjunto16 = {1,2,frozenset({"a","b"})} # CONJUNTOS CONGELADOS SI POR SER LA VERSIÓN INMUTABLE DE UN CONJUNTO
```

___
### No se puede accede a los conjuntos a través de un índice
**Solución**: Reconvertir a lista o tupla

```python
conjunto17 = {"texto1",6,7,8.5,"texto2"}

# print(conjunto9[0]) # ERROR: No se puede acceder a través de un índice
# Reconvertimos a lista
print(list(conjunto17)[0])
print(list(conjunto17)[1])
print(list(conjunto17)[2])
print(list(conjunto17)[3])
print(list(conjunto17)[4])

print("=======")
```

### Podemos acceder a través de un bucle for

```python
conjunto18 = {1,2,3,4,5}
for i in conjunto18:
    print(i,end=' ')
```

___
### No podemos hacer slicing
**Solución**: Reconvertiendo a lista previamente

```python
abecedario = {'a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k',
              'l', 'm', 'n', 'ñ', 'o', 'p', 'q', 'r', 's', 't', 'u', 
              'v', 'w', 'x', 'y', 'z'}

print(set(list(abecedario)[0:5]))
print(list(abecedario)[3:9])
print(list(abecedario)[9:])
print(list(abecedario)[:9])
print(list(abecedario)[-9:-4])

# print(list(abecedario[-9:-4])) #ERROR: Cuidado donde ponemos el paréntesis
```

Empleamos el operador **in** o el operador **not in**

```python
abecedario = {'a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k',
              'l', 'm', 'n', 'ñ', 'o', 'p', 'q', 'r', 's', 't', 'u', 
              'v', 'w', 'x', 'y', 'z'}

if "A" in abecedario: # Se diferencia entre mayúsculas y minúsculas
    print("Si está")
else:
    print("No está")
```

___
### Cambiar conjuntos
Aunque son inmutables y no podemos cambiar elementos ni individualmente, podemos reconvertirla con el mismo nombre una vez modificada pasándola a lista primero.

No se mantiene el orden haciendo esto. Es un nuevo literal y se añade al _final_ (orden no significativo).

```python
coches = {"seat","audi","citroen"}
print(coches)

lista_coches = list(coches)
lista_coches[0] = "mercedes"
print(lista_coches)

coches = set(lista_coches)
print(coches)

# De igual forma podríamos usar todos los métodos de List y posteriormente volvemos a reconvertir en tupla
```

___
### Añadir elementos
Aunque podemos reconvertir como hemos hecho antes, disponemos de dos métodos para añadir elementos:
* **add()**: Añade un elemento
* **update()**: Añade elementos

```python
coches = {"seat","audi","citroen"}
coches.add("bmw")
coches.add("seat") # Ya existe. No da error pero tampoco duplica
# coches.add(["opel","mercedes"]) # ERROR: No puedes añadir listas
coches.add(("opel","mercedes")) # Funciona pero añades una tupla, no los elementos separadamente
#coches.add({"opel","mercedes"}) # ERROR: No puedes añadir diccionarios
print(coches)

print("=====")
coches = {"seat","audi","citroen"}
coches.update(["opel","mercedes"]) # Admite una lista y los mete de forma individual
coches.update(("fiat","skoda")) # Admite una tupla y los mete de forma individual
coches.update({"hyundai","nissan"}) # Admite un conjunto y los mete de forma individual
print(coches)
print(len(coches))
```

___
### Eliminar elementos
Aunque podemos reconvertir como hemos hecho antes, disponemos de dos métodos para añadir elementos:
* **remove**: Elimina un elemento. Si no existe da error.
* **discard**: Elimina un elemento. Si no existe no da error.
* **pop()**: Elimina un elemento aleatoriamente. Devuelve el elemento eliminado.
* **clear()**: Elimina todos los elementos

```python
coches = {'citroen', 'audi', 'opel', 'fiat', 'seat', 'nissan', 'hyundai', 'mercedes', 'skoda'}

coches.remove("opel")
# coches.remove("kia") # ERROR: El valor no está en el conjunto

coches.discard("seat")
coches.discard("kia") # No lo elimina puesto que no está pero no da error

print(coches)

print(coches.pop())

print(coches)

print(coches.clear())
```

___
### Operadores sobre conjuntos

```python
numeros1 = {1,2,3,4,5,6}
numeros2 = {4,5,6,7,8,9}

# Operadores que devuelven otro conjunto
print("Unión:",numeros1 | numeros2) # Combina los elementos sin repetirlos
print("Intersección:",numeros1 & numeros2) # Devuelve los que están en ambos
print("Diferencia:",numeros1 - numeros2) # Devuelve los que están en el primero y sin que estén en el segundo
print("Diferencia:",numeros2 - numeros1) # Igual que el anterior
print("Diferencia asimétrica:",numeros1 ^ numeros2) # Devuelve los que está en uno o en otro pero no en ambos

# Operadores que devuelven un booleano
print("Pertenencia:", 1 in numeros1) # Comprueba si un elemento se encuentra dentro del conjunto
print("No pertenencia:", 8 not in numeros1) # Comprueba si un elemento no se encuentra dentro del conjunto

print("Igualdad:",numeros1 == numeros2) # Comprueba si dos conjuntos tiene exactamente los mismos elementos
print("Igualdad:",{1,2,3} == {1,2,3})

numeros3 = {2,3}
print("Subconjunto:",numeros3 <= numeros1)
print("Subconjunto:",numeros3 <= numeros2)

numeros4 = {1,2,3,4,5,6,7}
print("Superconjunto:",numeros4 >= numeros1)
print("Superconjunto:",numeros4 >= numeros2)
```

___
### Otros métodos
* union
* copy
* difference
* difference_update
* intersection
* intersection_update
* isdisjoint
* issubset
* issuperset
* symmetric_difference
* symmetric_difference_update

```python
print("\nUNION")
numeros1 = {1,2,3,4,5,6}
numeros2 = {4,5,6,7,8,9}
print(numeros1.union(numeros2)) # Igual que el operador |

print("\nCOPY")
numeros1 = {1,2,3,4,5,6}
numeros2 = {4,5,6,7,8,9}
copia_numeros1 = numeros1.copy()
print(copia_numeros1)

print("\nDIFFERENCE")
numeros1 = {1,2,3,4,5,6}
numeros2 = {4,5,6,7,8,9}
print(numeros1.difference(numeros2)) # Igual que el operador -, es decir, los que están en el primero y no en el segundo
print(numeros1)

print("\nDIFFERENCE_UPDATE")
numeros1 = {1,2,3,4,5,6}
numeros2 = {4,5,6,7,8,9}
numeros1.difference_update(numeros2) # Igual que el anterior pero actualiza el conjunto inicial
print(numeros1)

print("\nINTERSECTION")
numeros1 = {1,2,3,4,5,6}
numeros2 = {4,5,6,7,8,9}
print(numeros1.intersection(numeros2)) # Devuelve un conjunto con los elementos que comparten

print("\nINTERSECTION_UPDATE")
numeros1 = {1,2,3,4,5,6}
numeros2 = {4,5,6,7,8,9}
numeros1.intersection_update(numeros2) # Igual que el anterior pero actualiza el conjunto inicial
print(numeros1)

print("\nSYMMETRIC_DIFFERENCE")
numeros1 = {1,2,3,4,5,6}
numeros2 = {4,5,6,7,8,9}
print(numeros1.symmetric_difference(numeros2)) # Igual que el operador ^

print("\nSYMMETRIC_DIFFERENCE_UPDATE")
numeros1 = {1,2,3,4,5,6}
numeros2 = {4,5,6,7,8,9}
numeros1.symmetric_difference_update(numeros2) # Igual que el anterior pero actualiza el conjunto inicial
print(numeros1)

print("\nISDISJOINT")
numeros1 = {1,2,3,4,5,6}
numeros2 = {4,5,6,7,8,9}
print(numeros1.isdisjoint(numeros2)) # Devuelve verdadero si no hay elementos de un conjunto en el otro
print({"a","b"}.isdisjoint(numeros2))

print("\nISSUBSET")
numeros1 = {1,2,3,4,5,6}
numeros2 = {4,5,6,7,8,9}
print({1,2}.issubset(numeros1)) # Igual que el operador <=
print({"a","b"}.issubset(numeros1))

print("\nISSUPERSET")
numeros1 = {1,2,3,4,5,6}
numeros2 = {4,5,6,7,8,9}
print(numeros1.issuperset({1,2})) # Igual que el operador >=
print({"a","b"}.issuperset(numeros1))
```

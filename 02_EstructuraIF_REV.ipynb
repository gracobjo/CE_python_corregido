# **Estructura IF..ELSE**
Se emplea para ejecutar código condicionalmente.

Se combinan los siguientes elementos:
 * La palabra reservada **if**, a continuación u
 * Una **condición (todo lo compleja que se quiera)**
 * carácter dos puntos **:**
 * El **código a ejecutar** si la condición se ha cumplido

___
## **Sintaxis**
```Python
if condición:
    Bloque de código
```

````python
if 5 > 2:
    print("5 es mayor que 2")
````

Mientras la indentación se mantenga todas las sentencias pertenecen al bloque y se ejecutan si la condición es verdadera

````python
if 5 > 2:
    print("Resultado")
    print("5 es mayor que 2")

print("Otra sentencia") # esta sentencia se ejecuta siempre puesto que está fuera del bloque
````

___
## **elif**
La expresión **elif** permite comprobar otra condición si la primera no se cumple.


````python
a = 2
b = 2
if a > b:
    print("a es mayor que b")
elif a == b:
    print("a es igual a b")
````

___
# **else**
La expresión **else** permite ejecutar un código si las condiciones anteriores no se han cumplido

````python
a = 3
b = 3
if a > b:
    print("a es mayor que b")
elif a < b:
    print("a es menos que a b")
else:
    print("a es igual a b")
````

___
# **Abreviatura**
Si sólo se necesita una sentencia a ejecutar se puede colocar todo en la misma línea pero se siguen necesitando los dos puntos

````python
if 5 > 2: print("5 es mayor que 2")
````

___
# **Anidación**
Podemos tener varias estructuras anidadas, es decir, unas dentro de otras.

````python
a = 5
b = 2
if a > b:
    print("a es mayor que b")
    if a < b:
        print("a es menor que b")
    else:
        print("a y b son iguales")
````

___
# **Operador ternario**
Es a abreviatura del **if..else** en una sólo línea. Se usa cuando debemos elegir entre dos valores asociados a una condición.

````python
edad = 21
mayor = True if edad >= 18 else False
print(mayor)
````

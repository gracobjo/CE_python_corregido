# Funciones **filter**, **sorted** y **map**

# Filter
Esta función hace un filtrado de un conjunto quedándose con aquellos elementos que cumplen una condición.

## Sintaxis
```python
filter(funcion,iterable)
```
## Parámetros:
* **funcion**:Función que determina si un elemento cumple o no una condición en función si devuelve True o False
* **iterable**:El conjunto inicial

## Devuelve:
* Un objeto de tipo _filter_ por lo que hay que pasarlo a _list_, _tuple_, _set_ o _dict_.

````python
# No quedamos con los elementos pares
def es_par(x):
    return not (x % 2)

numeros=[2,4,3,5,6,7,9,1,8]

print(type(filter(es_par,numeros)))
print(filter(es_par,numeros))
print(list(filter(es_par,numeros)))
````

````python
# No quedamos con las palabras de longitud <= 5
def longitud_hasta5(x):
    return 0 if len(x)>5 else 1

palabras=["bicicleta","casa","boligrafo","cena","bocadillo","placa"]

print(list(filter(longitud_hasta5,palabras)))
````

___
Se suelen usar funciones lambda puesto que podemos poner la condición de filtrado en una única sentecia.

````python
palabras=["bicicleta","casa","boligrafo","cena","bocadillo","placa"]

print(list(filter(lambda x: 0 if len(x)>5 else 1 ,palabras)))
````

# Sorted
Esta función hace una ordenación de un conjunto según un criterio establecido.

## Sintaxis
```python
sorted(iterable,funcion,reverse)
```
## Parámetros:
* **iterable**:El conjunto inicial
* **funcion**:Función que determina el criterio para establecer el orden (opcional)
* **reverse**: Booleano que indica si el orden es ascendente o descendente

## Devuelve:
* Un objeto de tipo _list_

````python
#Ejemplo básico de sort sin funciones: ordenación alfanumérica

print("\nLista","-"*100)
lista = [4,2,7,5,9,3]
print(sorted(lista))
print(type(sorted(lista)))

print("\nTupla","-"*100)
tupla = (4,2,7,5,9,3)
print(sorted(tupla))
print(type(sorted(tupla))) #Devuelve una lista

print("\nConjunto","-"*100)
conjunto = {4,2,7,5,9,3}
print(sorted(conjunto))
print(type(sorted(conjunto))) #Devuelve una lista

print("\nDiccionario","-"*100)
diccionario = {
    "Valladolid":"Castilla y Leon",
    "Oviedo":"Asturias",
    "Sevilla":"Andalucia" }
print(sorted(diccionario))
print(type(sorted(diccionario))) #Devuelve una lista ordenada por el valor, no por la clave
````

## Sorted con función integrada

````python
print("\nPalabras por longitud","-"*100)
palabras = ["sol", "luna", "estrella", "nube", "viento", "mar", "tierra"]

print(sorted(palabras,key=len))
print(sorted(palabras,key=len,reverse=True))
````

## Sorted con función definida por el usuario

````python
print("\nNúmeros próximos a 5","-"*100)
def proximo_a_cinco(x):
    return abs(5-x)

numeros = [7, 2, 9, 1, 5, 10, 4, 3, 8, 6]
print(sorted(numeros,key=proximo_a_cinco))
````

## Sorted con función lambda

````python
print("\nNúmeros próximos a 5 con función lambda","-"*100)

numeros = [7, 2, 9, 1, 5, 10, 4, 3, 8, 6]
print(sorted(numeros,key=lambda x: abs(5-x)))
````

# Map
Esta función hace una transformación elemento a elemento de un conjunto d

## Sintaxis
```python
map(funcion,iterable)
```
## Parámetros:
* **funcion**:Función que determina el criterio para establecer el orden (opcional)
* **iterable**:Una colección de conjuntos. La función debe tener un parámetro por cada colección.


## Devuelve:
* Un objeto de tipo _map_

````python
lista = [1,2,3,4,5] En el ejemplo es una lista pero puede ser tupla, conjunto, etc.

print(type(map(lambda x: x*x,lista))) # tipo map
print(type(list(map(lambda x: x*x,lista)))) # convertido a list
print(map(lambda x: x*x,lista)) # Obtenemos la referencia al objeto map
print(list(map(lambda x: x*x,lista))) # Obtenemos la lista con sus elementos transformados
````

````python
#Podemos pasar dos parámetros
#Calculamos la diferencia entre los números de igual índice de dos listas

lista1 = [4,5,2,8,1]
lista2 = [7,1,2,7,9]

print(list(map(lambda x,y : x-y,lista1,lista2)))
````

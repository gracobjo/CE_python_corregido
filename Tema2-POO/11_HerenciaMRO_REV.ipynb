# Method Resolution Order
MRO (Method Resolution Order u Orden de Resolución de Métodos) es la secuencia que utiliza Python para buscar métodos y atributos en una jerarquía de clases, lo cual es fundamental cuando se trabaja con herencia múltiple.

Python utiliza el algoritmo de linealización C3 para calcular este orden, el cual cumple con tres reglas clave:
1. Las clases hijas se comprueban antes que sus padres.
2. Si se hereda de varias clases, se comprueban de izquierda a derecha según se declaren.
3. Una clase no se comprueba más de una vez en el recorrido.

Se puede consultar el MRO de dos formas:
1. Accediendo al atributo especial `__mro__` que devuelve una tupla.
2. Llamando al método `mro()` que devuelve una lista

___
## Ejemplo básico de MRO

La clase **A** es la superclase y **B** hereda de **A**.

Ambas clases implementan el método **imprimir**.

En este caso el método **imprimir** se llama en el contexto de **B**, que es clase hija de **A**. La clase **B** se comprueba antes que **A** por lo que se ejecuta **B**.

`Cadena MRO: B --> A`

```python
class A:
    def imprimir(self):
        print("En la clase A")

class B(A):
    def imprimir(self):
        print("En la clase B")
    
objeto = B()
objeto.imprimir()
```

___
## Ejemplo del diamante de MRO

La clase **A** es la superclase y tanto **B** como **C** heredan de **A**. A su vez **D** hereda de **B** y de **C** (herencia múltiple).

Todas las clases implementan el método **imprimir** salvo **D**.

En este caso el método **imprimir** se llama en el contexto de **D** pero no la implementa. Sigue la cadena y comprueba sus padres, que en este caso son dos, es decir, tanto **B** como **C**. Puesto que **B** se declara antes es el método llamado.

`Cadena MRO: D --> B --> C --> A`

```python
class A:
    def imprimir(self):
        print("En la clase A")

class B(A):
    def imprimir(self):
        print("En la clase B")

class C(A):
    def imprimir(self):
        print("En la clase C")

class D(B, C):
    pass

objeto = D()
objeto.imprimir()
```

___
## C3 Linearization en MRO

Python usa un algoritmo llamado `C3 Linearization` para determinar en qué orden se buscan las class cuando se llama un método.

Este algoritmo produce un único orden teniendo en cuenta la herencia y el orden de escritura de las clases.

Python sigue el orden que le marca el algoritmo, es decir, lo que le marca:
1. El atributo `__mro__`
2. El método `mro()`

```python
class A:
    def imprimir(self):
        print("En la clase A")

class B(A):
    def imprimir(self):
        print("En la clase B")

class C(A):
    def imprimir(self):
        print("En la clase C")

class D(B, C):
    pass

objeto = D()
objeto.imprimir()

# Atritubos y métodos de la clase
print(D.__mro__) # Tupla
print(D.mro()) #Lista
```

## Ejemplo con muchas clases y herencia múltiple
La linealización de una clase C con padres P1, P2, ... PN se calcula:

`L(C)=[C] + merge(L(P1),L(P2),....,L(PN),[P1,P2,...,PN])`

El operador **merge** hace lo siguiente:
1. Coge el primer elemento (la cabeza) de la primera lista.
2. Comprueba si ese elemento no aparece en la "cola" (el resto de elementos) de ninguna otra lista.
	* Si es así, lo saca de las listas y lo añade al resultado final del MRO.
	* Si no es así, mira la cabeza de la siguiente lista.
3. Repite el proceso hasta que todas las clases se han ordenado o lanza un error de tipo TypeError si la jerarquía es imposible de resolver (herencia inconsistente).
![MRO para las siguientes clases](../Imagenes/MRO.png)

```python
class A:
  pass

class B:
  pass 

class C:
  pass

class D:
  pass

class E:
  pass
    
class F(A, C):
  pass 

class G(D, B, E):
  pass 
    
class H(D, A): # Añadir E al principio de la herencia de H. Da error por no poder construir un MRO consistente class H(E, D, A)
  pass

class Z(F, G, H):
  pass

print(Z.__mro__)
print("=====")
print(Z.mro())

print("=====") # Atributo name de una clase
print([ x.__name__ for x in Z.mro() ])
print([ x.__class__ for x in Z.mro() ])
```

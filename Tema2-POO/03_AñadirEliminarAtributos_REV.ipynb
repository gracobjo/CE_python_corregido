# Añadir y eliminar atributos

En ambos casos se pueden hacer tanto para la clase como para cada objeto
* Añadir atributos:
  * Atritubo de objeto o instancia
  * Atributo de clase
* Eliminar atributos:
  * De un atributo de objeto o instancia
  * De un atributo de clase

___
## 1. Añadir atributos de objeto o instancia
Se puede hacer de dos formas:
* Asignación directa: Es la forma más sencilla puesto que se añade una propiedad nueva en la propia asignación del valor de la misma.
* Con `setattr(objeto,atributo,valor)`

```python
class Coche:
    def __init__(self,marca,precio=1000):
        self.marca = marca
        self.precio = precio

    def mostrar_datos(self):
        print(f"El coche es de la marca {self.marca} y el precio es {self.precio}") 

coche1 = Coche("seat",18000)
coche1.mostrar_datos()

coche1.tipo_cambio = "Manual" # Asignación directa
setattr(coche1,"combustible","gasolina") # Con el método setattr

print(f"El timpo de cambio es {coche1.tipo_cambio} y el combustible es {coche1.combustible}")

coche2 = Coche("fiat",15000) 
try:
    print(f"El timpo de cambio es {coche2.tipo_cambio} y el combustible es {coche2.combustible}") # coche2 no tiene tipo_cambio ni combustible
except AttributeError as e:
    print(f"Error: {e}")
```

En el caso más extremo, podríamos partir de una clase vacía e ir introduciendo propiedades.

```python
class Guitarra:
    pass

guitarra1 = Guitarra()
guitarra1.tipo = "Eléctrica"

print(guitarra1.tipo)
```

___
## 2. Añadir atributos de clase
Se hace igual que en el caso anterior pero en el contexto de la clase

```python
class Coche:
    def __init__(self,marca,modelo):
        self.marca = marca
        self.modelo = modelo

    def mostrar_datos(self):
        print({self.marca},{self.modelo}) 

coche1 = Coche("seat",18000)
coche1.mostrar_datos()

Coche.tipo_cambio = "Manual" # Añado el atributo incluso antes de crear una clase
print(f"El timpo de cambio del {coche1.marca} es {Coche.tipo_cambio} o lo que es 'lo mismo' {coche1.tipo_cambio}")

coche2 = Coche("fiat",15000) 
print(f"El timpo de cambio del {coche2.marca} es {Coche.tipo_cambio} o lo que es 'lo mismo' {coche2.tipo_cambio}")
```

___
## 3. Eliminar atributos de objeto o instancia

Se eliminan los atributos para las instancias sobre la que hemos hecho esta operación ya que para el resto seguirán existiendo.

Tenemos dos formas de hacerlo:
* Con `del`
* Con `delattr(objeto,atributo)`

```python
class Coche:
    def __init__(self,marca,modelo,precio,color):
        self.marca = marca
        self.modelo = modelo
        self.precio = precio
        self.color = color

c1 = Coche("seat","ibiza",15000,"rojo")
c2 = Coche("renault","clio",16000,"blanco")

# Eliminamos con del
del c1.precio
print(c1.marca,c1.modelo,c1.color,sep="---")
print(c2.marca,c2.modelo,c2.precio,c2.color,sep="---") # El precio no se ha eliminado de coche2, sólo de coche1
try:
    print(c1.precio) # Capturamos el error puesto que precio ya no existe en coche1
except AttributeError as e:
    print(f"Error: {e}")

print("===")

# Eliminamos con delattr
delattr(c2,"color")
print(c2.marca,c2.modelo,c2.precio,sep="---")
print(c1.marca,c1.modelo,c1.color,sep="---")
try:
    print(coche2.color) # Capturamos el error puesto que color ya no existe en coche2
except AttributeError as e:
    print(f"Error: {e}")
```

___
## 4. Eliminar de atributos de clase

Se eliminan igual que los de objeto o instancia pero se usa el contexto de la clase. Podemos usar tanto `del` como `delattr`

```python
class Bicicleta:
    sistema = "pedaleo"
    ruedas = "2"

b1 = Bicicleta()
b2 = Bicicleta()

print(b1.sistema,b1.ruedas)
print(b2.sistema,b2.ruedas)

del Bicicleta.sistema
delattr(Bicicleta,"ruedas")

try:
    print(b1.sistema,b1.ruedas)
    print(b2.sistema,b2.ruedas)
except AttributeError as e:
    print(f"Error {e}")
```

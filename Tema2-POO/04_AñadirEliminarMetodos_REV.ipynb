# Añadir y eliminar métodos

Al igual que con los atributos, podemos añadir y eliminar métodos una vez creada la clase

___
## 1. Añadir método
Se puede hacer de dos formas:
* Asignación directa: Es la forma más sencilla puesto que se añade el método asignándolo a un nuevo nombre de método como si fuera un atributo 
* Con `setattr(clase u objeto,nombremetodo,funcion_externa)`

**IMPORTANTE** : Em principio sólo lo hacemos en el contexto de la clase para que esté disponible para todos los objetos que se creen nuevos.

```python
class Coche:
    def __init__(self,marca,precio):
        self.marca = marca
        self.precio = precio

    def mostrar_datos(self):
        print(f"El coche es de la marca {self.marca} y el precio es {self.precio}") 

def reparar(self,coste):
    self.precio += coste

def repostar(self,coste):
    self.precio += coste

# Agregados como métodos de clase
Coche.reparar = reparar
setattr(Coche,"repostar",reparar)  

coche1 = Coche("seat",18000) 
coche1.reparar(300)
coche1.repostar(70)
coche1.mostrar_datos()
```

___
## 2. Eliminar método
Con `del` o `delattr`

```python
class Coche:
    def __init__(self,marca,precio):
        self.marca = marca
        self.precio = precio

    def mostrar_datos(self):
        print(f"El coche es de la marca {self.marca} y el precio es {self.precio}") 

    def reparar(self,coste):
        self.precio += coste

    def repostar(self,coste):
        self.precio += coste

del Coche.reparar
delattr(Coche,"repostar")

c1 = Coche("fiat",14000)
c1.mostrar_datos()

try:
    c1.reparar()
    c1.repostar()
except AttributeError as ae:
    print(f"Error: {ae}")
```

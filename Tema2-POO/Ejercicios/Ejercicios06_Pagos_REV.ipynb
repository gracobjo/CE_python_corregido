# Ejercicios 06
Crear una clase llamada MedioPago que gestione los tipos de pagos que se puede hacer con tarjetas, bizum y transferencias bancarias.

Estos tres tipos de pago serán clases hijas de MedioPago.

Cada subclase implementará un método `procesar_pago` que podrá ser llamado dependiendo del objeto que reciba.

```python
class MedioPago:
    def __init__(self, ordenante, concepto, cantidad):
        self.ordenante = ordenante
        self.concepto = concepto
        self.cantidad = cantidad


class TarjetaCredito(MedioPago):
    def __init__(self, numero_tarjeta, ordenante, concepto, cantidad):
        super().__init__(ordenante,concepto,cantidad)
        self.numero_tarjeta = numero_tarjeta

    def procesar_pago(self):
        print(f"Pago de {self.cantidad:.2f}€ realizado por {self.ordenante} con Tarjeta de Crédito (terminada en {self.numero_tarjeta[-4:]}) en concepto de {self.concepto}")


class Bizum(MedioPago):
    def __init__(self, telefono, ordenante, concepto, cantidad):
        super().__init__(ordenante,concepto,cantidad)
        self.telefono = telefono

    def procesar_pago(self):
        print(f"Pago de {self.cantidad:.2f}€ realizado por {self.ordenante} a través de Bizum: {self.telefono} en concepto de {self.concepto}.")


class TransferenciaBancaria(MedioPago):
    def __init__(self, iban, ordenante, concepto, cantidad):
        super().__init__(ordenante,concepto,cantidad)
        self.iban = iban

    def procesar_pago(self):
        print(f"Pago de {self.cantidad:.2f}€ realizado por {self.ordenante} mediante transferencia bancaria: {self.iban} en concepto de {self.concepto}.")


# Función que demuestra el polimórfismo
def realizar_compra(medio_pago):
    medio_pago.procesar_pago()

formas_de_pago = [TarjetaCredito("4532-8888-1234-9876","Ana","Ropa",120.50),
                  Bizum("611111111","Paco","Comida",70.50),
                  TransferenciaBancaria("ES91 2100 0418 4502 0005 1332","Luis","Repuestos",200.50)]

for pago in formas_de_pago:
    realizar_compra(pago)
```

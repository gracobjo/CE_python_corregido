# Ejercicios 03
Crear una carreta de coches. Cada coche se crea con tres atributos principales: nombre, potencia y combustible. Tiene otros dos que serán tiempo y recorrido.

La única función que hace el coche es acelerar. ¿Cómo funciona? Si el coche tiene combustible acelera en función de su potencia y valor aleatorio. Este resultado será directamente proporcional al consumo realizado y a la distancia recorrida.

El coche puede ser monitorizado consultando el valor de sus atributos.

La carrera de coches se compondrá de una lista de coches.

La carrera finaliza cuando algún coche ha recorrido la distancia completa.

**NOTA**: Usar función `all()` que devuelve `True` si todos los elementos de un iterable son verdaderos
Ejemplo:
```Python
all(i % 2 == 0 for i in [2,4,6])
```

````python
import random

CONSUMO = 0.00999
TRAYECTO = 1000

class Coche:
    def __init__(self,nombre,potencia,combustible):
        self._nombre = nombre
        self._potencia = potencia
        self._combustible = combustible
        self.tiempo = 0
        self.recorrido = 0

    def acelerar(self):
        if self._combustible > 0:
            acelerado = random.random()*self._potencia
            self._combustible -= acelerado*CONSUMO
            self.recorrido += acelerado
            self.tiempo += 1

    def monitorizar(self):
        print(f"Coche: {self._nombre} - Combustible: {self._combustible} - Tiempo: {self.tiempo} - Recorrido: {self.recorrido}")

carrera = []

carrera.append(Coche("coche1",200,100))
carrera.append(Coche("coche2",220,100))
carrera.append(Coche("coche3",240,100))

while(all(coche.recorrido < TRAYECTO for coche in carrera)):
    for i in carrera:
        i.acelerar()
        i.monitorizar()
    print("=====")
````

```python
import random

CONSUMO = 0.00999
TRAYECTO = 1000

class Coche:
    def __init__(self,nombre,potencia,combustible):
        self._nombre = nombre
        self._potencia = potencia
        self._combustible = combustible
        self.tiempo = 0
        self.recorrido = 0
        # self._tiempo_por_vuelva = []

    def acelerar(self):
        if self._combustible > 0:
            acelerado = random.random()*self._potencia
            self._combustible -= acelerado*CONSUMO
            self.recorrido += acelerado
            self.tiempo += 1
            

    def monitorizar(self):
        print(f"Coche: {self._nombre} - Combustible: {self._combustible} - Tiempo: {self.tiempo} - Recorrido: {self.recorrido}")

carrera = []

carrera.append(Coche("coche1",200,100))
carrera.append(Coche("coche2",220,100))
carrera.append(Coche("coche3",240,100))

while(all(coche.recorrido < TRAYECTO for coche in carrera)):
    for i in carrera:
        i.acelerar()
        i.monitorizar()
    print("=====")
            
```

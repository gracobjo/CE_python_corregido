# Ejercicios 02
Debes crear un programa en Python que modele el funcionamiento interno de una biblioteca mediante tres clases principales: Libro, Usuario y Biblioteca.

1. **Clase Libro**

* Atributos de instancia: titulo (str), autor (str), isbn (str) y disponible (bool, que por defecto debe ser True).
* Métodos:
	* prestar(): Si el libro está disponible, cambia su estado a no disponible (False) y muestra un mensaje confirmando el préstamo. Si ya estaba prestado, avisa que no se puede prestar.
	* devolver(): Cambia el estado del libro a disponible (True) y muestra un mensaje de confirmación.

2. **Clase Usuario**

* Atributos de instancia: nombre (str), id_usuario (str) y libros_prestados (una lista vacía por defecto).
* Métodos:
	* tomar_prestado(libro): Añade el objeto Libro a su lista de libros prestados, siempre y cuando el libro se pueda prestar con éxito.
	* devolver_libro(libro): Elimina el objeto Libro de su lista y ejecuta la devolución del libro.

3. **Clase Biblioteca**

* Atributos de instancia: nombre (str), catalogo (lista vacía de libros) y usuarios (lista vacía de usuarios registrados).
* Métodos:
	* registrar_usuario(usuario): Añade un objeto Usuario a la lista de la biblioteca.
	* agregar_libro(libro): Añade un objeto Libro al catálogo.
	* mostrar_catalogo(): Muestra en pantalla la lista de todos los libros de la biblioteca indicando si están disponibles o no.

```python
class Libro:
    def __init__(self, titulo, autor, isbn):
        self.titulo = titulo
        self.autor = autor
        self.isbn = isbn
        self.disponible = True

    def prestar(self):
        if self.disponible:
            self.disponible = False
            return True
        return False

    def devolver(self):
        self.disponible = True
        print(f"El libro '{self.titulo}' ha sido devuelto a la biblioteca.")


class Usuario:
    def __init__(self, nombre, id_usuario):
        self.nombre = nombre
        self.id_usuario = id_usuario
        self.libros_prestados = []

    def tomar_prestado(self, libro):
        if libro.prestar():
            self.libros_prestados.append(libro)
            print(f"{self.nombre} ha tomado prestado '{libro.titulo}'.")
        else:
            print(f"Lo sentimos, '{libro.titulo}' ya se encuentra prestado.")

    def devolver_libro(self, libro):
        if libro in self.libros_prestados:
            self.libros_prestados.remove(libro)
            libro.devolver()
        else:
            print(f"{self.nombre} no tiene asignado el libro '{libro.titulo}'.")


class Biblioteca:
    def __init__(self, nombre):
        self.nombre = nombre
        self.catalogo = []
        self.usuarios = []

    def registrar_usuario(self, usuario):
        self.usuarios.append(usuario)
        print(f"Usuario {usuario.nombre} registrado con éxito en {self.nombre}.")

    def agregar_libro(self, libro):
        self.catalogo.append(libro)

    def mostrar_catalogo(self):
        print(f"\n--- Catálogo de la Biblioteca: {self.nombre} ---")
        for libro in self.catalogo:
            estado = "Disponible" if libro.disponible else "Prestado"
            print(f" '{libro.titulo}' por {libro.autor} (ISBN: {libro.isbn}) - [{estado}]")


# 1. Creamos la biblioteca
mi_biblioteca = Biblioteca("Biblioteca Central")

# 2. Creamos libros y los añadimos
libro1 = Libro("Cien años de soledad", "Gabriel García Márquez", "978-0307474728")
libro2 = Libro("1984", "George Orwell", "978-0451524935")
mi_biblioteca.agregar_libro(libro1)
mi_biblioteca.agregar_libro(libro2)

# 3. Registramos un usuario
usuario1 = Usuario("Carlos Gómez", "USR001")
mi_biblioteca.registrar_usuario(usuario1)

# 4. Mostramos catálogo inicial
mi_biblioteca.mostrar_catalogo()

# 5. Flujo de préstamos y devoluciones
usuario1.tomar_prestado(libro1)
mi_biblioteca.mostrar_catalogo()

# Intentamos prestar el mismo libro de nuevo
usuario1.tomar_prestado(libro1)

# Devolvemos el libro
usuario1.devolver_libro(libro1)
mi_biblioteca.mostrar_catalogo()
```

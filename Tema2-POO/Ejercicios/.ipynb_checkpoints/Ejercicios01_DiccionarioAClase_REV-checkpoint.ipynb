```python
clase_diccionario = {
    "nombre":{
        "tipo":"atributo_clase"
    },
    "edad":{
        "tipo":"atributo_objeto"
    }
}

def mostrar(self):
    print(f"El nombre es {self.nombre} y la edad {self.edad}")

class MiClase:
    pass

mc = MiClase()

for k,v in clase_diccionario.items():
    if v["tipo"] == "atributo_clase":
        setattr(mc,k,"")
    if v["tipo"] == "atributo_objeto":
        setattr(MiClase,k,"")

dir(mc)
```

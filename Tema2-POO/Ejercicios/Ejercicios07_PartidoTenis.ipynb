# Ejercicios 07
1. Crea una clase externa llamada **PartidoTenis** que reciba los nombres de los jugadores (jugador1 y jugador2) en su constructor. Debe inicializar un atributo self.cancelado = False y una lista vacía para almacenar los sets (self.historial_sets).
2. Al inicializar el partido, este debe generar automáticamente el primer set instanciando la clase interna **_Set** y añadiéndolo a **self.historial_sets.**
3. Al inicializarse el set, debe recibir la referencia al partido (self.partido) y el número de set (por ejemplo, 1). Tendrá atributos para el marcador: **self.juegos_j1 = 0** y **self.juegos_j2 = 0**.
4. La clase interna debe tener un método **registrar_punto(self, jugador_num)**. Antes de sumar nada, debe verificar si el partido  sigue activo (**not self.partido.cancelado**). Si está activo, suma un juego al jugador correspondiente. Si el partido fue cancelado, debe bloquear el marcador.
5. La clase externa tendrá un método **anotar_en_set_actual(self, jugador_num)** que delegue la acción en el último set de la lista, y un método **suspender_por_lluvia(self)** que cambie **self.cancelado** a True.

```python
class PartidoTenis:
    def __init__(self, jugador1, jugador2):
        self.jugador1 = jugador1
        self.jugador2 = jugador2
        self.cancelado = False
        self.historial_sets = []
        
        primer_set = self._Set(self, 1) 
        self.historial_sets.append(primer_set)

    def anotar_en_set_actual(self, jugador_num):
        if self.historial_sets:
            set_actual = self.historial_sets[-1]
            set_actual.registrar_punto(jugador_num)
        else:
            print("[Error] No hay ningún set activo en juego.")

    def suspender_por_lluvia(self):
        print(f"\n[Torneo] Tormenta eléctrica detectada. Partido SUSPENDIDO.")
        self.cancelado = True

    # --- CLASE INTERNA ---
    class _Set:
        def __init__(self, partido_padre, numero_set):
            self.partido = partido_padre
            self.numero_set = numero_set
            self.juegos_j1 = 0
            self.juegos_j2 = 0

        def registrar_punto(self, jugador_num):
            if self.partido and not self.partido.cancelado:
                if jugador_num == 1:
                    self.juegos_j1 += 1
                elif jugador_num == 2:
                    self.juegos_j2 += 1
                
                print(f"[Set {self.numero_set}] Marcador actualizado: "
                      f"{self.partido.jugador1} [{self.juegos_j1}] vs "
                      f"[{self.juegos_j2}] {self.partido.jugador2}")
            else:
                print(f"[Árbitro] Punto denegado en Set {self.numero_set}: El partido no está activo.")

final_open = PartidoTenis("Carlos Alcaraz", "Jannik Sinner")
    
final_open.anotar_en_set_actual(1)
final_open.anotar_en_set_actual(2)
final_open.anotar_en_set_actual(1)
    
final_open.suspender_por_lluvia()
    
final_open.anotar_en_set_actual(1)
```

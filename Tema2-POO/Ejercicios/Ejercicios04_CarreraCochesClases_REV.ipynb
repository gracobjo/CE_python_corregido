# Ejercicios 04
Modificación del ejercicio anterior para que todo se haga con clases, tanto el coche como la carrera.

```python
import random

CONSUMO = 0.00999
TRAYECTO = 1000

class Coche:
    def __init__(self,nombre,potencia,combustible):
        self._nombre = nombre
        self._potencia = potencia
        self._combustible = combustible
        self.tiempo = 0
        self.recorrido = 0

    def acelerar(self):
        if self._combustible > 0:
            acelerado = random.random()*self._potencia
            self._combustible -= acelerado*CONSUMO
            self.recorrido += acelerado
            self.tiempo += 1

    def monitorizar(self):
        print(f"Coche: {self._nombre} - Combustible: {self._combustible} - Tiempo: {self.tiempo} - Recorrido: {self.recorrido}")

class Carrera:
    def __init__(self,participantes):
        self.participantes = participantes

    def registrar_participante(self,coche):
        self.participantes.append(coche)

    def iniciar_carrera(self):
        while(all(coche.recorrido < TRAYECTO for coche in self.participantes)):
            for i in self.participantes:
                i.acelerar()
                i.monitorizar()
            print("=====")


inscritos = []
inscritos.append(Coche("coche1",200,100))
inscritos.append(Coche("coche2",220,100))
inscritos.append(Coche("coche3",240,100))

carrera1 = Carrera(inscritos)

inscritos.append(Coche("coche4",260,100))
carrera1.registrar_participante(Coche("coche5",215,100))


carrera1.iniciar_carrera()
```

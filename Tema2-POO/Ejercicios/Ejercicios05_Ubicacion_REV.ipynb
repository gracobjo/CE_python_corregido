Crear una clase llamada `Ubicacion` que administre la latitud y longitud de un punto en la Tierra. Deberás sobrecargar los operadores necesarios para poder sumar un desplazamiento, comparar si dos ubicaciones son idénticas e imprimir la coordenada de forma entendible.

Requisitos de la clase:
1. __init__(self, latitud, longitud): El constructor debe validar que la latitud esté entre -90 y 90, y la longitud entre -180 y 180.
2. __str__(self): Debe devolver el formato legible: (Latitud° N/S, Longitud° E/O).
3. __add__(self, otro): Debe permitir "sumar" una ubicación y un desplazamiento devolviendo la nueva ubicación.
4. __sub__(self,otro): Debe permitir restar dos ubicaciones devolviendo un desplazamiento en términos (latitud, longitud)
5. __eq__(self, otro): Debe devolver True si ambas coordenadas son iguales (redondear con `round`).

```python
class Ubicacion:
    def __init__(self, latitud: float, longitud: float):
        if not (-90 <= latitud <= 90):
            raise ValueError("La latitud debe estar entre -90 y 90 grados.")
        if not (-180 <= longitud <= 180):
            raise ValueError("La longitud debe estar entre -180 y 180 grados.")
        
        self.latitud = latitud
        self.longitud = longitud

    def __str__(self) -> str:
        ns = "N" if self.latitud >= 0 else "S"
        eo = "E" if self.longitud >= 0 else "O"
        return f"({abs(self.latitud)}° {ns}, {abs(self.longitud)}° {eo})"

    def __add__(self, desplazamiento):
        if isinstance(desplazamiento, tuple) and len(desplazamiento) == 2:
            nueva_lat = self.latitud + desplazamiento[0]
            nueva_lon = self.longitud + desplazamiento[1]
            
            # Sin ajuste para desborde de los límites del mapa
            return Ubicacion(nueva_lat, nueva_lon)
        return NotImplemented

    def __sub__(self,otro):
        if isinstance(otro, Ubicacion):
            return (round(otro.latitud - self.latitud,5), round(otro.longitud - self.longitud,5))
        return NotImplemented
        
    def __eq__(self, otro):
        if isinstance(otro, Ubicacion):
            return (round(self.latitud, 5) == round(otro.latitud, 5) and round(self.longitud, 5) == round(otro.longitud, 5))
        return False


madrid = Ubicacion(40.4167, -3.7037)
desplazamiento_gps = (0.5, 1.0)
punto_b = Ubicacion(40.4167, -3.7037)
valladolid = Ubicacion(41.65541, -4.72353)

print(madrid)  # Llama a __str__

print(f"¿Son iguales las coordenadas?: {madrid == punto_b}")  # Llama a __eq__

nueva_posicion = madrid + desplazamiento_gps  # Llama a __add__
print(f"Nueva ubicación tras el viaje: {nueva_posicion}")

variacion = madrid - valladolid
print(f"Variación de latitud y longitud entre Madrid y Valladolid: {variacion}")
```

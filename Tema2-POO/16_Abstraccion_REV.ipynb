# Abstracción en Python

La abstracción es un pilar de la programación orientada a objetos (POO) que consiste en ocultar los detalles de implementación y mostrar solo la funcionalidad esencial al usuario. En Python, esto se logra principalmente mediante el módulo nativo abc (Abstract Base Classes).

Piensa en un coche: tú sabes usar el volante y los pedales (interfaz), pero no necesitas saber exactamente cómo inyecta combustible el motor (implementación).

Para crear abstracción necesitas dos componentes del módulo abc:
* `ABC`: La clase de la que debes heredar para que tu clase sea abstracta.
* `@abstractmethod`: El decorador que obliga a las clases hijas a implementar un método.

**Regla principal**: No se pueden crear instancias (objetos) directamente de una clase abstracta.

```python
from abc import ABC, abstractmethod

class MetodoPago(ABC):
    
    @abstractmethod
    def autenticar(self):
        """Cada método de pago debe validar al usuario a su manera."""
        pass
        
    @abstractmethod
    def procesar_pago(self, cantidad):
        """Cada método de pago ejecuta la transacción de forma distinta."""
        pass


class TarjetaCredito(MetodoPago):
    def autenticar(self):
        print("Validando el número de tarjeta y el CVV...")
        
    def procesar_pago(self, cantidad):
        print(f"Cobrando {cantidad}€ a través de la tarjeta bancaria.")


class Bizum(MetodoPago):
    def autenticar(self):
        print("Redirigiendo a Bizum para para comprobar teléfono...")
        
    def procesar_pago(self, cantidad):
        print(f"Transfiriendo {cantidad}€ desde por Bizum.")

# Por si queremos implementar también polimorfismo
# def realizar_compra(sistema_pago, total):
#    sistema_pago.autenticar()
#    sistema_pago.procesar_pago(total)

# Probando con Tarjeta de Crédito
pago_visa = TarjetaCredito()
pago_visa.autenticar()
pago_visa.procesar_pago(150)
realizar_compra(pago_visa, 150)

print("=" * 30)

# Probando con Bizum
pago_bizum = Bizum()
pago_bizum.autenticar()
pago_bizum.procesar_pago(85)
```

# Paso por referencia

Los objetos se pasan a las funciones por **referencia**, es decir, los cambios afectan al mismo objeto

```python
class CuentaBancaria:
    def __init__(self, saldo):
        self._saldo = saldo

    @property
    def saldo(self):
        return self._saldo

    # Setter con @nombre.setter
    @saldo.setter
    def saldo(self, nuevo_saldo):
        self._saldo = nuevo_saldo
        
def incentivar(cuenta):
    if cuenta._saldo > 1000:
        cuenta._saldo += 100

c1 = CuentaBancaria(600)
c2 = CuentaBancaria(1200)

incentivar(c1)
incentivar(c2)

print(c1.saldo)
print(c2.saldo)
```

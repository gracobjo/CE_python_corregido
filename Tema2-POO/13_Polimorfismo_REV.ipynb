# Polimorfismo
La palabra proviene del griego poly (muchos) y morphos (formas). En programación, significa que un mismo mensaje o llamada de método puede ser utilizado en diferentes tipos de objetos, y cada uno ejecutará su propia versión o lógica interna.

Ventajas:
* **Flexibilidad y extensibilidad**: Puedes añadir nuevas clases (por ejemplo, una clase Avion o Pajaro) sin modificar el código que ya procesa los objetos.
* **Reutilización de código**: Evita la saturación de condicionales if/elif/else para comprobar el tipo exacto de cada objeto (isinstance).
* **Código limpio**: Las funciones interactúan con una interfaz común en lugar de detalles de implementación específicos.

Tipos de polimorfismo:
* Polimorfismo con herencia
* Tipado dinámico
* Polimorfismo con funciones incorporadas y operadores

___
## Polimorfismo con herencia (Clases base y derivadas)

Ocurre cuando una superclase define un método general, y las subclases lo sobrescriben (override) para adaptarlo a su propia naturaleza.

````python
class Animal:
    def __init__(self, nombre):
        self.nombre = nombre

    def comer(self):
        return f"{self.nombre} está comiendo."
        
class Gato(Animal):
    def __init__(self, nombre):
        super().__init__(nombre)

    def comer(self):
        return f"{self.nombre} come pescado."

gato1 = Gato("Minino")
gato1.comer()
````

___
## Polimorfismo con Tipado dinámico
Si dos objetos tienen un método con igual nombre y firma los podemos usar indistintamente.

````python
class Auto:
    def mover(self):
        print("El auto se desplaza sobre 4 ruedas")

class Barco:
    def mover(self):
        print("El barco navega sobre el agua")

def iniciar_movimiento(transporte):
    transporte.mover()  # No importa si es Auto o Barco, mientras tenga .mover()

# El método iniciar_movimiento puede ser usado por diferentes objetos y en consecuencia realizar tareas distintas
iniciar_movimiento(Auto())
iniciar_movimiento(Barco())
````

### Con funciones integradas
La función `len()` es una función integrada en Python y es un ejemplo claro de polimorfismo porque realiza una acción diferente (cuenta elementos distintos) según el tipo de dato que recibe como argumento, adaptándose a través del método especial `__len__()` que cada objeto tiene implementado.

Cómo funciona internamente:

Al escribir `len(objeto)`, Python no calcula nada por sí mismo. Lo único que hace es llamar de forma interna al método especial `__len__()` de ese objeto:

```Python
len(objeto) ---> objeto.__len__()
```
Por lo tanto, el polimorfismo ocurre porque cada clase decide cómo responder a esa llamada:
* En un texto (str), cuenta el número de caracteres.
* En una lista o tupla, cuenta la cantidad de elementos.
* En un diccionario, cuenta cuántas claves (pares clave-valor) contiene.
* En una clase personalizada, se puede programar qué significa "su longitud" definiendo el método __len__().

````python
# Ejemplo con la función len()
print(len("Hola mundo"))  # Devuelve la cantidad de caracteres: 10
print(len([10, 20, 30, 40]))  # Devuelve la cantidad de elementos: 4
print(len({"a": 1, "b": 2}))  # Devuelve el número de claves: 2

# Cómo funciona internamente (método especial __len__)
# Python llama a objeto.__len__() de forma transparente para cada tipo.
````

### Usado con una clase personalizada

````python
import random

class Coche:
    def __init__(self,marca,modelo):
        self.marca = marca
        self.modelo = modelo

    def __len__(self):
        return random.randint(1,10)

    def __str__(self):
        return f"La marca es {self.marca} y el modelo es {self.modelo}" 

coche1 = Coche("seat","ibiza")

print(len(coche1))

print("=====")

print(str(coche1))

print("=====")

print(coche1) # Imprimes el coche directamente
````

## Operadores
El operador `+` con números suma (2 + 3 = 5) y con textos concatena ("Hola " + "mundo" = "Hola mundo") por poner dos ejemplo básicos.

Se denomina **sobrecarga de operadores**

````python
# Operador "+" con números (suma aritmética)
resultado_num = 5 + 3
print(resultado_num)  # Salida: 8

# Operador "+" con cadenas (concatenación)
resultado_str = "Hola, " + "¿cómo estás?"
print(resultado_str)  # Salida: Hola, ¿cómo estás?

# Operador "+" con listas (concatenación de listas)
resultado_list = [1, 2] + [3, 4]
print(resultado_list)  # Salida: [1, 2, 3, 4]
````

___
### Usado con una clase personalizada

````python
class Persona:
    def __init__(self,nombre,edad):
        self.nombre = nombre
        self.edad = edad

    def __add__(self,otro):
        return self.edad + otro.edad

    def __lt__(self,otro):
        return True if self.edad < otro.edad else False


p1 = Persona("Luis",60)
p2 = Persona("Ana",30)

print(p1 + p2)
print(p1 < p2)
````

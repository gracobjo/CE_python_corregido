# Generadores
 Un generador (**función generadora**) es una función especial que devuelve un iterador.
Un generador es un tipo especial de iterador que permite pausar la ejecución de la función, guardando su estado (valores de variables) y reanudarla más tarde, lo que permite generar valores sobre la marcha en lugar de calcularlos todos a la vez y almacenarlos en memoria. Esto es útil para trabajar con grandes conjuntos de datos o secuencias infinitas.

Si una función contiene la palabra clave **yield**, se convierte en una función generadora. Es como un **return** pero pausa la ejecución en ese punto y la siguiente vez que se llame continuará donde quedó.

Cuando se ejecuta se devuelve un objeto generador, que es un iterador que se puede recorrer utilizando la función next() o un bucle for.

````python
def migenerador():
    yield 1
    yield 2
    yield 3

gen = migenerador()

print(next(gen))
print(next(gen))
print(next(gen))
# print(next(gen)) # ERROR --> Si se ejecuta otra vez, se lanzará una excepción StopIteration.
````

___
### Tipos de los objetos generadores

````python
print(migenerador()) # Devuelve un objeto generador
print(type(migenerador())) # Devuelve <class 'generator'>, indicando que es un objeto generador
````

___
### Los generadores son iterables, lo que significa que se pueden recorrer utilizando bucles for.

````python
for i in migenerador():
    print(i)  # Imprime los valores generados por la función generadora
````

___
### EJERCICIO: Crear un contador de dos en dos con un generador que acabe en 20

````python
def contador():
    contador = 0
    while contador <= 20:
        yield contador
        contador += 2

#Los tres primeros los imprimimos directamente pero serán 0 todos porque en cada print estamos generando un generador nuevo
print(next(contador()))
print(next(contador()))
print(next(contador()))

print("=========")

# Para evitar eso asignamos a un generador único
con = contador()
print(next(con))
print(next(con))
print(next(con))

print("=========")

#El resto los imprimimos con un for puesto que 
for i in contador():
    print(i)
````

___
## Se pueden crear generadores sin crear una función
Para ello se usan expresiones creadoras entre paréntesis al igual con la misma sintaxis que las **list comprehension**.

### Sintaxis
```python
generador = (expresion for elemento in iterable)
```

````python
generador = (x*2 for x in range(1,10))

print(type(generador))

print(next(generador))
print(next(generador))
print(next(generador))
print(next(generador))

for i in generador:
    print(i)
````

___
### Uso directo del generador

````python
print(next((x*2 for x in range(1,10))))
print(next((x*2 for x in range(1,10))))
````

___
### Podemos aplicar funciones directamente a generadores y aplicarse al conjunto de todos los elementos

````python
numeros = (x for x in range(5)) # 0, 1, 2, 3, 4

print(type(numeros))

print(sum(numeros))
print(sum(numeros)) # Da cero porque ya se ha generado toda la secuencia

print("======")

listaNumeros = [0,1,2,3,4]
print(sum(listaNumeros))
print(sum(listaNumeros))
````

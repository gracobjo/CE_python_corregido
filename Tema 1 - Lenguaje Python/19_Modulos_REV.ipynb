# Módulos en Python: Uso de `import`

Un módulo en Python es un archivo con extensión `.py` que contiene código reutilizable, como funciones, clases y variables.
#### ¿Para qué sirve?
* **Organizar el código:** Permite dividir programas largos en archivos más pequeños y ordenados.
* **Reutilizar código:** Evita repetir las mismas líneas en diferentes proyectos.
* **Facilitar el mantenimiento:** Ayuda a encontrar y corregir errores con mayor rapidez.
#### Tipos de módulos
* **Módulos propios:** Archivos creados por uno mismo dentro de un proyecto (por ejemplo, utilidades.py).
* **Módulos estándar:** Vienen instalados por defecto con Python (como `math` para operaciones matemáticas o `random` para números aleatorios).
* **Módulos de terceros:** Creados por la comunidad y que puedes instalar usando gestores como pip (por ejemplo, `pandas`).

#### Cómo se utiliza
Para usar un módulo en tu programa, debes importarlo con la palabra clave `import` y escribiendo el nombre del módulo, que en los personalizados será el nombre del archivo sin la extensión `.py`.

___
## Uso simple de `import`
```Python
import modulo
```
Importado de esta forma, para usarlo hay que utilizar el prefijo antes de los objetos del módulo y un punto.

```Python
modulo.objeto
```

````python
import modulopersonalizado

print(modulopersonalizado.VELOCIDAD_MAX)
print(modulopersonalizado.sumar(5,3))
print(modulopersonalizado.restar(5,3))
c1 = modulopersonalizado.Coche("Toyota","Corolla")
print(c1.ver_coche())
````

___
## Uso con un alias: `as` 
```Python
import modulo as alias
```
Importado de esta forma, para usarlo hay que utilizar el alias en lugar del nombre del módulo

```Python
modulo.objeto
```

````python
import modulopersonalizado as mp

# print(modulopersonalizado.VELOCIDAD_MAX) # ERROR: Se ha importado con as y debemos usar el alias
# print(VELOCIDAD_MAX) # ERROR: Se ha importado con as y debemos usar el alias

print(mp.VELOCIDAD_MAX)
print(mp.sumar(5,3))
print(mp.restar(5,3))
c1 = mp.Coche("Toyota","Corolla")
print(c1.ver_coche())
````

___
## Importar sólo determinados objetos del módulo: `from` 
```Python
from modulo import objeto
```
Importado de esta forma, para usarlo hay que escribirlo directamente

```Python
modulo.objeto
```

````python
from modulopersonalizado import sumar

# Al hacerlo de esta forma no se usa el prefijo antes del objeto a emplear
#print(VELOCIDAD_MAX) # ERROR: No se ha importado la constante

print(sumar(5,3))
# print(modulopersonalizado.sumar(5,3)) # ERROR: No se escribe el nombre del módulo

#print(restar(5,3)) # ERROR: No se ha importado la función
#c1 = Coche("Toyota","Corolla") # ERROR: No se ha importado la clase
#print(c1.ver_coche())
````

## Uso de `from` y `as`
```Python
from modulo import objeto as alias
```
Importado de esta forma, para usarlo podemos:
* Se usa el alias únicamente
* Se usa el nombre original únicamente

```Python
modulo.objeto
```

````python
from modulopersonalizado import sumar as sm

# Al hacerlo de esta forma no se usa el prefijo antes del objeto a emplear
#print(VELOCIDAD_MAX) # ERROR: No se ha importado la constante

print(sm(5,3))
# print(modulopersonalizado.sm(5,3)) # ERROR: No poner el nombre del módulo

print(sumar(5,3))
# print(modulopersonalizado.sumar(5,3)) # ERROR: No poner el nombre del módulo

#print(restar(5,3)) # ERROR: No se ha importado la función
#c1 = Coche("Toyota","Corolla") # ERROR: No se ha importado la clase
#print(c1.ver_coche())
````

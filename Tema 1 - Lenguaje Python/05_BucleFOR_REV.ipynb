# Bucle FOR
El bucle **for** permite ejecutar un bloque de código un número determinado de veces.

Se usa principalmente para iterar a través de una secuencia o conjunto.

___
## Sintaxis
```python
for elemento in conjunto
    sentencia1
    sentencia2
    ...
    sentenciaN
```
La variable _elemento_ tomará el valor de cada uno de los elementos del conjunto o secuencia en cada una de las iteraciones.

````python
for i in [1,2,3]:
    print(i)

for i in "Python":
    print(i,end="")
````

___
## Sentencia _break_
Permite romper el bucle en un determinado momento. Una vez se ejecuta salimos del bucle.



````python
for i in [1,2,3]:
    print(i)
    if i == 2: break
````

___
## Sentencia _continue_
Permite detener la iteración actual y continuar con la siguiente.


````python
for i in [1,2,3,4,5]:
    if i == 4: continue
    print(i)
````

## Sentencia _else_ en el bucle _for_
Permite ejecutar un código una vez ha finalizado finalice salvo que se haya ejecutado una sentencia _break_.


````python
for i in [1,2,3]:
    if i == 9: break
    print(i)
else:
    print("Ejecutado al completo")
        
````

___
## Bucle con *range*
Con range hacemos un bucle _for_ muy sencillo controlando el número de iteraciones.

````python
for i in range(10):
    print(i)
````

___
## Podemos iterar sobre una cadena
Al iterar sobre una cadena, los caracteres son los elementos de la cadena por lo que en cada iteración el valor con el que trabajamos será el del carácter correspondiente de dicha cadena.

````python
for i in "Hola mundo":
    print(i)
````

___
## Bucles _for_ anidados

````python
## Bucles _for_ anidados
for i in range(1,10):
    print(f"Tabla de multiplicar del {i}")
    for j in range(1,10):
        print(f"{i*j:>4}",end="   ")
    print("\n")
````

___
Lo podemos usar para iterar sobre listas, tuplas y conjuntos

````python
print("LISTA")
for i in [1,2,3]:
    print(i)

print("TUPLA")
for i in (1,2,3):
    print(i)

print("CONJUNTO")
for i in {1,2,3}:
    print(i)
````

___
En el caso de diccionarios, el contenido de la variable será la _clave_.

Para acceder al _valor_ se indica con corchetes sobre el nombre del diccionario y dentro el contenido de la variable.

````python
persona = {"nombre":"juan","edad":20}
for i in persona:
    print(i,persona[i],sep=" --- ")
````

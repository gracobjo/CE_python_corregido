# Diccionarios
Almacenan múltiples elementos en forma de pares **clave-valor**
* Se crean con corchetes {}
* Los pares clave-valor se separan por comas **,**
* La clave se separa del valor por dos puntos **:**
* Cada clave debe ser única
* El valor puede ser cualquier tipo de dato
* Sus elementos se consideran ordenados
* Son modificables: Podemos añadir, elminar, cambiar los elementos de un diccionario.
* Las claves no se pueden repetir

**Corrección:** Las claves de un diccionario no tienen que ser siempre cadenas: pueden ser de cualquier tipo *hashable* (por ejemplo, cadenas, números o tuplas cuyos elementos sean hashables). Lo que debe ser único son las claves, no los valores; se pueden repetir valores y estos pueden ser de cualquier tipo.

````python
persona = {
    "nombre": "Ana",
    "edad": 30,
    "ciudad": "Madrid",
    "es_estudiante": False,
    "hobbies": ["leer", "viajar", "cocinar"]
}
print(persona)
````

___
### El tipo de dato es _dict_

````python
print(type(persona))
````

___
### Acceso a los elementos del diccionario
Se accede entre corchetes y poniendo la cadena correspondiente a la clave tanto si queremos leer un valor como modificarlo

````python
# Acceso para leer
print(persona["nombre"])
print(persona["edad"])

# Acceso para escribir
persona["ciudad"] = "Valladolid"
print(persona)
````

___
### No se permiten elementos duplicados
Se queda con el último valor especificado de esa clave

````python
coche = {
    "marca":"seat",
    "modelo":"ibiza",
    "color":"negro",
    "color":"azul"
}    
print(coche)
````

___
### La longitud la consultamos con la funcion _len()_
Hace referencia al número de pares _clave-valor_

````python
persona = {
    "nombre": "Ana",
    "edad": 30,
    "ciudad": "Madrid",
    "es_estudiante": False,
    "hobbies": ["leer", "viajar", "cocinar"]
}

print(len(persona))
````

___
### Podemos crear un diccionario con su constructor _dict()_

````python
estanteria = dict(material="madera",alto=200,ancho=100,fondo=30)
print(estanteria)
````

___
### Recorrer un diccionario

````python
persona = {
    "nombre": "Ana",
    "edad": 30,
    "ciudad": "Madrid",
    "es_estudiante": False,
    "hobbies": ["leer", "viajar", "cocinar"]
}

for i in persona:
    print(i)
    
print("======")

for i in persona:
    print(persona[i])

print("======")

# Recorrer un bucle for con 2 valores por iteración, la clave y el valor
for i,j in persona.items():
    print(i,j,sep="--")
````

___
### Añadir un elemento a un diccionario
Se accede con un nombre de clave nuevo para ser escrita.

````python
coche = {
    "marca":"seat",
    "modelo":"ibiza",
    "color":"negro",
}

coche["precio"] = 23000

print(coche)

# print(coche["combustible"]) # ERROR: Si lees una clave que no exite da error
````

___
### Eliminar un elemento
Se accede a una clave y se usa **del** como con cualquier variable

````python
coche = {
    "marca":"seat",
    "modelo":"ibiza",
    "color":"negro",
}

del coche["modelo"]
print(coche)

del coche # Elimina todo el diccionario
````

___
### Comprobar si una clave existe
Se emplea el operador **in**

````python
print("Existe" if "modelo" in coche else "No existe")
print("Existe" if "cilindrada" in coche else "No existe")
````

___
### Método _get(clave)_
Se obtiene para obtener un valor a partir de una clave.

````python
persona = {
    "nombre": "Ana",
    "edad": 30,
    "ciudad": "Madrid",
    "es_estudiante": False,
    "hobbies": ["leer", "viajar", "cocinar"]
}

print(persona.get("nombre"))
````

___
### Método _keys()_
Para obtener un listado de las claves del diccionario

Se obtiene una clase **dict_keys** que puede iterarse

````python
print(persona.keys())
print(type(persona.keys()))
for i in persona.keys():
    print(i)
````

___
### Método _values()_
Para obtener un listado de los valores del diccionario

Se obtiene una clase **dict_values** que puede iterarse

````python
print(persona.values())
print(type(persona.values()))
for i in persona.values():
    print(i)
````

___
### Método _items()_
Para obtener una lista de tuplas de dos elementos, que son cada clave y cada valor

Se obtiene una clase **dict_items** que puede iterarse

````python
print(persona.items())
print(type(persona.items()))
for i in persona.items():
    print(i)
````

___
### Método _update(elementos)_
Inserta un nuevo elemento (par clave-valor) en el diccionario.

Los elementos a insertar pueden ser un diccionario o un iterable con claves-valores.

````python
persona = {
    "nombre": "Ana",
    "edad": 30,
    "ciudad": "Madrid",
    "es_estudiante": False,
    "hobbies": ["leer", "viajar", "cocinar"]
}

persona.update({"pais":"España","telefono":55345315})
persona.update({"edad":35})
persona.update([("apellido1","perez"),("apellido2","garcia")])
print(persona)
````

___
### Método _pop(elemento)_
Elimina el elemento con la clave indicada devolviendo el valor de la clave eliminada

````python
coche = {
    "marca":"seat",
    "modelo":"ibiza",
    "color":"negro",
}

print(coche.pop("marca"))
print(coche)
````

___
### Método _popitem()_
Elimina el último elemento insertado devolviendo el par que se elimina

````python
coche = {
    "marca":"seat",
    "modelo":"ibiza",
    "color":"negro",
}

print(coche.popitem())
print(coche)
````

___
### Método _clear()_
Vacía el diccionario

````python
coche = {
    "marca":"seat",
    "modelo":"ibiza",
    "color":"negro",
}

print(coche.clear())
print(coche)
````

___
### Método _copy()_
Copia diccionarios.

No se puede hacer **dict2 = dict1**

````python
dic1 = {"a":1,"b":2}

dic2 = dic1.copy()
print(dic2)
````

___
### Método _fromkeys(claves,valores)_
Crea un diccionario a partir de un conjunto de claves y un único valor.

Funciona como un método de clase, es decir, no se llama desde un diccionario, sino desde la clase _dict_

````python
claves = ("nombre1","nombre2","nombre3")
valor = "pepe"

nuevo_diccionario = dict.fromkeys(claves,valor)
print(nuevo_diccionario)
````

___
### Método _setdefault(clave,valor)_
Devuelve el valor del elemento con la clave especificada.

Si la clave no existe lo inserta

````python
coche = {
    "marca":"seat",
    "modelo":"ibiza",
    "color":"negro",
}

print(coche.setdefault("modelo","ibiza"))
print(coche.setdefault("modelo","cordoba"))
coche.setdefault("combustible","diesel")
print(coche)
````

## Dictionary comprehension
Funciona parecido a las listas y con la misma sintaxis salvo:
* Usamos llaves
* Introducimos clave y valor separados por **:** antes del for

```python
diccionario = { clave:valor for elemento in conjunto if condición_verdadera }
```

````python
lista = ["coche","casa","instituto"]

diccionario = { x:len(x) for x in lista }

print(diccionario)
print(type(diccionario))
````

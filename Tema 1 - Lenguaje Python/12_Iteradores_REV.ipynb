# Iteradores
## Concepto
Un objeto iterable es aquel que puede ser recorrido elemento por elemento, como una lista, una tupla, un diccionario, un conjunto o un string.

Es un objeto sobre el que se puede iterar.

```python
listaMuebles = [ "mesa" , "silla" , "armario"] # listaMuebles es una lista, es decir, es iterable.
print(listaMuebles)
print(type(listaMuebles))  # <class 'list'>, es un objeto iterable.
```

___
### Un bucle for recorre un objeto iterable, pero no es un iterador en sí mismo.

```python
listaMuebles = [ "mesa" , "silla" , "armario"]

for mueble in listaMuebles:
    print(mueble)
```

___
### Un iterador es un objeto que se usa para iterar sobre un iterable.
Un iterador es un objeto que implementa los métodos __iter__() y __next__().

Creación de un iterador a partir de la lista emplando __iter__()

```python
listaMuebles = [ "mesa" , "silla" , "armario"]

iteradorMuebles = iter(listaMuebles)

print(next(iteradorMuebles))
print(next(iteradorMuebles))
print(next(iteradorMuebles))

# print(next(listaMuebles))   #ERROR --> No es un iterador, no tiene el método __next__(). Es iterable, pero no es un iterador.
```

___
### Todo iterador es iterable, pero no todo iterable es un iterador.

```python
listaMuebles = [ "mesa" , "silla" , "armario"]

iteradorMuebles = iter(listaMuebles)

for mueble in iteradorMuebles:
    print(mueble)
```

___
### Si no hay más elemento y ejecutamos next se genera una excepción StopIteration

```python
listaMuebles = [ "mesa" , "silla" , "armario"]

iteradorMuebles = iter(listaMuebles)

try:
    print(next(iteradorMuebles))
    print(next(iteradorMuebles))
    print(next(iteradorMuebles))
    print(next(iteradorMuebles))
except StopIteration:
    print("No hay más elementos en el iterador.")
```

```python
listaMuebles = [ "mesa" , "silla" , "armario"]

iteradorMuebles = iter(listaMuebles)

for mueble in iteradorMuebles:
    print(mueble)

try:
    print(next(iteradorMuebles))
except StopIteration:
    print("No hay más elementos en el iterador.")
```

## Los tipos de datos iterables en python son:
* Listas
* Tuplas
* Conjuntos
* Diccionarios
* Rangos
* Cadenas de texto

```python
texto = "Estamos en clase de Python"

for i in texto:
    print(i,end="")

# print(next(texto)) #ERROR: str object is not an iterator. Efectivamente, es iterable pero no iterador

print("\n");

iteradorTexto = iter(texto)
print(next(iteradorTexto))
print(next(iteradorTexto))
print(next(iteradorTexto))
```

___
### Comprobar si un objeto es iterable o un iterador
* Si es iterable puede ejecutar el método **iter**()
* Si es iterador puede ejecutar el método **next**()
En ambos caso se lanza una excepción de tipo TypeError

```python
def es_iterable(ob):
  try:
      iter(ob)
      return True
  except TypeError:
      return False


def es_iterador(ob):
  try:
      next(ob)
      return True
  except TypeError:
      return False

#Ejemplo 1: Una lista
listaMuebles = [ "mesa" , "silla" , "armario"]
print("Ejemlo 1: Una lista")
print(es_iterable(listaMuebles))  # True, es iterable
print(es_iterador(listaMuebles))  # False, no es un iterador
print(es_iterador(iter(listaMuebles)))  # True, es un iterador

#Ejemplo 2: Un entero
numero = 6
print("Ejemlo 2: Un entero")
print(es_iterable(numero))  # False, no es iterable
print(es_iterador(numero))  # False, no es un iterador

#Ejemlo 3: Una cadena
cadena = "cadena de texto"
print("Ejemlo 3: Una cadena")
print(es_iterable(cadena))  # True, es iterable
print(es_iterador(cadena))  # False, no es un iterador
# next(cadena)  # ERROR, no es un iterador, no tiene el método __next__()
print(es_iterador(iter(cadena)))  # True, es un iterador

#Ejemplo 4: Un conjunto
conjuntoMuebles = { "mesa" , "silla" , "armario"}
print("Ejemlo 4: Un conjunto")
print(es_iterable(conjuntoMuebles))  # True, es iterable
print(es_iterador(conjuntoMuebles))  # False, no es un iterador
print(es_iterador(iter(conjuntoMuebles)))  # True, es un iterador

#Ejemplo 5: Un rango
miRango = range(5)
print("Ejemlo 5: Un rango")
print(es_iterable(miRango))  # True, es iterable
print(es_iterador(miRango))  # False, no es un iterador
print(es_iterador(iter(miRango)))  # True, es un iterador
```

___
### Creación de un iterador genérico en Python

```python
def mi_iterador(iterable):
    # Creamos un iterador a partir del iterable
    iterador = iter(iterable)
    
    while True:
        try:
            # Obtenemos el siguiente elemento del iterador
            elemento = next(iterador)
            yield elemento  # Devolvemos el elemento actual
        except StopIteration:
            break  # Salimos del bucle cuando no hay más elementos

listaMuebles = [ "mesa" , "silla" , "armario"]

# En realidad estamos reescribiendo la función iter() y next() para crear un iterador genérico que funcione con cualquier iterable.
iteradorMuebles = mi_iterador(listaMuebles)

print(next(iteradorMuebles))
print(next(iteradorMuebles))
print(next(iteradorMuebles))
```

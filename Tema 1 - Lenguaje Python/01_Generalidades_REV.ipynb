# Introducción a Python
## 1. Sintaxis básica
En Python una sentencia termina por lo general cuando la línea que la contiene finaliza. No se necesita un punto y como como en otros lenguajes de programación.

````python
print("Hola mundo")
````

En Python, el punto y coma (;) no es obligatorio al final de las líneas, pero sirve para separar dos instrucciones distintas en una misma línea.
¿Cómo funciona?
* __Fin de línea normal:__ Python usa los saltos de línea para saber dónde termina cada instrucción. No necesitas escribir nada especial.
* __En la misma línea:__ Si quieres escribir dos comandos seguidos en el mismo renglón, los separas con un punto y coma.

````python
a = 5; b = 10; print(a+b)
````

### Buenas prácticas
* Evitar usarlo: Los expertos en Python recomiendan no usar el punto y coma. Poner cada instrucción en su propia línea mejora la lectura del código.
* No es como C o Java: En esos lenguajes el punto y coma es obligatorio para cerrar cada orden. En Python, olvidarlo o usarlo en exceso rompe el estilo visual estándar (PEP 8).

### Indentación
La indentación en Python son los espacios al principio de una línea de código que sirven para agrupar instrucciones y definir qué partes del programa pertenecen a una estructura.

A diferencia de otros lenguajes de programación que usan llaves {} para marcar el inicio y el final de un bloque, Python utiliza la sangría (espacios en blanco) de manera obligatoria:
* __Define la estructura:__ Indica qué líneas se ejecutan dentro de una condición, un bucle o una función
* __Evita errores:__ Si dejas mal los espacios, Python mostrará un IndentationError y el programa no funcionará.


````python
edad = 18
if edad >= 18:
    print("Eres mayor de edad")  # Esta línea pertenece al if
print("Esto está fuera del if y se ejecuta siemre")  # Se ejecuta siempre
````

### Reglas principales
* __Cantidad de espacios:__ La regla estándar recomienda usar cuatro espacios por cada nivel de indentación.
* __Consistencia:__ Todas las líneas de un mismo bloque deben tener exactamente la misma cantidad de espacios. En distintos bloques es posible tener distinta cantidad de espacios.

````python
edad = 20
if edad >= 18:
    print("Eres mayor de edad")  # Esta línea pertenece al if
    print("Eres mayor de edad")  # Esta línea de un error
````

## 3. Salida por pantalla y entrada por teclado
### print
Se usa el método integrado __print__ para sacar información por pantalla.

````python
print("hola") # Recibe una cadena de texto principalmente.

print(5) # También puede recibir otros datos: numéricos, etc.

print(2 + 3) # Puede recibir expresiones.

print("Hola","Adios") # También puede recibir varios parámetros para imprimir en la misma sentencia.

# Se puede combinar con variables.
nombre = "Ana"
edad = 25
print("Me llamo", nombre, "y tengo", edad, "años.")

print(f"Me llamo {nombre} y tengo {edad} años.") #Se puede usar con cadenas f (f-string) que se verán más adelante.
````

También puede recibir varios parámetros para imprimir en la misma sentencia.

Parámetros especiales:
* __end:__ define qué carácter o texto se imprime al final de la salida, siendo el salto de línea el que se usa por defecto.
* __sep:__ define que carácter o texto se coloca entre los diferentes elementos que se van a imprimir, es decir, el elemento separador. Por defecto es el espacio en blanco.

````python
print("Hola", end=" ")
print("Mundo")
# Salida: Hola Mundo
````

````python
print("rojo","verde","azul", sep="-")
# Separar con un guion: rojo,verde,azul

print("rojo","verde","azul", sep="\n")
# Separar con un salto de línea
# rojo
# verde
# azul

print("rojo","verde","azul", sep="")
# Juntar sin separar usando la cadena vacía: rojoverdeazul
````

## 4. Variables
Una variable es un espacio en la memoria del ordenador que permite almacenar un valor. Principales características:
* Permite guardar cualquier tipo de datos.
* Se identifica con un nombre.
* Su contenido puede cambiar cuantas veces sea necesario.

Partes de una variable:
* __Nombre o identificador:__ Es la palabra para referirse a la misma.
* __Tipo de datos que almacena:__ El tipo de información que guarda (número, textos, booleanos…)
* __Valor:__ El contenido de la misma.

En Python no hay una sentencia o comando para declarar una variable.

En Python una variable se crea en el momento en que se le asigna un valor.

En Python no se necesita declararlas con un tipo particular.

````python
edad = 25
edad = 40
print(edad)
# Variable llamada edad que almacena un número cuyo valor es 25
````

### Sintaxis de nombramiento
* Debe empezar por letra, o guion bajo. Puede contener letras, números y guiones bajos.
* No puede comenzar con un número ni contener espacios o caracteres especiales (@,-,$...)
* Se distingue entre mayúsculas y minúsculas.
* No puede usar palabras reservadas del lenguaje (**class**, **for**, **import**…)

### Tipos de datos
* Toda variable llevará asociada un tipo de dato que contiene.
* Los básicos con __int__, __float__, __str__, __bool__, etc, pero hay unos cuantos más.
* Se puede verificar el tipo de una variable usanto la función __type()__

### Estilo de nombrado
* __snake_case:__ Se separan las palabras de las variables por guiones bajos y se escribe todo en minúsculas.

        mi_variable_principal


* __camelCase:__ No separan las palabras de las variables por guiones bajos y se escribe todo en minúsculas salvo la primera letra de cada palabra a partir de la segunda palabra.

        miVariablePrincipal


* __PascalCase:__ Es como camelCase para también en mayúsculas la primera letra de la primera palabra.

        MiVariablePrincipal


* __Screaming Snake Case:__ Todo en mayúsculas y se separan las palabras de las variables con guiones bajos.

        MAX_VALOR


Recomendaciones para Python:
* __Variables__ y funciones: snake_case
* __Clases:__ PascalCase
* __Constantes:__ Screaming Snake Case (recordamos que las constantes no existen como tal)

### Asignación de variables
La asignación puede ser de varios tipos:
* **Asignación simple:** Es la asignación en la que una variable recibe un valor.
* **Asignación múltiple:** Cada variable recibe un valor en función del orden de las variables y de los valores que se van a asignar.
* **Asignación múltiple de valor único:** Cada variable recibe un valor en función del orden de las variables y de los valores que se van a asignar.

````python
# Simple
x = 10

# Múltiple
nombre, apellido, edad = "Juan", "García", 30

# Múltiple de valor único
ancho = alto = 50
````

### Ámbito de las variables
El lugar donde se crea una variable determina dónde puedes usarla.
* **Ámbito local:** Se crean dentro de una función y sólo existen dentro de ella.
* **Ámbito global:** Se crean en el cuerpo principal del código y se pueden usar en cualquier parte.

````python
# Ejemplo de ámbito local
def mi_funcion():
    local_var = "Solo existo aquí dentro"
    print(local_var)

mi_funcion()
# print(local_var)  # Esto daría un ERROR fuera de la función.
````

````python
# Ejemplo de ámbito global
global_var = "Existo en todo el archivo"

def otra_funcion():
    print(global_var)  # Funciona perfectamente

otra_funcion()
print(global_var)
````

#### list
Una lista es una estructura de datos:
* Se crean con corchetes y los elementos separados por comas.
* Está ordenada.
* Es mutable.
* Permite almacenar elementos de cualquier tipo.
* Se pueden añadir y eliminar elementos.

````python
lista1 = [10, "Hola", True, 3.14]
lista2 = list([1, 2, 3, 4, 5]) # A partir de una lista
lista3 = list((1, 2, 3, 4, 5)) # A partir de una tupla
lista4 = list({1, 2, 3, 4, 5}) # a partir de un conjunto
lista5 = list("Hola")
lista5[2] # Acceso al tercer elemento de la lista
lista5[2] = "L"
print(lista5)
````

#### set

* No están ordenadas (no se accede con un índice).
* No se pueden cambiar los elementos aunque se pueden añadir o eliminar.
* No permite duplicados aunque la asignación lo permita.
* No permite cualquier tipo de datos ya que solo permite básicos e inmutables.

````python
conjunto1 = {1, 2, 3, 4, 5}
conjunto2 = set([1, 2, 3, 4, 5])
conjunto3 = set((1, 2, 3, 4, 5))
conjunto4 = set({1, 2, 3, 4, 5})

conjunto5 = {1, 2, 3, 1, 2} #Sí permite asignar pero no duplica elementos
conjunto5 # {1, 2, 3}

tupla1 = (10, "Hola", True, 3.14)
conjunto4 = set(tupla1)
````

* **Mutables:** Permiten ser modificados una vez creados.

El método **id()** permite ver el identificador de cualquier objeto por lo que nos sirve para diferenciar entre tipos mutables o inmutables.

````python
# Creamos una lista
mi_lista = [1, 2, 3]
print("Lista original:", mi_lista)
print("ID original   :", id(mi_lista))

# Modificamos la lista (añadimos un elemento)
mi_lista.append(4)
print("\nLista modificada:", mi_lista)
print("ID modificado   :", id(mi_lista)) 
# ¡El ID sigue siendo exactamente el mismo! por lo que la lista es mutable porque puede cambiarse y seguir siendo el mismo objeto.

print("=====================")

# Tenemos un entero
x = 5
print("Entero original   :", x)
print("ID original   :", id(x))

# Modificamos el entero
x = 6
print("Entero modificado   :", x)
print("ID original   :", id(x))
# ¡El ID sigue ha cambiado! por lo que crea uno nuevo al modificar su valor. El entero es inmutable porque si se cambia crea un objeto nuevo.
````

Esta técnica se realiza empleando las funciones constructor. Algunas funciones son:

## 8. Booleanos
Un booleano representa uno de los siguientes valores:
* **True** (verdadero)
* **False** (falso)

Podemos evaluar cualquier expresión para obtener cualquiera de estos dos valores.

````python
print(10 > 9) # True
print(10 == 9) # False
print(10 < 9) # False
````

Estas expresiones son típicas de estructuras de control como **if-else**.

Podemos usar el constructor **bool()** para evaluar una expresión y obtener *True* o *False*.

````python
print(bool(1))
print(bool(1>4))
print(bool("hola"))
````

Casi todos los valores evaluados son True si tienen contenido.
* Una cadena es True salvo la cadena vacía.
* Un número es True salvo el cero.
* Una lista, tupla, conjunto o diccionario es True salvo que estén vacíos.
* Por supuesto True será True pero False y None serán False.

````python
print(bool(0))
print(bool(""))
print(bool([1,2]))
print(bool([]))
print(bool({}))
print(bool(()))
print(bool(False))
print(bool(None))
print(bool(True))
````

## 9. Cadenas
Una cadena de texto en Python es un conjunto de caracteres entre comillas simples o dobles.

La cadena **"hola mundo"** es igual que la cadena **'hola mundo'**. Ambas cadenas son del tipo str.

Cuando en una cadena de caracteres queremos meter más información estamos acostumbrados a concatenar con el operador **+**.

````python
persona = "Luis"
print("Buenos días, " + persona)
````

Sin embargo, no se recomienda abusar de este operador.

Al contrario que en otros lenguajes, no concatena con otros tipos de datos, por ejemplo, un simple entero.

Lo solucionamos usanto el método **str**.

````python
resultado = 7

# print("El resultado es " + resultado) # ERROR: Sólo se pueden concatenar directamente las cadenas.

print("El resultado es " + str(resultado))
````

Existen alternativas al uso del operador concatenación:
* Operados **%**
* Método **format()**
* Uso de **FStrings**

#### Operador **%**:
* Es la forma clásica de antiguos lenguajes de programación.
* En el texto se escriben caracteres de conversión dependiendo del tipo de dato que queremos poner.
* A continuación, va el operador % y después entre paréntesis y separadas por comas los valores que sustituirán a los caracteres de conversión.
* Los caracteres de conversión serán $d para enteros, %f para decimales, %s para cadenas, etc.
* Esta conversión es muy laxa, ya que no se requiere coincidencia total entre carácter y valor.

````python
print("Hola, me llamo %s y tengo %d años." % ("Ana", 30))
````

#### Método **format()**:
* Se aplica este método al final de la cadena y como parámetros se le indican los valores a sustituir.
* Dentro de la cadena se hacen corresponder con los fragmentos de texto {} en orden correlativo.
* Es posible darle un nombre a cada elemento.

````python
print("Hola, me llamo {} y tengo {} años.".format("Ana", 30))

print("Hola, me llamo {nombre} y tengo {edad} años.".format(nombre="Ana", edad=30))
````

#### F-strings
Permite introducir modificaciones y formateos dentro de la cadena de texto. La cadena lleva una letra **f** antes de la primera comilla.

Se compone básicamente de dos partes:
* **Placehoders:** El un espacio entre llaves **{ }** donde ponemos variables, código u otra expresión que dará contenido a la cadena. Es opcional.
* **Modifiers:** Es una modificación de lo anterior para indicar un formato concreto del placeholder. Se emplea el carácter de dos puntos **:**. Es opcional.

Si se omite la letra **f** al principio no se realiza la sustitución.

````python
nombre = "Ana"
edad = 30
print(f"Hola, me llamo {nombre} y tengo {nombre} años.")

print("Hola, me llamo {nombre} y tengo {nombre} años.") # Si omitimos la f no se hace la sustitución en los placeholders
````

En un f-string podemos "*incluir todo tipo de expresiones*":
* Expresiones que devuelvan un valor
* Uso de métodos
* Uso del operador ternario:  resultado = *valor_si_verdadero* **if** *condicion* **else** *valor_si_falso*

````python
# Expresiones que devuelvan un valor
print(f"La suma de 2 + 3 es {2 + 3}")

# Uso de métodos
palabra = "casa"
print(f"En mayúsculas {palabra.upper()}")

# Uso del operador ternario
edad = 20
print(f"{nombre} es { 'mayor de edad' if edad >= 18 else 'menor de edad' }")

edad = 6
print(f"{nombre} es { 'mayor de edad' if edad >= 18 else 'menor de edad' }")
````

#### Modificadores en las FStrings

|:>|Alineación a la derecha dejando los espacios que se indiquen|
|:^|Alineación centrada dejando espacios a ambos lados|
|:=|Espacios que se dejan entre el signo y el valor|
|:+|Indicar si el resultado es positivo o negativo|
|:-|Usa el signo - sólo para valores negativos|
|:,|Uso de la coma como separador de miles|
|:_|Uso de subrayado como separador de miles|
|:e,:E|Notación exponencial de un número con la e minúscula o e mayúscula|
|:b|Valor binario de un número|
|:d|Valor decimal de un número|
|:x,:X|Valor hexadecimal de un número|
|:o|Valor octal de un número|

#### Identidad

|Operador|Función|
|--------|-------|
|is|Devuelve True si ambas variables son el mismo objeto|
|is not|Devuelve True si ambas variables no son el mismo objeto|

En Python dos asignaciones simples al mismo número entero suelen producir el mismo objeto con el mismo idenfiticador (**id**). Esto se debe a un mecanismo interno de optimización llamado internado de enteros (**integer interning**).

Python preasigna en memoria un rango de números enteros de uso muy frecuente para ahorrar recursos.
* **El rango mágico**: Python guarda en caché todos los enteros entre -5 y 256.
* **Resultado**: Si creas variables con valores dentro de este rango, apuntarán exactamente al mismo objeto en memoria.

Fuera de ese rango el comportamiento depende de cómo se ejecute el código.
Suponemos que **a=300** y **b=300**:
* *En la consola*: a **is** b derá **False**
* *En un IDE*: a **is** b dará **True**

````python
a = 7
b = 8
print(id(a) == id(b)) # Tienen valores distintos por lo que tienen identificador distinto

a = 7
b = 7
print(id(a) == id(b)) # Tienen valores iguales por lo que tienen identificador distinto

a = 300
b = 300
print(id(a) == id(b)) # Tienen valores iguales fuera del rango mágico. En este caso de False pero en PyCharm sería True
````

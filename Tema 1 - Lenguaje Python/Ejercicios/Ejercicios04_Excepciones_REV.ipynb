## Ejercicio 1 - Enunciado
Crea una función llamada `procesar_compras(inventario, carrito, presupuesto)` que reciba:
* **inventario**: Un diccionario donde las claves son nombres de productos y los valores son sus precios (enteros O decimales).
* **carrito**: Una lista de tuplas, donde cada tupla contiene (nombre_producto, cantidad).
* **presupuesto**: Un número que representa el dinero disponible.
La función debe calcular el coste total del carrito y restar el dinero del presupuesto. Hay que gestionar mediante un único bloque `try-except` (o bloques anidados pero sin clases) las siguientes situaciones:
1. Si un producto del carrito no existe en el inventario, lanzar un `KeyError`, mostrar un mensaje avisando qué producto no está y continuar procesando el resto del carrito.
2. Si la cantidad de un producto es menor o igual a cero, o el precio en el inventario es negativo, lanzar un `ValueError` con un mensaje descriptivo.
3. Si el coste total supera el presupuesto, lanza un `ArithmeticError` indicando que los fondos son insuficientes.
4. Si ocurre cualquier otro error inesperado, capturar una excepción genérica.
Al final, la función debe usar obligatoriamente un bloque finally para imprimir el saldo restante (o el presupuesto intacto si no se completó la compra).

___
### Ejercicio 1 - Solución

```python
def procesar_compras(inventario, carrito, presupuesto):
    total_gastado = 0
    saldo_final = presupuesto
    
    print("--- Iniciando procesamiento de compra ---")
    
    for producto, cantidad in carrito:
        try:
            # Validar cantidad
            if cantidad <= 0:
                raise ValueError(f"La cantidad para '{producto}' debe ser mayor que cero.")
            
            # Obtener precio (puede lanzar KeyError)
            precio = inventario[producto]
            
            # Validar precio
            if precio < 0:
                raise ValueError(f"El precio de '{producto}' no puede ser negativo.")
            
            total_gastado += precio * cantidad
            
        except KeyError:
            print(f"Error: El producto '{producto}' no existe en el inventario. Se omite.")
        except ValueError as ve:
            print(f"Error de validación: {ve}. Se omite este producto.")
            
    try:
        # Verificar presupuesto al final
        if total_gastado > presupuesto:
            raise ArithmeticError(f"Fondos insuficientes. Total: {total_gastado}€, Presupuesto: {presupuesto}€.")
        
        saldo_final = presupuesto - total_gastado
        print(f"Compra realizada con éxito. Total cobrado: {total_gastado}€.")
        
    except ArithmeticError as ae:
        print(f"Error financiero: {ae} No se realizó ningún cargo.")
    except Exception as e:
        print(f"Error inesperado: {e}")
    finally:
        print(f"OPERACIÓN FINALIZADA. Saldo actual en cuenta: {saldo_final}€\n")

# --- Prueba del código ---
mi_inventario = {"manzana": 1.5, "platano": 2.0, "ordenador": 800.0, "producto_roto": -5}
mi_carrito = [
    ("manzana", 3), 
    ("pera", 2),         # Provocará KeyError
    ("platano", -1),     # Provocará ValueError (cantidad)
    ("producto_roto", 1) # Provocará ValueError (precio)
]

# Prueba 1: Compra con éxito dentro del presupuesto
procesar_compras(mi_inventario, [("manzana", 2)], 50)

# Prueba 2: Intento de compra que supera el presupuesto
procesar_compras(mi_inventario, [("ordenador", 1)], 500)

# Prueba 3: Carrito con errores mixtos
procesar_compras(mi_inventario, mi_carrito, 100)
```

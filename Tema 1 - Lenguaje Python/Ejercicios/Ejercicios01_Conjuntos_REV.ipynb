### Ejercicio 1 - Punto 1 - Solución
A partir de los dos conjuntos iniciales, crear otros dos conjuntos. Uno debe contener los nombres que empiezan por consonante y el otro los que comienzan por vocal.

```python
nombres_1 = {
    "Laura", "David", "Sofia", "Javier", "Elena", 
    "Manuel", "Carmen", "Jose", "Isabel", "Francisco", 
    "Paula", "Antonio", "Lucia", "Diego", "Martina"
}

nombres_2 = {
    "Daniel", "Marta", "Luis", "Alba", "Pablo", 
    "Sara", "Adrian", "Julia", "Alvaro", "Irene", 
    "Hugo", "Valeria", "Jorge", "Emma", "Ivan"
}

nombres_vocal = set()
nombres_consonante = set()

for x in nombres_1 | nombres_2:
    nombres_vocal.add(x) if x[0] in {"A","E","I","O","U"} else nombres_consonante.add(x)

print(nombres_vocal)
print("=====")
print(nombres_consonante)
```

### Ejercicio 1 - Punto 3 - Solución
A partir de la lista obtenida en el punto 2, construir un conjunto de 40 tuplas donde cada elemento del conjunto es el nombre y un número del 1 al 40.


```python
nombres_1 = {
    "Laura", "David", "Sofia", "Javier", "Elena", 
    "Manuel", "Carmen", "Jose", "Isabel", "Francisco", 
    "Paula", "Antonio", "Lucia", "Diego", "Martina"
}

nombres_2 = {
    "Daniel", "Marta", "Luis", "Alba", "Pablo", 
    "Sara", "Adrian", "Julia", "Alvaro", "Irene", 
    "Hugo", "Valeria", "Jorge", "Emma", "Ivan"
}

conjunto_nombres = set()
x = 1

for i in sorted(list(nombres_1 | nombres_2)): # Lo obtenido en el punto 2
    conjunto_nombres.add((i,x))
    x += 1

print(conjunto_nombres)
```

### Ejercicio 1 - Punto 4 - Solución
A partir de la lista obtenida en el punto 2, obtener listas por cantidad de letras en un nombre. Una lista con los nombres de tres letras, otra con los de cuatro y así sucesivamente.

```python
nombres_1 = {
    "Laura", "David", "Sofia", "Javier", "Elena", 
    "Manuel", "Carmen", "Jose", "Isabel", "Francisco", 
    "Paula", "Antonio", "Lucia", "Diego", "Martina"
}

nombres_2 = {
    "Daniel", "Marta", "Luis", "Alba", "Pablo", 
    "Sara", "Adrian", "Julia", "Alvaro", "Irene", 
    "Hugo", "Valeria", "Jorge", "Emma", "Ivan"
}

lista_nombres = [[],[],[],[],[],[],[]]

for i in sorted(list(nombres_1 | nombres_2)): # Lo obtenido en el punto 2
    match len(i):
        case 0|1|2|3:
            lista_nombres[0].append(i)
        case 4:
            lista_nombres[1].append(i)
        case 5:
            lista_nombres[2].append(i)
        case 6:
            lista_nombres[3].append(i)
        case 7:
            lista_nombres[4].append(i)
        case 8:
            lista_nombres[5].append(i)
        case _:
            lista_nombres[6].append(i)

print(lista_nombres)
```

## Ejercicio 1
Generar 10 números aleatorios entre 1 y 10 usando generadores, es decir, cada número se tiene que obtener con la función next sobre el generador.

```python
import random

def generar_aleatorio():
    numeros = []
    while len(numeros) < 10:
        n = random.randint(1,10)
        if n not in numeros:
            numeros.append(n)
            yield n

gen = generar_aleatorio()

print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
# print(next(gen)) # ERROR: Ya se han generado todos

print("===")

for i in generar_aleatorio():
    print(i)
```

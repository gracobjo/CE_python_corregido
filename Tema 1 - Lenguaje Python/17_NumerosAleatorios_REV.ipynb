# Números aleatorios

El módulo **`random`** forma parte de la librería estándar de Python y se utiliza para generar números pseudoaleatorios. Es ideal para simulaciones, juegos, pruebas y tareas automatizadas donde se necesita variabilidad.

Los números generados por este módulo son **pseudoaleatorios**, lo que significa que siguen un orden determinado por un valor inicial (semilla o *seed*).

---

## 1. Configuración Inicial e Importación
Para empezar a usarlo, simplemente hay que importar el módulo.

```python
import random

# Verificamos que se haya importado correctamente imprimiendo la versión o info del módulo
print("Módulo 'random' cargado correctamente.")
```

## 2. Generación de Números Aleatorios Básicos

Generar decimales entre 0 y 1, y cómo controlar la aleatoriedad usando una semilla (*seed*).

* `random.random()`: Devuelve un float aleatorio en el rango `[0.0, 1.0)`.
* `random.seed(x)`: Inicializa el generador de números aleatorios. Si se usa la misma semilla, se obtiene exactamente los mismos resultados en cada ejecución.

```python
import random

print("--- Uso de random.random() ---")
# Generar 3 números decimales aleatorios
for i in range(3):
    print(f"Número {i+1}: {random.random()}")

print("\n--- Uso de random.seed() ---")
# Fijamos la semilla en 42
random.seed(42)
print(f"\nCon semilla 42: {random.random()}")
print(f"Siguiente número: {random.random()}")

# Si reiniciamos la semilla a 53, los números se repiten idénticamente
random.seed(53)
print(f"\nReiniciando semilla 53: {random.random()}")
print(f"Siguiente número: {random.random()}")

# Si reiniciamos la semilla a 42, los números se repiten idénticamente
random.seed(42)
print(f"\nReiniciando semilla 42: {random.random()}")
print(f"Siguiente número: {random.random()}")
```

## 3. Generación de Números Enteros
* `random.randint(a, b)`: Devuelve un entero aleatorio **incluyendo** tanto a `a` como a `b` `(a <= N <= b)`.
* `random.randrange(start, stop, step)`: Funciona igual que la función `range()`. Devuelve un entero del rango seleccionado (excluye el valor `stop`).


```python
print("--- Lanzamiento de un dado (randint) ---")
# Genera un número entero entre 1 y 6 (ambos incluidos)
for _ in range(100):
    print(random.randint(1, 6), end=" ")
print("\n")

print("--- Números pares aleatorios entre 0 y 10 (randrange) ---")
# start=0, stop=11 (no se incluye), step=2 (solo números pares)
for _ in range(100):
    print(random.randrange(0, 11, 2), end=" ")
print()
```

## 4. Números Decimales en Rangos Específicos

Número decimal pero fuera del rango estándar de 0 a 1. Dos opciones:

* `random.uniform(a, b)`: Devuelve un float aleatorio entre `a` y `b`. El valor de `b` puede o no estar incluido dependiendo del redondeo flotante.

```python
print("--- Temperatura aleatoria entre 15.0 y 25.0 (uniform) ---")
print(random.uniform(15.0, 25.0))
```

## 5. Trabajando con Listas y Secuencias

El módulo `random` es extremadamente potente para interactuar con colecciones de datos.

* `random.choice(lista)`: Elige un único elemento al azar de una lista.
* `random.choices(lista, k=n)`: Elige múltiples elementos de una lista **con reemplazo** (pueden repetirse). Permite añadir "pesos" para dar más prioridad a unos elementos que a otros.
* `random.sample(lista, k=n)`: Elige elementos de una lista **sin reemplazo** (no se pueden repetir).
* `random.shuffle(lista)`: Mezcla la lista original modificándola.

```python
opciones = ['Piedra', 'Papel', 'Tijera', 'silla', 'mesa']

print("--- Elección única (choice) ---")
print(f"La computadora eligió: {random.choice(opciones)}")

print("\n--- Elección múltiple con repetición (choices) ---")
# Extraemos 3 elementos de la lista (pueden repetirse)
print(random.choices(opciones, k=3))

print("\n--- Elección múltiple con pesos (choices con weights) ---")
# Le damos un 80% de probabilidad al 'Papel' y 5% a los demás
pesos = [0.01, 0.95, 0.01, 0.01, 0.01]
print(random.choices(opciones, weights=pesos, k=5))

print("\n--- Selección única sin repetición (sample) ---")
# Extraer 3 elementos únicos de la lista (no se repetirán)
baraja = ['As', 'Rey', 'Reina', 'Jota', '10', '9']
print(random.sample(baraja, k=3))

print("\n--- Mezclar una lista (shuffle) ---")
numeros = [1, 2, 3, 4, 5]
random.shuffle(numeros)
print(f"Lista mezclada: {numeros}")
```

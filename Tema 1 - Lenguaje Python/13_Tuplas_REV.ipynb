# Tuplas
Almacenan múltiples elementos en una variable:
* Se crean con paréntesis (), aunque se podrían omitir
* Sus elementos están:
    * Ordenados: Hay un orden definido intrínseco, es decir, un primer elemento, segundo, etc.
    * No son modificables.
    * Permite almacenar valores duplicados
    * Son accesibles mediante índices, siendo el 0 el primero

___
### Definición básica de una tupla

```python
tupla1 = (1,2,3,4,5)
tupla2 = 1,2,3,4,5 # Podríamos omitir los paréntesis
#tupla3 = tuple(1,2,3,4,5) # ERROR: El construcctor tupla no admite el paso de argumentos individuales para cada elemento de la tupla
tupla4 = tuple((1,2,3,4,5)) # Tupla a partir de una tupla
tupla5 = tuple([1,2,3,4,5]) # Tupla a partir de una lista
tupla6 = tuple({1,2,3,4,5}) # Tupla a partir de una conjunto
tupla7 = (1,) # Tupla de un único elemento. Imprescindible añadir la coma al final
tupla8 = (1) # No es una tuple porque falta la coma final
tupla9 = ()

print(tupla1,type(tupla1))
print(tupla2,type(tupla2))
#print(tupla3,type(tupla3))
print(tupla4,type(tupla4))
print(tupla5,type(tupla5))
print(tupla6,type(tupla6))
print(tupla7,type(tupla7))
print(tupla8,type(tupla8))
print(tupla9,type(tupla9))
```

___
### Podemos tener diferentes tipos de datos incluso mezclados

```python
tupla10 = (1,2,3)
tupla11 = ("texto1","texto2","texto3")
tupla12 = ("texto1",6,7,8.5,[1,2],"texto2")
```

___
### Se accede a las tuplas a través de un índice
* Se empieza en 0.
* Se admiten números negativos siendo el -1 el último elemento.

```python
tupla13 = ("texto1",6,7,8.5,[1,2],"texto2")

print(tupla13[0])
print(tupla13[1])
print(tupla13[-4])
```

___
### Podemos hacer slicing al igual que en las listas

```python
abecedario = ('a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm', 'n', 'ñ', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z')

print(abecedario[0:5])
print(abecedario[3:9])
print(abecedario[9:])
print(abecedario[:9])
print(abecedario[-9:-4])
```

___
### Comprobar la existencia o no de un elemento
Empleamos el operador **in** o el operador **not in**

```python
abecedario = ('a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm', 'n', 'ñ', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z')

if "A" in abecedario: # Se diferencia entre mayúsculas y minúsculas
    print("Si está")
else:
    print("No está")
```

___
### Cambiar tuplas
Aunque son inmutables y no podemos cambiar elementos ni individualment ni la tupla entera, podemos reconvertirla con el mismo nombre una vez modificada pasándola a lista primero.

```python
coches = ("seat","audi","citroen")
print(coches)
print(id(coches))
# coches[0] = "mercedes" # ERROR: No puede asignarse un valor

lista_coches = list(coches)
lista_coches[0] = "mercedes"

coches = tuple(lista_coches)
print(coches)
print(id(coches)) # El identificador es distinto al original

# De igual forma podríamos usar todos los métodos de List y posteriormente volvemos a reconvertir en tupla
coches2 = list(coches)
coches2.append("renault")
coches = tuple(coches2)
print(coches)
```

### Añadir una tupla a una tupla

```python
numeros1 = (1,2,3)
numeros2 = (4,5,6)
print(id(numeros1))

numeros1 += numeros2
print(numeros1)
print(id(numeros1))
```

### Recorrer una tupla
Las tuplas son iterables por lo que podemos recorrerlas con un bucle **for**

```python
letras = ('a', 'b', 'c', 'd', 'e')

for i in letras:
    print(i)
```

___
### Desempaquetado de tuplas
Al crear una tupla juntamos varios elementos en uno solo, es decir, en la tupla. Esto se llama empaquetado.

El desempaquetado es lo contrario. Se trata de extraer los valores de dicha tupla.

Sintaxis:

**variables separadas por comas entre paréntesis o no = tupla**

```python
letras = ('a', 'b', 'c', 'd', 'e')

letra1,letra2,letra3,letra4,letra5 = letras
print(letra1,letra2,letra3,letra4,letra5,sep="---")

(letra1,letra2,letra3,letra4,letra5) = letras
print(letra1,letra2,letra3,letra4,letra5,sep="---")
```

El asterisco __*__ se usa cuando faltan variables. Se asignan al elemento que lo lleva en forma de lista.

Si el asterisco __*__ se asigna a una variable que no es la última, se asignan para que encaje correctamente.

```python
letras = ('a', 'b', 'c', 'd', 'e')

(letra1,letra2,*letra3) = letras
print(letra1)
print(letra2)
print(letra3)

print("=====")

(letra1,*letra2,letra3,letra4) = letras
print(letra1)
print(letra2)
print(letra3)
print(letra4)
```

___
### Unir tuplas
Podemos unir tuplas con el operador **+**

También podemos multiplicar el contenido de una tupla un número determinado de veces con el operador __*__

```python
letras1 = ('a', 'b', 'c')
letras2 = ('d', 'e')

letras3 = letras1 + letras2
print(letras3)

letras4 = letras1 * 3
print(letras4)
```

# Métodos
* count
* index

#### count(elemento)
Cuenta el número de veces que aparece un elemento en la tupla.

```python
numeros = (1,4,5,3,6,7,3,4,5,2,7,4,5,2,1,4,5,2,4,6,9)

print(numeros.count(4))
print(numeros.count(2))
print(numeros.count(8))
```

#### index(elemento)
Busca en una tupla y devuelve el índice en que se encuentra dicho elemento.

**IMPORTANTE:** Produce un error de tipo ValueError si el elemento no está

```python
abecedario = ('a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm', 'n', 'ñ', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z')

print(abecedario.index('h'))

try:
    print(abecedario.index('A')) # No está
except ValueError:
    print("El valor no está")
```

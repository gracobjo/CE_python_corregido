# Decoradores
Los decoradores añaden código adicional a las funciones.
Sabiendo dos cuestiones:
* En Python las funciones se pueden asignar a variables. No confundir llamada o uso de la función con la asignación de la misma
* En Python se pueden definir funciones dentro de otras funciones

```python
# Asignación de una función a una variable
def mi_funcion():
    return "Texto"

var_funcion = mi_funcion

print(mi_funcion())
print(mi_funcion)

print(var_funcion())
print(var_funcion)
```

```python
# Funcion que devuelve una función

def operar(orden):
    def incrementar(n):
        return n + 1
    def decrementar(n):
        return n -1

    match orden:
        case "i":
            return incrementar
        case "d":
            return decrementar

var_funcion = operar("i")
print(var_funcion(6))

var_funcion = operar("d")
print(var_funcion(15))
```

# Construimos nuestro propio decorador
Dos elementos:
* Decorador: Recibe una función como entrada y la devuelve decorada.
* La función decorada: Es la función principal que no queremos cambiar en su modo básico.

```python
def decorador(f):
    def antes_y_despues(texto):
        print("Ejecución antes")
        f(texto)
        print("Ejecución después")
    return antes_y_despues

def imprimir(t):
    print(f"En la función imprimir: {t}")

funcion_decorada = decorador(imprimir)
funcion_decorada("HOLA")
```

___
Las dos últimas líneas anteriores las podemos evitar usando **@nombre_decorador** justo antes de la función decorada

```python
def decorador(f):
    def antes_y_despues(texto):
        print("Ejecución antes")
        f(texto)
        print("Ejecución después")
    return antes_y_despues

@decorador
def imprimir(t):
    print(f"En la función imprimir: {t}")

imprimir("HOLA")
```

___
Los decoradores pueden recibir parámetros envolviendo una vez más la función

```python
def decorador(caja):
    def decorador_interior(f):
        def antes_y_despues(texto):
            print("Ejecución antes")
            if caja=="mayusculas":
                f(texto.upper())
            if caja=="minusculas":
                f(texto.lower())
            print("Ejecución después")
        return antes_y_despues
    return decorador_interior

@decorador("minusculas") #Cambiar entre mayúsculas y minúsculas
def imprimir(t):
    print(f"En la función imprimir: {t}")

imprimir("Hola")
```

___
Si la función interna devuelve un valor el decorador tendrá más sentido y no se limitará a poner código antes y después.

```python
def decorador(f):
    def interna(texto):
        return f(texto.capitalize())
    return interna

@decorador
def imprimir(t):
    return f"En la función imprimir: {t}"

print(imprimir("ESTO ES UN TEXTO PARA PROBAR LA FUNCIÓN."))
```

___
Un decorador puede usarse con distintas funciones

```python
def decorador(f):
    def interna():
        print("ANTES")
        f()
        print("DESPUÉS")
    return interna

@decorador
def decir_hola():
    print("Hola")

@decorador
def decir_adios():
    print("Adios")

decir_hola()
decir_adios()
```

# EJERCICIO
Decorar las 4 operaciones aritméticas básicas para que cada vez que se ejecute se imprima el resultado

```python
def decorador(func):
    def interna(x,y):
        print(f"El resultado es: {func(x,y)}")
    return interna


@decorador
def sumar(a,b):
    return a + b

@decorador
def restar(a,b):
    return a - b

@decorador
def multiplicar(a,b):
    return a * b


@decorador
def dividir(a,b):
    return a / b

sumar(3,4)
restar(8,5)
multiplicar(3,4)
dividir(25,5)
```

___
Una función puede tener varios decoradores.
Para que se llame el segundo decorador es necesario llamarlo desde el primero

```python
#Una función puede tener varios decoradores

def decorador1(func):
    def interna():
        print("Decorador 1: ANTES")
        func()
        print("Decorador 1: DESPUÉS")
    return interna


def decorador2(func):
    def interna():
        print("Decorador 2: ANTES")
        func()
        print("Decorador 2: DESPUÉS")
    return interna


@decorador1
@decorador2
def decir_hola():
    print("Hola")

decir_hola()
```

# EJERCICIO
En el ejercicio de las operaciones aritméticas, añadir un decorador a la función de dividir para que en caso de intentar dividir entre cero produzca un mensaje de error.

```python
def decorador(func):
    def interna(x,y):
        print(f"El resultado es: {func(x,y)}")
    return interna

def decorador_comprobar(func):
    def interna(x,y):
        if y == 0:
            print("Se intenta dividir entre 0")
        else:
            func(x,y)
    return interna

@decorador_comprobar
@decorador
def dividir(a,b):
    return a / b

dividir(25,5)
dividir(25,0)
```

En el ejercicio de las operaciones aritméticas, debe indicarse en el texto la operación realizada. No se puede modificar la función original.

```python
# EJERCICIO: Añadir la función realizada

def decorador(operacion):
    def decorador_interior(func):
        def interna(x,y):
            print(f"El resultado la {operacion} es: {func(x,y)}")
        return interna
    return decorador_interior


@decorador("suma")
def sumar(a,b):
    return a + b

@decorador("resta")
def restar(a,b):
    return a - b

@decorador("multiplicación")
def multiplicar(a,b):
    return a * b

@decorador("división")
def dividir(a,b):
    return a / b

sumar(3,4)
restar(8,5)
multiplicar(3,4)
dividir(25,5)
```

# Listas
Almacenan múltiples elementos en una variable:
* Se crean con corchetes []
* Sus elementos están:
    * Ordenados: Hay un orden definido intrínseco, es decir, un primer elemento, segundo, etc.
    * Son modificables: Podemos añadir, elminar, cambiar los elementos de una lista.
    * Pueden almacenar valores duplicados
    * Son accesibles mediante índices, siendo el 0 el primero

___
### Definición básica de una lista

````python
muebles = [ "mesa" , "silla" , "armario"]
print(muebles)
````

___
### Acceso por índice a los elementos de una lista siendo el primer índice el 0

````python
print(muebles[0])
print(muebles[1])
print(muebles[2])
````

___
### Puede haber elementos repetidos

````python
muebles = [ "mesa" , "silla" , "armario" , "silla"]
print(muebles)
````

___
### La función _len()_ nos permite obtener el número de elementos de la lista

````python
print(len(muebles))
````

___
### Los tipos de datos de una lista pueden ser los que queramos

````python
lista1 = ["coche","moto","bici"] #cadenas
lista2 = [23,45,67,89] #números
lista3 = [True,False,True] #booleanos
lista4 = [lista1,lista2] #listas
print(lista4)
````

___
### Podemos tener diferentes tipos de datos

````python
lista5 = ["coche",True,45,"moto"]
print(lista5)
````

___
### Python define una lista como un objeto de tipo list

````python
print(type(lista5))
````

___
### Uso del constructor list() para crear listas

````python
lista6 = list(["a","b","c","d","e"])
print(type(lista6))
````

___
### Se permite indexación negativa
Significa empezar por el final siendo el -1 el último, el -2 el penúltimo, etc.

````python
muebles = [ "mesa" , "silla" , "armario", "puerta" , "estanteria" , "cama" , "sillón"]
print(muebles[-1])
print(muebles[-2])
````

___
### Es posible especificar un rango de índices
No se incluye el último, es decir, en el siguiente ejemplo el primero será el 2 y el último será el 4 porque el 5 (el que se indica) no se incluye

````python
print(muebles[2:5])
````

___
### Si se omite alguna parte del rango se supone el final o el principio

````python
muebles = [ "mesa" , "silla" , "armario", "puerta" , "estanteria" , "cama" , "sillón"]
print(muebles[2:])
print(muebles[:3]) # Recordamos que el 3 no se incluye
````

___
### Es posible usar rangos con índices negativos

````python
print(muebles[-4:-2]) # El -2 no se incluye
````

___
### Es posible comprobar si un elemento existe con el operador in

````python
muebles = [ "mesa" , "silla" , "armario", "puerta" , "estanteria" , "cama" , "sillón"]

if "silla" in muebles:
    print("Silla está en muebles")
````

___
### Podemos cambiar un elemento

````python
muebles = [ "mesa" , "silla" , "armario", "puerta" , "estanteria" , "cama" , "sillón"]
muebles[1] = "butaca"
print(muebles[1])
print(muebles)
````

___
### Es posible cambiar un rango de valores a la vez

````python
muebles = [ "mesa" , "silla" , "armario", "puerta" , "estanteria" , "cama" , "sillón"]
muebles[1:3] = ["M1","M2"]
muebles[1],muebles[2] = "m1","m2"

print(muebles)
````

___
### Si asignas más indices de los que reemplazas, los nuevos se insertarán donde se indica y los restantes se mueven en consecuencia

````python
muebles = [ "mesa" , "silla" , "armario", "puerta" , "estanteria" , "cama" , "sillón"]
muebles[1:3] = ["M1","M2","M3"]
print(muebles)
````

___
### Si asignas menos indices de los que reemplazas, los nuevos se insertarán donde se indica y los restantes desaparecen

````python
muebles = [ "mesa" , "silla" , "armario", "puerta" , "estanteria" , "cama" , "sillón"]
muebles[1:6] = ["M1","M2"]
print(muebles)
````

# Métodos
* append
* insert
* extend
* remove
* pop
* sort
* copy
* clear
* reverse

___
### append(elemento)
Añade un elemento al final de una lista

````python
numeros = [1,2,3]
print(numeros)
print(id(numeros))
numeros.append(4)

print(numeros)
print(id(numeros))
````

___
### insert(indice,elemento)
Insertar un elemento sin reemplazar uno existente en el índice especificado


````python
muebles = [ "mesa" , "silla" , "armario", "puerta" , "estanteria" , "cama" , "sillón"]
muebles.insert(3,"taburete")
print(muebles)
````

___
### extend(lista)
Añade elementos de una lista a otra

````python
numeros = [1,2,3]
muebles = [ "mesa" , "silla" , "armario", "puerta" , "estanteria" , "cama" , "sillón"]

numeros.extend(muebles)
print(numeros)

#Podemos extender la lista con sólo una parte de los elementos de la lista original
lista_a = [1,2,3,4,5]
lista_b = [6,7,8,9,10]
lista_a.extend(lista_b[2:4])
print(lista_a)
````

___
### remove(elemento)
Elimina el elemento especificado de la lista

````python
muebles = ["mesa" , "silla" , "armario", "puerta" , "estanteria" , "cama" , "sillón"]

muebles.remove("silla")
print(muebles)
````

___
### pop(indice)
Elimina el elemento del índice especificado

````python
muebles = ["mesa" , "silla" , "armario", "puerta" , "estanteria" , "cama" , "sillón"]

muebles.pop(4)
print(muebles)
````

___
### sort(indice)
Ordena la lista en principio de forma ascendente

````python
muebles = ["mesa" , "silla" , "armario", "puerta" , "estanteria" , "cama" , "sillón"]
muebles.sort()
print(muebles)

muebles = ["mesa" , "silla" , "armario", "puerta" , "estanteria" , "cama" , "sillón"]
muebles.sort(reverse=True)
print(muebles)

muebles = ["mesa" , "silla" , "5", "armario", "puerta" , "estanteria" , "2", "cama" , "sillón"]
muebles.sort()
print(muebles)

muebles = ["mesa" , "silla" , 5, "armario", "puerta" , "estanteria" , 2, "cama" , "sillón"]
#muebles.sort() #ERROR -> No puede ordenar enteros con cadenas
print(muebles)
````

___
### copy()
Copia una lista creando un objeto nuevo

````python
muebles = ["mesa" , "silla" , "armario", "puerta" , "estanteria" , "cama" , "sillón"]
muebles_copia = muebles.copy() #Crea un elemento nuevo
print(muebles_copia)

muebles_copia_2 = muebles # Es el mismo objeto, no una copia
muebles_copia_3 = list(muebles) # Crea un elemento nuevo

print(id(muebles)) # Igual a copia_2
print(id(muebles_copia))
print(id(muebles_copia_2)) # Igual a la original
print(id(muebles_copia_3))
````

___
### clear()
Vacía una lista

````python
muebles = ["mesa" , "silla" , "armario", "puerta" , "estanteria" , "cama" , "sillón"]
muebles.clear()
print(muebles)
````

___
### reverse()
Invierte el orden de una lista

````python
muebles = ["mesa" , "silla" , "armario", "puerta" , "estanteria" , "cama" , "sillón"]
muebles.reverse()
print(muebles)
````

# Unir listas
Podemos usar el operador +, append o extend

````python
lista1 = [1,2,3]
lista2 = [4,5,6]
lista3 = lista1 + lista2 #Genera una nueva lista
print(lista3)

lista1 = [1,2,3]
lista2 = [4,5,6]
for i in lista2:
    lista1.append(i)
print(lista1)

lista1 = [1,2,3]
lista2 = [4,5,6]
lista1.extend(lista2)
print(lista1)
````

# LIST COMPREHENSION
Permite crear lista a partir de una expresión
```python
lista = [ elemento for elemento in conjunto if condición_verdadera ]
```
Devuelve una **nueva** lista.

**elemento** es el valor en la iteración actual pero también la salida que puede ser modificada.

**conjunto** es un elemento iterable.

La parte del **if** es opcional y funciona a modo de filtro para evaluar si el elemento se incluye o no en la lista.

````python
lista = [ x for x in range(10) ]
print(lista)
print(type(lista))
````

````python
lista = [ x for x in range(10) if x != 4 ]
print(lista)
````

````python
lista = [ x*x for x in range(10) if x != 4 ]
print(lista)
````

## Ejercicio 1
Partiendo de una lista de elementos de distinto tipo:
1. Obtener una lista que contenga únicamente los tipos de datos de cada elemento en su posición correspondiente.
2. Obtener el tipo de dato que más aparece.
3. Ordenar la lista agrupando los tipos de datos

```python
#Ejemplo para la lista siguiente
[1,8.6,"cadena",[3,4],7,"texto",9.2,"palabra",5,["otro texto",9],7]

# Punto 1 
[<class 'int'>, <class 'float'>, <class 'str'>, <class 'list'>, <class 'int'>, <class 'str'>, <class 'float'>, <class 'str'>, <class 'int'>, <class 'list'>, <class 'int'>]

# Punto 2 
El tipo de dato que más aparece es <class 'int'> con 4 veces

# Punto 3 
[[3, 4], ['otro texto', 9], 1, 7, 5, 7, 'cadena', 'texto', 'palabra', 8.6, 9.2]
```

````python
# Punto 1
lista = [1,8.6,"cadena",[3,4],7,"texto",9.2,"palabra",5,["otro texto",9],7]
lista1 = list(map(lambda x : type(x),lista))
print("Punto 1\n",lista1)


# Punto 2
lista = [1,8.6,"cadena",[3,4],7,"texto",9.2,"palabra",5,["otro texto",9],7]
max_numero , max_tipo = 0 , ""

for i in list(set(map(lambda x : type(x),lista))): #Para eliminar duplicados de una lista, la reconvertimos a set
    actual = len([x for x in lista if i == type(x)])
    if max_numero <= actual:
        max_numero , max_tipo = actual , i
print(f"\nPunto 2\nEl tipo de dato que más aparece es {max_tipo} con {max_numero} veces")


# Punto 3
lista = [1,8.6,"cadena",[3,4],7,"texto",9.2,"palabra",5,["otro texto",9],7]
lista_agrupada = []

for i in list(set(map(lambda x : type(x),lista))):
    [ lista_agrupada.append(x) for x in lista if i == type(x) ]

print(f"\nPunto 3\n{lista_agrupada}")
````

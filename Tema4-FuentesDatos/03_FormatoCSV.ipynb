## Objeto `reader`
Crea un objeto lector que itera sobre las líneas del archivo CSV. Cada línea es una lista de cadenas

Propiedades:
* `line_num`: Número de líneas que se han leído hasta el momento.
* `dialect`: Objeto Dialect (dialecto) empleado (se verá en el cuaderno siguiente).

Formas de recorrer el archivo:
* Empleando el método `next`
* Empleando un bucle `for`

### Usando **next**

```python
import csv

with open('./recursos/datos.csv', mode='r', encoding='utf-8') as archivo:
    lector = csv.reader(archivo)
    
    fila = next(lector)
    print(lector.line_num,fila)
    
    fila = next(lector)
    print(lector.line_num,fila)
    
    fila = next(lector)
    print(lector.line_num,fila)
    
    fila = next(lector)
    print(lector.line_num,fila)
    
    # ...

    print(lector.dialect.delimiter)
```

___
### Usando **for**

```python
import csv

with open('./recursos/datos.csv', mode='r', encoding='utf-8') as archivo:
    lector = csv.reader(archivo)
    for fila in lector:
        print(fila)  # ['columna1', 'columna2', ...] Todas las columnas

    archivo.seek(0) # Vuelve el puntero al principio aunque se recomienda volve a instanciar csv.reader(archivo)
    
    for fila in lector:
        print(fila[1:3]) # Sólo nombre y edad
        # print(fila["nombre"]) # ERROR: Sólo acceso por índices
```

## DictReader
Mapea cada línea en un diccionario Python donde las claves son los nombres de las columnas y los valores los datos de cada celda.
Parámetros:
* `archivo`
* `fieldnames`
* `restkey`
* `restval`
* `dialect`: Se verá más adelante.
* `delimiter`: Se verá más adelante.
* `quotechar`: Se verá más adelante.

```python
import csv

with open('recursos/datos.csv', mode='r', encoding='utf-8') as archivo:
    lector = csv.DictReader(archivo)

    print(lector.fieldnames)
    
    for fila in lector:
        print(fila)
        print(fila["nombre"])
```

___
### `fieldnames`

Una lista o secuencia (opcional) con los nombres de las columnas. Si no se pasa, DictReader toma por defecto la primera fila del archivo CSV como cabecera. Si el fichero CSV no tiene cabecera, se debe definir manualmente aquí.

**IMPORTANTE**: No se trata de indicar los campos con los que te quieres quedar, sino de definir la cabecera si no existe.

```python
import csv

#with open('recursos/datos.csv', mode='r', encoding='utf-8') as archivo:
with open('recursos/datos-sincabecera.csv', mode='r', encoding='utf-8') as archivo:
    lector = csv.DictReader(archivo,fieldnames=["dni","nombre","edad","ciudad","telefono"])

    for fila in lector:
        print(fila)
```

___
### `restkey`y `restval`
* `restkey`: Si una fila tiene más campos que la cabecera, los valores sobrantes se guardan en una lista bajo esta clave (por defecto es None)
* `restval`: Si una fila tiene menos campos que la cabecera, a las columnas faltantes se les asigna este valor (por defecto es None)

```python
import csv

#with open('recursos/datos.csv', mode='r', encoding='utf-8') as archivo:
# En este fichero a la primera línea le sobra un campo y a la segunda le falta un campo
with open('recursos/datos-faltantes-sobrantes.csv', mode='r', encoding='utf-8') as archivo:
    lector = csv.DictReader(archivo,restkey="sobrante",restval="faltante")

    for fila in lector:
        print(fila)
```

## Objeto `writer`

Consideraciones al abrir el archivo con `open`:
* **mode='w'**: Abre el archivo para escritura (si no existe, lo crea; si existe, lo sobrescribe). Si se quiere agregar datos al final sin borrar lo anterior, se usa el modo 'a' (append).
* **newline=''**: Es crucial en Python. Evita que el sistema operativo agregue líneas en blanco adicionales al gestionar los saltos de línea (\r\n o \n).
* **encoding='utf-8**': Asegura que los caracteres especiales (tildes, ñ) se guarden correctamente.

### Métodos principales
* `writerow`: Escribe una fila a partir de un iterable.
* `writerows`: Escribe múltiples filas a partir de una lista de listas (u otro tipo).

```python
import csv

# Abrir el archivo en modo escritura
with open('recursos/salida.csv', mode='w', newline='', encoding='utf-8') as archivo:
# with open('recursos/salida.csv', mode='w', encoding='utf-8') as archivo: # Sin newline deja una línea en blanco entre medias de las líneas
    escritor = csv.writer(archivo)
    
    # Escribir filas individualmente o múltiples filas
    escritor.writerow(['Nombre', 'Edad', 'Ciudad'])
    escritor.writerow(['Ana', 30, 'Madrid'])
    escritor.writerow(['Carlos', 25, 'Barcelona'])

    filas_adicionales = [
        ["Luis",40,"Oviedo"],
        ["María",50,"Logroño"]
    ]

    escritor.writerows(filas_adicionales)
```

## Objeto `DictWriter`

A diferencia de `csv.writer`, `DictWriter` mapea las claves de un diccionario Python directamente con los nombres de las columnas del archivo CSV.

Parámetros del constructor:
* `archivo`: El objeto archivo abierto en modo de escritura `(open(..., 'w', newline=''))`.
* `fieldnames`: Una secuencia (lista o tupla) con los nombres de las columnas en el orden exacto en que se desean escribir.
* `restval`: El valor que se escribe si un diccionario no contiene alguna de las claves especificadas en fieldnames (por defecto es '').
* `extrasaction`: Define qué hacer si un diccionario pasado contiene claves que no están en fieldnames. Puede valer `raise` (lanza una excepción `ValueError`, por defecto) o `ignore` (ignora las claves sobrantes).

Métodos principales del objeto DictWriter

* `writeheader()`: Escribe la primera fila del archivo CSV utilizando los nombres definidos en fieldnames como encabezados.
* `writerow(rowdict)`: Escribe un único diccionario (rowdict) donde las claves coinciden con fieldnames.
* `writerows(rowdicts)`: Escribe una lista o iterable de múltiples diccionarios de una sola vez.

```python
import csv

# Se definen los nombres de las columnas (fieldnames)
nombres_columnas = ['nombre', 'edad', 'ciudad']

# Se crea una lista de datos en formato de lista de diccionarios
datos = [
    {'nombre': 'Ana', 'edad': 30, 'ciudad': 'Madrid'},
    {'nombre': 'Carlos', 'edad': 25, 'ciudad': 'Barcelona'},
    {'nombre': 'Elena', 'edad': 35, 'ciudad': 'Valencia'}
]

# Se abre el archivo CSV en modo escritura ('w')
# Se recomienda usar 'newline=""' para evitar saltos de línea extra en Windows/Mac
with open('recursos/personas.csv', mode='w', newline='', encoding='utf-8') as archivo:
    
    # Se crea la instancia de DictWriter
    writer = csv.DictWriter(archivo, fieldnames=nombres_columnas)
    
    # Se escribe la fila de encabezados
    writer.writeheader()  # Genera: nombre,edad,ciudad

    # Se escribe una fila de datos
    writer.writerow({'nombre': 'Juan', 'edad': 37, 'ciudad': 'Zamora'})
    
    # Se escriben todas las filas de datos de una vez
    writer.writerows(datos)

print("El archivo CSV se ha creado correctamente.")
```

### Uso de `restval`

```python
import csv

nombres_columnas = ['nombre', 'edad', 'ciudad']

datos = [
    {'nombre': 'Ana', 'edad': 30, 'ciudad': 'Madrid'},
    {'nombre': 'Carlos', 'edad': 25},
    {'nombre': 'Elena', 'edad': 35, 'ciudad': 'Valencia'}
]

with open('recursos/personas.csv', mode='w', newline='', encoding='utf-8') as archivo:
    writer = csv.DictWriter(archivo, fieldnames=nombres_columnas,restval="---")
    writer.writeheader()  # Genera: nombre,edad,ciudad
    writer.writerows(datos)
```

### Uso de `extrasaction` con valor `ignore`

```python
import csv

nombres_columnas = ['nombre', 'edad']

datos = [
    {'nombre': 'Ana', 'edad': 30, 'ciudad': 'Madrid'},
    {'nombre': 'Carlos', 'edad': 25, 'ciudad': 'Barcelona'},
    {'nombre': 'Elena', 'edad': 35, 'ciudad': 'Valencia'}
]

with open('recursos/personas.csv', mode='w', newline='', encoding='utf-8') as archivo:
    writer = csv.DictWriter(archivo, fieldnames=nombres_columnas,extrasaction="ignore")
    writer.writeheader()  # Genera: nombre,edad
    writer.writerows(datos)
```

### Uso de `extrasaction` con valor `raise`

```python
import csv

nombres_columnas = ['nombre', 'edad']

datos = [
    {'nombre': 'Ana', 'edad': 30, 'ciudad': 'Madrid'},
    {'nombre': 'Carlos', 'edad': 25, 'ciudad': 'Barcelona'},
    {'nombre': 'Elena', 'edad': 35, 'ciudad': 'Valencia'}
]

try:
    with open('recursos/personas.csv', mode='w', newline='', encoding='utf-8') as archivo:
        writer = csv.DictWriter(archivo, fieldnames=nombres_columnas,extrasaction="raise")
        writer.writeheader()  # Genera: nombre,edad
        writer.writerows(datos)
except ValueError as e:
    print(f"Error: {e}")
```

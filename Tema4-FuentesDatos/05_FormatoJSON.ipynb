# JSON
___
Para trabajar con datos JSON en Python se utiliza el módulo incorporado `json`. A diferencia de otros lenguajes orientados a objetos, en Python el JSON se manipula principalmente a través de funciones y de la correspondencia directa con tipos de datos nativos (como diccionarios y listas).

La función principal del módulo integrado json de Python es serializar y deserializar datos entre estructuras nativas de Python y el formato estándar JSON (JavaScript Object Notation). Esto permite intercambiar información fácilmente con APIs web o guardar configuraciones en archivos de texto legible.

#### Relación de datos entre JSON y Python
| Tipo en JSON | Tipo en Python |
| :--- | :--- |
| `object` | `dict` |
| `array` | `list` |
| `string` | `str` |
| `number (int)` | `int` |
| `number (real)` | `float` |
| `true` | `True` |
| `false` | `False` |
| `null` | `None` |



## Métodos
El módulo realiza dos procesos fundamentales mediante funciones clave:
* **Serialización (Python a JSON)**:  Convierte diccionarios, listas u otros tipos básicos de Python en cadenas de texto o archivos con formato JSON.
* **Deserialización (JSON a Python)**: Lee texto o archivos en formato JSON y los transforma de vuelta en objetos manipulables de Python.

| Función | Dirección de conversión | Uso principal |
| :--- | :--- | :--- |
| **`json.loads()`** | Cadena `str` → Python | Analiza un texto JSON y lo devuelve como un diccionario o lista de Python. |
| **`json.load()`** | Archivo `.json` → Python | Lee un archivo JSON desde el disco y lo carga como una estructura de Python. |
| **`json.dumps()`** | Python → Cadena `str` | Convierte un diccionario o lista en un texto JSON (ideal para enviar datos por red). |
| **`json.dump()`** | Python → Archivo `.json` | Escribe directamente los datos de Python dentro de un archivo abierto en disco. |

## Método `json.loads()`
Analiza un texto (str) JSON y lo devuelve como un diccionario o lista de Python.

```python
import json

# Cadena que se compone de una LISTA DE DICCIONARIOS. Empieza por corchetes por lo que a primer nivel es una lista (una lista es un json).
personas_json = """[
  {
    "dni": "12345678A",
    "nombre": "Alejandro García",
    "edad": 28,
    "ciudad": "Madrid",
    "telefono": "600123456"
  },
  {
    "dni": "87654321B",
    "nombre": "María Rodríguez",
    "edad": 34,
    "ciudad": "Barcelona",
    "telefono": "611234567"
  },
  {
    "dni": "23456789C",
    "nombre": "Carlos Martínez",
    "edad": 42,
    "ciudad": "Valencia",
    "telefono": "622345678"
  }
]"""

print(type(personas_json))
print(personas_json)
objeto_json = json.loads(personas_json)
print(type(objeto_json))
print(objeto_json)

print("="*150)

# Cadena que se compone de un DICCIONARIO: 
vehiculo_json = '''
{
  "vehiculo": {
    "marca": "Toyota",
    "modelo": "RAV4",
    "anio": 2024,
    "tipo_carroceria": "SUV",
    "color": "Gris Grafito",
    "kilometraje": 15200,
    "especificaciones_tecnicas": {
      "motor": "2.5L 4 cilindros",
      "transmision": "Automática eCVT",
      "tipo_combustible": "Híbrido",
      "potencia_hp": 219,
      "traccion": "AWD (Tracción integral)"
    },
    "equipamiento": {
      "techo_solar": true,
      "asientos_cuero": true,
      "conectividad": ["Apple CarPlay", "Android Auto", "Bluetooth"],
      "camara_reversa": true
    },
    "estado_legal": {
      "matricula": "ABC-1234",
      "seguro_vigente": true,
      "ultimo_mantenimiento": "2026-08-15"
    }
  }
}'''

print(type(vehiculo_json))
print(vehiculo_json)
objeto_json = json.loads(vehiculo_json)
print(type(objeto_json))
print(objeto_json)
```

## Método `json.dumps()`
Convierte un diccionario o lista en un texto JSON.

```python
import json

deportes_diccionario = {
    "futbol": {
        "nombre": "Fútbol",
        "categoria": "Deporte de equipo",
        "jugadores_por_equipo": 11,
        "equipamiento": ["Balón", "Porterías", "Botas de tacos", "Espinilleras"],
        "es_olimpico": True
    },
    "baloncesto": {
        "nombre": "Baloncesto",
        "categoria": "Deporte de equipo",
        "jugadores_por_equipo": 5,
        "equipamiento": ["Balón de baloncesto", "Canastas", "Zapatillas de pista"],
        "es_olimpico": True
    },
    "tenis": {
        "nombre": "Tenis",
        "categoria": "Deporte individual o parejas",
        "jugadores_por_equipo": 1,
        "equipamiento": ["Raqueta", "Pelotas de tenis", "Red"],
        "es_olimpico": True
    },
    "natacion": {
        "nombre": "Natación",
        "categoria": "Deporte acuático",
        "jugadores_por_equipo": 1,
        "equipamiento": ["Bañador", "Gafas de natación", "Gorro"],
        "es_olimpico": True
    },
    "padel": {
        "nombre": "Pádel",
        "categoria": "Deporte de parejas",
        "jugadores_por_equipo": 2,
        "equipamiento": ["Pala de pádel", "Pelotas", "Red"],
        "es_olimpico": False
    }
}

print("SIN PARÁMETROS")
deportes_json = json.dumps(deportes_diccionario)
print(type(deportes_json))
print(deportes_json)

print("="*150)

# ensure_ascii
print("ENSURE_ACSII")
deportes_json = json.dumps(deportes_diccionario,ensure_ascii=False)
print(deportes_json)

print("="*150)

# INDENT
print("INDENT")
deportes_json = json.dumps(deportes_diccionario,ensure_ascii=False,indent=8)
print(deportes_json)

print("="*150)

# separators
print("SEPARATORS")
deportes_json = json.dumps(deportes_diccionario,ensure_ascii=False,separators=('  ,  ','  :  ')) # Separador de elementos y de claves
print(deportes_json)

print("="*150)

# sort_keys
print("SORT_KEYS")
deportes_json = json.dumps(deportes_diccionario,ensure_ascii=False,sort_keys=True,indent=4) # Ordena las claves
print(deportes_json)

print("="*150)

# 
print("SKIPKEYS")
datos_torneo = {
    "deporte": "Tenis",
    ("Rafa", "Carlos"): "Dobles",  # La tupla no es una clave válida en JSON
    1: "Pista Central"             # Los enteros sí se convierten a texto ("1")
}
torneo_json = json.dumps(datos_torneo,ensure_ascii=False,skipkeys=True) # Cambiar a False para comprobar que da error
print(torneo_json)
```

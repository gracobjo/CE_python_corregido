# Módulo `os`
El módulo os de Python permite interactuar con el sistema operativo en el que se está ejecutando el código. Viene integrado en Python (no necesita instalación) y es fundamental para realizar tareas de gestión de archivos, carpetas y configuraciones del sistema de forma automática.

* `os.name`: Te dice en qué sistema estás ('posix' para Linux/Mac o 'nt' para Windows).
* `os.environ`: Un diccionario con todas las variables de entorno del sistema.
* `os.system('comando')`: Te permite ejecutar comandos directamente en la terminal del sistema desde el código Python. Se emplea el módulo `subprocess` especializado más en ejecuciones de comandos.
* `os.getpid()`: Obtiene el ID del proceso (PID) actual de Python.
* `os.getppid()`: Obtiene el ID del proceso padre (Parent PID).
* `os.getcwd()`: Devuelve la ruta de la carpeta actual donde estás trabajando.
* `os.listdir(path='.')`: Lista los archivos y carpetas dentro del directorio indicado.
* `os.chdir('ruta')`: Cambia la carpeta de trabajo actual a la ruta que le indiques.
* `os.walk`: Recorre de forma recursiva un arbol de directorios. **Es un generador**. Comúnmente se usa en un bucle **for**.
* `os.mkdir('nombre')`: Crea una nueva carpeta.
* `os.rmdir('carpeta')`: Elimina una carpeta (debe estar vacía).
* `os.makedirs('ruta/nueva')`: Crea varias carpetas anidadas de golpe (como carpetas dentro de otras carpetas).
* `os.removedirs(path)`: Elimina carpetas vacías de forma recursiva hacia atrás.
* `os.stat(path)`: Obtiene metadatos del archivo (tamaño, fecha de modificación, permisos, etc.).
* `os.remove('archivo.txt')`: Elimina un archivo.
* `os.rename(src, dst)`: Renombra o mueve un archivo o carpeta de una ruta origen (src) a una destino (dst).
* `os.replace(src, dst)`: Renombra o mueve un archivo forzando el reemplazo si el destino ya existe (comportamiento multiplataforma más seguro que rename).

Excepciones comunes:
* `FileNotFoundError`: Si la ruta especificada no existe.
* `NotADirectoryError`: Si apuntas a un archivo en lugar de a una carpeta.
* `PermissionError`: Si tu script no tiene permisos de lectura para ese directorio.

## Recorrido recursivo de una carpeta `walk`
Se utiliza para recorrer de forma recursiva un árbol de directorios (carpetas y subcarpetas) a partir de una ruta inicial. Es un **generador** que recorre un árbol de directorios de forma recursiva, produciendo una tupla de tres elementos (dirpath, dirnames, filenames) por cada carpeta que visita.
Genera una tupla de tres elementos:
* **raíz**: Ruta actual que se está analizando
* **directorios**: Lista con las subcarpeta que hay dentro de `raíz`.
* **ficheros**: Lista con los ficheros que hay dentro de `raíz`.

```python
import os

# Se crea el generador para el directorio actual '.'
generador_carpetas = os.walk('.')

try:
    print("--- (Directorio inicial) ---")
    # Consumimos la primera iteración del generador usando next()
    raiz, carpetas, archivos = next(generador_carpetas)
    print(f"Examinando: {raiz}")
    print(f"Subcarpetas encontradas aquí: {carpetas}")
    print(f"Archivos encontrados aquí: {archivos}\n")

    print("--- (Entrando en la primera subcarpeta) ---")
    # Al llamar a next() de nuevo, avanza automáticamente a la siguiente subcarpeta del árbol
    raiz_sub, carpetas_sub, archivos_sub = next(generador_carpetas)
    print(f"Examinando subnivel: {raiz_sub}")
    print(f"Subcarpetas en este subnivel: {carpetas_sub}")
    print(f"Archivos en este subnivel: {archivos_sub}\n")

    print("--- (Entrando en la segunda subcarpeta) ---")
    # Al llamar a next() de nuevo, avanza automáticamente a la siguiente subcarpeta del árbol
    raiz_sub, carpetas_sub, archivos_sub = next(generador_carpetas)
    print(f"Examinando subnivel: {raiz_sub}")
    print(f"Subcarpetas en este subnivel: {carpetas_sub}")
    print(f"Archivos en este subnivel: {archivos_sub}")

except StopIteration:
    print("El directorio no tiene más subcarpetas para explorar.")

```

## Información de archivos

`stat`

Información que devuelve:
* `st_size`: El tamaño del archivo en bytes.
* `st_mtime`: La fecha y hora de la última modificación (en formato timestamp de Unix).
* `st_atime`: La fecha y hora del último acceso al archivo.
* `st_ctime`: En Linux/Unix es la fecha del último cambio en los metadatos; en Windows es la fecha de creación del archivo.
* `st_mode`: Los permisos del archivo y el tipo de archivo (representados como un número entero). Hay que pasarlo a octal para comprobar los permisos.
* `st_uid` / `st_gid`: El ID del usuario y del grupo propietario del archivo (más relevante en sistemas Unix/Linux ya que en Windows será 0 siempre).

```python
import os
from datetime import datetime

# Definimos la ruta de un archivo de prueba que tiene que existir previamente en el directorio actual
ruta_archivo = "recursos/ejemplo.txt"

try:
    info_archivo = os.stat(ruta_archivo)
    
    tamaño = info_archivo.st_size
    
    # Se convierten los timestamps a fechas legibles
    fecha_modificacion = datetime.fromtimestamp(info_archivo.st_mtime)
    fecha_acceso = datetime.fromtimestamp(info_archivo.st_atime)
    
    # En Windows st_ctime es la creación, en Linux es el cambio de metadatos
    fecha_cambio = datetime.fromtimestamp(info_archivo.st_ctime)

    # 4. Mostrar los resultados en pantalla
    print(f"--- Metadatos de: {ruta_archivo} ---")
    print(f"Tamaño: {tamaño} bytes")
    print(f"Última modificación: {fecha_modificacion}")
    print(f"Último acceso: {fecha_acceso}")
    print(f"Creación / Cambio de metadatos: {fecha_cambio}")
    print(f"Permisos (st_mode): {info_archivo.st_mode} y en octal {oct(info_archivo.st_mode)}")
    print(f"Propietario (st_uid): {info_archivo.st_uid}")

except FileNotFoundError:
    print("El archivo especificado no existe.")
```

## El submódulo os.path

Es uno de los componentes más utilizados de os para gestionar rutas de archivos de forma segura.
* `os.path.split(path)`: Divide una ruta en una tupla (directorio, nombre_archivo).
* `os.path.splitext(path)`: Obtiene el nombre y la extensión de un archivo.
* `os.path.join(path, *paths)`: Une varios componentes de ruta usando el separador correcto (/ o \\) según el sistema operativo.
* `os.path.abspath(path)`: Devuelve la ruta absoluta de un archivo o carpeta.
* `os.path.exists(path)`: Verifica si una ruta existe (ya sea archivo o carpeta). Devuelve True o False.
* `os.path.isfile(path)`: Devuelve True si la ruta apunta a un archivo.
* `os.path.isdir(path)`: Devuelve True si la ruta apunta a una carpeta.
* `os.path.getsize(path)`: Devuelve el tamaño del archivo en bytes.

```python
import os

# split
ruta = "C:\\Users\\Jesus\\Dropbox\\Docencia\\Python/Temario/Cuadernos/Tema4/recursos/fichero01.txt"
print(f"Carpeta actual (split): {ruta}")
carpeta,archivo = os.path.split(ruta)
print(f"Carpeta: {carpeta}")
print(f"Archivo: {archivo}")

print("="*100)

# splitext
archivo = "documento.pdf"
print(f"Carpeta actual (split): {ruta}")
nombre,extension = os.path.splitext(archivo)
print(f"Nombre: {nombre}")
print(f"Extensión: {extension}")

print("="*100)

# join
print(f"Carpeta actual (join): {os.path.join(carpeta,archivo)}")

print("="*100)

# abspath
print(f"Ruta absoluta (abspath): {os.path.abspath("./recursos/fichero01.txt")}")

print("="*100)

# exists
if os.path.exists("./recursos/fichero01.txt"):
    print(f"Sí existe dicha ruta (exists): {os.path.abspath("./recursos/fichero01.txt")}")
else:
    print("No existe el fichero")

print("="*100)

# isfile
ruta = "C:\\Users\\Jesus\\Dropbox\\Docencia\\Python/Temario/Cuadernos/Tema4-FuentesDatos/recursos/fichero01.txt"
if os.path.isfile(ruta):
    print(f"Es un fichero y ocupa {os.path.getsize(ruta)}")
else:
    print("No es un fichero")

print("="*100)

# isdir
if os.path.isdir("C:\\Users\\Jesus\\Dropbox\\Docencia\\Python/Temario/Cuadernos/Tema4-FuentesDatos/recursos/fichero01.txt"):
    print("Es un directorio")
else:
    print("No es un directorio")

# getsize
tam = os.path.getsize("C:\\Users\\Jesus\\Dropbox\\Docencia\\Python/Temario/Cuadernos/Tema4-FuentesDatos/recursos/fichero01.txt")
print(f"Tiene un tamaño de {tam} bytes")
```

# Módulo subprocess

Método `subprocess.run`

Parámetros:
* `capture_output`: Booleano. Por defecto, la salida del comando se imprime directamente en la consola de tu script. Si se pone a `True` esta opción, Python interceptará la salida y la guardará en los atributos `.stdout` y `.stderr` del objeto devuelto.
* `text`: Booleano. Por defecto, las salidas capturadas (stdout y stderr) se devuelven como objetos de tipo bytes (b'...'). Si se pone a `True` esta opción, Python las decodificará automáticamente a cadenas de texto (``) utilizando la codificación del sistema (usualmente UTF-8).
* `check`: Booleano. Si el comando falla (devuelve un `returncode` distinto de cero), Python no lanzará ninguna excepción por defecto; simplemente continuará con el script. Si se pone a `True`, Python lanzará una excepción del tipo `CalledProcessError` si el comando falla, lo cual es ideal para manejar errores de forma controlada.
* `shell`: Booleano. Permite ejecutar el comando directamente a través del shell del sistema (como bash en Linux/macOS o cmd.exe/PowerShell en Windows). La ventaja es que permite usar características del shell como tuberías (|), comodines (*) o variables de entorno. Puede abrir la puerta a ataques de inyección de código. Siempre que sea posible, se debe poner a `False`. Si se ejecutan comando internos de la shell debe ser `True`.
* `timeout`: Permite definir un tiempo límite en segundos para la ejecución del comando. Si el proceso tarda más de ese tiempo, se interrumpe y lanza una excepción `TimeoutExpired`.

```python
import subprocess

# Ejecutar mediante cmd.exe
resultado_cmd = subprocess.run(["dir"],capture_output=True,text=True,shell=True)
print("Salida del comando de CMD:")
print(resultado_cmd.stdout)

# Mostrar errores si los hubiera
if resultado_cmd.stderr:
    print("Error:")
    print(resultado_cmd.stderr)
    
# Ejecutar mediante powershell.exe
resultado_ps = subprocess.run(["powershell.exe", "-Command", "Get-LocalUser"],capture_output=True,text=True)
print("Salida del comando de PowerShell:")
print(resultado_ps.stdout)

# Mostrar errores si los hubiera
if resultado_ps.stderr:
    print("Error:")
    print(resultado_ps.stderr)

```

## Apertura de ficheros
Se emplea el método `open`. Recibe dos parámetros:
* **Fichero**: Ruta y nombre del fichero.
* **Modo**: Hay 4 tipos de modos de apertura:
  * **r**: Lectura (read). Es el valor por defecto y abre el fichero para leerlo, dando error si no existe.
  * **a**: Añadidura (append). Abre el fichero para añadir al final, creándolo si no existe.
  * **w**: Escritura (write). Abre el fichero para escribir creándolo si no existe.
  * **x**: Creación (create). Crea el fichero devolviendo error si el fichero existe.
  * **+**: Actualización (lectura y escrituras combinadas `r+`, `w+`)

Adicionalmente se puede especificar cómo queremos tratar el fichero:
* **t**: Modo texto.
* **b**: Modo binario.

Otros parámetros: 
* **encoding**: (Opcional, por defecto None): Nombre de la codificación usada para decodificar o codificar el archivo (ej. 'utf-8', 'latin-1'). Solo se utiliza en modo texto
* **buffering**: (Opcional, por defecto -1): Entero que fija la política de almacenamiento en búfer. El almacenamiento en búfer es una técnica de optimización: en lugar de escribir o leer datos del disco duro (operación que es muy lenta) byte por byte, los datos se acumulan temporalmente en la memoria RAM (el búfer) y se transfieren en bloques grandes, lo que acelera enormemente el rendimiento.:
    * **0**: Sin búfer (solo permitido en modo binario).
    * **1**: Búfer de línea (solo en modo texto).
    * **Negativo (-1)**: Usa el tamaño de búfer por defecto del sistema operativo (modos binario y texto).
    * **Otro valor**: Tamaño del buffer (modos binario y texto).
* **newline**: (Opcional, por defecto None): Controla cómo se manejan los finales de línea en modo texto (puede ser None, '', '\n', '\r', o '\r\n').


## Expresión `With`

Se utiliza para la gestión de contexto con objeto de simplificar el manejo de recursos como los ficheros (bases de datos, etc.), asegurando que se cierren de forma automática y correcta una vez que se han terminado de usar, incluso si ocurren errores o excepciones durante el proceso. A este mecanismo se le conoce técnicamente como un Gestor de Contexto (Context Manager).

**Sintaxis básica**

```Python
with open("nombre_fichero", "r") as fichero:
    ## Acceder al fichero a través de la variable fichero
```

Se pueden especificar varios ficheros a la vez separándolos por comas.
```Python
with open("origen.txt", "r") as origen, open("destino.txt", "w") as destino:
    contenido = origen.read()
    destino.write(contenido)
```
El bloque with funciona los métodos especiales que tienen los objetos en Python (llamados "métodos dunder"):
1. __enter__(): Se ejecuta al entrar al bloque with. En el caso de los ficheros, este método los abre y te los entrega en la variable después del `as`.
2. __exit__(): Se ejecuta obligatoriamente al salir del bloque (ya sea porque el código terminó con éxito o porque saltó un error/excepción). En los ficheros, este método ejecuta internamente el cierre del archivo.

````python
# Abrir fichero normal sin with
archivo = open("recursos/fichero01.txt", "r",encoding="utf-8")
contenido = archivo.read()
archivo.close()  # Si ocurre un error antes, esto no se ejecuta
print("ARCHIVO\n",contenido)

print("="*100)

# Se abre el fichero con with
with open("recursos/fichero01.txt","r") as archivo: # No se pone encoding por lo que las tildes, ñ, etc no se verán correctamente
    contenido = archivo.read()
    print("ARCHIVO\n",contenido)
# El archivo ya está cerrado aquí de forma segura
````

## Información de un archivo
`closed`, `mode`, `name`, `encoding`

````python
with open("recursos/fichero01.txt", "w", encoding="utf-8") as archivo:
    print(archivo.name)
    print(archivo.mode)
    print(archivo.closed)
    print(archivo.encoding)
````

**Para leer**

````python
# Se lee un archivo sin especificar la codificación
with open("recursos/fichero02.txt","r") as archivo: 
    contenido = archivo.read()
    print("ARCHIVO\n",contenido)

with open("recursos/fichero02.txt","r",encoding="ascii") as archivo: 
    #contenido = archivo.read() # Error: Intenta leer un carácter con tilde
    #print("ARCHIVO\n",contenido)
    pass

# Se lee un archivo especificando la codificación, utf-8 en este caso
with open("recursos/fichero02.txt","r",encoding="utf-8") as archivo:
    contenido = archivo.read()
    print("ARCHIVO\n",contenido)
````

**Para escribir**

````python
# Asegurarse que no existen los ficheros que se van a crear

try:
    # Sin especificar codificación
    with open("recursos/fichero51.txt","w") as archivo:
        archivo.write("Python es un lenguaje de programación de alto nivel, interpretado, de propósito general y multiplataforma.")

    # utf-8
    with open("recursos/fichero52.txt","w",encoding="utf-8") as archivo:
        archivo.write("Python es un lenguaje de programación de alto nivel, interpretado, de propósito general y multiplataforma.")

    # ascii con tildes: genera un error
    with open("recursos/fichero53.txt","w",encoding="ascii") as archivo:
        archivo.write("Python es un lenguaje de programación de alto nivel, interpretado, de propósito general y multiplataforma.")

except UnicodeEncodeError as e:
    print(f"Error: {e}")
````

## Leer: `read`

````python
# Usar fichero02.txt

with open("recursos/fichero02.txt","r",encoding="utf-8") as archivo:
    contenido = archivo.read() # Se lee todo el contenido
    print("ARCHIVO 1\n",contenido)

print("="*150)

with open("recursos/fichero02.txt","r",encoding="utf-8") as archivo:
    contenido = archivo.read(-1) # Se lee todo el contenido
    print("ARCHIVO 2\n",contenido)

print("="*150)

with open("recursos/fichero02.txt","r",encoding="utf-8") as archivo:
    contenido = archivo.read(15) # Se leen 15 caracteres
    print("ARCHIVO 3\n",contenido)

print("="*150)

with open("recursos/fichero02.txt","r",encoding="utf-8") as archivo:
    print("ARCHIVO 4")
    while True:
        caracter = archivo.read(1) # Se lee de 1 en 1 hasta el final del fichero
        if not caracter:
            break
        print(caracter,end="")
````

## Leer: `readline`

````python
# Usar fichero01.txt

with open("recursos/fichero01.txt","r",encoding="utf-8") as archivo:
    contenido = archivo.readline() # Se lee una línea
    print("ARCHIVO 1\n",contenido)

print("="*150)

with open("recursos/fichero01.txt","r",encoding="utf-8") as archivo:
    print("ARCHIVO 2")
    while True:
        caracter = archivo.readline() # Se lee de 1 en 1 línea hasta el final del fichero
        if not caracter:
            break
        print(caracter,end="")
````

## Leer: `readlines`

* Lee las líneas del archivo y las devuelve como una **lista de cadenas**.

* Sin parámetros (0 o negativo) lee todo el archivo.

* Lee líneas completas hasta la suma total de bytes leidos sin cortar líneas a la mitad.

````python
# Usar fichero01.txt

with open("recursos/fichero01.txt","r",encoding="utf-8") as archivo:
    contenido = archivo.readlines(10) # Se lee una línea. Leería 10 bytes de la primera línea pero como no la corta la lee entera.
    print("Ejecución 1\n",contenido)

with open("recursos/fichero01.txt","r",encoding="utf-8") as archivo:
    contenido = archivo.readlines(60) # Se lee una línea. Leería 150 bytes que es más de la primera por lo que lee las dos primeras
    print("Ejecución 2\n",contenido)
````

## Lectura por líneas con bucle for

````python
with open("recursos/fichero01.txt","r",encoding="utf-8") as archivo:
    for linea in archivo:
        print(linea,end="")
````

## Escribir: `write`

* Escribe una cadena en un archivo

````python
# Usar fichero31.txt
with open("recursos/fichero31.txt", "w", encoding="utf-8") as archivo:
    bytes_escritos = archivo.write("Hola, mundo!\n¿Qué tal todo?")
    print(f"Se escribieron {bytes_escritos} caracteres.")

# Usar fichero32.txt. No escribe los saltos de línea automàticamente
with open("recursos/fichero32.txt", "w", encoding="utf-8") as archivo:
    archivo.write("linea1")
    archivo.write("linea2")
    archivo.write("linea3")

# Hay que indicar el \n
with open("recursos/fichero33.txt", "w", encoding="utf-8") as archivo:
    archivo.write("linea1\n")
    archivo.write("linea2\n")
    archivo.write("linea3\n")
````

## Escribir: `writelines`

Escribe una lista o tupla en un fichero en forma de líneas.

**IMPORTANTE**: No añade los saltos de página

````python
# Usar fichero34.txt
lineas_sin_saltos = ["Esta es la primera linea","Esta es la segunda linea","Esta es la tercera linea"]

with open("recursos/fichero34.txt", "w", encoding="utf-8") as archivo:
    archivo.writelines(lineas_sin_saltos)

# Usar fichero35.txt
lineas_con_saltos = ["Esta es la cuarta linea\n","Esta es la quinta linea\n","Esta es la sexta linea\n"]
with open("recursos/fichero35.txt", "w", encoding="utf-8") as archivo:
    archivo.writelines(lineas_con_saltos)
````

## Cerrar un archivo `close`

Si no usamos `with` hay que cerrar el archivo.

````python
archivo = open("recursos/fichero01.txt", "r",encoding="utf-8")
contenido = archivo.read()
archivo.close()  # Si ocurre un error antes, esto no se ejecuta
print("ARCHIVO\n",contenido)
````

## Puntero de lectura/escritura

Para leer y establecer la posición del puntero de lectura/escritura usamos los métodos `tell`y `seek`.
* **tell**: Sin parámetros.
* **seek**:
    * **offset**: Desplazamiento.
    * **whence**: Desde dónde desplaza: 0 inicio del archivo (por defecto), 1 desde la posición actual y 2 desde el final de archivo

````python
with open("recursos/fichero01.txt", "r") as archivo:
    print(archivo.tell())
    contenido = archivo.read(5)
    print(contenido)
    print(archivo.tell())

    archivo.seek(10,0)
    contenido = archivo.read(5)
    print(contenido)

    archivo.seek(10,1) # Da error. En utf-8 un carácter puede ocupar más de un byte
    contenido = archivo.read(5)
    print(contenido)

    archivo.seek(0,2) # Va al final
    contenido = archivo.read(5)
    print(contenido)
````

## Escritura en una posición concreta

````python
# Se crea un fichero de ejemplo con contenido inicial
with open("recursos/ejemplo.txt", "w", encoding="utf-8") as archivo:
    archivo.write("Hola, mundo! Esto es una prueba.")

# Se abre el fichero en modo 'r+' (lectura y escritura sin borrar el contenido)
with open("recursos/ejemplo.txt", "r+", encoding="utf-8") as archivo:
    # Nos movemos a la posición 6 (justo después de "Hola, ")
    archivo.seek(6)
    
    # Escribimos el nuevo texto en esa posición (sobrescribirá lo que había)
    archivo.write("Python")

# Se verifica el resultado leyendo el archivo completo
with open("recursos/ejemplo.txt", "r", encoding="utf-8") as archivo:
    contenido = archivo.read()
    print("Contenido final:", contenido)
````

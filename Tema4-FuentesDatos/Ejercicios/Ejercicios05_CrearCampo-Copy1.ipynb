# Ejercicios 05
Crear un campo de categoría a partir de edad para posible aplicación de descuento.

```python
import csv

with open("../recursos/datos.csv","r",encoding="utf-8") as archivo, open("../recursos/datos_campo_nuevo.csv","w",encoding="utf-8") as fsalida:
    lector = csv.DictReader(archivo)
    
    escritor = csv.DictWriter(fsalida, fieldnames=["dni","nombre","edad","ciudad","telefono","descuento"])
    escritor.writeheader()
    
    for fila in lector:
        if int(fila["edad"]) < 45:
            fila["descuento"] = "Joven"
        elif int(fila["edad"]) < 65:
            fila["descuento"] = "Adulto"
        else:
            fila["descuento"] = "Jubilado"
        escritor.writerow(fila)
```

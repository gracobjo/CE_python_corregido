# Ejercicios 02
A partir de una carpeta donde nos encontramos distintos ficheros con distintos tamaños y nombre, se pide:
* Una lista con los nombres de ficheros ordenados por tamaño.
* Una lista por cada tipo de fichero de acuerdo a su extensión.

```python
import os

tam = []
ruta = "../recursos"
lista = os.listdir(path='../recursos')
absoluta = os.path.abspath(ruta)
for i in lista:
    info = os.stat(os.path.join(absoluta,i))
    tam.append(info.st_size)

sin_ordenar = list(zip(lista,tam))
ordenada = [t for t in sorted(sin_ordenar, key=lambda x: x[1])]

lista_final = []
for x,y in ordenada:
    lista_final.append(x)

print(lista_final)

print("="*150)
     
ruta = "../recursos"
lista = os.listdir(path='../recursos')
absoluta = os.path.abspath(ruta)

conjunto_extensiones = set()

for i in lista:
    nombre_completo = os.path.join(absoluta,i)
    if os.path.isfile(nombre_completo):
        n,e = os.path.splitext(i)
        conjunto_extensiones.add(e)
print(conjunto_extensiones)


listas = [];
num = 0
for i in conjunto_extensiones:
    listas.append([])
    for n in lista:
        nombre_completo = os.path.join(absoluta,n)
        if os.path.isfile(nombre_completo):
            nom,ext = os.path.splitext(n)
            if ext == i:
                listas[num].append(n)
    num += 1
        
print(listas)      
```

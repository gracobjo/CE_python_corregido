A partir del ejercicio anterior, ordenar el filtro al hacer clic en la cabecera de cada campo.

```python
import csv
import tkinter as tk
from tkinter import ttk
from tkinter import messagebox

campo_orden = -1
filas_lista = []

filtro = {
    "edad":[40,60],
    "ciudad":["Sevilla","Bilbao","Madrid","Barcelona"]
}

def actualizar_control(campo):
    global filas_lista, campo_orden

    if campo != -1:
        if campo_orden == campo:
            filas_lista.reverse()
        else:
            filas_lista = sorted(filas_lista, key=lambda x: x[campo])

    lista.delete(*lista.get_children())
       
    for fila in filas_lista:
        lista.insert("", tk.END, values=fila)

    campo_orden = campo


ventana = tk.Tk()
ventana.title("Widget Treeview como lista")
ventana.geometry("600x400")

columnas = ["dni", "nombre", "edad", "ciudad","telefono"]

lista = ttk.Treeview(ventana, columns=columnas, show="headings", displaycolumns="#all", selectmode="browse")

with open("../recursos/datos.csv","r",encoding="utf-8") as archivo:#, open("../recursos/datos_filtrado.csv","w",encoding="utf-8") as fsalida:
    lector = csv.DictReader(archivo)

    for fila in lector:
        pasa_filtro = True
        if len(filtro["edad"]) == 1:
            if int(filtro["edad"][0]) != int(fila["edad"]):
                pasa_filtro = False
        if len(filtro["edad"]) == 2:
            if int(filtro["edad"][0]) > int(fila["edad"]) or int(filtro["edad"][1]) < int(fila["edad"]):
                pasa_filtro = False

        for ciudad in filtro["ciudad"]:
            if fila["ciudad"] not in filtro["ciudad"]:
                pasa_filtro = False
                
        if pasa_filtro:
            filas_lista.append(list(fila.values()))
            lista.insert("", tk.END, values=list(fila.values()))


# Se configuran los encabezados (headings) de las columnas
lista.heading("dni", text="DNI",command=lambda: actualizar_control(0))
lista.heading("nombre", text="Nombre",command=lambda: actualizar_control(1))
lista.heading("edad", text="Edad",command=lambda: actualizar_control(2))
lista.heading("ciudad", text="Ciudad",command=lambda: actualizar_control(3))
lista.heading("telefono", text="Teléfono",command=lambda: actualizar_control(4))

# Se configura el ancho y la alineación de las columnas
lista.column("dni", width=50, anchor="center")
lista.column("nombre", width=150, anchor="w")
lista.column("edad", width=80, anchor="center")
lista.column("ciudad", width=150, anchor="w")
lista.column("telefono", width=150, anchor="w")


def guardar():
    with open("../recursos/datos_filtrado.csv","w",encoding="utf-8") as fsalida:
        escritor = csv.writer(fsalida)
        escritor.writerow(["dni","nombre","edad","ciudad","telefono"])

        for fila in filas_lista:
            escritor.writerow(fila)


lista.pack(fill=tk.BOTH, expand=True, padx=10, pady=10)

boton = tk.Button(ventana, text="Guardar", command=guardar)
boton.pack(pady=10)

ventana.mainloop()
```

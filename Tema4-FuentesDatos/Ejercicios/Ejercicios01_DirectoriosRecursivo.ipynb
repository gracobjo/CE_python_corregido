# Ejercicios 01
Imprimir por pantalla de forma recursiva todos los archivos y carpetas que se encuentran en el directorio actual.

```python
import os

for directorio_actual, directorios, ficheros in os.walk(".."):
    print(f"[DIR] {os.path.abspath(directorio_actual)}")

    for directorio in directorios:
        ruta = os.path.join(directorio_actual, directorio)
        print(f"    [D] {ruta}")

    for fichero in ficheros:
        ruta = os.path.join(directorio_actual, fichero)
        print(f"    [F] {ruta}")
```

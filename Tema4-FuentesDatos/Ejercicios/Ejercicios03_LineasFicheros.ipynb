# Ejercicios 03
A partir de una lista de nombres de ficheros y una lista de números de líneas, se pide crear un fichero que contenga la líneas correspondientes a los números de línea de la lista de todos los ficheros de la lista de ficheros.

```python
import os

lista_archivos = ["fichero01.txt","fichero03.txt","fichero02.txt"]
lista_numeros = [2,6,4]

absoluta = os.path.abspath("../recursos")

lineas_finales = []

for i in lista_archivos:
    nombre_completo = os.path.join(absoluta,i)
    if os.path.isfile(nombre_completo):
        with open(nombre_completo,"r",encoding="utf-8") as archivo:
            lineas = archivo.readlines(10000) # En lineas tienes una lista de líneas
            for lin in lista_numeros:
                lineas_finales.append(lineas[lin-1])

with open("../recursos/fichero_lineas.txt","w",encoding="utf-8") as archivo:
    archivo.writelines(lineas_finales)
```

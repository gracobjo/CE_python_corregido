# Ejercicios 06
Crear un filtro para quedarnos las filas que cumplan las condiciones del filtro. Sólo con **edad** y **ciudad**.
* `edad`: Puede:
    * Estar vacía: No se tiene en cuenta.
    * Tener un único valor ej: \[45\]: Las personas que tengan exactamente 45.
    * Tener dos valores ej: \[36,52\]: Las personas que tengan entre 36 y 52

* `ciudad`: Puede:
    * Estar vacía: No se tiene en cuenta.
    * Lista de valores: Las personas que sean de alguna de esas ciudades.

```python
import csv

filtro = {
    "edad":[40,70],
    "ciudad":["Valladolid","Madrid","Oviedo","Barcelona"]
}

with open("../recursos/datos.csv","r",encoding="utf-8") as archivo, open("../recursos/datos_filtrado.csv","w",encoding="utf-8") as fsalida:
    lector = csv.DictReader(archivo)
    
    escritor = csv.DictWriter(fsalida, fieldnames=["dni","nombre","edad","ciudad","telefono"])
    escritor.writeheader()

    for fila in lector:
        pasa_filtro = True
        
        if len(filtro["edad"]) == 1:
            if int(filtro["edad"][0]) != int(fila["edad"]):
                pasa_filtro = False
        
        if len(filtro["edad"]) == 2:
            if int(filtro["edad"][0]) > int(fila["edad"]) or int(filtro["edad"][1]) < int(fila["edad"]):
                pasa_filtro = False

        for ciudad in filtro["ciudad"]:
            if fila["ciudad"] not in filtro["ciudad"]:
                pasa_filtro = False
                
        if pasa_filtro:
            escritor.writerow(fila)
```

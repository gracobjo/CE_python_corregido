# Ejercicios 04
Ordenar un archivo cvs a partir de alguno de sus campos.

### Solución 1
* Usando sorted y una función lambda
* Más elegante y sencillo

```python
import csv

campo = "edad" # se asigna una de los que hay
filas_sin_ordenar = []
filas_ordenadas = []

with open("../recursos/datos.csv","r",encoding="utf-8") as archivo:
    lector = csv.DictReader(archivo)
    for fila in lector:
        filas_sin_ordenar.append(fila)

filas_ordenadas = sorted(filas_sin_ordenar, key=lambda x: x[campo])

# ORDENA DE MAYOR A MENOR. LE DAMOS LA VUELTA SI ASÍ SE REQUIERE
# filas_ordenadas.reverse()

with open("../recursos/datos_ordenados.csv","w",encoding="utf-8") as archivo:
    writer = csv.DictWriter(archivo, fieldnames=["dni","nombre","edad","ciudad","telefono"])
    writer.writeheader()
    writer.writerows(filas_ordenadas)
```

### Solución 2
* Usando bucles for anidados para recorrer la lista original y la que se va construyendo
* Menos elegante

```python
import csv

campo = "nombre" # se asigna una de los que hay
filas_ordenadas = []

with open("../recursos/datos.csv","r",encoding="utf-8") as archivo:
    lector = csv.DictReader(archivo)

    if campo in lector.fieldnames:
        # PRINCIPIO DEL ALGORITMO DE ORDENACIÓN
        for fila in lector:
            if len(filas_ordenadas) == 0:
                filas_ordenadas.append(fila)
            else:
                indice = 0
                insertado = False
                for i in filas_ordenadas:
                    if fila[campo] > i[campo]:
                        filas_ordenadas.insert(indice,fila)
                        insertado = True
                        break
                    else:
                        indice += 1
                if not insertado:
                    filas_ordenadas.insert(len(filas_ordenadas),fila)
                # FIN DEL ALGORITMO DE ORDENACIÓN
    else:
        print("Ese campo no existe")

# ORDENA DE MAYOR A MENOR. LE DAMOS LA VUELTA SI ASÍ SE REQUIERE
filas_ordenadas.reverse()

with open("../recursos/datos_ordenados.csv","w",encoding="utf-8") as archivo:
    writer = csv.DictWriter(archivo, fieldnames=["dni","nombre","edad","ciudad","telefono"])
    writer.writeheader()
    writer.writerows(filas_ordenadas)
```
